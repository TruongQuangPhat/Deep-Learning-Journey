# Why ML Strategy Matters and Orthogonalization

## 1. Core Idea

Machine Learning Strategy is the discipline of deciding **what to do next** when an ML system is not good enough.

In Course 1, we learned how neural networks work internally: forward propagation, backward propagation, cost functions, gradient descent, vectorization, activations, and deep network implementation.

In Course 2, we learned how to improve and debug deep neural networks using techniques such as regularization, dropout, initialization, Adam, Batch Normalization, learning rate decay, and hyperparameter tuning.

Course 3 shifts the focus from implementation-level questions to project-level decision-making:

> Given the current evidence, what is the highest-impact next step for improving the ML system?

This is the essence of ML Strategy. It helps us avoid randomly trying techniques and instead make decisions based on diagnosis, metrics, data distribution, error analysis, and project objectives.

A strong ML engineer does not simply ask:

> “Which technique should I try?”

They ask:

> “What is the current bottleneck, and which action is most likely to remove it?”

## 2. Why ML Strategy Is Necessary

In real ML projects, there are many possible actions when a model performs poorly:

* collect more data
* clean labels
* change the model architecture
* train longer
* tune hyperparameters
* add regularization
* reduce regularization
* use dropout
* use data augmentation
* change the optimizer
* change the learning rate
* redesign the dev/test set
* change the evaluation metric
* handle data mismatch
* use transfer learning
* switch from a pipeline system to end-to-end learning, or vice versa

The problem is that not all actions are useful in every situation.

If the model has high bias, adding stronger dropout may make the model even worse. If the model has high variance, simply increasing model size may worsen overfitting. If the dev set does not represent production, then improving dev performance may not improve the real system. If the metric is wrong, then the team may optimize a number that does not reflect the product goal.

Therefore, ML Strategy is important because it turns ML development from random experimentation into **evidence-based decision-making**.

Instead of saying:

> “Let’s try Adam, dropout, more layers, or more data and see what happens.”

A strategy-driven engineer says:

> “Let’s inspect the errors, compare training/dev/test performance, check whether the metric is aligned with the goal, and then decide the next action.”

## 3. The Main Project-Level Question

The central question of Course 3 is:

> What should I do next?

This question sounds simple, but it is difficult in real ML projects because model failure can come from many sources:

* the model is too simple
* the model is too complex
* optimization is not working
* the training data is insufficient
* the training data has wrong labels
* the dev/test sets are poorly designed
* the metric is misleading
* the training distribution differs from the production distribution
* the model overfits the dev set
* the system has latency, memory, fairness, or safety constraints
* the problem is close to human-level performance and further improvement is hard

ML Strategy gives us a structured way to diagnose these situations.

The goal is not to memorize a list of techniques. The goal is to choose the right technique for the right reason.

## 5. Basic Thinking Framework: From Metric to Action

A useful ML Strategy workflow can be summarized as:

1. define the right metric
2. design the right train/dev/test split
3. measure errors at multiple levels
4. perform manual error analysis
5. choose the action with the highest expected impact

This workflow is important because model improvement should not start from techniques. It should start from diagnosis.

### Step 1: Define the Right Metric

Before improving a model, we must ask:

> What are we optimizing?

Possible metrics include:

* accuracy
* precision
* recall
* F1-score
* AUC
* latency
* memory usage
* inference cost
* fairness
* user satisfaction
* revenue
* safety
* false positive rate
* false negative rate

If the metric is wrong, the entire training process may go in the wrong direction.

For example, in a medical diagnosis system, accuracy may be misleading if the positive class is rare. A model can achieve high accuracy by predicting most patients as healthy, but this may be unacceptable if it misses real disease cases. In this case, recall or false negative rate may be more important.

In a real-time speech recognition system, word error rate may be the optimizing metric, while latency must stay below a fixed threshold. In that case, latency is not necessarily the main quantity to optimize, but it is a constraint that every acceptable model must satisfy.

This leads to an important distinction:

* an **optimizing metric** is the metric we try to make as good as possible
* a **satisficing metric** is a metric that only needs to meet an acceptable threshold

For example:

| Metric Type        | Example                         |
| ------------------ | ------------------------------- |
| Optimizing metric  | Minimize word error rate        |
| Satisficing metric | Latency must be below 200 ms    |
| Satisficing metric | Model size must be below 100 MB |

This prevents confusion when comparing models.

Without a clear metric strategy, the team may argue endlessly because one model has better accuracy, another has better latency, another has better recall, and another has lower memory usage.

### Step 2: Design Train/Dev/Test Sets Correctly

After choosing the metric, we need to design the data splits correctly.

The training set is used to learn parameters. The dev set is used to choose models, tune hyperparameters, and guide project decisions. The test set is used for final unbiased evaluation.

The dev set is especially important because it acts as the team’s compass.

If the dev set points in the wrong direction, the entire team will optimize the wrong objective.

For example, suppose we are building a cat detector for mobile users. If the dev set contains clean internet images but production contains blurry phone images, then high dev accuracy may not mean the model works well in reality.

A good dev set should represent the distribution that we actually care about improving.

Usually, the dev and test sets should come from the same distribution, especially when they are meant to represent production. The training set may sometimes come from a broader or different distribution, especially when target-distribution data is limited. However, if the training distribution differs from the dev/test distribution, then we must explicitly diagnose data mismatch.

The key idea is:

> The dev set defines what the team is trying to improve.

If the dev set is wrong, the project direction is wrong.

### Step 3: Measure Error at Multiple Levels

We should not rely only on one accuracy number. To diagnose an ML system, we need to compare errors across multiple references:

* human-level error
* training error
* train-dev error, if needed
* dev error
* test error
* error by data subgroup
* error by error category
* error by distribution

Consider the following example:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Training error    |    3% |
| Train-dev error   |    4% |
| Dev error         |   12% |
| Test error        |   13% |

This table gives much more information than a single dev error.

Human-level error is 1%, while training error is 3%. This suggests that there is about 2% avoidable bias. The model is still worse than the human-level baseline on the training set.

Training error is 3%, while train-dev error is 4%. This gap is small, so ordinary variance is not the main problem.

Train-dev error is 4%, while dev error is 12%. This large gap suggests data mismatch. The model performs well on data from the training distribution but poorly on the dev distribution.

Dev error is 12%, while test error is 13%. This suggests that dev and test are reasonably consistent.

The recommended next action is not simply to add dropout or tune the learning rate. The priority should be to investigate and reduce the mismatch between the training distribution and the dev/test distribution.

This example shows why ML Strategy matters: the right action depends on the pattern of errors, not just on the fact that the model performs poorly.

### Step 4: Perform Manual Error Analysis

After measuring errors, we need to inspect the mistakes manually.

Error analysis means opening misclassified examples and grouping them into meaningful categories.

For example, in a cat classifier, we may find:

| Error Category           | Share of Errors |
| ------------------------ | --------------: |
| Blurry images            |             35% |
| Occluded cats            |             20% |
| Dogs that look like cats |             15% |
| Incorrect labels         |             10% |
| Low-light images         |              8% |
| Other cases              |             12% |

This tells us where improvement may come from.

If blurry images account for 35% of the errors, then collecting more blurry images or applying blur-based data augmentation may produce a large improvement.

If incorrect labels account for only 1% of the errors, then fixing labels may not be the highest-priority action. But if incorrect labels account for 10% or 20% of errors, especially in the dev/test sets, then label correction may become important.

Manual error analysis is important because aggregate metrics hide the structure of failure.

A model does not simply “have 12% error.” It makes specific kinds of mistakes. ML Strategy asks us to identify which mistakes are common, costly, and fixable.

### Step 5: Choose the Highest-Impact Action

After diagnosis, we choose the action with the highest expected impact.

Different diagnoses imply different actions:

| Diagnosis              | Priority Actions                                                                                          |
| ---------------------- | --------------------------------------------------------------------------------------------------------- |
| High bias              | Use a larger model, train longer, improve architecture, reduce regularization, improve optimization       |
| High variance          | Add more data, use regularization, dropout, data augmentation, early stopping                             |
| Data mismatch          | Collect data similar to dev/test, use domain adaptation, build a train-dev set, analyze distribution gaps |
| Wrong metric           | Redesign the metric or dev set                                                                            |
| Serious label noise    | Fix labels in dev/test first, then consider fixing training labels                                        |
| Poor optimization      | Tune learning rate, use Adam, use BatchNorm, check initialization                                         |
| Dev set too small      | Increase dev/test size or use cross-validation when data is small                                         |
| Overfitting to dev set | Create a new dev set, reduce repeated tuning on the same dev set                                          |

This table should not be used mechanically, but it captures the main principle:

> Do not choose an action before diagnosing the bottleneck.

For example, if the main issue is high bias, adding more data may not help much because the model is not even fitting the current training data well. If the main issue is high variance, making the model larger may worsen overfitting. If the main issue is data mismatch, tuning hyperparameters may give only small gains.

Good strategy is about matching the action to the diagnosis.

## 6. Orthogonalization

Orthogonalization is one of the central strategic principles in Course 3.

In mathematics, two directions are orthogonal if they are independent. In ML Strategy, orthogonalization means designing the development workflow so that each “knob” or decision mainly controls one type of problem.

The intuition is simple:

> When the system fails, we want to know which part to adjust.

A well-designed car has separate controls: the steering wheel controls direction, the accelerator controls speed, and the brake slows the car down. If pressing the brake also turned the steering wheel, driving would be extremely difficult.

An ML project has a similar issue. We want different tools to control different problems as clearly as possible.

For example:

| Problem                                                    | Main Knob                                                                        |
| ---------------------------------------------------------- | -------------------------------------------------------------------------------- |
| Training error is high                                     | Improve model capacity, architecture, optimization, or training time             |
| Dev error is much higher than training error               | Improve generalization using more data, regularization, dropout, or augmentation |
| Train-dev error is low but dev error is high               | Investigate data mismatch                                                        |
| Dev performance is good but test performance is poor       | Recheck dev/test split and possible dev overfitting                              |
| Offline evaluation is good but production behavior is poor | Check production distribution, preprocessing, monitoring, latency, and drift     |
| Metric score is good but product behavior is bad           | Redesign the metric or evaluation protocol                                       |

Orthogonalization does not mean that every technique affects only one thing. In deep learning, most techniques interact. For example, data augmentation can reduce variance and also reduce data mismatch. Batch Normalization can improve optimization and may have a mild regularization effect. Increasing model size can reduce bias but may increase variance.

Therefore, orthogonalization is not a strict mathematical law in ML project management. It is a strategic design principle:

> Try to make each action have a clear primary purpose.

This makes debugging easier and prevents the team from changing many things at once without knowing what actually caused improvement.

## 7. Orthogonalization as a Debugging Framework

Orthogonalization gives us a structured way to debug ML systems.

Instead of asking vaguely:

> Why is the model bad?

We ask a sequence of more precise questions:

1. Does the model fit the training set well?
2. Does the model generalize from training set to dev set?
3. Does the dev set reflect the test or production distribution?
4. Does the metric reflect the real product objective?
5. Does the deployed system behave like the offline evaluation?

Each question corresponds to a different part of the ML workflow.

### Fitting the Training Set

If training error is high, the first issue is that the model has not learned the training set well enough.

Possible causes include:

* the model is too small
* the architecture is not suitable
* the learning rate is poor
* the optimizer is not effective
* regularization is too strong
* the model has not been trained long enough
* there may be an implementation bug

In this case, the main actions are:

* use a larger model
* train longer
* improve the architecture
* tune the learning rate
* use Adam or another suitable optimizer
* use Batch Normalization
* reduce regularization
* check the training loop
* run a small-subset overfitting test

This is mainly a bias or optimization problem.

### Generalizing to the Dev Set

If training error is low but dev error is high, the model has learned the training set but does not generalize well.

If train and dev come from the same distribution, this is usually a variance problem.

Possible actions include:

* collect more data
* use L2 regularization
* use dropout
* apply data augmentation
* use early stopping
* reduce model complexity if necessary

This is where Course 2 techniques such as regularization and dropout become useful.

However, if train and dev do not come from the same distribution, dev error may be high because of data mismatch rather than ordinary variance. In that case, a train-dev set is needed.

### Diagnosing Data Mismatch

A train-dev set is a set drawn from the same distribution as the training set, but not used for training.

It helps separate variance from data mismatch.

Consider this example:

| Set           | Error |
| ------------- | ----: |
| Training set  |    1% |
| Train-dev set |    2% |
| Dev set       |   15% |

The gap from training error to train-dev error is small, so ordinary variance is not the main issue. The gap from train-dev error to dev error is large, so the main issue is data mismatch.

The correct strategic action is not simply to add dropout or tune the learning rate. Better actions include:

* collect data closer to the dev/test distribution
* use data augmentation that mimics the target distribution
* fine-tune on target-distribution data
* inspect preprocessing differences
* check whether the dev/test sets truly represent production

### Checking Dev/Test Consistency

If dev error is low but test error is high, the problem may be in the evaluation setup.

Possible causes include:

* dev and test come from different distributions
* the dev set is too small
* the team has overfit to the dev set through repeated tuning
* the test set is more representative of production than the dev set

In this case, the main action is not to modify the optimizer or architecture immediately. The first priority is to re-examine the data split and evaluation protocol.

A good ML workflow keeps the test set as an unbiased final estimate and avoids repeatedly using it for model selection.

### Checking Production Behavior

A model can perform well offline but fail in production.

This may happen because:

* production data differs from the test set
* preprocessing differs between training and serving
* there is data drift
* latency is too high
* the model is too expensive to serve
* user behavior changes over time
* the system violates a safety, fairness, or reliability constraint

In this situation, the problem is not necessarily the neural network itself. The issue may be the production environment.

Orthogonalization tells us to separate offline model quality from production system quality.

## 8. Orthogonalization and Metrics

Metric design is one of the most important forms of orthogonalization.

If the metric is wrong, then every model improvement based on that metric may push the project in the wrong direction.

For example, in fraud detection, accuracy may be misleading because fraud is rare. A model that predicts “not fraud” most of the time can achieve high accuracy but be useless. In this case, recall, precision, F1-score, PR-AUC, or cost-sensitive metrics may be more meaningful.

Similarly, in a real-time application, the best model is not always the most accurate model. A model with slightly lower accuracy but much lower latency may be better for production.

This is why Course 3 emphasizes the distinction between optimizing and satisficing metrics.

For example:

| Role               | Metric                                        |
| ------------------ | --------------------------------------------- |
| Optimizing metric  | Maximize F1-score                             |
| Satisficing metric | Latency must be below 200 ms                  |
| Satisficing metric | Memory must be below 100 MB                   |
| Satisficing metric | Cost per request must be below a fixed budget |

This structure prevents the team from comparing models emotionally or inconsistently.

Instead of saying:

> Model A is more accurate, but Model B is faster, so which one is better?

We define the decision rule first:

> Among models that satisfy latency, memory, and cost constraints, choose the one with the best F1-score.

That is orthogonalization applied to metric design.

## 9. Example: Cat Classifier

Suppose we are building a cat classifier for a mobile app.

The training set contains clean images from the internet. The dev and test sets contain photos taken by real users on mobile phones. These images may be blurry, dark, cropped, or taken from unusual angles.

Assume the model has the following errors:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    2% |
| Training error    |    1% |
| Dev error         |   15% |

A beginner may conclude that this is high variance because training error is low and dev error is high. They may immediately add dropout or regularization.

However, a strategist asks:

> Are the training and dev sets from the same distribution?

In this case, they are not. The training set contains clean internet images, while the dev set contains real mobile photos.

A better diagnostic step is to create a train-dev set from the same distribution as the training set.

Suppose the result is:

| Set           | Error |
| ------------- | ----: |
| Training set  |    1% |
| Train-dev set |    2% |
| Dev set       |   15% |

Now the diagnosis becomes clearer.

The gap from training error to train-dev error is small, so ordinary variance is not the main issue. The gap from train-dev error to dev error is large, so data mismatch is likely the main problem.

The better next actions include:

* collect more mobile phone images
* augment internet images to simulate blur, darkness, cropping, and rotation
* fine-tune the model on target-distribution data
* inspect preprocessing differences between training and production
* ensure that dev/test sets represent the real production environment

This example shows how ML Strategy and orthogonalization work together. We do not simply see high dev error and apply regularization. We identify which assumption is failing, then choose the right knob.

## 10. Example: Spam Detection

Suppose we build a spam classifier.

Model A has 99% accuracy but misses many dangerous spam emails. Model B has 97% accuracy but catches many more dangerous spam emails.

If accuracy is the only metric, Model A looks better.

But if the product goal is to protect users from harmful spam, Model B may be better because recall on spam is more important.

This is not primarily an architecture problem. It is a metric design problem.

The lesson is:

> A model is only “good” relative to the metric and distribution we care about.

A high score on the wrong metric can be worse than a lower score on the right metric.

## 11. Common Mistakes

### Mistake 1: Tuning Hyperparameters Too Early

Many beginners see high dev error and immediately tune:

* learning rate
* batch size
* number of layers
* hidden units
* dropout rate
* regularization strength
* optimizer

Hyperparameter tuning is useful, but it is not always the right first action.

If the dev error is high because the metric is wrong, the dev set is not representative, or the training and dev distributions are mismatched, then tuning hyperparameters will not solve the root cause.

### Mistake 2: Failing to Distinguish Bias, Variance, and Data Mismatch

Another common mistake is using only training error and dev error to make a quick conclusion.

If train and dev come from the same distribution, then the gap between training error and dev error can often be interpreted as variance.

But if train and dev come from different distributions, dev error may be high because of mismatch rather than ordinary overfitting.

This is why Course 3 introduces the train-dev set.

A train-dev set is drawn from the same distribution as the training set, but it is not used for training. It helps separate variance from data mismatch.

### Mistake 3: Using a Dev Set That Does Not Represent Production

A dev set that does not represent production can make the whole team optimize in the wrong direction.

For example:

* Dev set: clean internet images
* Production data: night-time security camera images

A model can perform well on the dev set and still fail in production.

In this case, the core problem may not be the model architecture, the optimizer, or the number of layers. The core problem is that the dev set gives the wrong signal.

The dev set should represent the distribution that the team actually wants to improve.

### Mistake 4: Treating Too Many Metrics as Equal Priorities

Consider the following model comparison:

| Model | Accuracy | Latency | Memory | Recall | Precision |
| ----- | -------: | ------: | -----: | -----: | --------: |
| A     |      95% |  500 ms | 300 MB |    92% |       94% |
| B     |      93% |  100 ms |  80 MB |    90% |       91% |
| C     |      94% |  200 ms | 120 MB |    95% |       88% |

Without a clear distinction between optimizing and satisficing metrics, it is difficult to choose among these models.

One model has the best accuracy, another has the best latency and memory, and another has the best recall.

A better strategy is to define constraints first, such as latency below 200 ms and memory below 150 MB, then optimize the most important performance metric among models that satisfy those constraints.

### Mistake 5: Changing Too Many Things at Once

Orthogonalization also teaches us not to change too many things in the same experiment.

If we change architecture, learning rate, batch size, dropout, data augmentation, and preprocessing at the same time, then even if the model improves, we do not know which change caused the improvement.

This makes the project harder to debug and harder to reproduce.

In research exploration, changing many things at once may sometimes help discover a promising direction quickly. But in serious debugging and production-oriented ML, we should keep experiments controlled enough to understand cause and effect.

### Mistake 6: Fixing the Model When the Problem Is the Data or Metric

Many ML failures are not caused by the model architecture.

Sometimes the metric is wrong. Sometimes the dev set is not representative. Sometimes the train and dev distributions are mismatched. Sometimes labels are inconsistent. Sometimes the offline test set does not match production.

A common mistake is to keep modifying the neural network when the actual bottleneck is outside the neural network.

Orthogonalization helps avoid this mistake by asking:

> Which part of the system is actually failing?

## 12. Practical Project Mindset

A good ML engineer should think in terms of assumptions and bottlenecks.

An ML system relies on several assumptions:

1. the metric reflects the real goal
2. the dev set represents what the team wants to improve
3. the test set provides an unbiased estimate of final performance
4. the training set contains enough useful signal
5. the model has enough capacity
6. optimization can find good parameters
7. the model generalizes beyond the training set
8. the production environment matches the offline evaluation closely enough

When the model fails, we should ask:

> Which assumption is currently broken?

This question is more useful than randomly asking:

> Should I add dropout?
> Should I tune the learning rate?
> Should I use a bigger model?
> Should I collect more data?

The right action depends on the broken assumption.

If the metric is broken, fix the metric.
If the dev set is broken, fix the dev set.
If training error is high, fix capacity or optimization.
If dev error is high relative to training error, fix variance or mismatch.
If production fails despite good offline evaluation, fix the production data pipeline or monitoring setup.

This is the practical value of orthogonalization.

## 13. Essence

Machine Learning Strategy matters because real ML development is not mainly about knowing many techniques. It is about knowing **which problem to solve first**.

Course 1 gives us the mathematical and implementation foundation of neural networks.

Course 2 gives us tools for optimization, regularization, initialization, and hyperparameter tuning.

Course 3 teaches us how to use these tools wisely inside a real project.

Orthogonalization is one of the core ideas that makes this possible.

The essence of orthogonalization is:

> Design the ML workflow so that each type of problem has a relatively clear knob to adjust.

If training error is high, focus on capacity and optimization.
If dev error is much higher than training error, focus on generalization.
If train-dev error is low but dev error is high, focus on data mismatch.
If dev performance is good but test performance is poor, inspect the evaluation setup.
If metric performance is good but product behavior is poor, redesign the metric.
If offline evaluation is good but production fails, inspect the serving pipeline and production distribution.

The broader essence of ML Strategy is:

> Do not improve blindly. Diagnose first, then act.

A strong ML engineer asks:

> What is the metric?
> Is the dev set representative?
> What does the error pattern say?
> Is the issue bias, variance, data mismatch, label noise, optimization, metric design, or production mismatch?
> Which action has the highest expected impact?

ML Strategy and Orthogonalization turn model development from random trial-and-error into a systematic engineering process.
