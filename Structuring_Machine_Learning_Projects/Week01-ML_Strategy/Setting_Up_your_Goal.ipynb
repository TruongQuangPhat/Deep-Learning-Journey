# Single Number Evaluation Metric

## 1. Core Idea

A **Single Number Evaluation Metric** is a single main metric used to compare models and decide which experiment is better.

In Machine Learning Strategy, this idea is important because model development usually involves many possible experiments: changing the model architecture, tuning hyperparameters, adding regularization, collecting more data, modifying preprocessing, or changing the training procedure. Without a clear metric, it becomes difficult to know whether an experiment actually improves the system.

The key question is:

> How do we decide which model is better?

A single number metric gives the team a clear and consistent answer.

It does not mean that we ignore all other metrics. It means that when we need to make a model selection decision, we should have one primary metric that guides the choice.

## 2. Why a Single Number Metric Matters

In real projects, models often have trade-offs across different evaluation criteria.

For example, suppose we are comparing three classifiers:

| Model | Precision | Recall |
|---|---:|---:|
| A | 95% | 90% |
| B | 98% | 85% |
| C | 92% | 95% |

It is not immediately clear which model is best.

Model B has the highest precision, meaning that when it predicts positive, it is usually correct. However, it has lower recall, meaning that it misses more true positive examples. Model C has the highest recall but lower precision. Model A is more balanced.

If the team does not define a single main metric, different people may prefer different models for different reasons.

One person may choose Model B because precision is highest. Another may choose Model C because recall is highest. Another may prefer Model A because it looks balanced.

This creates confusion and slows down decision-making.

A single number metric solves this by giving the team one agreed-upon criterion for comparison.

## 3. Example: Using F1-score

When both precision and recall matter, a common single number metric is the **F1-score**.

The F1-score combines precision and recall into one value:

$$
F1 = \frac{2 \cdot Precision \cdot Recall}{Precision + Recall}
$$

Equivalently, it can be written as the harmonic mean:

$$
F1 = \frac{2}{\frac{1}{Precision} + \frac{1}{Recall}}
$$

The F1-score is high only when both precision and recall are high. If either precision or recall is low, the F1-score will also be limited.

For example:

| Model | Precision | Recall | F1-score |
|---|---:|---:|---:|
| A | 95% | 90% | 92.4% |
| B | 98% | 85% | 91.0% |
| C | 92% | 95% | 93.5% |

If F1-score is the main evaluation metric, then Model C is the best model among the three.

The important point is not that F1-score is always the correct metric. The important point is that the team should define one main number that determines model comparison.

## 4. Multiple Metrics for Analysis, One Metric for Decision

A single number evaluation metric does not mean that we stop looking at other information.

In practice, we may still inspect:

- accuracy
- precision
- recall
- F1-score
- confusion matrix
- per-class error
- subgroup error
- latency
- memory usage
- inference cost
- qualitative error examples

These metrics are useful for understanding the model. However, when choosing between models, we should avoid having too many equally important criteria without a decision rule.

A useful distinction is:

> Multiple metrics can be used for analysis, but one main metric should guide model selection.

For example, a team may use F1-score as the main metric for choosing models, but still inspect the confusion matrix to understand which classes are being confused.

The single number metric provides clarity. The additional metrics provide diagnosis.

## 5. Relationship to ML Strategy

Single Number Evaluation Metric is directly connected to ML Strategy.

ML Strategy is about choosing the highest-impact next action based on evidence. But to know whether an action helps, we need a clear way to measure improvement.

If the evaluation metric is unclear, then experiment results become ambiguous.

For example, suppose an experiment improves recall but reduces precision. Is that better or worse? Without a defined metric, the answer depends on personal preference. With a single number metric such as F1-score, the decision becomes more systematic.

This is why metric design should happen before extensive model tuning.

If the metric is wrong, the entire project may optimize in the wrong direction. If the metric is clear and aligned with the goal, experiments become easier to compare.

In this sense, the single number metric acts as a quantitative compass for the project.

## 6. The Metric Must Match the Real Objective

A single number metric is useful only if it reflects the real goal of the system.

For example, accuracy may be a poor metric for imbalanced classification.

Suppose we are building a fraud detection system where only 1% of transactions are fraudulent. A model that always predicts “not fraud” can achieve 99% accuracy, but it detects no fraud. In this case, accuracy is misleading.

Better metrics may include:

- precision
- recall
- F1-score
- PR-AUC
- cost-sensitive metrics
- false negative rate
- false positive rate

Similarly, in a medical diagnosis system, false negatives may be much more dangerous than false positives. A metric that treats all errors equally may not reflect the real cost of mistakes.

Therefore, the first question should not be:

> Which metric is common?

The better question is:

> Which metric reflects the real objective and cost structure of this project?

## 7. Connection to Orthogonalization

Single Number Evaluation Metric also connects to orthogonalization.

Orthogonalization means designing the ML workflow so that each part of the system has a clear role. The metric has one important role:

> It tells us what we are optimizing.

If the metric is unclear, then every later decision becomes unclear. Model selection, hyperparameter tuning, error analysis, and project prioritization all depend on what the team considers “better.”

For example, if a model improves accuracy but worsens latency, we need a clear decision rule. This leads naturally to the next idea in Course 3: separating **optimizing metrics** and **satisficing metrics**.

For now, the key point is simple:

> A single number metric gives the project one clear target for model comparison.

## 8. Common Mistakes

A common mistake is choosing a metric because it is easy to compute rather than because it reflects the real goal. Accuracy is often easy to understand, but it can be misleading in imbalanced datasets or high-risk applications.

Another mistake is using too many metrics as equal priorities. If accuracy, precision, recall, latency, memory, and cost are all treated equally, model comparison becomes confusing. A team should define which metric drives model selection and which metrics are used for analysis or constraints.

A third mistake is forgetting that the metric and the dev set must work together. Even a good metric can be misleading if the dev set does not represent the distribution that matters. For example, a model may achieve high F1-score on clean dev images but fail on blurry production images.

A fourth mistake is optimizing the metric blindly without doing error analysis. A single number tells us which model is better according to the metric, but it does not fully explain why the model fails. Error analysis is still necessary for deciding what to improve next.

## 9. Essence

The essence of Single Number Evaluation Metric is:

> To improve a machine learning system systematically, we need one clear number that tells us whether one model is better than another.

This metric helps the team compare experiments, reduce subjective debate, automate model selection, and keep development focused.

However, the metric must be chosen carefully. A single number metric is powerful only when it reflects the real objective of the project.

A good ML engineer does not only ask:

> What is the model accuracy?

They ask:

> Is this the right metric for the problem?
> Does this metric reflect the product goal?
> Does the dev set represent the data we care about?
> Are we using other metrics for analysis without confusing the model selection process?

Single Number Evaluation Metric is therefore not just a reporting tool. It is a strategic tool for making ML project decisions clearer and more consistent.

# Satisficing and Optimizing Metrics

## 1. Core Idea

When a machine learning project has multiple important evaluation criteria, we should not treat all metrics as equal priorities. Instead, Course 3 introduces two useful concepts:

- **Optimizing metric**
- **Satisficing metric**

An **optimizing metric** is the metric we want to make as good as possible.

A **satisficing metric** is a metric that only needs to reach an acceptable threshold.

The basic decision rule is:

> First, filter out models that do not satisfy the required constraints. Then, among the remaining models, choose the model with the best optimizing metric.

This idea helps make model selection clearer when a project has multiple requirements such as accuracy, latency, memory, cost, or safety.

## 2. Why This Idea Is Needed

In the previous topic, we learned about the **Single Number Evaluation Metric**. The purpose of a single number metric is to make model comparison simpler and more consistent.

However, real ML systems usually do not care about only one thing.

For example, a speech recognition system may need:

- high accuracy
- low latency
- reasonable memory usage
- low inference cost

A mobile computer vision model may need:

- high F1-score
- small model size
- fast inference
- low battery consumption

If the team tries to optimize all metrics equally, model selection becomes confusing.

For example:

| Model | Accuracy | Latency |
|---|---:|---:|
| A | 96% | 500 ms |
| B | 94% | 120 ms |
| C | 93% | 80 ms |

If we only care about accuracy, Model A is best. But if this is a real-time application, 500 ms may be too slow. In that case, Model A may not be acceptable even though it has the highest accuracy.

This is why we separate metrics into two roles:

| Metric Type | Meaning |
|---|---|
| Optimizing metric | The metric we want to optimize as much as possible |
| Satisficing metric | A metric that only needs to pass a minimum acceptable threshold |

## 3. Example: Accuracy and Latency

Suppose we define:

- optimizing metric: maximize accuracy
- satisficing metric: latency must be below 200 ms

Now consider the models again:

| Model | Accuracy | Latency |
|---|---:|---:|
| A | 96% | 500 ms |
| B | 94% | 120 ms |
| C | 93% | 80 ms |

Model A is removed because its latency is above 200 ms.

Models B and C both satisfy the latency requirement.

Among the remaining models, Model B has higher accuracy than Model C.

Therefore, we choose Model B.

The decision is now clear:

> Among models with latency below 200 ms, choose the one with the highest accuracy.

This is the essence of combining satisficing and optimizing metrics.

## 4. Example with Multiple Constraints

Suppose we are building a classifier for a mobile app.

| Model | F1-score | Latency | Memory |
|---|---:|---:|---:|
| A | 92% | 300 ms | 80 MB |
| B | 90% | 120 ms | 70 MB |
| C | 89% | 90 ms | 150 MB |
| D | 88% | 80 ms | 60 MB |

Assume the product requirements are:

- latency must be below 150 ms
- memory must be below 100 MB
- among acceptable models, choose the model with the highest F1-score

Then:

| Model | Latency < 150 ms? | Memory < 100 MB? | Accepted? |
|---|---|---|---|
| A | No | Yes | No |
| B | Yes | Yes | Yes |
| C | Yes | No | No |
| D | Yes | Yes | Yes |

Only Model B and Model D satisfy both constraints.

Between them, Model B has the higher F1-score.

Therefore, Model B is selected.

In this example:

- F1-score is the optimizing metric
- latency is a satisficing metric
- memory is a satisficing metric

## 5. Relationship to Single Number Evaluation Metric

The idea of satisficing and optimizing metrics extends the previous idea of a single number evaluation metric.

A single number metric is useful because it gives the team one main criterion for model comparison. But when there are practical constraints, we may not want to combine everything into one complicated formula.

Instead, we can define a simple rule:

> Among models that satisfy the required constraints, choose the model with the best main metric.

For example:

> Among models with latency below 200 ms and memory below 100 MB, choose the model with the highest F1-score.

This keeps model selection simple while still respecting real-world requirements.

## 6. When to Use Satisficing Metrics

Satisficing metrics are useful when a requirement has a clear acceptable threshold.

Examples include:

- latency must be below 200 ms
- memory usage must be below 100 MB
- cost per request must be below a fixed budget
- model size must fit on a mobile device
- false negative rate must be below a maximum threshold
- safety violation rate must be below an acceptable level

These metrics may not always need to be optimized endlessly.

For example, if users cannot notice the difference between 80 ms and 120 ms latency, and both are below the acceptable threshold, then it may be better to choose the model with higher accuracy or F1-score.

The key idea is:

> A satisficing metric must be good enough, but it does not necessarily need to be the best possible.

## 7. When to Use an Optimizing Metric

The optimizing metric should represent the main goal of the ML system.

Examples include:

- maximize accuracy
- maximize F1-score
- maximize recall
- minimize word error rate
- minimize mean squared error
- maximize user satisfaction
- minimize ranking loss

This is the metric that the team tries to improve after all required constraints are satisfied.

For example, in a spam detection system:

- optimizing metric: maximize F1-score or spam recall
- satisficing metric: latency must be below 100 ms
- satisficing metric: false positive rate must stay below an acceptable threshold

The exact choice depends on the product goal and the cost of different types of mistakes.

## 8. Common Mistakes

A common mistake is trying to optimize too many metrics at the same time. If accuracy, recall, latency, memory, cost, and model size are all treated as equally important, the team may struggle to decide which model is best.

Another mistake is setting the satisficing threshold too loosely or too strictly. If the threshold is too loose, it does not meaningfully constrain the system. If it is too strict, it may eliminate many useful models unnecessarily.

A third mistake is choosing the wrong optimizing metric. For example, using accuracy as the optimizing metric in a highly imbalanced fraud detection problem may lead to a model that looks good numerically but fails to detect actual fraud.

A fourth mistake is thinking that satisficing metrics are unimportant. A satisficing metric is still critical. If a model fails a satisficing requirement, it should not be selected, even if its optimizing metric is very strong.

For example, a model with excellent F1-score may still be unusable if its latency is too high for a real-time application.

## 9. Essence

The essence of satisficing and optimizing metrics is:

> Not every metric should be optimized equally.

A good ML project should define:

- one main metric to optimize
- several constraints that must be satisfied

The practical rule is:

> First satisfy the constraints. Then optimize the main metric.

This makes model selection clearer, reduces subjective debate, and connects evaluation to real product requirements.

In ML Strategy, this idea is important because it prevents the team from comparing models emotionally or inconsistently. Instead of saying:

> Model A is more accurate, but Model B is faster, so which one should we choose?

We define a decision rule:

> Among models that are fast enough, small enough, and cheap enough, choose the one with the best main performance metric.

This is a simple but powerful way to structure model selection in real machine learning projects.

# Train/Dev/Test Distributions

## 1. Core Idea

The key idea of **train/dev/test distributions** is:

> The dev set and test set should come from the same distribution and should represent the data we actually care about in production.

The training set is used to learn parameters. The dev set is used to choose models and guide project decisions. The test set is used for final evaluation.

Therefore, the dev set acts as the project’s compass. If the dev set does not represent the real target distribution, the team may optimize the model in the wrong direction.

A model is not truly good just because it performs well on a convenient dataset. It is good only if it performs well on the distribution that matters.

## 2. Roles of Train, Dev, and Test Sets

| Dataset | Main Role |
|---|---|
| Train set | Learn model parameters |
| Dev set | Select models, tune hyperparameters, guide decisions |
| Test set | Estimate final generalization performance |

The dev set and test set should usually be drawn from the same distribution. This makes model selection reliable: the model that performs best on the dev set is more likely to also perform well on the test set.

If dev and test come from different distributions, dev performance may no longer predict test performance. In that case, the team may choose the wrong model.

## 3. Why Dev/Test Should Match Production

In real projects, the goal is not to perform well on an arbitrary benchmark. The goal is to perform well on the data the system will actually see.

For example, if a mobile cat detector will be used on blurry phone images, then the dev/test sets should contain blurry phone images, not only clean internet images.

If a Legal QA system is designed for real legal questions, then the dev/test sets should contain realistic user questions, require correct citations, and reflect the legal documents the system is expected to handle.

The principle is:

> Choose dev/test from the distribution you want to optimize for.

The training set may sometimes come from a broader or different distribution, especially when target-distribution data is limited. However, if train and dev/test differ, we must explicitly diagnose data mismatch.

## 4. Train Set Can Differ, but Be Careful

Ideally, train/dev/test all come from the production distribution. But this is not always possible.

For example:

- We may have many clean internet images.
- We may have only a small number of real mobile images.
- Production will be mobile images.

In this case, it may be reasonable to train on internet images plus some mobile images, while keeping dev/test focused on mobile images.

This can make dev error look worse, but that is useful because it reflects the real target problem.

The danger is that if train and dev/test come from different distributions, a large gap between training error and dev error may not mean ordinary overfitting. It may mean **data mismatch**.

## 5. The Train-Dev Set

A **train-dev set** is drawn from the same distribution as the training set, but it is not used for training.

Its purpose is to distinguish:

- high variance
- data mismatch

A typical setup is:

| Dataset | Distribution | Purpose |
|---|---|---|
| Train | Source/training distribution | Learn parameters |
| Train-dev | Same as train distribution | Diagnose variance |
| Dev | Target/production-like distribution | Select model |
| Test | Target/production-like distribution | Final evaluation |

The train-dev set is especially useful when the training data and dev/test data come from different sources.

## 6. Error Pattern Diagnosis

This is the most important practical part.

### Pattern 1: High Bias or Optimization Problem

| Set | Error |
|---|---:|
| Human-level | 1% |
| Training | 15% |
| Train-dev | 16% |
| Dev | 17% |

The training error is already high. The model has not fit the training set well.

Likely problem:

> high bias or poor optimization

Priority actions:

- use a larger model
- improve architecture
- train longer
- tune learning rate
- reduce excessive regularization
- check the training loop
- test whether the model can overfit a small subset

### Pattern 2: High Variance

| Set | Error |
|---|---:|
| Human-level | 1% |
| Training | 1% |
| Train-dev | 14% |
| Dev | 15% |

The model fits the training set well, but performs poorly even on train-dev data from the same distribution.

Likely problem:

> high variance

Priority actions:

- add more data
- use regularization
- use dropout
- apply data augmentation
- use early stopping
- reduce model complexity if needed

### Pattern 3: Data Mismatch

| Set | Error |
|---|---:|
| Human-level | 1% |
| Training | 1% |
| Train-dev | 2% |
| Dev | 15% |

The model generalizes well to train-dev data, but fails on the dev set.

Likely problem:

> data mismatch between train distribution and dev/test distribution

Priority actions:

- collect more target-like data
- augment training data to resemble dev/test data
- fine-tune on target-distribution data
- inspect preprocessing differences
- analyze how train and dev/test distributions differ

This is a very important Course 3 lesson:

> A large train-dev gap suggests variance.  
> A large train-dev-to-dev gap suggests data mismatch.

### Pattern 4: Dev/Test Mismatch or Dev Overfitting

| Set | Error |
|---|---:|
| Dev | 4% |
| Test | 12% |

The model performs well on dev but poorly on test.

Likely problem:

> dev/test mismatch, dev set too small, or overfitting to the dev set

Priority actions:

- check whether dev and test come from the same distribution
- increase dev/test size if too small
- avoid using the test set for model selection
- create a new dev set if the old one has been overused

## 7. Common Mistakes

A common mistake is using a random split when the data comes from multiple sources. If production data comes from mobile users but most of the dataset is clean internet data, a random split may create dev/test sets dominated by internet data. The model may look good offline but fail in production.

Another mistake is choosing dev/test data because it is easy to collect, not because it represents the target distribution.

A third mistake is concluding “high variance” too early. Training error low and dev error high is not enough. If train and dev come from different distributions, the real problem may be data mismatch.

A fourth mistake is using the test set repeatedly for model selection. Once the test set is used many times to guide decisions, it becomes similar to another dev set and no longer gives an unbiased final estimate.

## 8. Relationship to Metrics

A good metric is not enough if it is measured on the wrong distribution.

For example, F1-score may be a good metric, but if the dev/test sets contain clean images while production contains blurry images, then high F1-score is misleading.

The principle is:

> Right metric + wrong distribution = misleading evaluation.

Metric design and data distribution must work together.

## 9. Essence

The essence of train/dev/test distributions is:

> The dev and test sets should represent the distribution you actually care about.

The training set can sometimes come from a different or broader distribution, but then we need careful diagnosis.

The most important diagnostic tool is the train-dev set:

> Training error → train-dev error tells us about variance.  
> Train-dev error → dev error tells us about data mismatch.

Before asking:

> How can I reduce dev error?

A good ML engineer first asks:

> Is this dev set really the distribution I want the model to perform well on?

This is why train/dev/test distribution design is a strategic decision, not just a data preprocessing step.

# Size of the Dev and Test Sets

## 1. Core Idea

The size of the dev and test sets should not be chosen mechanically by a fixed percentage such as 60/20/20 or 70/15/15 in every project.

The key principle is:

> The dev and test sets should be large enough to give reliable evaluation, but they do not necessarily need to be a large percentage of the whole dataset.

In traditional machine learning with small datasets, it was common to allocate a relatively large fraction of data to dev and test sets. However, in modern deep learning, datasets can be very large. In that case, even 1% of the data may already contain thousands or millions of examples, which can be enough for reliable evaluation.

The question is not only:

> What percentage should dev/test be?

The better question is:

> Are the dev and test sets large enough and representative enough to support reliable decisions?

## 2. Traditional Splits vs Large-Scale Deep Learning

In smaller datasets, common splits such as 60/20/20 or 70/15/15 can make sense.

For example, if we have 1,000 examples:

| Train | Dev | Test |
|---:|---:|---:|
| 600 | 200 | 200 |

Here, using 20% for dev and 20% for test may be reasonable because the total dataset is small.

However, if we have 1,000,000 examples, using 20% for dev and 20% for test may be unnecessary.

For example:

| Train | Dev | Test |
|---:|---:|---:|
| 980,000 | 10,000 | 10,000 |

This is a 98/1/1 split, but the dev and test sets still contain 10,000 examples each. That may be enough for stable model comparison and final evaluation.

Therefore, for large datasets, the absolute number of dev/test examples matters more than the percentage.

## 3. How Large Should the Dev Set Be?

The dev set should be large enough to let us compare models reliably.

If the dev set is too small, the metric becomes noisy. A model may appear better simply because of random variation in the dev examples.

For example, if the dev set has only 100 examples, then one mistake changes the error rate by 1%. It becomes difficult to confidently compare models whose performance differs by a small amount.

A good dev set should be large enough to:

- compare models consistently
- tune hyperparameters with confidence
- support meaningful error analysis
- represent important data subgroups
- reduce random noise in evaluation

There is no universal number that works for all projects. The required size depends on the task, metric, number of classes, class imbalance, data noise, and how small the performance differences between models are.

The practical rule is:

> Make the dev set large enough that changes in the metric reflect real improvements, not random noise.

## 4. How Large Should the Test Set Be?

The test set is used to estimate final performance after model selection.

It should be large enough to give a trustworthy final estimate of generalization performance.

If the test set is too small, the final performance estimate may be unstable. For example, with only 100 test examples, one mistake changes accuracy by 1%. This is too noisy for many projects.

The test set does not always need to be larger than the dev set. In many projects, the dev set may even be larger because it is used repeatedly for model selection and error analysis, while the test set is used mainly for final evaluation.

The practical rule is:

> The test set should be large enough that the final reported performance is credible.

## 5. Dev/Test Size Should Depend on the Metric

The required dev/test size depends strongly on what we are measuring.

For a balanced classification problem, a few thousand examples may be enough to estimate accuracy reasonably well.

But for imbalanced classification, the total number of examples can be misleading.

For example, suppose fraud cases are only 1% of all transactions. If the dev set has 1,000 examples, it may contain only about 10 fraud cases. Measuring fraud recall on only 10 positive examples is extremely noisy.

In that case, the dev/test sets need either:

- more total examples
- enough examples from the rare but important class
- a carefully designed evaluation set focused on the critical class

The key idea is:

> Dev/test sets must be large enough not only overall, but also for the important subgroups and error types we care about.

## 6. Small Data Settings

When the dataset is small, allocating too much data to dev/test can leave too little data for training.

In such cases, we may need to use strategies such as:

- smaller dev/test sets with careful interpretation
- cross-validation
- k-fold validation
- collecting more data if possible

For very small datasets, evaluation will naturally be less stable. The important thing is to be aware of that uncertainty instead of over-trusting small dev/test results.

## 7. Relationship to Train/Dev/Test Distributions

Size alone is not enough.

A large dev set is still not useful if it comes from the wrong distribution. A small but well-targeted dev set may sometimes be more informative than a large dev set that does not represent production.

The dev and test sets should satisfy two conditions:

1. They should be large enough for reliable evaluation.
2. They should represent the distribution we actually care about.

This connects directly to the previous lesson on train/dev/test distributions.

A metric measured on the wrong distribution can still be misleading, even if the dev/test set is very large.

## 8. Essence

The essence of choosing dev/test set sizes is:

> Do not choose dev/test sizes by fixed percentages alone. Choose them based on whether they are large enough and representative enough to support reliable decisions.

For small datasets, traditional splits such as 60/20/20 or 70/15/15 may be reasonable.

For large datasets, very small percentages such as 98/1/1 or 99/0.5/0.5 may be enough, because the absolute number of dev/test examples is still large.

The dev set should be large enough to guide model selection reliably.

The test set should be large enough to provide a credible final estimate.

Most importantly:

> A useful dev/test set is not just a portion of the data. It is an evaluation instrument for making trustworthy ML project decisions.

# When to Change Dev/Test Sets and Metrics

## 1. Core Idea

Dev/test sets and evaluation metrics are not just reporting tools. They are the project’s decision-making system.

The dev set tells the team which model or experiment is better.  
The metric defines what “better” means.  
The test set gives the final estimate of performance.

Therefore, if the metric or dev/test set does not reflect the real project goal, the team may optimize in the wrong direction.

The core principle is:

> Change the metric or dev/test set when it helps measure the real goal better, not when it merely makes the current model look better.

## 2. When to Change the Metric

A metric should be changed when it no longer captures what the system is supposed to optimize.

One common case is when the metric improves, but the product does not improve. For example, a fraud detection model may have high accuracy because most transactions are normal, but still fail to detect actual fraud. In that case, accuracy is misleading.

Better metrics may include:

- fraud recall
- precision
- F1-score
- PR-AUC
- false negative rate
- cost-sensitive metrics

Another reason to change the metric is when different types of errors have different costs.

For example, in medical screening, a false negative can be much more dangerous than a false positive. A metric that treats all errors equally may not reflect the real risk.

A metric may also need to change when practical constraints become important, such as:

- latency
- memory
- inference cost
- safety
- fairness
- hallucination rate
- citation correctness

For example, a model with high F1-score may still be unusable if it is too slow for a real-time application.

The key idea is:

> If the metric rewards behavior that is not actually useful for the product, the metric should be redesigned.

## 3. When to Change Dev/Test Sets

A dev or test set should be changed when it no longer represents the distribution the project actually cares about.

For example, suppose a cat detector is developed for mobile users, but the dev/test sets contain mostly clean internet images. The model may perform well on dev/test but fail on blurry phone images in production. In this case, the dev/test sets should be rebuilt using more production-like mobile images.

Another reason to change dev/test sets is distribution shift over time.

Production data may change because:

- user behavior changes
- new devices or sensors are introduced
- fraud patterns evolve
- language usage changes
- the product expands to new user groups
- legal or domain knowledge changes

If the dev/test sets represent an old distribution, they may no longer guide the project correctly.

The dev set may also need to be refreshed if the team has overfit to it through repeated tuning. If dev performance improves but test or production performance does not, the dev set may no longer be a reliable guide.

Dev/test sets should also be corrected if they contain serious problems such as:

- wrong labels
- data leakage
- duplicated examples
- corrupted samples
- missing important subgroups
- too few examples for rare but important cases

The principle is:

> Dev/test sets should represent the real target distribution and provide reliable evaluation.

## 4. When Not to Change Them

Metric and dev/test set changes are powerful but dangerous if used carelessly.

We should not change them simply because the current model performs poorly.

For example, if the model fails on the dev set, that may mean the dev set is doing its job. It is revealing a real weakness. Changing the dev set just to make results look better is not valid ML strategy.

Do not change the metric or dev/test set when:

- the metric still reflects the real objective
- the dev/test sets still represent production
- the only reason is that the model score is low
- the goal is to make a report look better
- there is no evidence that the current evaluation is misleading

The correct reason to change evaluation is not:

> The model does not look good.

The correct reason is:

> The current evaluation does not measure the real goal correctly.

## 5. How to Change Them Properly

When changing metrics or dev/test sets, the process should be controlled.

First, clearly document the reason for the change.

For example:

> The old dev set contains mostly clean internet images, but production data consists of mobile phone images. Therefore, we create a new dev set from real mobile data.

Or:

> Accuracy does not reflect the cost of false negatives, so we switch to recall and F1-score.

Second, version the evaluation setup:

- metric v1, metric v2
- dev set v1, dev set v2
- test set v1, test set v2
- reason for the change
- date or project phase

Third, re-evaluate important baselines on the new setup.

This may include:

- the current production model
- the previous best model
- a simple baseline
- the new candidate model

This is important because results from the old evaluation setup may not be directly comparable with results from the new setup.

## 6. Example: Cat Detector

Suppose the original dev/test sets contain clean internet images.

The model achieves high accuracy on dev/test, but users complain that it fails on blurry, dark, real mobile photos.

This suggests that the dev/test sets are not representative of production.

A better evaluation setup would include:

- real mobile images
- blurry images
- low-light images
- unusual angles
- realistic user-uploaded photos

After changing the dev set, the reported accuracy may drop. However, this lower score is more useful because it reflects the real problem better.

This is a good change because it measures the actual target distribution more accurately.

## 7. Example: Legal QA/RAG

In a Legal QA/RAG system, an early metric might focus on general answer correctness or fluency.

However, in a real legal setting, a fluent answer is not enough. The answer must be grounded in the correct legal source.

Important evaluation criteria may include:

- answer correctness
- citation correctness
- retrieval of the correct article or clause
- hallucination rate
- legal grounding
- effective-date awareness

If the system gives fluent answers but cites the wrong law or hallucinates legal claims, then the original metric is insufficient.

The dev/test sets should also contain realistic legal questions, including ambiguous questions, multi-step questions, questions requiring citations, and cases where the system should refuse to answer without enough evidence.

This is a case where both the metric and dev/test set may need to change to reflect the real product goal.

## 8. Essence

The essence of this topic is:

> Metrics and dev/test sets should be stable enough to compare models fairly, but flexible enough to be corrected when they no longer measure the real objective.

A good ML engineer should not blindly keep an evaluation setup just because it was chosen at the beginning of the project.

But they also should not change it casually whenever a model performs badly.

The right mindset is:

> Keep the evaluation setup stable when it measures the right goal.  
> Change it when it clearly measures the wrong goal.

In ML Strategy, changing the metric or dev/test set is not a trick to improve reported numbers. It is a strategic correction to make sure the team is optimizing the right target.

# Weighted Loss and Class Weight

## 1. Core Idea

**Weighted loss** is a technique that makes some samples or classes contribute more strongly to the training loss.

The main idea is:

> If a class is rare, important, or costly to misclassify, we can make mistakes on that class more expensive.

For example, in binary classification:

```python
weights = {0: 1, 1: 2.8}
```

This means:

- class `0` has normal importance
- class `1` is treated as 2.8 times more important in the loss

So, when the model makes an error on class `1`, that error contributes more to the total loss and produces a stronger gradient during training.

## 2. Why Weighted Loss Is Useful

Weighted loss is often used when the dataset is imbalanced.

For example:

| Class | Count |
|---|---:|
| Class 0 | 28,000 |
| Class 1 | 10,000 |

If we train normally, the model may focus more on class `0` because class `0` appears more often. As a result, the model may achieve high overall accuracy while performing poorly on the minority class.

Weighted loss tells the model:

> Do not ignore the minority or important class. Errors on that class are more costly.

This is especially useful in problems such as:

- fraud detection
- medical diagnosis
- spam detection
- churn prediction
- rare event detection

## 3. How It Works Mathematically

For binary classification, the standard binary cross-entropy loss is:

$$
L(y, \hat{y}) = -[y\log(\hat{y}) + (1-y)\log(1-\hat{y})]
$$

where:

- $y \in \{0, 1\}$ is the true label
- $\hat{y}$ is the predicted probability of class `1`

With weighted loss, we multiply the loss by the weight of the true class:

$$
L_{\text{weighted}}(y, \hat{y}) = w_y \cdot L(y, \hat{y})
$$

If $y = 1$, then:

$$
L_{\text{weighted}} = -w_1 \log(\hat{y})
$$

If $y = 0$, then:

$$
L_{\text{weighted}} = -w_0 \log(1-\hat{y})
$$

So if:

$$
w_0 = 1
$$

and

$$
w_1 = 2.8
$$

then an error on class `1` is penalized 2.8 times more heavily than a comparable error on class `0`.

## 4. Simple Example

Suppose two samples are both badly misclassified.

For a class `0` sample, the true label is:

$$
y = 0
$$

and the model predicts:

$$
\hat{y} = 0.9
$$

The normal loss is:

$$
L = -\log(1 - 0.9) = -\log(0.1) \approx 2.302
$$

With weight $w_0 = 1$:

$$
L_{\text{weighted}} = 1 \cdot 2.302 = 2.302
$$

For a class `1` sample, the true label is:

$$
y = 1
$$

and the model predicts:

$$
\hat{y} = 0.1
$$

The normal loss is:

$$
L = -\log(0.1) \approx 2.302
$$

With weight $w_1 = 2.8$:

$$
L_{\text{weighted}} = 2.8 \cdot 2.302 \approx 6.446
$$

So the two mistakes are equally bad in terms of raw prediction error, but the class `1` mistake is treated as more important during training.

## 5. Effect on Training

Neural networks learn by minimizing loss through gradients.

When a sample has a higher loss weight, its gradient contribution becomes larger. This means the sample has more influence on parameter updates.

Intuitively:

> A class with a larger weight “pulls” the model more strongly during training.

Therefore, increasing the weight of class `1` usually makes the model more sensitive to class `1`.

This often leads to:

| Metric | Typical Effect |
|---|---|
| Recall of minority class | Increases |
| False negatives of minority class | Decreases |
| False positives | May increase |
| Precision | May decrease |
| Overall accuracy | May decrease |

This trade-off is normal. If class `1` is important, improving recall may be worth a small drop in accuracy or precision.

## 6. How to Choose Class Weights

A common heuristic is to set weights inversely proportional to class frequency.

If class `0` has 28,000 samples and class `1` has 10,000 samples, then class `0` is about 2.8 times more frequent:

$$
\frac{28000}{10000} = 2.8
$$

So one simple choice is:

```python
weights = {0: 1, 1: 2.8}
```

This means each class `1` sample has about the same total influence as 2.8 class `0` samples.

Another common formula is:

$$
w_c = \frac{N}{K \cdot N_c}
$$

where:

- $N$ is the total number of samples
- $K$ is the number of classes
- $N_c$ is the number of samples in class $c$

The exact scale of weights is less important than the relative ratio between classes.

## 7. Class Weight vs Oversampling

Class weighting and oversampling both try to solve class imbalance, but they do it differently.

| Method | Idea |
|---|---|
| Class weight | Increase the loss contribution of minority-class samples |
| Oversampling | Make minority-class samples appear more often during training |

Class weighting does not create new data. It only changes how strongly each sample affects the loss.

Oversampling changes the composition of the training batches by repeating or generating more minority-class samples.

A simple way to remember:

> Class weight makes minority samples louder.  
> Oversampling makes minority samples appear more often.

## 8. Weighted Loss vs Threshold Tuning

Weighted loss and threshold tuning are related but different.

**Weighted loss** affects training. It changes how the model learns its parameters.

**Threshold tuning** happens after training. It changes how predicted probabilities are converted into class labels.

For binary classification, the default rule is usually:

```text
if y_hat >= 0.5 → class 1
else → class 0
```

If we want higher recall for class `1`, we can lower the threshold:

```text
if y_hat >= 0.3 → class 1
else → class 0
```

The distinction is:

| Technique | When It Happens | Effect |
|---|---|---|
| Weighted loss | During training | Changes learned parameters |
| Threshold tuning | After training | Changes decision boundary |
| Class weight | Training-time correction | Makes important classes influence learning more |
| Threshold adjustment | Inference-time correction | Changes precision/recall trade-off |

In practice, both can be used together.

## 9. Limitations

Weighted loss is useful, but it does not solve every problem.

It may not help much if:

- the minority class has noisy labels
- the minority class has too few representative examples
- features do not contain enough signal
- train and dev/test distributions are different
- the metric is poorly chosen
- the model is too weak
- the model overfits rare minority samples

Weighted loss should be used together with proper evaluation.

After applying class weights, we should inspect:

- confusion matrix
- precision and recall per class
- F1-score
- PR-AUC
- false positives and false negatives
- error analysis
- dev/test distribution quality

## 10. Relationship to ML Strategy

Weighted loss connects directly to ML Strategy because it reflects the idea that not all errors have equal importance.

If false negatives are more costly than false positives, or if the minority class is more important, we can express that preference through:

- weighted loss
- class weights
- sample weights
- cost-sensitive metrics
- threshold tuning

However, we should separate three ideas clearly:

| Component | Role |
|---|---|
| Weighted loss | Controls how the model learns |
| Metric | Controls how the model is evaluated |
| Threshold | Controls how predictions become decisions |

Weighted loss is a training tool, not a replacement for good metrics or proper dev/test design.

## 11. Essence

The essence of weighted loss is:

> Make important or underrepresented examples count more during training.

With class weights such as:

```python
weights = {0: 1, 1: 2.8}
```

we are saying:

> A mistake on class `1` should be treated as 2.8 times more costly than a comparable mistake on class `0`.

Mathematically:

$$
L_{\text{weighted}} = w_y \cdot L(y, \hat{y})
$$

Practically, this makes class `1` produce stronger gradients and influence the model more during training.

The key takeaway is:

> Weighted loss does not create more data.  
> It makes mistakes on important classes more expensive.

# Why Human-Level Performance and Avoidable Bias?

## 1. Core Idea

**Human-level performance** is used as a reference point for understanding how good or bad a model’s error really is.

A training error number alone is not very meaningful without context.

For example:

| Error Type     | Value |
| -------------- | ----: |
| Training error |    5% |
| Dev error      |    7% |

At first, this may look good. But whether 5% training error is high or low depends on the task.

If humans achieve 1% error, then 5% training error is still quite high.
If humans also achieve around 5% error, then the model may already be close to the practical limit of the task.

The key idea is:

> Human-level performance gives context to training error.

It helps us answer:

> Is the model still far from what is realistically achievable, or is it already close to the best possible performance?

## 2. Human-Level Performance and Bayes Error

In machine learning, **Bayes error** is the lowest possible error that any model can theoretically achieve on a task.

Bayes error exists because some examples are inherently ambiguous, noisy, or impossible to classify perfectly.

For example:

* an image may be too blurry to classify correctly
* an audio clip may be too noisy to transcribe
* a medical image may be ambiguous even for experts
* a legal question may lack enough information for a definite answer
* labels may be inconsistent or incorrect

In practice, we usually do not know the true Bayes error.

Human-level performance is useful because it often acts as a rough proxy for Bayes error, especially in tasks where skilled humans perform well.

For example:

| Error Type            |   Value |
| --------------------- | ------: |
| Estimated Bayes error | unknown |
| Human-level error     |      1% |
| Training error        |      5% |
| Dev error             |      7% |

If human-level error is around 1%, then we can infer that the model still has significant room to improve.

The relationship can be understood as:

$$
\text{Bayes error} \leq \text{Human-level error} \leq \text{Training error}
$$

This is not always exact, but it is a useful practical approximation.

## 3. Avoidable Bias

Human-level performance helps us estimate **avoidable bias**.

Avoidable bias is the part of the model’s training error that is still higher than human-level performance.

A useful approximation is:

$$
\text{Avoidable bias} = \text{Training error} - \text{Human-level error}
$$

For example:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Training error    |   10% |
| Dev error         |   12% |

Here:

$$
\text{Avoidable bias} = 10% - 1% = 9%
$$

This means the model is still far from human-level performance on the training set. It has not learned the task well enough, even on the data it was trained on.

The likely problem is high bias or poor optimization.

Possible actions include:

* using a larger model
* improving the architecture
* training longer
* tuning the learning rate
* improving optimization
* reducing excessive regularization

The main intuition is:

> Avoidable bias measures how much the model still fails to fit the task compared with a strong reference point.

## 4. Avoidable Bias vs Ordinary Bias

In earlier courses, we often describe **bias** as underfitting.

A model has high bias when it is too simple or poorly optimized and therefore cannot fit the training set well.

However, Course 3 refines this idea by comparing training error with human-level performance.

| Concept                         | Meaning                                                                                |
| ------------------------------- | -------------------------------------------------------------------------------------- |
| Bias                            | The model’s tendency to underfit or simplify the task too much                         |
| High bias                       | Training error is high                                                                 |
| Avoidable bias                  | Training error is higher than human-level error                                        |
| Bayes error / irreducible error | The part of error that cannot be avoided because the task itself is noisy or ambiguous |

This distinction matters because training error alone can be misleading.

Consider two cases:

| Case | Human-level error | Training error |
| ---- | ----------------: | -------------: |
| A    |                1% |             5% |
| B    |              4.5% |             5% |

In both cases, training error is 5%.

But in Case A, the model is still far from human-level performance. The avoidable bias is:

$$
5% - 1% = 4%
$$

In Case B, the model is already close to human-level performance. The avoidable bias is:

$$
5% - 4.5% = 0.5%
$$

So the same training error can mean different things depending on the human-level baseline.

The key idea is:

> Bias tells us the model underfits.
> Avoidable bias tells us how much of that underfitting is still realistically reducible.

## 5. Human-Level Performance Helps Compare Bias and Variance

Human-level performance also helps us decide whether bias or variance is the bigger issue.

Consider this example:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Training error    |   10% |
| Dev error         |   12% |

The avoidable bias is:

$$
10% - 1% = 9%
$$

The variance gap is:

$$
12% - 10% = 2%
$$

Here, avoidable bias is much larger than variance. So the priority should be improving the model’s ability to fit the training set.

A better next step may be:

* use a bigger model
* improve the architecture
* train longer
* tune optimization
* reduce excessive regularization

Now compare with this case:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Training error    |    2% |
| Dev error         |   10% |

The avoidable bias is:

$$
2% - 1% = 1%
$$

The variance gap is:

$$
10% - 2% = 8%
$$

Here, the model is already close to human-level on the training set, but performs much worse on the dev set.

The main issue is more likely variance or possibly data mismatch.

Possible actions include:

* adding more data
* using regularization
* using dropout
* applying data augmentation
* checking whether train and dev come from the same distribution

Human-level performance therefore helps us understand whether training error is actually high or already close to the best achievable level.

## 6. Why Improvement Becomes Harder Near Human-Level Performance

When a model is far worse than human-level performance, there are usually many obvious errors to fix.

For example:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Model error       |   15% |

The model still has a lot of room to improve.

But when the model is close to human-level performance:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Model error       |  1.5% |

the remaining errors may be much harder.

They may come from:

* ambiguous examples
* noisy labels
* extremely difficult edge cases
* examples where even humans disagree
* information missing from the input

As a model approaches human-level performance, it may also approach Bayes error. At that point, further improvement usually requires deeper error analysis, better data, better labels, or a more careful understanding of the remaining errors.

The intuition is:

> When a model is far from human-level performance, many errors may be easy to fix.
> When a model is close to human-level performance, the remaining errors are often much harder.

## 7. Choosing a Human-Level Baseline

Human-level performance is not always a single obvious number.

For example, in medical diagnosis:

| Human Performer    | Error |
| ------------------ | ----: |
| Non-expert         |   20% |
| Junior doctor      |    8% |
| Experienced doctor |    4% |
| Expert panel       |    2% |

Which one should we use?

It depends on the project goal.

If the system is meant to assist non-experts, comparing it to a junior doctor may already be useful. If the goal is expert-level diagnosis, then comparison to experienced doctors or an expert panel is more appropriate.

When human-level performance is used as a proxy for Bayes error, we usually care about the best reliable human performance available.

The reason is that we want to estimate how low the error could realistically go.

## 8. When Human-Level Performance Is Hard to Define

Human-level performance is useful, but not every task has a clear human benchmark.

It is easier to define for tasks such as:

* image classification
* speech recognition
* medical diagnosis
* translation
* document classification

It is harder to define for tasks such as:

* recommendation systems
* stock prediction
* long-term forecasting
* user behavior prediction
* some reinforcement learning problems

When human-level performance is unclear, we can use other baselines instead:

* an existing production system
* a simple heuristic
* a traditional machine learning baseline
* a rule-based system
* historical performance

The purpose is still the same: we need a reference point to understand whether the model is performing poorly, reasonably, or near the practical limit.

## 9. Diagnostic View

A useful diagnostic view is:

| Gap                                | Meaning                                              |
| ---------------------------------- | ---------------------------------------------------- |
| Human-level error → Training error | Avoidable bias                                       |
| Training error → Dev error         | Variance or generalization gap                       |
| Train-dev error → Dev error        | Data mismatch, if train and dev distributions differ |
| Dev error → Test error             | Dev/test mismatch or overfitting to the dev set      |

For this topic, the most important gap is:

$$
\text{Training error} - \text{Human-level error}
$$

This tells us whether the model still has significant avoidable bias.

If this gap is large, the model is not fitting the task well enough.

If this gap is small but dev error is much higher, the problem is less likely to be bias and more likely to be variance, data mismatch, or evaluation issues.

## 10. Essence

The essence of **Why Human-Level Performance and Avoidable Bias** is:

> Human-level performance helps us interpret model error.

It gives context to training error and helps estimate avoidable bias.

Avoidable bias is:

$$
\text{Avoidable bias} = \text{Training error} - \text{Human-level error}
$$

If training error is much higher than human-level error, the model likely still has avoidable bias.

If training error is close to human-level error but dev error is much higher, the main problem is more likely variance, data mismatch, or evaluation issues.

The key takeaway is:

> Human-level performance tells us how much room the model still has to improve on the task.

And the key distinction is:

> Bias means the model underfits.
> Avoidable bias means the part of that underfitting that is still realistically reducible compared with human-level performance.


# Estimating Human-Level Error

## 1. Core Idea

Human-level error is usually **estimated**, not measured exhaustively.

If a dataset has 1 million images, we normally do not ask humans to label or predict every single image just to compute human-level performance. Instead, we select a smaller but representative evaluation subset and measure human performance on that subset.

The goal is not to know the exact human-level error with perfect precision. The goal is to get a useful reference point for understanding model performance.

The key idea is:

> Human-level error gives us a practical baseline for judging whether the model still has avoidable bias.

## 2. How Human-Level Error Is Estimated

A typical process is:

1. Select a representative subset from the dev/test or production-like distribution.
2. Ask appropriate humans or domain experts to label or evaluate the examples.
3. Compare their answers with trusted labels, consensus labels, or expert judgment.
4. Compute the human error rate on that subset.

For example:

| Evaluation Subset | Human Mistakes |
|---:|---:|
| 1,000 images | 20 |

Then:

$$
\text{Human-level error} = \frac{20}{1000} = 2\%
$$

This 2% is an estimate of human-level error.

It may not be perfect, but it can still be useful for deciding whether a model with 15% training error is still far from human-level performance.

## 3. The Subset Must Be Representative

The subset used to estimate human-level error should reflect the distribution that matters.

It should include:

- common cases
- difficult cases
- important classes
- rare but critical examples
- examples similar to production data

If the subset contains only easy examples, human-level error may look unrealistically low. If it contains only hard examples, it may look unrealistically high.

The principle is:

> Human-level error should be estimated on data that represents the real evaluation target.

## 4. Choosing the Right Human Baseline

The “human” in human-level performance depends on the task.

| Task | Suitable Human Baseline |
|---|---|
| Simple image classification | ordinary humans |
| Medical diagnosis | doctors or expert panel |
| Legal QA | legal experts |
| Speech recognition | fluent native speakers |
| Defect detection | trained inspectors |

For expert tasks, using non-experts as the human baseline can be misleading.

For example, in medical diagnosis, the relevant human baseline may be an experienced doctor or expert panel, not an average person.

When human-level performance is used as a proxy for Bayes error, we usually care about the best reliable human performance available.

## 5. Human-Level Error and Ambiguous Labels

Sometimes labels are not perfectly clear.

Different humans may disagree because:

- the example is ambiguous
- the input lacks enough information
- the label definition is unclear
- the task is subjective
- the dataset label may be wrong

In such cases, human-level performance may be estimated using:

- multiple annotators
- majority vote
- expert adjudication
- consensus labels
- inter-annotator agreement

This helps distinguish model errors from cases that are genuinely difficult even for humans.

## 6. How Accurate Does the Estimate Need to Be?

The estimate does not always need to be extremely precise.

If the model is far from human-level performance, a rough estimate is often enough.

For example:

| Human-Level Error Estimate | Training Error |
|---:|---:|
| 1–3% | 20% |

Even if human-level error is not exact, the conclusion is clear:

> The model still has large avoidable bias.

But if the model is already close to human-level performance:

| Human-Level Error Estimate | Training Error |
|---:|---:|
| around 2% | 2.3% |

then the estimate matters much more. In that case, we may need:

- a larger evaluation subset
- better expert evaluation
- cleaner labels
- more careful analysis of ambiguous cases

The closer the model is to human-level performance, the more carefully we need to estimate human-level error.

## 7. When Human-Level Performance Is Hard to Define

Some tasks do not have a clear human-level baseline.

Examples include:

- recommendation systems
- stock prediction
- long-term forecasting
- user behavior prediction
- some reinforcement learning tasks

In these cases, we can use other reference points instead:

- an existing production system
- a simple baseline model
- a rule-based system
- expert heuristics
- historical performance

The purpose is the same:

> We need a meaningful baseline to understand whether the model is poor, reasonable, or near the practical limit.

## 8. Essence

Human-level error is usually estimated from a representative subset, not measured on the entire dataset.

The purpose is to provide a reference point for interpreting model error.

A good estimate of human-level error helps us answer:

> Is the model still far from what humans can do?  
> Does the model still have significant avoidable bias?  
> Is the model already close to the practical limit of the task?

The key takeaway is:

> Human-level error does not need to be perfect. It needs to be useful enough to guide ML strategy.

\text{Avoidable bias} = \text{Training error} - \text{Human-level error}
$$

For example:

| Error Type | Value |
|---|---:|
| Best reliable human-level error | 2% |
| Training error | 8% |
| Dev error | 10% |

Then:

$$
\text{Avoidable bias} = 8\% - 2\% = 6\%
$$

and:

$$
\text{Variance gap} = 10\% - 8\% = 2\%
$$

This suggests that avoidable bias is larger than variance, so the priority should be improving the model’s ability to fit the task.

Possible actions include:

- using a larger model
- improving the architecture
- training longer
- tuning optimization
- reducing excessive regularization

If we used a weak human baseline, such as non-expert error of 10%, we might incorrectly conclude that the model has no avoidable bias. That would lead to a wrong project decision.

## 6. Multiple Human Baselines Can Be Useful

Sometimes it is helpful to keep several human baselines.

For example:

| Human Baseline | Error |
|---|---:|
| Non-expert | 20% |
| Trained annotator | 7% |
| Expert | 4% |
| Expert panel | 2% |

This gives a richer understanding of the model.

The model may already be better than non-experts, close to trained annotators, but still far from experts.

This is useful because “human-level” is not always one point. It can be a range of meaningful reference levels.

## 7. When Human-Level Performance Is Hard to Define

Some tasks have clear human-level baselines:

- image classification
- speech recognition
- medical diagnosis
- translation
- document classification
- legal QA with expert evaluation

Other tasks are harder:

- recommendation systems
- stock prediction
- user behavior prediction
- long-term forecasting
- some reinforcement learning tasks

When human-level performance is unclear, we can use other baselines instead:

- an existing production system
- a simple baseline model
- a rule-based system
- historical performance
- expert heuristics
- business KPIs

The purpose is still the same:

> We need a meaningful reference point to understand whether the model is poor, reasonable, or close to the practical limit.

## 8. Example: Legal QA

In a Legal QA system, there can be several possible human baselines:

| Human Baseline | Meaning |
|---|---|
| Ordinary user | understands the question but may not know the law |
| Law student | can perform basic legal lookup |
| Lawyer | has professional legal knowledge |
| Expert panel | gives more reliable legal judgment |

If the system is only meant to help ordinary users get initial guidance, comparing it to a law student or junior legal expert may be useful.

But if the system is expected to provide reliable legal answers with correct citations, it should be compared with legal experts or a strong expert-designed rubric.

For Legal QA, human-level performance may involve:

- answer correctness
- citation correctness
- legal grounding
- hallucination rate
- effective-date awareness

So the human baseline must match the real product goal.

## 9. Essence

The essence of **Understanding Human-Level Performance** is:

> Human-level performance is not just “what a random person can do.” It is the relevant human baseline for the goal of the ML project.

It helps us interpret model error, estimate avoidable bias, and understand how much room the model still has to improve.

The most important rule is:

> Choose the human baseline according to the decision you want to make.

If the goal is product usefulness, compare against the relevant user or practitioner.

If the goal is expert-level performance, compare against experts.

If the goal is to approximate Bayes error, use the best reliable human-level performance.

Human-level performance is therefore not merely a number. It is a strategic reference point for diagnosing model performance.

# Surpassing Human-Level Performance

## 1. Core Idea

**Surpassing human-level performance** means that a model performs better than a chosen human baseline on a specific metric and evaluation distribution.

For example:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    2% |
| Training error    |    1% |
| Dev error         |  1.5% |

In this case, the model has lower error than the human baseline.

However, this does not mean the model is perfect. It only means:

> The model has surpassed a particular human baseline under a particular evaluation setup.

The important strategic lesson is:

> Human-level performance is very useful when the model is worse than humans, but it becomes less useful for diagnosing progress once the model surpasses humans.

## 2. Why Human-Level Performance Becomes Less Useful After Being Surpassed

Earlier, we used human-level error to estimate avoidable bias:

$$
\text{Avoidable bias} = \text{Training error} - \text{Human-level error}
$$

This works well when training error is higher than human-level error.

For example:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    2% |
| Training error    |    8% |
| Dev error         |   10% |

Here:

$$
\text{Avoidable bias} = 8% - 2% = 6%
$$

The model is clearly still worse than humans, so there is likely a lot of room to improve.

But when the model surpasses human-level performance:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    2% |
| Training error    |    1% |
| Dev error         |  1.5% |

the formula no longer gives a meaningful positive gap. The model is already better than the human reference point.

The issue is that human-level error is only a proxy for Bayes error, not Bayes error itself.

$$
\text{Bayes error} \leq \text{Human-level error}
$$

Once the model is better than the human baseline, we no longer know clearly how far the model is from Bayes error. The remaining possible improvement becomes harder to estimate.

## 3. Why Models Can Surpass Human-Level Performance

Models can outperform humans in some tasks because they may have advantages that humans do not have.

For example, models can:

* learn from very large datasets
* detect statistical patterns humans may not notice
* process many features consistently
* optimize directly for a specific metric
* make predictions without fatigue
* use large-scale feedback from users or environments

This often happens in tasks such as:

| Domain                 | Why Models Can Do Well                  |
| ---------------------- | --------------------------------------- |
| Online advertising     | large-scale user behavior data          |
| Recommendation systems | patterns across millions of users       |
| Credit scoring         | many statistical features               |
| Game playing           | massive self-play or simulation         |
| Ranking systems        | direct optimization on interaction logs |

However, surpassing humans on a metric does not necessarily mean the model understands the task like humans do. It means the model is better according to the chosen evaluation criterion.

## 4. What Becomes Harder After Surpassing Human-Level Performance

When a model is worse than humans, diagnosis is often clearer.

For example:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Training error    |    5% |
| Dev error         |    7% |

Here, avoidable bias is large enough to be useful for diagnosis.

But if the model has:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Training error    |  0.7% |
| Dev error         |  0.9% |

then human-level performance no longer tells us how much reducible error remains.

The model may still have room to improve, but we need other tools to understand the remaining errors.

At this stage, improvement usually depends more on:

* detailed error analysis
* better labels
* identifying ambiguous examples
* subgroup evaluation
* production monitoring
* metric alignment
* data quality
* distribution shift analysis

In other words, once human-level performance is surpassed, progress becomes more dependent on careful analysis of the remaining errors.

## 5. Surpassing Human-Level Does Not Mean the Project Is Finished

A model may beat human-level performance overall but still fail in important ways.

Possible issues include:

* poor performance on rare but important cases
* weak robustness under distribution shift
* wrong or incomplete metric
* poor performance on safety-critical examples
* high latency or serving cost
* label leakage or evaluation flaws
* failure on important subgroups
* unreliable production behavior

For example, a Legal QA system may outperform non-experts on general answer correctness, but still hallucinate citations or cite outdated laws. In that case, surpassing a weak human baseline is not enough.

The correct interpretation is:

> Surpassing human-level performance is evidence of strong model performance, but it is not proof that the system is production-ready.

## 6. How to Make a Claim of Surpassing Human-Level Performance

A claim that a model surpasses humans should be stated carefully.

A vague claim like:

> The model beats humans.

is not enough.

A better claim is:

> The model achieves 1.5% error on the production-like test set, compared with 2.0% error from experienced human annotators under the same evaluation protocol.

A trustworthy claim should specify:

| Question              | Why It Matters                                       |
| --------------------- | ---------------------------------------------------- |
| Which human baseline? | non-expert, annotator, expert, or expert panel       |
| Which metric?         | accuracy, error, F1, WER, citation correctness, etc. |
| Which distribution?   | benchmark, dev set, production-like test set         |
| Which protocol?       | same inputs, same labels, same evaluation rules      |
| Which groups?         | overall only or also important subgroups             |

This prevents overclaiming.

## 7. Strategic Implication

Before surpassing human-level performance, human-level error is useful for estimating avoidable bias.

After surpassing human-level performance, the focus shifts.

Instead of mainly asking:

> How far is the model from human-level performance?

we ask:

> What errors still remain, and do they matter for the product?

This changes the workflow.

The team should pay more attention to:

* error categories
* rare cases
* label quality
* subgroup performance
* production behavior
* whether the metric still reflects the real goal

This is why Course 3 emphasizes that human-level performance is a useful tool, but not the final goal.

## 8. Essence

The essence of **Surpassing Human-Level Performance** is:

> Human-level performance is a useful reference point until the model reaches or surpasses it. After that, it becomes less useful for estimating how much improvement is still possible.

When the model is worse than humans, human-level error helps estimate avoidable bias.

When the model is better than humans, Bayes error is still unknown, so the remaining room for improvement becomes harder to measure.

The key takeaway is:

> Surpassing human-level performance does not mean the system is perfect. It means we must shift from broad bias diagnosis to careful analysis of the remaining errors, metrics, subgroups, and production behavior.


# Improving Your Model Performance

## 1. Core Idea

Improving model performance is not mainly about trying every technique we know. It is about diagnosing the current bottleneck and choosing the right action.

A weak model can fail for many different reasons:

* it may not fit the training set well
* it may overfit the training set
* train and dev/test distributions may differ
* the dev set may not represent production
* the metric may not reflect the real goal
* labels or data may be noisy
* the model may already be close to human-level performance

Therefore, the key question is not:

> Which technique should I try?

The better question is:

> What is the current bottleneck?

The basic workflow is:

```text
Measure errors
→ Diagnose the main bottleneck
→ Choose the appropriate action
→ Run an experiment
→ Re-evaluate
→ Repeat
```

## 2. Diagnostic Framework

A useful way to diagnose model performance is to compare different error gaps.

| Gap                                | Meaning                                         |
| ---------------------------------- | ----------------------------------------------- |
| Human-level error → Training error | Avoidable bias                                  |
| Training error → Train-dev error   | Variance on the training distribution           |
| Train-dev error → Dev error        | Data mismatch                                   |
| Dev error → Test error             | Dev/test mismatch or overfitting to the dev set |

Not every project requires every split, but this framework helps avoid premature conclusions.

For example, if training error is low and dev error is high, the issue may be high variance. But if train and dev come from different distributions, the issue may instead be data mismatch. In that case, a train-dev set helps separate the two.

## 3. Reducing Avoidable Bias

Avoidable bias is large when the training error is much higher than human-level error or a strong baseline.

Example:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Training error    |   12% |
| Dev error         |   14% |

Here, the model is still far from human-level performance even on the training set. The main issue is likely high bias or poor optimization.

Useful actions include:

* use a larger model
* improve the architecture
* train longer
* tune the learning rate
* use a better optimizer
* use Batch Normalization if appropriate
* check initialization
* reduce excessive regularization
* check for bugs in the training loop
* test whether the model can overfit a small subset

The key idea is:

> If the model cannot fit the training set well, improve capacity or optimization before focusing on regularization.

## 4. Reducing Variance

Variance is large when the model performs well on the training set but poorly on new data from the same distribution.

Example:

| Error Type        | Value |
| ----------------- | ----: |
| Human-level error |    1% |
| Training error    |    2% |
| Dev error         |   12% |

If train and dev come from the same distribution, this suggests high variance.

Useful actions include:

* collect more training data
* use L2 regularization
* use dropout
* apply data augmentation
* use early stopping
* reduce model complexity if necessary
* inspect leakage or split problems
* check whether important subgroups lack enough data

The key idea is:

> If the model fits train well but generalizes poorly to the same distribution, focus on regularization and data.

## 5. Handling Data Mismatch

Data mismatch appears when the model generalizes well within the training distribution but performs poorly on the dev/test distribution.

Example:

| Set             | Error |
| --------------- | ----: |
| Training error  |    1% |
| Train-dev error |    2% |
| Dev error       |   15% |

The small gap between training and train-dev suggests variance is not the main issue. The large gap between train-dev and dev suggests data mismatch.

Useful actions include:

* collect more target-like data
* add target-distribution data to training
* fine-tune on target-distribution data
* design augmentation that mimics the dev/test distribution
* inspect preprocessing differences
* analyze how train and dev/test distributions differ
* check whether dev/test truly represent production

The key idea is:

> If train-dev performance is good but dev performance is poor, focus on the distribution gap.

## 6. Fixing Dev/Test Mismatch

If dev performance is good but test performance is poor, the issue may be in the evaluation setup rather than the model itself.

Example:

| Set        | Error |
| ---------- | ----: |
| Dev error  |    4% |
| Test error |   10% |

Possible causes include:

* dev and test come from different distributions
* the dev set is too small
* the team has tuned too much on the dev set
* the test set is more representative of production than the dev set

Useful actions include:

* recheck the dev/test split
* ensure dev and test come from the same distribution
* increase dev/test size if needed
* create a new dev set if the old one has been overused
* avoid using the test set for model selection

The key idea is:

> If dev is good but test is bad, inspect the evaluation setup before changing the model.

## 7. Fixing Metric or Dev Set Problems

Sometimes the model scores well on the metric, but the product still performs badly.

Examples:

* accuracy is high, but the model misses an important rare class
* F1-score is high, but latency is too slow
* a Legal QA system sounds fluent but gives wrong citations
* offline recommendation metrics improve, but users are not more satisfied
* dev performance is good, but production performance is poor

In these cases, the metric or dev set may not reflect the real objective.

Useful actions include:

* redesign the metric
* define optimizing and satisficing metrics clearly
* add metrics for important classes or subgroups
* check whether the dev set represents production
* analyze production errors
* update dev/test sets if the target distribution has changed

The key idea is:

> If the metric improves but the product does not, you may be optimizing the wrong target.

## 8. Summary of Diagnosis and Actions

| Diagnosis         | Main Signal                                            | Priority Actions                                                                                      |
| ----------------- | ------------------------------------------------------ | ----------------------------------------------------------------------------------------------------- |
| Avoidable bias    | Training error is much higher than human-level error   | bigger model, better architecture, better optimization, train longer, reduce excessive regularization |
| Variance          | Training error is low, but dev/train-dev error is high | more data, regularization, dropout, augmentation, early stopping                                      |
| Data mismatch     | Train-dev error is low, but dev error is high          | target-like data, fine-tuning, distribution-aware augmentation, preprocessing audit                   |
| Dev/test mismatch | Dev error is low, but test error is high               | recheck dev/test split, avoid dev overfitting, improve evaluation setup                               |
| Wrong metric      | Metric improves, but product quality does not          | redesign metric, add constraints, align evaluation with product goal                                  |
| Label/data issue  | Errors come from wrong labels or dirty data            | audit labels, clean dev/test first, fix data pipeline                                                 |

This table is not a rigid formula. It is a guide for choosing the most likely high-impact direction.

## 9. Practical Workflow

A practical workflow for improving model performance is:

1. Check whether the metric reflects the real goal.
2. Check whether dev/test sets represent production.
3. Compare training error with human-level performance or a strong baseline.
4. Compare training, train-dev, dev, and test errors if needed.
5. Identify whether the main issue is bias, variance, mismatch, metric design, or data quality.
6. Choose one focused group of actions.
7. Run controlled experiments and re-evaluate.

This workflow prevents random experimentation and helps the team improve the actual bottleneck.

## 10. Essence

The essence of **Improving Your Model Performance** is:

> Do not blindly improve the model. Improve the bottleneck.

If avoidable bias is high, improve capacity and optimization.
If variance is high, improve generalization with data and regularization.
If data mismatch is high, fix the distribution gap.
If dev/test mismatch exists, fix the evaluation setup.
If the metric is wrong, redesign the metric.
If labels or data are unreliable, audit the data.

A strong ML engineer does not simply ask:

> What technique should I try next?

They ask:

> What evidence tells me where the bottleneck is?

This is the strategic mindset behind improving model performance in real ML projects.
