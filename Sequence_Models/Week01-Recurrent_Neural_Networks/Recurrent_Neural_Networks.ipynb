# Why Sequence Models?

Sequence models are designed for learning problems where the data has a **sequential structure**. Many real-world problems involve inputs and/or outputs that naturally occur as ordered sequences, such as speech, text, DNA, music, and video.

The key motivation is:

$$
\boxed{
\text{Sequential Structure}
\rightarrow
\text{Sequence Learning Problems}
\rightarrow
\text{Need for Sequence Models}
}
$$

---

## 1. Sequence Data

A sequence consists of ordered elements:

$$
x^{\langle 1\rangle},
x^{\langle 2\rangle},
\ldots,
x^{\langle T\rangle}
$$

where:

* $T$ is the sequence length.
* $x^{\langle t\rangle}$ is the element at position or timestep $t$.

The index $t$ does not necessarily represent physical time. In language, for example, it can represent the position of a word in a sentence.

The important property is:

$$
\boxed{
\text{The order of the elements carries information.}
}
$$

---

## 2. Examples of Sequence Learning Problems

Sequence models can be applied to many different domains.

### 2.1. Speech Recognition

A speech signal is a sequence, and the output is a text sequence:

$$
\boxed{
\text{Speech Sequence}
\rightarrow
\text{Text Sequence}
}
$$

Example:

```text
Speech signal
      ↓
"The quick brown fox jumped over the lazy dog."
```

---

### 2.2. Music Generation

The goal is to generate a sequence of music or sound:

$$
\boxed{
\text{Initial Condition}
\rightarrow
\text{Music Sequence}
}
$$

This is an example where the desired output itself is sequential.

---

### 2.3. Sentiment Classification

The input is a text sequence, while the output is a single value such as a sentiment rating:

$$
\boxed{
\text{Text Sequence}
\rightarrow
\text{Single Output}
}
$$

Example:

```text
"There is nothing to like in this movie."
                ↓
              1 star
```

---

### 2.4. DNA Sequence Analysis

DNA can be represented as an ordered sequence of nucleotides:

$$
A,\;C,\;G,\;T
$$

A model may process a DNA sequence and predict labels associated with different positions.

$$
\boxed{
\text{DNA Sequence}
\rightarrow
\text{Label Sequence}
}
$$

---

### 2.5. Machine Translation

The input is a sentence in one language and the output is a sentence in another language:

$$
\boxed{
\text{Source Text Sequence}
\rightarrow
\text{Target Text Sequence}
}
$$

For example:

```text
French sentence
      ↓
English sentence
```

The input and output sequences do not have to have the same length.

---

### 2.6. Video Activity Recognition

A video can be viewed as a sequence of frames:

$$
x^{\langle1\rangle},
x^{\langle2\rangle},
\ldots,
x^{\langle T\rangle}
$$

The model may predict a single activity label:

$$
\boxed{
\text{Video Sequence}
\rightarrow
\text{Activity Label}
}
$$

For example:

```text
Video frames
     ↓
"Running"
```

---

### 2.7. Named Entity Recognition

A sentence is a sequence, and the model predicts a label for each position:

$$
\boxed{
\text{Text Sequence}
\rightarrow
\text{Label Sequence}
}
$$

For example:

```text
John     lives     in     Paris
 ↓         ↓        ↓       ↓
Person     O        O     Location
```

---

## 3. Different Input/Output Structures

The examples above show that sequence learning does not correspond to a single input/output pattern.

Common structures include:

### Sequence-to-Sequence

$$
\boxed{
x^{\langle1\rangle},\ldots,x^{\langle T_x\rangle}
\rightarrow
y^{\langle1\rangle},\ldots,y^{\langle T_y\rangle}
}
$$

Examples:

* Speech recognition
* Machine translation
* DNA sequence analysis
* Named entity recognition

---

### Sequence-to-One

$$
\boxed{
x^{\langle1\rangle},\ldots,x^{\langle T\rangle}
\rightarrow
y
}
$$

Examples:

* Sentiment classification
* Video activity recognition

---

### One-to-Sequence

$$
\boxed{
x
\rightarrow
y^{\langle1\rangle},\ldots,y^{\langle T\rangle}
}
$$

Example:

* Music generation

The important point is:

> **Sequence learning problems can have different input and output structures, while still sharing the need to model sequential data.**

---

Although text, speech, DNA, music, and video are very different types of data, they share a common property:

$$
\boxed{
\text{The elements are ordered.}
}
$$

For example:

### Text

$$
\text{word}_1
\rightarrow
\text{word}_2
\rightarrow
\text{word}_3

### Speech

$$

\boxed{

}
$$

# Notation

Sequence models process data that is naturally represented as an ordered sequence of elements. To describe these sequences and the computations performed on them, we use the following notation.

## 1. Sequence Representation

An input sequence is written as:

$$
x^{\langle 1\rangle},
x^{\langle 2\rangle},
\ldots,
x^{\langle T_x\rangle}
$$

where:

* $x^{\langle t\rangle}$ is the input at timestep $t$.
* $t$ denotes the current timestep.
* $T_x$ is the length of the input sequence.

For example, for the sentence:

```text
I → love → AI
```

we have:

$$
x^{\langle1\rangle}=\text{I}
$$

$$
x^{\langle2\rangle}=\text{love}
$$

$$
x^{\langle3\rangle}=\text{AI}
$$

and therefore:

$$
T_x=3
$$

---

## 2. Input and Output Sequences

The output can also be a sequence:

$$
y^{\langle1\rangle},
y^{\langle2\rangle},
\ldots,
y^{\langle T_y\rangle}
$$

where:

* $y^{\langle t\rangle}$ is the target/output at timestep $t$.
* $T_y$ is the output sequence length.

Importantly:

$$
\boxed{
T_x \neq T_y \text{ is possible}
}
$$

For example, in machine translation, the number of words in the source sentence and translated sentence can be different.

---

## 3. Dimensions at Each Timestep

Each input at one timestep can itself be a vector:

$$
x^{\langle t\rangle}\in\mathbb{R}^{n_x}
$$

where $n_x$ is the input dimension.

Similarly:

$$
y^{\langle t\rangle}\in\mathbb{R}^{n_y}
$$

where $n_y$ is the output dimension.

Thus, a sequence can be viewed as:

$$
\boxed{
T_x \text{ input vectors, each of dimension } n_x
}
$$

rather than simply a list of scalar values.

---

## 4. RNN Notation

When we later introduce recurrent neural networks, an important intermediate representation is:

$$
a^{\langle t\rangle}
$$

where $a^{\langle t\rangle}$ denotes the hidden activation at timestep $t$.

The prediction at timestep $t$ is written as:

$$
\hat y^{\langle t\rangle}
$$

Thus, a single timestep can be represented conceptually as:

$$
\boxed{
x^{\langle t\rangle}
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
}
$$

The key notation to remember is:

$$
\boxed{
x^{\langle t\rangle},
\quad
a^{\langle t\rangle},
\quad
\hat y^{\langle t\rangle}
}
$$

---

## 5. Whole Sequence View

The complete input and output sequences can be written as:

$$
x=
\left(
x^{\langle1\rangle},
\ldots,
x^{\langle T_x\rangle}
\right)
$$

and:

$$
y=
\left(
y^{\langle1\rangle},
\ldots,
y^{\langle T_y\rangle}
\right)
$$

Conceptually:

```text
Input Sequence
x¹ → x² → x³ → ... → xᵀˣ
                      ↓
               Sequence Model
                      ↓
y¹ → y² → y³ → ... → yᵀʸ
Output Sequence
```

---

## 6. Key Takeaway

The notation used throughout Course 5 can be summarized as follows:

| Symbol                      | Meaning                           |
| --------------------------- | --------------------------------- |
| $t$                         | Current timestep                  |
| $T_x$                       | Input sequence length             |
| $T_y$                       | Output sequence length            |
| $x^{\langle t\rangle}$      | Input at timestep $t$             |
| $y^{\langle t\rangle}$      | Target/output at timestep $t$     |
| $a^{\langle t\rangle}$      | Hidden activation at timestep $t$ |
| $\hat y^{\langle t\rangle}$ | Prediction at timestep $t$        |
| $n_x$                       | Input dimension per timestep      |
| $n_y$                       | Output dimension per timestep     |

The essential mental model is:

$$
\boxed{
\text{Sequence}
=
\text{Ordered elements indexed by }t
}
$$

and, for recurrent models:

$$
\boxed{
x^{\langle t\rangle}
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
}
$$

This notation provides the foundation for describing the RNN architecture and its forward computation.


# Recurrent Neural Network Model

A **Recurrent Neural Network (RNN)** is a neural network architecture designed to process sequential data by maintaining a hidden state that carries information from previous timesteps.

The core idea is:

$$
\boxed{
\text{Previous Hidden State}
+
\text{Current Input}
\rightarrow
\text{Current Hidden State}
}
$$

and then:

$$
\boxed{
\text{Current Hidden State}
\rightarrow
\text{Prediction}
}
$$

Therefore, at timestep $t$:

$$
x^{\langle t\rangle}
+
a^{\langle t-1\rangle}
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
$$

The important idea is that the same computation is repeatedly applied across timesteps while sharing the same parameters.

---

## 1. RNN Architecture

At a single timestep, the RNN receives:

* the current input $x^{\langle t\rangle}$;
* the previous hidden activation $a^{\langle t-1\rangle}$.

It combines them to produce the current hidden activation:

$$
a^{\langle t\rangle}
=
f\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

The current hidden activation is then used to produce the prediction:

$$
a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
$$

A single recurrent cell can therefore be viewed as:

![RNN single-timestep architecture](https://media.geeksforgeeks.org/wp-content/uploads/20250523171309383561/recurrent_neural_network.webp)

The essential structure is:

$$
\boxed{
a^{\langle t-1\rangle}
+
x^{\langle t\rangle}
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
}
$$

The hidden activation $a^{\langle t\rangle}$ is the mechanism through which information is carried from one timestep to the next.

---

## 2. Hidden State as Recurrent Memory

The key difference between an RNN and a standard feedforward network is the recurrent state.

A feedforward network can be viewed as:

$$
x \rightarrow \hat y
$$

while an RNN has the form:

$$
(a^{\langle t-1\rangle},x^{\langle t\rangle})
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
$$

Thus, the current representation depends on both:

$$
\boxed{
\text{Current Input}
+
\text{Previous State}
}
$$

Conceptually, $a^{\langle t\rangle}$ contains a learned representation of relevant information from the sequence processed so far.

For example:

$$
a^{\langle1\rangle}
\rightarrow
a^{\langle2\rangle}
\rightarrow
a^{\langle3\rangle}
$$

means that information can flow from earlier parts of the sequence toward later parts.

The hidden state should not be interpreted as a literal copy of the entire past. Instead:

$$
\boxed{
a^{\langle t\rangle}
=
\text{learned representation of sequence history}
}
$$

---

## 3. RNN Forward Propagation

The standard vanilla RNN equations are:

$$
\boxed{
a^{\langle t\rangle}
=
\tanh
\left(
W_{aa}a^{\langle t-1\rangle}
+
W_{ax}x^{\langle t\rangle}
+
b_a
\right)
}
$$

and:

$$
\boxed{
\hat y^{\langle t\rangle}
=
g_y
\left(
W_{ya}a^{\langle t\rangle}
+
b_y
\right)
}
$$

These equations describe the complete computation at one timestep.

### 3.1. Previous-State Contribution

The term:

$$
W_{aa}a^{\langle t-1\rangle}
$$

transforms the previous hidden state into the hidden representation space.

If:

$$
a^{\langle t-1\rangle}\in\mathbb R^{n_a}
$$

then:

$$
W_{aa}\in\mathbb R^{n_a\times n_a}
$$

and therefore:

$$
W_{aa}a^{\langle t-1\rangle}\in\mathbb R^{n_a}
$$

This is the part of the computation that carries information from the past.

### 3.2. Current-Input Contribution

The term:

$$
W_{ax}x^{\langle t\rangle}
$$

maps the current input into the hidden-state space.

If:

$$
x^{\langle t\rangle}\in\mathbb R^{n_x}
$$

then:

$$
W_{ax}\in\mathbb R^{n_a\times n_x}
$$

and:

$$
W_{ax}x^{\langle t\rangle}\in\mathbb R^{n_a}
$$

### 3.3. Nonlinear Transformation

The two contributions are combined with the bias:

$$
z^{\langle t\rangle}
=
W_{aa}a^{\langle t-1\rangle}
+
W_{ax}x^{\langle t\rangle}
+
b_a
$$

and then:

$$
a^{\langle t\rangle}
=
\tanh(z^{\langle t\rangle})
$$

The $\tanh$ activation introduces nonlinearity into the recurrent state transition.

Thus:

$$
\boxed{
a^{\langle t\rangle}
=
\text{nonlinear transformation of past state and current input}
}
$$

---

## 4. Why Are There Two Weight Matrices?

The two matrices have different roles:

$$
W_{aa}
$$

controls how the previous hidden representation contributes to the current hidden representation.

$$
W_{ax}
$$

controls how the current input contributes to the current hidden representation.

Their dimensions make this explicit:

$$
W_{aa}\in\mathbb R^{n_a\times n_a}
$$

$$
W_{ax}\in\mathbb R^{n_a\times n_x}
$$

Therefore:

$$
W_{aa}a^{\langle t-1\rangle}
\in\mathbb R^{n_a}
$$

and:

$$
W_{ax}x^{\langle t\rangle}
\in\mathbb R^{n_a}
$$

so they can be added together.

This gives the basic state-transition equation:

$$
\boxed{
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
}
$$

where:

$$
\theta=
\{W_{aa},W_{ax},b_a\}
$$

---

## 5. Output Computation

Once the hidden state has been computed:

$$
a^{\langle t\rangle}
$$

the RNN can generate an output:

$$
\hat y^{\langle t\rangle}
=
g_y
\left(
W_{ya}a^{\langle t\rangle}
+
b_y
\right)
$$

If:

$$
a^{\langle t\rangle}\in\mathbb R^{n_a}
$$

then:

$$
W_{ya}\in\mathbb R^{n_y\times n_a}
$$

and:

$$
b_y\in\mathbb R^{n_y}
$$

so:

$$
\hat y^{\langle t\rangle}\in\mathbb R^{n_y}
$$

The output activation $g_y$ depends on the task.

For example, a classification problem may use a suitable probabilistic output function.

---

## 6. Unrolling the RNN Through Time

The real power of an RNN appears when the same recurrent cell is applied repeatedly across the sequence.

For a sequence:

$$
x^{\langle1\rangle},
x^{\langle2\rangle},
\ldots,
x^{\langle T\rangle}
$$

we compute:

$$
a^{\langle1\rangle}
=
f_\theta
\left(
a^{\langle0\rangle},
x^{\langle1\rangle}
\right)
$$

$$
a^{\langle2\rangle}
=
f_\theta
\left(
a^{\langle1\rangle},
x^{\langle2\rangle}
\right)
$$

$$
a^{\langle3\rangle}
=
f_\theta
\left(
a^{\langle2\rangle},
x^{\langle3\rangle}
\right)
$$

and so on until:

$$
a^{\langle T\rangle}
=
f_\theta
\left(
a^{\langle T-1\rangle},
x^{\langle T\rangle}
\right)
$$

The RNN can therefore be visualized as an unrolled recurrent computation:

![Unrolled RNN through time](https://www.d2l.ai/_images/unfolded-rnn.svg)

The crucial point is:

$$
\boxed{
\text{One RNN cell}
\rightarrow
\text{Repeated across all timesteps}
}
$$

---

## 7. Parameter Sharing Across Time

The RNN does **not** learn a separate set of parameters for every timestep.

There is only one:

$$
W_{aa}
$$

one:

$$
W_{ax}
$$

and one:

$$
W_{ya}
$$

which are reused at every timestep.

Conceptually:

$$
\theta^{\langle1\rangle}
=
\theta^{\langle2\rangle}
=
\cdots
=
\theta^{\langle T\rangle}
=
\theta
$$

This is called:

$$
\boxed{
\text{Parameter Sharing Across Time}
}
$$

This is an extremely important property.

If the sequence becomes longer:

$$
T\rightarrow T+1
$$

the number of recurrent parameters does not increase.

Only the number of times the same computation is executed increases.

Therefore:

$$
\boxed{
\text{Longer Sequence}
\neq
\text{More Recurrent Parameters}
}
$$

---

## 8. Initial Hidden State

To compute the first hidden state, we need:

$$
a^{\langle0\rangle}
$$

A common initialization is:

$$
\boxed{
a^{\langle0\rangle}=0
}
$$

Then:

$$
a^{\langle1\rangle}
=
\tanh
\left(
W_{aa}a^{\langle0\rangle}
+
W_{ax}x^{\langle1\rangle}
+
b_a
\right)
$$

Once $a^{\langle1\rangle}$ is computed, it becomes the state used by timestep 2:

$$
a^{\langle1\rangle}
\rightarrow
a^{\langle2\rangle}
$$

and so on.

---

## 9. Loss Over a Sequence

For many sequence tasks, the network produces an output at multiple timesteps:

$$
\hat y^{\langle1\rangle},
\hat y^{\langle2\rangle},
\ldots,
\hat y^{\langle T\rangle}
$$

Each prediction can have its own loss:

$$
L^{\langle1\rangle},
L^{\langle2\rangle},
\ldots,
L^{\langle T\rangle}
$$

The sequence-level loss is commonly written as:

$$
\boxed{
L
=
\sum_{t=1}^{T}
L^{\langle t\rangle}
}
$$

The exact form of each $L^{\langle t\rangle}$ depends on the task.

The important concept is that the total loss receives contributions from multiple timesteps.

---

## 10. Why Parameter Sharing Changes Backpropagation

Suppose the same parameter matrix $W_{aa}$ is used at every timestep.

Then the total loss depends on $W_{aa}$ through multiple computation paths:

$$
L
=
L^{\langle1\rangle}
+
L^{\langle2\rangle}
+
\cdots
+
L^{\langle T\rangle}
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial W_{aa}}
=
\sum_{t=1}^{T}
\frac{\partial L^{\langle t\rangle}}
{\partial W_{aa}}
}
$$

This equation captures an essential consequence of parameter sharing:

> The same parameter receives gradient contributions from multiple timesteps.

Thus:

$$
\boxed{
\text{Shared Parameters}
\rightarrow
\text{Accumulated Gradient Contributions}
}
$$

This will be the foundation for understanding Backpropagation Through Time.

---

## 11. Backpropagation Through Time

Once the RNN is unrolled, its computational graph contains dependencies across time:

$$
a^{\langle1\rangle}
\rightarrow
a^{\langle2\rangle}
\rightarrow
\cdots
\rightarrow
a^{\langle T\rangle}
$$

During backpropagation, gradients can therefore flow in the reverse direction:

$$
T
\rightarrow
T-1
\rightarrow
\cdots
\rightarrow
1
$$

This procedure is called:

$$
\boxed{
\text{Backpropagation Through Time (BPTT)}
}
$$

The key distinction is that ordinary backpropagation operates through network depth, whereas BPTT also propagates gradients through the recurrent time dimension.

---

## 12. Why Does the Gradient Propagate Across Time?

Consider:

$$
a^{\langle t-1\rangle}
\rightarrow
a^{\langle t\rangle}
\rightarrow
a^{\langle t+1\rangle}
$$

Because:

$$
a^{\langle t\rangle}
$$

depends on:

$$
a^{\langle t-1\rangle}
$$

a later loss can depend indirectly on an earlier state.

For example:

$$
\frac{\partial L^{\langle t+1\rangle}}
{\partial a^{\langle t-1\rangle}}
=
\frac{\partial L^{\langle t+1\rangle}}
{\partial a^{\langle t\rangle}}
\frac{\partial a^{\langle t\rangle}}
{\partial a^{\langle t-1\rangle}}
$$

For a longer dependency:

$$
\frac{\partial L^{\langle T\rangle}}
{\partial a^{\langle1\rangle}}
=
\frac{\partial L^{\langle T\rangle}}
{\partial a^{\langle T-1\rangle}}
\cdots
\frac{\partial a^{\langle2\rangle}}
{\partial a^{\langle1\rangle}}
$$

Thus, long-range dependencies introduce repeated multiplication of derivatives or Jacobians.

This observation will later explain why vanilla RNNs suffer from vanishing and exploding gradients.

---

## 13. Local Jacobian of the RNN State Transition

Consider the scalar intuition:

$$
a_t
=
\tanh
\left(
w a_{t-1}
+
u x_t
+
b
\right)
$$

The derivative with respect to the previous state is:

$$
\frac{\partial a_t}{\partial a_{t-1}}
=
(1-a_t^2)w
$$

because:

$$
\frac{d}{dz}\tanh(z)
=
1-\tanh^2(z)
$$

For a long sequence:

$$
\frac{\partial a_T}{\partial a_1}
=
\prod_{t=2}^{T}
\frac{\partial a_t}{\partial a_{t-1}}
$$

Therefore:

$$
\boxed{
\frac{\partial a_T}{\partial a_1}
=
\prod_{t=2}^{T}
(1-a_t^2)w
}
$$

This reveals the fundamental mathematical structure behind long-term gradient propagation:

$$
\boxed{
\text{Repeated Multiplication Across Time}
}
$$

If the magnitude of the factors is consistently smaller than one, the product can become extremely small.

If the factors are sufficiently large, the product can grow rapidly.

The detailed consequences will be studied in the dedicated **Vanishing and Exploding Gradients** topic.

---

## 14. RNN as a Deep Computational Graph

Although the RNN cell itself may look simple, unrolling it over a long sequence creates a deep computational graph in the time dimension:

$$
x^{\langle1\rangle}
\rightarrow
a^{\langle1\rangle}
\rightarrow
a^{\langle2\rangle}
\rightarrow
\cdots
\rightarrow
a^{\langle T\rangle}
$$

Therefore:

$$
\boxed{
\text{RNN Unrolled Through Time}
\approx
\text{Deep Computational Graph Along the Time Dimension}
}
$$

The important distinction is that the transformations at different timesteps share the same parameters.

This perspective is essential for understanding BPTT.

---

## 15. Complete RNN Training Flow

The complete training process can be summarized as:

$$
\boxed{
\text{Input Sequence}
\rightarrow
\text{Forward Pass}
\rightarrow
\text{Sequence Loss}
\rightarrow
\text{BPTT}
\rightarrow
\text{Parameter Update}
}
$$

During the forward pass:

$$
x^{\langle1\rangle}
\rightarrow
a^{\langle1\rangle}
\rightarrow
a^{\langle2\rangle}
\rightarrow
\cdots
\rightarrow
a^{\langle T\rangle}
$$

During backward propagation:

$$
T
\rightarrow
T-1
\rightarrow
\cdots
\rightarrow
1
$$

And finally:

$$
\theta
\leftarrow
\theta
-
\alpha
\frac{\partial L}{\partial\theta}
$$

where $\alpha$ is the learning rate.

---

## 16. A Small Forward Example

Suppose:

$$
n_x=2,\qquad n_a=3
$$

Then:

$$
x^{\langle t\rangle}\in\mathbb R^2
$$

and:

$$
a^{\langle t\rangle}\in\mathbb R^3
$$

The parameter shapes are:

$$
W_{ax}\in\mathbb R^{3\times2}
$$

$$
W_{aa}\in\mathbb R^{3\times3}
$$

$$
b_a\in\mathbb R^3
$$

The two matrix products are:

$$
W_{ax}x^{\langle t\rangle}\in\mathbb R^3
$$

and:

$$
W_{aa}a^{\langle t-1\rangle}\in\mathbb R^3
$$

so:

$$
z^{\langle t\rangle}
=
W_{aa}a^{\langle t-1\rangle}
+
W_{ax}x^{\langle t\rangle}
+
b_a
\in\mathbb R^3
$$

Then:

$$
a^{\langle t\rangle}
=
\tanh(z^{\langle t\rangle})
\in\mathbb R^3
$$

Every operation is therefore dimensionally consistent.

---

## 17. The Deepest Mental Model

A vanilla RNN can be understood as a learned state-transition system:

$$
\boxed{
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
}
$$

The state is updated repeatedly:

$$
a^{\langle0\rangle}
\rightarrow
a^{\langle1\rangle}
\rightarrow
a^{\langle2\rangle}
\rightarrow
\cdots
\rightarrow
a^{\langle T\rangle}
$$

while the same parameter set is reused:

$$
\boxed{
\theta
=
\{W_{aa},W_{ax},W_{ya},b_a,b_y\}
}
$$

This gives the complete conceptual picture:

$$
\boxed{
\text{Input}
+
\text{Previous State}
\rightarrow
\text{Current State}
\rightarrow
\text{Prediction}
}
$$

and during training:

$$
\boxed{
\text{Shared Parameters Across Time}
\rightarrow
\text{Multiple Gradient Contributions}
\rightarrow
\text{BPTT}
}
$$

---

## 18. Key Takeaway

The essential mechanism of a vanilla RNN is:

$$
\boxed{
a^{\langle t\rangle}
=
\tanh
\left(
W_{aa}a^{\langle t-1\rangle}
+
W_{ax}x^{\langle t\rangle}
+
b_a
\right)
}
$$

followed by:

$$
\boxed{
\hat y^{\langle t\rangle}
=
g_y
\left(
W_{ya}a^{\langle t\rangle}
+
b_y
\right)
}
$$

The same parameters are reused across all timesteps:

$$
\boxed{
W_{aa},W_{ax},W_{ya}
\text{ are shared across time}
}
$$

The resulting computational graph is recurrent:

$$
\boxed{
x^{\langle t\rangle}
+
a^{\langle t-1\rangle}
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
}
$$

and training requires gradients to propagate through this unrolled graph:

$$
\boxed{
\text{Forward}
\rightarrow
\text{Loss}
\rightarrow
\text{BPTT}
\rightarrow
\text{Parameter Update}
}
$$

The most important causal chain to remember is:

$$
\boxed{
\text{Recurrent State}
\rightarrow
\text{Information Across Time}
\rightarrow
\text{Parameter Sharing}
\rightarrow
\text{BPTT}
\rightarrow
\text{Repeated Gradient Products}
}
$$

This final term is what leads naturally to the next major question:

$$
\boxed{
\text{Why do gradients vanish or explode in a vanilla RNN?}
}
$$


# Backpropagation Through Time (BPTT)

**Backpropagation Through Time (BPTT)** is the procedure used to compute gradients for a Recurrent Neural Network (RNN) by **unrolling the recurrent computation through time and applying backpropagation to the resulting computational graph**.

The central idea is:

$$
\boxed{
\text{Unroll RNN}
\rightarrow
\text{Forward Through Time}
\rightarrow
\text{Compute Sequence Loss}
\rightarrow
\text{Backward Through Time}
\rightarrow
\text{Accumulate Gradients}
\rightarrow
\text{Update Parameters}
}
$$

BPTT is essentially standard backpropagation applied to an RNN after it has been unrolled along the time dimension.

The two properties that make BPTT distinctive are:

1. hidden states are connected across timesteps;
2. the same parameters are reused at every timestep.

Therefore, an earlier hidden state can receive gradient contributions from multiple future losses, while a shared parameter receives gradient contributions from multiple timesteps.

---

## 1. RNN Setup

For a vanilla RNN, define the recurrent pre-activation:

$$
z^{\langle t\rangle}
=
W_{aa}a^{\langle t-1\rangle}
+
W_{ax}x^{\langle t\rangle}
+
b_a
$$

The hidden activation is:

$$
\boxed{
a^{\langle t\rangle}
=
\tanh
\left(
z^{\langle t\rangle}
\right)
}
$$

The output layer is:

$$
z_y^{\langle t\rangle}
=
W_{ya}a^{\langle t\rangle}
+
b_y
$$

and:

$$
\boxed{
\hat y^{\langle t\rangle}
=
g_y
\left(
z_y^{\langle t\rangle}
\right)
}
$$

For a sequence of length $T$:

$$
x^{\langle1\rangle},
x^{\langle2\rangle},
\ldots,
x^{\langle T\rangle}
$$

the RNN produces:

$$
a^{\langle1\rangle},
a^{\langle2\rangle},
\ldots,
a^{\langle T\rangle}
$$

and:

$$
\hat y^{\langle1\rangle},
\hat y^{\langle2\rangle},
\ldots,
\hat y^{\langle T\rangle}
$$

---

## 2. Sequence Loss

Suppose each timestep has a corresponding loss:

$$
L^{\langle1\rangle},
L^{\langle2\rangle},
\ldots,
L^{\langle T\rangle}
$$

The total sequence loss is:

$$
\boxed{
L
=
\sum_{t=1}^{T}
L^{\langle t\rangle}
}
$$

For $T=3$:

$$
\boxed{
L
=
L^{\langle1\rangle}
+
L^{\langle2\rangle}
+
L^{\langle3\rangle}
}
$$

This defines the loss over the complete sequence.

The important point for BPTT is that these losses are connected through the recurrent hidden states.

---

## 3. Unrolling the RNN Through Time

An RNN cell is reused at every timestep:

```text
x¹ ──→ [RNN] ──→ a¹ ──→ ŷ¹ ──→ L¹
              │
x² ──→ [RNN] ──→ a² ──→ ŷ² ──→ L²
              │
x³ ──→ [RNN] ──→ a³ ──→ ŷ³ ──→ L³
              │
             ...
```

The hidden-state dependency is:

$$
\boxed{
a^{\langle1\rangle}
\rightarrow
a^{\langle2\rangle}
\rightarrow
\cdots
\rightarrow
a^{\langle T\rangle}
}
$$

After unrolling, the RNN becomes a computational graph that is deep along the time dimension.

Therefore:

$$
\boxed{
\text{BPTT}
=
\text{Backpropagation on the unrolled RNN}
}
$$

The forward pass moves from:

$$
1\rightarrow2\rightarrow\cdots\rightarrow T
$$

while the backward pass moves from:

$$
\boxed{
T\rightarrow T-1\rightarrow\cdots\rightarrow1
}
$$

---

## 4. Parameter Sharing Across Time

The same parameters are used at every timestep.

The recurrent parameters are:

$$
W_{aa},\quad W_{ax},\quad b_a
$$

and the output parameters are:

$$
W_{ya},\quad b_y
$$

There is not a separate parameter matrix for every timestep.

Instead:

$$
\boxed{
W_{aa}^{\langle1\rangle}
=
W_{aa}^{\langle2\rangle}
=
\cdots
=
W_{aa}^{\langle T\rangle}
=
W_{aa}
}
$$

and similarly for the other parameters.

Therefore:

$$
\boxed{
\text{One shared parameter}
\rightarrow
\text{Multiple computational paths}
}
$$

This means that gradient contributions generated at different timesteps must ultimately be accumulated onto the same parameter.

---

## 5. Backward Propagation at the Hidden State

Consider $T=4$.

The recurrent chain is:

$$
a^{\langle1\rangle}
\rightarrow
a^{\langle2\rangle}
\rightarrow
a^{\langle3\rangle}
\rightarrow
a^{\langle4\rangle}
$$

Consider:

$$
a^{\langle3\rangle}
$$

It affects the loss through two paths.

### 5.1. Local Path

The hidden state directly produces the output at timestep 3:

$$
a^{\langle3\rangle}
\rightarrow
z_y^{\langle3\rangle}
\rightarrow
\hat y^{\langle3\rangle}
\rightarrow
L^{\langle3\rangle}
$$

### 5.2. Future Recurrent Path

The hidden state also affects the next hidden state:

$$
a^{\langle3\rangle}
\rightarrow
a^{\langle4\rangle}
\rightarrow
z_y^{\langle4\rangle}
\rightarrow
\hat y^{\langle4\rangle}
\rightarrow
L^{\langle4\rangle}
$$

Therefore:

$$
\boxed{
\frac{\partial L}
{\partial a^{\langle3\rangle}}
=
\frac{\partial L^{\langle3\rangle}}
{\partial a^{\langle3\rangle}}
+
\frac{\partial L^{\langle4\rangle}}
{\partial a^{\langle3\rangle}}
}
$$

The first term is the **local gradient contribution**.

The second term is the **future gradient contribution propagated through the recurrent connection**.

---

## 6. General Two-Source Gradient

For a general timestep $t<T$:

$$
\boxed{
\frac{\partial L}
{\partial a^{\langle t\rangle}}
=
\frac{\partial L^{\langle t\rangle}}
{\partial a^{\langle t\rangle}}
+
\frac{\partial L}
{\partial a^{\langle t+1\rangle}}
\frac{\partial a^{\langle t+1\rangle}}
{\partial a^{\langle t\rangle}}
}
$$

This gives the recursive backward rule:

$$
\boxed{
\text{Total Gradient at }a^{\langle t\rangle}
=
\text{Local Contribution}
+
\text{Future Contribution}
}
$$

For the final timestep:

$$
\boxed{
\frac{\partial L}
{\partial a^{\langle T\rangle}}
=
\frac{\partial L^{\langle T\rangle}}
{\partial a^{\langle T\rangle}}
}
$$

because there is no future timestep.

---

## 7. Output-Layer Gradient

Suppose the output uses sigmoid activation:

$$
\hat y^{\langle t\rangle}
=
\sigma
\left(
z_y^{\langle t\rangle}
\right)
$$

with binary cross-entropy:

$$
L^{\langle t\rangle}
=
-y^{\langle t\rangle}
\log\hat y^{\langle t\rangle}
-
(1-y^{\langle t\rangle})
\log
\left(
1-\hat y^{\langle t\rangle}
\right)
$$

Define:

$$
\delta_y^{\langle t\rangle}
=
\frac{\partial L^{\langle t\rangle}}
{\partial z_y^{\langle t\rangle}}
$$

For sigmoid activation combined with binary cross-entropy:

$$
\boxed{
\delta_y^{\langle t\rangle}
=
\hat y^{\langle t\rangle}
-
y^{\langle t\rangle}
}
$$

Since:

$$
z_y^{\langle t\rangle}
=
W_{ya}a^{\langle t\rangle}
+
b_y
$$

the local gradient with respect to the hidden state is:

$$
\boxed{
\frac{\partial L^{\langle t\rangle}}
{\partial a^{\langle t\rangle}}
=
W_{ya}^{T}
\delta_y^{\langle t\rangle}
}
$$

---

## 8. Complete Gradient at a Hidden State

The local output gradient is not the complete gradient because future timesteps also depend on the current hidden state.

For $t<T$:

$$
\boxed{
\frac{\partial L}
{\partial a^{\langle t\rangle}}
=
W_{ya}^{T}
\delta_y^{\langle t\rangle}
+
W_{aa}^{T}
\delta_z^{\langle t+1\rangle}
}
$$

where:

$$
\delta_z^{\langle t+1\rangle}
=
\frac{\partial L}
{\partial z^{\langle t+1\rangle}}
$$

Thus:

$$
\boxed{
\underbrace{
W_{ya}^{T}\delta_y^{\langle t\rangle}
}_{\text{Local output contribution}}
+
\underbrace{
W_{aa}^{T}\delta_z^{\langle t+1\rangle}
}_{\text{Future recurrent contribution}}
}
$$

At the final timestep:

$$
\boxed{
\frac{\partial L}
{\partial a^{\langle T\rangle}}
=
W_{ya}^{T}
\delta_y^{\langle T\rangle}
}
$$

---

## 9. Gradient Through the Recurrent $\tanh$

Recall:

$$
a^{\langle t\rangle}
=
\tanh
\left(
z^{\langle t\rangle}
\right)
$$

Since:

$$
\frac{d}{dz}\tanh(z)
=
1-\tanh^2(z)
$$

we have:

$$
\boxed{
\frac{\partial a^{\langle t\rangle}}
{\partial z^{\langle t\rangle}}
=
1-
\left(
a^{\langle t\rangle}
\right)^2
}
$$

Define:

$$
\delta_z^{\langle t\rangle}
=
\frac{\partial L}
{\partial z^{\langle t\rangle}}
$$

Then:

$$
\boxed{
\delta_z^{\langle t\rangle}
=
\frac{\partial L}
{\partial a^{\langle t\rangle}}
\odot
\left[
1-
\left(
a^{\langle t\rangle}
\right)^2
\right]
}
$$

where $\odot$ denotes element-wise multiplication.

The gradient is then propagated to the previous hidden state:

$$
\boxed{
\frac{\partial L}
{\partial a^{\langle t-1\rangle}}
=
W_{aa}^{T}
\delta_z^{\langle t\rangle}
}
$$

This is the recurrent step that allows the gradient to move backward through time.

---

## 10. Gradient Contributions to the Shared Parameters

For:

$$
z^{\langle t\rangle}
=
W_{aa}a^{\langle t-1\rangle}
+
W_{ax}x^{\langle t\rangle}
+
b_a
$$

the parameter-gradient contribution at timestep $t$ is:

$$
\boxed{
dW_{aa}^{\langle t\rangle}
=
\delta_z^{\langle t\rangle}
\left(
a^{\langle t-1\rangle}
\right)^T
}
$$

$$
\boxed{
dW_{ax}^{\langle t\rangle}
=
\delta_z^{\langle t\rangle}
\left(
x^{\langle t\rangle}
\right)^T
}
$$

$$
\boxed{
db_a^{\langle t\rangle}
=
\delta_z^{\langle t\rangle}
}
$$

For the output layer:

$$
\boxed{
dW_{ya}^{\langle t\rangle}
=
\delta_y^{\langle t\rangle}
\left(
a^{\langle t\rangle}
\right)^T
}
$$

$$
\boxed{
db_y^{\langle t\rangle}
=
\delta_y^{\langle t\rangle}
}
$$

Because the parameters are shared:

$$
\boxed{
dW_{aa}
=
\sum_{t=1}^{T}
dW_{aa}^{\langle t\rangle}
}
$$

$$
\boxed{
dW_{ax}
=
\sum_{t=1}^{T}
dW_{ax}^{\langle t\rangle}
}
$$

$$
\boxed{
dW_{ya}
=
\sum_{t=1}^{T}
dW_{ya}^{\langle t\rangle}
}
$$

$$
\boxed{
db_a
=
\sum_{t=1}^{T}
db_a^{\langle t\rangle}
}
$$

$$
\boxed{
db_y
=
\sum_{t=1}^{T}
db_y^{\langle t\rangle}
}
$$

This is the direct mathematical consequence of parameter sharing.

---

$$

with targets:

y^{\langle1\rangle}=0
$$

$$
y^{\langle2\rangle}=1
$$

$$
y^{\langle3\rangle}=1
$$

$$

=

\left(

+

+

\right)
$$

and:

Because this is a scalar example, matrix multiplication reduces to ordinary multiplication.

The recurrent pre-activation is:

$$
z_1
=

+

+
b_a
$$

$$
=
0.5(0)+1(0.5)+0
$$

Therefore:

\boxed{

}
$$

The hidden activation is:

a_1
=
\tanh(0.5)
\approx
0.4621
$$

\hat y_1
=
\sigma(W_{ya}a_1+b_y)
=
\sigma(0.4621)
$$

so:

$$
\boxed{
\hat y_1\approx0.6135
}
$$

Since $y_1=0$:

$$
L_1
=
-\log(1-\hat y_1)
\approx
0.9507
$$

---

### 12.2. Timestep 2

The previous hidden state is now used by the next timestep:

$$
z_2
=
W_{aa}a_1
+
W_{ax}x_2
+
b_a
$$

$$
=
0.5(0.4621)+1(1)+0
$$

Therefore:

\boxed{

}
$$

Then:

a_2
=

\approx

$$

and:

=

\approx
0.6991
$$

}
$$

### 12.3. Timestep 3

Now:

z_3
=

+

+

$$

$$
=
0.5(0.8429)+1(-0.5)
$$

Therefore:

\boxed{

}

## 14. Backward Pass

We now propagate the gradients from right to left:

$$
\boxed{
3\rightarrow2\rightarrow1
}
$$

### 14.1. Timestep 3

Since timestep 3 is the last timestep, there is no future recurrent contribution.

Therefore:

$$
\frac{\partial L}{\partial a_3}
=
\frac{\partial L_3}{\partial a_3}
$$

Because $W_{ya}=1$ and $b_y=0$:

$$
\frac{\partial L_3}{\partial a_3}
=
\hat y_3-y_3
$$

Thus:

$$
\frac{\partial L}{\partial a_3}
=
0.4804-1
$$

so:

$$
\boxed{
\frac{\partial L}{\partial a_3}
\approx
-0.5196
}
$$

Now pass through the $\tanh$ activation:

$$
\delta_3
=
\frac{\partial L}{\partial z_3}
=
\frac{\partial L}{\partial a_3}
(1-a_3^2)
$$

Using:

$$
a_3\approx-0.0784
$$

gives:

$$
\boxed{
\delta_3\approx-0.5164
}
$$

---

### 14.2. Propagating the Gradient from $t=3$ to $t=2$

The future loss influences $a_2$ through:

$$
a_2
\rightarrow
a_3
\rightarrow
L_3
$$

Therefore:

$$
\frac{\partial L_3}{\partial a_2}
=
\frac{\partial L_3}{\partial z_3}
\frac{\partial z_3}{\partial a_2}
$$

Since:

$$
z_3
=
W_{aa}a_2+W_{ax}x_3+b_a
$$

we have:

$$
\frac{\partial z_3}{\partial a_2}
=
W_{aa}
=
0.5
$$

Hence:

$$
\boxed{
\frac{\partial L_3}{\partial a_2}
=
(-0.5164)(0.5)
\approx
-0.2582
}
$$

This is the future gradient contribution arriving at $a_2$.

---

### 14.3. Timestep 2: Two Gradient Sources

Now $a_2$ receives gradient from two paths.

**Local contribution:**

$$
a_2
\rightarrow
\hat y_2
\rightarrow
L_2
$$

so:

$$
\frac{\partial L_2}{\partial a_2}
=
\hat y_2-y_2
$$

$$
=
0.6991-1
=
-0.3009
$$

**Future contribution:**

$$
a_2
\rightarrow
a_3
\rightarrow
L_3
$$

which gives:

$$
\frac{\partial L_3}{\partial a_2}
\approx
-0.2582
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial a_2}
=
\frac{\partial L_2}{\partial a_2}
+
\frac{\partial L_3}{\partial a_2}
}
$$

and:

$$
\frac{\partial L}{\partial a_2}
=
-0.3009-0.2582
$$

so:

$$
\boxed{
\frac{\partial L}{\partial a_2}
\approx
-0.5591
}
$$

This is the characteristic **two-source gradient** in BPTT.

---

### 14.4. Continue Backward Through $\tanh$

Since:

$$
a_2=\tanh(z_2)
$$

we have:

$$
\delta_2
=
\frac{\partial L}{\partial z_2}
=
\frac{\partial L}{\partial a_2}
(1-a_2^2)
$$

Using:

$$
a_2\approx0.8429
$$

gives:

$$
\boxed{
\delta_2\approx-0.1619
}
$$

The gradient can now propagate from $a_2$ to $a_1$.

---

### 14.5. Propagating the Gradient from $t=2$ to $t=1$

The hidden state $a_1$ receives a local contribution from $L_1$ and a future contribution from the later timesteps.

The local contribution is:

$$
\frac{\partial L_1}{\partial a_1}
=
\hat y_1-y_1
=
0.6135
$$

The future contribution is:

$$
\delta_2W_{aa}
=
(-0.1619)(0.5)
\approx
-0.0810
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial a_1}
=
\frac{\partial L_1}{\partial a_1}
+
\delta_2W_{aa}
}
$$

giving:

$$
\boxed{
\frac{\partial L}{\partial a_1}
\approx
0.6135-0.0810
=
0.5326
}
$$

---

### 14.6. Timestep 1

Finally:

$$
a_1=\tanh(z_1)
$$

so:

$$
\delta_1
=
\frac{\partial L}{\partial z_1}
=
\frac{\partial L}{\partial a_1}
(1-a_1^2)
$$

Using:

$$
a_1\approx0.4621
$$

gives:

$$
\boxed{
\delta_1\approx0.4188
}
$$

The backward traversal through the recurrent states is now complete.

## 15. Gradient Accumulation

The backward pass gives a gradient contribution for the shared parameters at every timestep.

### 15.1. Gradient of $W_{aa}$

For each timestep:

$$
dW_{aa}^{(t)}
=
\delta_t a_{t-1}
$$

Thus:

$$
dW_{aa}^{(1)}
=
(0.4188)(0)
=
0
$$

$$
dW_{aa}^{(2)}
=
(-0.1619)(0.4621)
\approx
-0.0748
$$

$$
dW_{aa}^{(3)}
=
(-0.5164)(0.8429)
\approx
-0.4353
$$

Since the same $W_{aa}$ is shared:

$$
\boxed{
dW_{aa}
=

## 17. Parameter Update

Once BPTT has computed the accumulated gradients, the parameters are updated using the usual gradient descent rule.

For learning rate $\alpha$:

$$
\boxed{
\theta
\leftarrow
\theta-\alpha\nabla_\theta L
}
$$

For example:

$$
W_{aa}
\leftarrow
W_{aa}
-
\alpha dW_{aa}
$$

The important distinction is:

$$
\boxed{
\text{BPTT determines how the gradient is computed}
}
$$

while:

$$
\boxed{
\text{Gradient Descent determines how parameters are updated}
}
$$

## 18. BPTT for $T=3$ at a Glance

The forward process is:

$$
x_1
\rightarrow
a_1
\rightarrow
\hat y_1
\rightarrow
L_1
$$

$$
x_2+a_1
\rightarrow
a_2
\rightarrow
\hat y_2
\rightarrow
L_2
$$

$$
x_3+a_2
\rightarrow
a_3
\rightarrow
\hat y_3
\rightarrow
L_3
$$

The total loss is:

$$
\boxed{
L=L_1+L_2+L_3
}
$$

The backward paths are:

$$
L_3
\rightarrow
a_3
\rightarrow
a_2
\rightarrow
a_1
$$

$$
L_2
\rightarrow
a_2
\rightarrow
a_1
$$

$$
L_1
\rightarrow
a_1
$$

Therefore:

$$
\boxed{
a_3:\quad L_3
}
$$

$$
\boxed{
a_2:\quad L_2+L_3\text{ contributions}
}
$$

$$
\boxed{
a_1:\quad L_1+L_2+L_3\text{ contributions}
}
$$

More precisely, later losses reach earlier hidden states through the recurrent chain.

---

## 19. Why Does the Gradient Become a Product Across Time?

Consider the influence of the final loss on an early hidden state:

$$
\frac{\partial L^{\langle T\rangle}}
{\partial a^{\langle1\rangle}}
$$

By the chain rule:

$$
\boxed{
\frac{\partial L^{\langle T\rangle}}
{\partial a^{\langle1\rangle}}
=
\frac{\partial L^{\langle T\rangle}}
{\partial a^{\langle T\rangle}}
\prod_{k=2}^{T}
\frac{\partial a^{\langle k\rangle}}
{\partial a^{\langle k-1\rangle}}
}
$$

Each recurrent transition contributes another Jacobian.

Thus:

$$
\boxed{
\text{Longer temporal dependency}
\rightarrow
\text{More Jacobian multiplications}
}
$$

For the scalar RNN:

$$
a_t
=
\tanh(wa_{t-1}+ux_t+b)
$$

we have:

$$
\frac{\partial a_t}{\partial a_{t-1}}
=
(1-a_t^2)w
$$

and therefore:

$$
\boxed{
\frac{\partial a_T}{\partial a_1}
=
\prod_{t=2}^{T}
(1-a_t^2)w
}
$$

This repeated multiplication is the mathematical foundation of the vanishing and exploding gradient problem.

---

## 20. BPTT and Vanishing / Exploding Gradients

BPTT itself is not the problem.

Rather, BPTT exposes the recurrent structure that can create the problem.

In matrix form:

$$
\boxed{
\frac{\partial a_T}{\partial a_1}
=
J_TJ_{T-1}\cdots J_2
}
$$

where:

$$
J_t
=
\frac{\partial a_t}
{\partial a_{t-1}}
$$

When many Jacobians are multiplied together, the magnitude of the gradient may become:

$$
\boxed{
\text{Very small}
}
$$

or:

$$
\boxed{
\text{Very large}
}
$$

This leads naturally to:

$$
\boxed{
\text{BPTT}
\rightarrow
\text{Repeated Jacobian Products}
\rightarrow
\text{Vanishing / Exploding Gradients}
}
$$

The detailed analysis of matrix norms, eigenvalues, singular values, and gradient clipping belongs to the next topic.

---

## 21. Final Mental Model

The complete BPTT mechanism is:

$$
\boxed{
\text{Forward Through Time}
}
$$

$$
x_t+a_{t-1}
\rightarrow
a_t
\rightarrow
\hat y_t
$$

for:

$$
t=1,\ldots,T
$$

Then:

$$
\boxed{
L
=
\sum_{t=1}^{T}L_t
}
$$

Then:

$$
\boxed{
\text{Backward Through Time}
}
$$

At every hidden state:

$$
\boxed{
\text{Total Gradient}
=
\text{Local Contribution}
+
\text{Future Contribution}
}
$$

Because parameters are shared across time:

$$
\boxed{
dW
=
\sum_{t=1}^{T}
dW^{(t)}
}
$$

Finally:

$$
\boxed{
\theta
\leftarrow
\theta-\alpha\nabla_\theta L
}
$$

The deepest conceptual chain is:

$$
\boxed{
\text{Recurrent Connection}
\rightarrow
\text{Future Losses Influence Earlier Hidden States}
}
$$

and:

$$
\boxed{
\text{Parameter Sharing}
\rightarrow
\text{Gradient Contributions Accumulate}
}
$$

Together:

$$
\boxed{
\text{Unrolled RNN}
\rightarrow
\text{BPTT}
\rightarrow
\text{Repeated Jacobian Products}
\rightarrow
\text{Vanishing / Exploding Gradients}
}
$$

> **BPTT is ordinary backpropagation applied to the RNN unrolled through time. The distinctive behavior comes from the recurrent connections and shared parameters: future losses propagate to earlier hidden states, while gradient contributions from all timesteps accumulate onto the same parameter set.**

# Different Types of RNN Architectures

RNN architectures can be classified according to how the **input sequence** is mapped to the **output sequence**.

The main architectures covered in this section are:

1. Many-to-Many
2. Many-to-One
3. One-to-Many
4. Encoder-Decoder

The key difference between these architectures is the relationship between the input sequence length $T_x$ and the output sequence length $T_y$.

---

## 1. Many-to-Many Architecture

In a **many-to-many** architecture, the model receives a sequence as input and produces a sequence as output.

In the simplest case:

$$
T_x = T_y
$$

There is an output corresponding to each input timestep.

### Structure

$$
x^{\langle 1\rangle}
\rightarrow
a^{\langle 1\rangle}
\rightarrow
\hat{y}^{\langle 1\rangle}
$$

$$
x^{\langle 2\rangle}
\rightarrow
a^{\langle 2\rangle}
\rightarrow
\hat{y}^{\langle 2\rangle}
$$

$$
\vdots
$$

$$
x^{\langle T\rangle}
\rightarrow
a^{\langle T\rangle}
\rightarrow
\hat{y}^{\langle T\rangle}
$$

At each timestep, the RNN computes:

$$
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

and then:

$$
\hat{y}^{\langle t\rangle}
=
g_\theta
\left(
a^{\langle t\rangle}
\right)
$$

Therefore, the model produces:

$$
\hat{y}^{\langle 1\rangle},
\hat{y}^{\langle 2\rangle},
\dots,
\hat{y}^{\langle T\rangle}
$$

The overall mapping is:

$$
\boxed{
\text{Sequence}
\rightarrow
\text{Sequence}
}
$$

### Example: Named Entity Recognition

Consider the sentence:

> John lives in Paris

The model can assign one label to every word:

| Input | Output |
|---|---|
| John | PERSON |
| lives | O |
| in | O |
| Paris | LOCATION |

The RNN processes the sentence sequentially and produces one prediction at each timestep.

Thus:

$$
x^{\langle t\rangle}
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat{y}^{\langle t\rangle}
$$

for every timestep $t$.

### Main Idea

Many-to-Many is useful when:

- each timestep requires an output,
- the input and output positions are naturally aligned,
- the model performs sequence labeling.

Typical examples include:

- Named Entity Recognition
- Part-of-Speech Tagging
- Sequence labeling

---

## 2. Many-to-One Architecture

In a **many-to-one** architecture, the model receives a sequence but produces only one output.

Therefore:

$$
T_x > 1,
\qquad
T_y = 1
$$

The RNN processes the input sequence step by step:

$$
x^{\langle 1\rangle}
\rightarrow
a^{\langle 1\rangle}
$$

$$
x^{\langle 2\rangle}
\rightarrow
a^{\langle 2\rangle}
$$

$$
\vdots
$$

$$
x^{\langle T\rangle}
\rightarrow
a^{\langle T\rangle}
$$

The final hidden state is then used to produce the output:

$$
a^{\langle T\rangle}
\rightarrow
\hat{y}
$$

Therefore:

$$
\boxed{
\text{Sequence}
\rightarrow
\text{One Output}
}
$$

### Example: Sentiment Classification

Consider:

> The movie was very interesting

The model reads the entire sequence and produces a single prediction:

$$
\hat{y} = \text{Positive}
$$

The hidden states evolve as:

$$
a^{\langle 1\rangle}
\rightarrow
a^{\langle 2\rangle}
\rightarrow
\cdots
\rightarrow
a^{\langle T\rangle}
$$

The final hidden state $a^{\langle T\rangle}$ acts as a learned representation of the sequence.

The final prediction is then:

$$
\hat{y}
=
g_\theta
\left(
a^{\langle T\rangle}
\right)
$$

### Why Can the Final Hidden State Represent the Sequence?

The recurrent update is:

$$
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

Therefore, the final hidden state depends recursively on the previous inputs:

$$
a^{\langle T\rangle}
=
f_\theta
\left(
a^{\langle T-1\rangle},
x^{\langle T\rangle}
\right)
$$

and:

$$
a^{\langle T-1\rangle}
=
f_\theta
\left(
a^{\langle T-2\rangle},
x^{\langle T-1\rangle}
\right)
$$

This means information from earlier timesteps can influence the final hidden state.

Therefore:

$$
\boxed{
a^{\langle T\rangle}
\approx
\text{learned representation of the sequence}
}
$$

The word "representation" is important: the hidden state does **not** literally store the complete sequence. It learns a compact representation that is useful for the task.

---

## 3. One-to-Many Architecture

In a **one-to-many** architecture, the model receives one input and produces a sequence of outputs.

Therefore:

$$
T_x = 1,
\qquad
T_y > 1
$$

The basic structure is:

$$
x
\rightarrow
a^{\langle 1\rangle}
\rightarrow
a^{\langle 2\rangle}
\rightarrow
\cdots
\rightarrow
a^{\langle T\rangle}
$$

with outputs generated at multiple timesteps:

$$
a^{\langle 1\rangle}
\rightarrow
\hat{y}^{\langle 1\rangle}
$$

$$
a^{\langle 2\rangle}
\rightarrow
\hat{y}^{\langle 2\rangle}
$$

$$
\vdots
$$

$$
a^{\langle T\rangle}
\rightarrow
\hat{y}^{\langle T\rangle}
$$

Therefore:

$$
\boxed{
\text{One Input}
\rightarrow
\text{Sequence}
}
$$

### Example: Music Generation

Suppose the model receives one input describing the initial condition for music generation.

The model can then generate a sequence:

$$
\hat{y}^{\langle 1\rangle},
\hat{y}^{\langle 2\rangle},
\dots,
\hat{y}^{\langle T\rangle}
$$

Each timestep produces the next element in the generated sequence.

The recurrent state allows the model to maintain information about what has already been generated.

Conceptually:

$$
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

and:

$$
\hat{y}^{\langle t\rangle}
=
g_\theta
\left(
a^{\langle t\rangle}
\right)
$$

During generation, the previously generated output can be used as the next input.

Conceptually:

$$
\hat{y}^{\langle t\rangle}
\rightarrow
x^{\langle t+1\rangle}
$$

This allows the model to generate the sequence step by step.

---

# 4. Many-to-Many with Different Sequence Lengths

There is another important sequence-to-sequence setting where:

$$
T_x \neq T_y
$$

For example, consider machine translation:

$$
\text{English sentence}
\rightarrow
\text{Vietnamese sentence}
$$

The number of words in the input sentence does not necessarily equal the number of words in the translated sentence.

Therefore, a simple timestep-by-timestep mapping is not sufficient.

We need an architecture that can:

1. read the entire input sequence,
2. build a representation of that sequence,
3. generate a new output sequence.

This leads to the **Encoder-Decoder architecture**.

---

# 5. Encoder-Decoder Architecture

The encoder-decoder architecture separates sequence processing into two stages:

$$
\boxed{
\text{Encoder}
\rightarrow
\text{Context}
\rightarrow
\text{Decoder}
}
$$

The encoder reads the input sequence.

The decoder then uses the encoded information to generate the output sequence.

![Encoder-Decoder Architecture](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRzmhrqpt5o0CCglvcgu5_2eUZnSAQbjBKHSvSwF5mtdl7k5d-qUV-VpvMZ&s=10)

---

## 5.1 Encoder

The encoder processes:

$$
x^{\langle 1\rangle},
x^{\langle 2\rangle},
\dots,
x^{\langle T_x\rangle}
$$

and updates its hidden state:

$$
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

After reading the complete input sequence, the encoder produces a representation of the input.

At Course 5 level, this can be represented by the final hidden state:

$$
c
=
a^{\langle T_x\rangle}
$$

where $c$ is the **context vector**.

Therefore:

$$
\boxed{
(x^{\langle 1\rangle},\dots,x^{\langle T_x\rangle})
\rightarrow
c
}
$$

The encoder's job is therefore:

$$
\boxed{
\text{Input Sequence}
\rightarrow
\text{Learned Representation}
}
$$

---

## 5.2 Decoder

The decoder takes the encoded representation and generates the output sequence.

Let the decoder hidden states be:

$$
s^{\langle 1\rangle},
s^{\langle 2\rangle},
\dots,
s^{\langle T_y\rangle}
$$

The decoder generates outputs sequentially.

Conceptually:

$$
c
\rightarrow
s^{\langle 1\rangle}
\rightarrow
\hat{y}^{\langle 1\rangle}
$$

$$
s^{\langle 2\rangle}
\rightarrow
\hat{y}^{\langle 2\rangle}
$$

$$
\vdots
$$

$$
s^{\langle T_y\rangle}
\rightarrow
\hat{y}^{\langle T_y\rangle}
$$

Therefore:

$$
\boxed{
c
\rightarrow
(\hat{y}^{\langle 1\rangle},
\dots,
\hat{y}^{\langle T_y\rangle})
}
$$

The decoder can therefore generate an output sequence whose length is independent of the input sequence length.

---

## 5.3 Why Do We Need Encoder-Decoder?

The main motivation is:

$$
T_x \neq T_y
$$

Suppose the input contains five words:

$$
T_x = 5
$$

while the translated sentence contains seven words:

$$
T_y = 7
$$

A simple aligned many-to-many architecture would have difficulty because there is no natural one-to-one correspondence between the input and output timesteps.

The encoder-decoder architecture solves this by separating:

$$
\text{Input Processing}
$$

from:

$$
\text{Output Generation}
$$

Therefore, the encoder and decoder can operate over different numbers of timesteps:

$$
\boxed{
T_x \neq T_y
}
$$

---

# 6. Comparison of RNN Architectures

| Architecture | Input | Output | Typical Task |
|---|---|---|---|
| Many-to-Many | Sequence | Sequence | NER |
| Many-to-One | Sequence | One output | Sentiment Classification |
| One-to-Many | One input | Sequence | Music Generation |
| Encoder-Decoder | Sequence | Sequence | Machine Translation |

A useful summary is:

$$
\boxed{
\begin{aligned}
\text{Many-to-Many}
&:\quad
\text{Sequence}
\rightarrow
\text{Sequence}
\\[4pt]
\text{Many-to-One}
&:\quad
\text{Sequence}
\rightarrow
\text{One}
\\[4pt]
\text{One-to-Many}
&:\quad
\text{One}
\rightarrow
\text{Sequence}
\\[4pt]
\text{Encoder-Decoder}
&:\quad
\text{Sequence}
\rightarrow
\text{Sequence}
\end{aligned}
}
$$

The important distinction between the two sequence-to-sequence cases is:

$$
\boxed{
\begin{aligned}
\text{Simple Many-to-Many}
&:\quad
T_x = T_y
\text{ with timestep-level alignment}
\\[4pt]
\text{Encoder-Decoder}
&:\quad
T_x \neq T_y
\text{ can be handled naturally}
\end{aligned}
}
$$

---

# 7. Relationship to the RNN Hidden State

All of these architectures are based on the same fundamental recurrent mechanism:

$$
\boxed{
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
}
$$

The main difference is **how the hidden states are connected to the inputs and outputs**.

### Many-to-Many

$$
x^{\langle t\rangle}
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat{y}^{\langle t\rangle}
$$

### Many-to-One

$$
x^{\langle 1\rangle},
\dots,
x^{\langle T\rangle}
\rightarrow
a^{\langle T\rangle}
\rightarrow
\hat{y}
$$

### One-to-Many

$$
x
\rightarrow
a^{\langle 1\rangle}
\rightarrow
a^{\langle 2\rangle}
\rightarrow
\cdots
\rightarrow
a^{\langle T\rangle}
\rightarrow
\hat{y}^{\langle 1:T\rangle}
$$

### Encoder-Decoder

$$
(x^{\langle 1\rangle},
\dots,
x^{\langle T_x\rangle})
\rightarrow
c
\rightarrow
(\hat{y}^{\langle 1\rangle},
\dots,
\hat{y}^{\langle T_y\rangle})
$$

---

# 8. Important Terminology

The terms **"many"** and **"one"** refer to the number of **timesteps in a sequence**, not the number of training samples.

For example, suppose one training sample contains 24 hours of measurements.

Then:

$$
T = 24
$$

This means that one sample contains:

$$
x^{\langle 1\rangle},
x^{\langle 2\rangle},
\dots,
x^{\langle 24\rangle}
$$

These are **24 timesteps of one sequence**, not 24 independent training samples.

The hierarchy is:

$$
\boxed{
\text{Dataset}
\rightarrow
\text{Samples}
\rightarrow
\text{Sequences}
\rightarrow
\text{Timesteps}
\rightarrow
\text{Features}
}
$$

This distinction becomes important when interpreting RNN input shapes.

For example, a batch of sequence data can have shape:

$$
(B,T,D)
$$

where:

- $B$ = batch size
- $T$ = number of timesteps
- $D$ = number of features at each timestep

---

# 9. Mental Model

The easiest way to identify an RNN architecture is to ask:

> **How does the input sequence map to the output sequence?**

### Many-to-Many

$$
\boxed{
\text{Many Timesteps}
\rightarrow
\text{Many Timesteps}
}
$$

Use this when there is an output at each timestep.

### Many-to-One

$$
\boxed{
\text{Many Timesteps}
\rightarrow
\text{One Output}
}
$$

Use this when the entire sequence must be summarized into one prediction.

### One-to-Many

$$
\boxed{
\text{One Input}
\rightarrow
\text{Many Timesteps}
}
$$

Use this when a single input is used to generate a sequence.

### Encoder-Decoder

$$
\boxed{
\text{Input Sequence}
\rightarrow
\text{Output Sequence}
}
$$

Use this when the input and output sequences may have different lengths:

$$
T_x \neq T_y
$$

---

# 10. Final Mental Model

The core idea is not that there are completely different RNN equations for every architecture.

Instead, the same recurrent mechanism can be organized in different ways depending on the required input-output structure.

$$
\boxed{
\text{RNN Cell}
+
\text{Different Input/Output Connections}
=
\text{Different RNN Architectures}
}
$$

The key relationships to remember are:

$$
\boxed{
\begin{aligned}

&:\quad
\text{Sequence}\rightarrow\text{Sequence}
\\
\text{Many-to-One}
&:\quad
\text{Sequence}\rightarrow\text{One}
\\
\text{One-to-Many}
&:\quad
\text{One}\rightarrow\text{Sequence}

&:\quad
\text{Sequence}\rightarrow\text{Sequence},
\quad

}
$$

# Language Model and Sequence Generation

A language model learns how likely a sequence of tokens is to occur.

For a sequence:

$$
x_1,x_2,\dots,x_T
$$

the fundamental objective is to model:

$$
P(x_1,x_2,\dots,x_T)
$$

A recurrent neural network can be used to model this probability distribution by predicting the probability of the next token given the previous tokens.

The core progression is:

$$
\boxed{
\text{Language Modeling}
\rightarrow
\text{Next-Token Prediction}
\rightarrow
\text{Sequence Probability}
\rightarrow
\text{Training}
\rightarrow
\text{Sequence Generation}
}
$$

---

# 1. What Is a Language Model?

A language model assigns probabilities to sequences of tokens.

For example:

$$
P(\text{I, love, machine, learning})
$$

represents the probability assigned to the complete sequence.

Therefore:

$$
\boxed{
\text{Language Model}
=
\text{A model of probability distributions over token sequences}
}
$$

The model should assign:

- high probability to plausible sequences;
- low probability to unlikely sequences.

For example, conceptually:

$$
P(\text{I love machine learning})
>
P(\text{I love machine banana learning})
$$

The important point is that a language model is fundamentally a **probability model**, not merely a word predictor.

---

# 2. From Sequence Probability to Next-Token Prediction

Directly modeling:

$$
P(x_1,x_2,\dots,x_T)
$$

is difficult.

The chain rule of probability allows us to decompose the sequence probability:

$$
P(x_1,x_2,\dots,x_T)
=
P(x_1)
P(x_2|x_1)
P(x_3|x_1,x_2)
\cdots
P(x_T|x_1,\dots,x_{T-1})
$$

More compactly:

$$
\boxed{
P(x_{1:T})
=
\prod_{t=1}^{T}
P(x_t|x_{1:t-1})
}
$$

This equation is the mathematical foundation of language modeling.

Instead of directly predicting an entire sequence, the model learns:

$$
\boxed{
P(x_t|x_1,\dots,x_{t-1})
}
$$

This is the probability of the current token conditioned on all previous tokens.

Therefore:

$$
\boxed{
\text{Language Modeling}
=
\text{Learning the conditional distribution of the next token}
}
$$

---

# 3. Next-Token Prediction

Suppose the sequence is:

> I love machine learning

The model can be trained using the following input-target pairs:

| Input | Target |
|---|---|
| `<START>` | I |
| I | love |
| I love | machine |
| I love machine | learning |
| I love machine learning | `<END>` |

Thus, the target sequence is shifted by one position relative to the input.

Conceptually:

$$
\boxed{
x_t
\rightarrow
\text{predict }x_{t+1}
}
$$

For example:

$$
P(\text{love}|\text{I})
$$

$$
P(\text{machine}|\text{I,love})
$$

$$
P(\text{learning}|\text{I,love,machine})
$$

and finally:

$$
P(\text{<END>}|\text{I,love,machine,learning})
$$

The model is therefore trained to predict what comes next.

---

# 4. How an RNN Implements a Language Model

The vanilla RNN recurrence is:

$$
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

The hidden state summarizes information extracted from the previous context.

The output layer then predicts a probability distribution over the vocabulary:

$$
\hat{y}^{\langle t\rangle}
=
g_\theta
\left(
a^{\langle t\rangle}
\right)
$$

For a language model:

$$
\boxed{
\hat{y}^{\langle t\rangle}
\approx
P
\left(
x^{\langle t+1\rangle}
\mid
x^{\langle 1\rangle},
\dots,
x^{\langle t\rangle}
\right)
}
$$

Therefore, the role of the hidden state is:

$$
\boxed{
\text{Previous Context}
\rightarrow
a^{\langle t\rangle}
\rightarrow
\text{Next-Token Distribution}
}
$$

The RNN does not need to pass the entire previous sequence directly to the output layer. Instead, it maintains a learned state containing information useful for predicting the future.

---

# 5. The Output Is a Probability Distribution

Suppose the vocabulary size is:

$$
|V|=10,000
$$

Then at each timestep:

$$
\hat{y}^{\langle t\rangle}
\in
\mathbb{R}^{10,000}
$$

After applying softmax:

$$
\sum_{j=1}^{|V|}
\hat{y}^{\langle t\rangle}_j
=
1
$$

Each vocabulary token receives a probability.

For example:

| Token | Probability |
|---|---:|
| the | 0.32 |
| machine | 0.18 |
| model | 0.12 |
| learning | 0.08 |
| ... | ... |

Thus:

$$
\boxed{
\hat{y}^{\langle t\rangle}
=
\text{Probability distribution over the vocabulary}
}
$$

The model does not necessarily output a single token directly.

It first produces a probability distribution.

---

# 6. Training Objective

At timestep $t$, the model predicts:

$$
P(x_{t+1}|x_{1:t})
$$

The corresponding cross-entropy loss is:

$$
L^{\langle t\rangle}
=
-\log
P(x_{t+1}|x_{1:t})
$$

For the complete sequence:

$$
\boxed{
L
=
\sum_{t=1}^{T}
L^{\langle t\rangle}
}
$$

or:

$$
L
=
-\sum_{t=1}^{T}
\log
P(x_{t+1}|x_{1:t})
$$

The model is therefore encouraged to assign high probability to the correct next token.

For example, if the correct next token is `machine`:

$$
P(\text{machine}|x_{1:t})=0.9
$$

then:

$$
L^{\langle t\rangle}
=
-\log(0.9)
$$

which is small.

If:

$$
P(\text{machine}|x_{1:t})=0.01
$$

then:

$$
L^{\langle t\rangle}
=
-\log(0.01)
$$

which is large.

Therefore:

$$
\boxed{
\text{High probability for the correct token}
\Rightarrow
\text{Low loss}
}
$$

---

# 7. Sequence Probability

Recall the chain rule:

$$
P(x_{1:T})
=
\prod_{t=1}^{T}
P(x_t|x_{1:t-1})
$$

Taking logarithms:

$$
\log P(x_{1:T})
=
\sum_{t=1}^{T}
\log P(x_t|x_{1:t-1})
$$

Therefore:

$$
-\log P(x_{1:T})
=
-\sum_{t=1}^{T}
\log P(x_t|x_{1:t-1})
$$

This is exactly the form of the language-model training objective.

Hence:

$$
\boxed{
\text{Minimizing Negative Log-Likelihood}
\iff
\text{Maximizing the Probability of the Observed Sequence}
}
$$

This is one of the most important mathematical connections in language modeling.

---

# 8. Why the RNN Hidden State Is Important

Consider:

> The cat is sleeping because it is tired.

When predicting a later token, information from earlier words can be useful.

The RNN recursively updates:

$$
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

Therefore:

$$
a^{\langle t\rangle}
$$

depends recursively on:

$$
x^{\langle 1\rangle},
x^{\langle 2\rangle},
\dots,
x^{\langle t\rangle}
$$

Conceptually:

$$
a^{\langle t\rangle}
\approx
\text{learned representation of the previous context}
$$

The language model then uses this representation to estimate:

$$
P(x_{t+1}|x_{1:t})
$$

Thus:

$$
\boxed{
\text{Past Context}
\rightarrow
\text{Hidden State}
\rightarrow
\text{Next-Token Probability}
}
$$

The hidden state is therefore a compressed representation of the information from the past that is relevant to future prediction.

---

# 9. Sequence Generation

Training and generation are different processes.

During training, the correct sequence is available.

During generation, the correct future sequence is unknown.

Suppose the model receives:

> I

The model produces:

$$
P(x_2|x_1)
$$

For example:

| Token | Probability |
|---|---:|
| love | 0.60 |
| like | 0.30 |
| am | 0.10 |

The model then selects or samples a token.

Suppose it generates:

> love

The current sequence becomes:

> I love

Now the model predicts:

$$
P(x_3|x_1,x_2)
$$

Suppose it generates:

> machine

The sequence becomes:

> I love machine

The process continues.

Thus sequence generation is iterative:

$$
\boxed{
\text{Predict}
\rightarrow
\text{Select/Sample}
\rightarrow
\text{Feed Back}
\rightarrow
\text{Repeat}
}
$$

---

# 10. Autoregressive Generation

The generated token becomes part of the input for the next step.

Conceptually:

$$
x_t
\rightarrow
\hat{x}_{t+1}
$$

then:

$$
\hat{x}_{t+1}
\rightarrow
\hat{x}_{t+2}
$$

and so on.

Therefore:

$$
\boxed{
\text{Generated Output}
\rightarrow
\text{Next Input}
}
$$

This is called **autoregressive generation**.

The complete process can be represented as:

$$
x_1
\rightarrow
P(x_2|x_1)
\rightarrow
x_2
$$

$$
x_1,x_2
\rightarrow
P(x_3|x_1,x_2)
\rightarrow
x_3
$$

$$
\vdots
$$

$$
x_1,\dots,x_t
\rightarrow
P(x_{t+1}|x_{1:t})
\rightarrow
x_{t+1}
$$

The process terminates when the model generates a special token such as:

$$
\texttt{<END>}
$$

---

# 11. Greedy Selection vs Sampling

The model produces a probability distribution, so we need a rule for choosing the next token.

## 11.1 Greedy Selection

Choose the token with the highest probability:

$$
x_{t+1}
=
\arg\max_x
P(x|x_{1:t})
$$

For example:

| Token | Probability |
|---|---:|
| cat | 0.50 |
| dog | 0.30 |
| bird | 0.15 |
| car | 0.05 |

Greedy selection chooses:

> cat

This is deterministic for a fixed model state.

---

## 11.2 Sampling

Instead of always choosing the most probable token, sample from the probability distribution.

For example:

$$
P(\text{cat})=0.50
$$

$$
P(\text{dog})=0.30
$$

$$
P(\text{bird})=0.15
$$

$$
P(\text{car})=0.05
$$

Then:

- `cat` is most likely;
- `dog` can also be generated;
- `bird` is less likely;
- `car` is unlikely.

Therefore, different sampling runs can generate different sequences.

This makes generation stochastic rather than deterministic.

---

# 12. Complete Sequence Generation Example

Suppose the vocabulary contains:

$$
V=
\{
I,
love,
like,
machine,
learning,
AI,
\texttt{<END>}
\}
$$

Start with:

$$
x_1=I
$$

The model estimates:

$$
P(x_2|I)
$$

Suppose:

$$
P(love|I)=0.6
$$

$$
P(like|I)=0.3
$$

$$
P(AI|I)=0.1
$$

Suppose the generated token is:

$$
x_2=love
$$

Now the sequence is:

> I love

The model estimates:

$$
P(x_3|I,love)
$$

Suppose:

$$
P(machine|I,love)=0.7
$$

Then:

$$
x_3=machine
$$

Now:

> I love machine

Next:

$$
P(x_4|I,love,machine)
$$

Suppose:

$$
P(learning|I,love,machine)=0.8
$$

Then:

$$
x_4=learning
$$

Finally:

$$
P(\texttt{<END>}|I,love,machine,learning)
$$

If the model generates:

$$
x_5=\texttt{<END>}
$$

generation stops.

The final sequence is:

> I love machine learning

Thus:

$$
\boxed{
\text{Generation}
=
\text{Repeated next-token prediction}
}
$$

# 13. Character-Level Language Modeling

A language model does not have to operate on words.

It can operate on characters.

For example:

> hello

can be represented as:

$$
h,e,l,l,o
$$

The model learns:

$$
P(e|h)
$$

$$
P(l|h,e)
$$

$$
P(l|h,e,l)
$$

$$
P(o|h,e,l,l)
$$

and so on.

### Advantages

The vocabulary is relatively small.

For example, English may use only a few dozen basic characters.

Therefore:

$$
|V|
\ll
\text{Word Vocabulary Size}
$$

### Disadvantages

The sequence becomes much longer.

A sentence of 100 words may correspond to hundreds or thousands of character-level timesteps.

Therefore:

$$
\boxed{
\text{Character-level}
\rightarrow
\text{Small Vocabulary}
+
\text{Longer Sequences}
}
$$

---

# 14. Word-Level Language Modeling

In a word-level model:

> I love machine learning

becomes:

$$
[I,\ love,\ machine,\ learning]
$$

The model learns:

$$
P(love|I)
$$

$$
P(machine|I,love)
$$

$$
P(learning|I,love,machine)
$$

Compared with character-level modeling:

$$
\boxed{
\text{Word-level}
\rightarrow
\text{Large Vocabulary}
+
\text{Shorter Sequences}
}
$$

Therefore:

$$
\boxed{
\begin{aligned}
\text{Character-level}
&:\quad
\text{Small vocabulary, long sequence}
\\
\text{Word-level}
&:\quad
\text{Large vocabulary, short sequence}
\end{aligned}
}
$$

This is a fundamental trade-off.

---

# 15. Training vs Generation

It is important not to confuse these two processes.

| | Training | Generation |
|---|---|---|
| Ground-truth sequence | Available | Not available |
| Target token | Known | Unknown |
| Loss | Computed | Not computed |
| Next-token prediction | Yes | Yes |
| Previous token | Usually ground truth | Usually previously generated token |
| Main purpose | Learn parameters | Generate a sequence |

## During Training

The model receives the known sequence:

$$
x_1,x_2,\dots,x_T
$$

and learns:

$$
P(x_{t+1}|x_{1:t})
$$

The loss is computed against the known target.

Therefore:

$$
\boxed{
\text{Training}
=
\text{Predict}
+
\text{Compare with Ground Truth}
+
\text{Compute Loss}
+
\text{Backpropagate}
}
$$

## During Generation

The model does not know the correct future token.

It must use its own previous prediction:

$$
\boxed{
\text{Predict}
+
\text{Select/Sample}
+
\text{Feed Back}
+
\text{Repeat}
}
$$

This difference is fundamental.

---

# 16. Connection to BPTT

Language modeling fits naturally with the RNN training procedure studied earlier.

There is a loss at every timestep:

$$
L
=
\sum_{t=1}^{T}
L^{\langle t\rangle}
$$

The recurrent structure connects:

$$
a^{\langle 1\rangle}
\rightarrow
a^{\langle 2\rangle}
\rightarrow
\cdots
\rightarrow
a^{\langle T\rangle}
$$

Therefore, gradients must propagate backward through time.

This gives:

$$
\boxed{
\text{Language Modeling}
\rightarrow
\text{RNN}
\rightarrow
\text{BPTT}
}
$$

The same recurrent connections that allow the model to use temporal context also create the repeated Jacobian products responsible for vanishing and exploding gradients.

Therefore:

$$
\boxed{
\text{BPTT}
\rightarrow
\text{Repeated Gradient Propagation}
\rightarrow
\text{Vanishing / Exploding Gradients}
}
$$

This is the bridge from language modeling back to the motivation for LSTM and GRU.

---

# 17. The Most Important Mathematical Picture

The entire language-modeling idea can be summarized by:

$$
\boxed{
P(x_{1:T})
=
\prod_{t=1}^{T}
P(x_t|x_{1:t-1})
}
$$

The RNN provides a mechanism for representing the previous context:

$$
x_{1:t}
\rightarrow
a_t
$$

and then predicting:

$$
a_t
\rightarrow
P(x_{t+1}|x_{1:t})
$$

Thus:

$$
\boxed{
\text{Past Context}
\rightarrow
\text{Hidden State}
\rightarrow
\text{Next-Token Distribution}
}
$$

During training:

$$
\boxed{
\text{Next-Token Prediction}
\rightarrow
\text{Cross-Entropy Loss}
\rightarrow
\text{BPTT}
}
$$

During generation:

$$
\boxed{
\text{Next-Token Distribution}
\rightarrow
\text{Select/Sample Token}
\rightarrow
\text{Feed Back}
\rightarrow
\text{Repeat}
}
$$

---

# 18. Final Mental Model

The complete causal chain is:

$$
\boxed{
\text{Sequence}
}
$$

$$
\downarrow
$$

$$
\boxed{
P(x_t|x_{1:t-1})
}
$$

$$
\downarrow
$$

$$
\boxed{
\text{RNN Hidden State}
}
$$

$$
\downarrow
$$

$$
\boxed{
\text{Next-Token Probability Distribution}
}
$$

$$
\downarrow
$$

$$
\boxed{
\text{Cross-Entropy Training}
}
$$

and during inference:

$$
\boxed{
\text{Probability Distribution}
\rightarrow
\text{Sample Token}
\rightarrow
\text{Feed Back}
\rightarrow
\text{Repeat}
}
$$

The deepest idea to remember is:

$$
\boxed{
\text{A language model learns how probable a sequence is by decomposing it into a sequence of next-token conditional probabilities.}
}
$$

An RNN makes this practical by maintaining a recurrent hidden state that summarizes information from previous timesteps.

# Sampling Novel Sequences

After training an RNN language model, the model can be used not only to predict the next token, but also to **generate an entirely new sequence**.

The key idea is:

$$
\boxed{
\text{Predict Probability Distribution}
\rightarrow
\text{Sample a Token}
\rightarrow
\text{Feed It Back}
\rightarrow
\text{Repeat}
}
$$

The generated sequence can be different each time because sampling is probabilistic.

---

# 1. From Language Modeling to Sequence Generation

A language model learns the conditional probability of the next token:

$$
P(x_{t+1}\mid x_1,\ldots,x_t)
$$

At timestep $t$, the RNN processes the current token and its previous hidden state:

$$
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

The output layer then produces a probability distribution over the vocabulary:

$$
\hat{y}^{\langle t\rangle}
=
P
\left(
x^{\langle t+1\rangle}
\mid
x^{\langle 1\rangle},
\dots,
x^{\langle t\rangle}
\right)
$$

Therefore:

$$
\boxed{
\text{Context}
\rightarrow
\text{Hidden State}
\rightarrow
\text{Next-Token Distribution}
}
$$

During generation, this distribution is used to select the next token.

---

# 2. Starting the Generation

Generation usually begins with a special token:

$$
\texttt{<START>}
$$

The model receives:

$$
x^{\langle 1\rangle}
=
\texttt{<START>}
$$

and computes:

$$
P(x^{\langle 2\rangle}\mid\texttt{<START>})
$$

For example:

| Token | Probability |
|---|---:|
| I | 0.50 |
| You | 0.30 |
| We | 0.15 |
| They | 0.05 |

The model now has a probability distribution over possible next tokens.

It has **not yet generated a token**.

---

# 3. What Does Sampling Mean?

Sampling means selecting a token according to the probability distribution produced by the model.

Suppose:

$$
P(\text{I})=0.50
$$

$$
P(\text{You})=0.30
$$

$$
P(\text{We})=0.15
$$

$$
P(\text{They})=0.05
$$

Then:

- `I` has the highest probability;
- `You` is also possible;
- `We` is less likely;
- `They` is least likely.

A sample might therefore produce:

$$
x^{\langle 2\rangle}=\text{I}
$$

But another sampling run could produce:

$$
x^{\langle 2\rangle}=\text{You}
$$

The token is sampled from the distribution rather than always taking the maximum-probability token.

Mathematically:

$$
\boxed{
x^{\langle t+1\rangle}
\sim
P
\left(
x^{\langle t+1\rangle}
\mid
x^{\langle 1:t\rangle}
\right)
}
$$

---

# 4. One Generation Step

A single generation step consists of three main operations.

## 4.1 Compute the Hidden State

$$
a^{\langle t\rangle}
=
f_\theta
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

## 4.2 Compute the Probability Distribution

$$
z^{\langle t\rangle}
=
W_{ya}a^{\langle t\rangle}+b_y
$$

Then:

$$
\hat{y}^{\langle t\rangle}
=
\operatorname{softmax}
\left(
z^{\langle t\rangle}
\right)
$$

Therefore:

$$
\hat{y}^{\langle t\rangle}
\in
\mathbb{R}^{|V|}
$$

where $|V|$ is the vocabulary size.

## 4.3 Sample the Next Token

$$
x^{\langle t+1\rangle}
\sim
\operatorname{Categorical}
\left(
\hat{y}^{\langle t\rangle}
\right)
$$

Thus:

$$
\boxed{
\text{RNN}
\rightarrow
\text{Softmax Distribution}
\rightarrow
\text{Sampled Token}
}
$$

---

# 5. Feeding the Sampled Token Back

This is the most important mechanism in sequence generation.

Suppose the model generates:

$$
x^{\langle 2\rangle}=\text{I}
$$

The new sequence is now:

> I

The model uses this generated token as part of the context for the next step.

It computes:

$$
P(x^{\langle 3\rangle}\mid x^{\langle 1\rangle},x^{\langle 2\rangle})
$$

Therefore:

$$
\boxed{
\text{Generated Token}
\rightarrow
\text{Next Input}
}
$$

This feedback mechanism allows the sequence to grow one token at a time.

---

# 6. Autoregressive Generation

The complete generation process is:

$$
x^{\langle 1\rangle}
\rightarrow
P(x^{\langle 2\rangle}\mid x^{\langle 1\rangle})
\rightarrow
x^{\langle 2\rangle}
$$

then:

$$
x^{\langle 1\rangle},x^{\langle 2\rangle}
\rightarrow
P(x^{\langle 3\rangle}\mid x^{\langle 1:2\rangle})
\rightarrow
x^{\langle 3\rangle}
$$

and so on.

In general:

$$
\boxed{
x^{\langle 1:t\rangle}
\rightarrow
P(x^{\langle t+1\rangle}\mid x^{\langle 1:t\rangle})
\rightarrow
x^{\langle t+1\rangle}
}
$$

This is called **autoregressive generation** because the model uses previously generated tokens to generate future tokens.

---

# 7. Complete Example

Suppose the model starts with:

$$
\texttt{<START>}
$$

### Step 1

The model computes:

$$
P(x_1\mid\texttt{<START>})
$$

Suppose the distribution is:

| Token | Probability |
|---|---:|
| I | 0.60 |
| You | 0.20 |
| We | 0.15 |
| They | 0.05 |

Suppose sampling gives:

$$
x_1=\text{I}
$$

---

### Step 2

The context is now:

> I

The model computes:

$$
P(x_2\mid\text{I})
$$

Suppose:

| Token | Probability |
|---|---:|
| love | 0.50 |
| like | 0.30 |
| am | 0.15 |
| eat | 0.05 |

Suppose sampling gives:

$$
x_2=\text{love}
$$

The generated sequence is now:

> I love

---

### Step 3

The model computes:

$$
P(x_3\mid\text{I,love})
$$

Suppose:

| Token | Probability |
|---|---:|
| machine | 0.60 |
| deep | 0.20 |
| AI | 0.10 |
| coding | 0.10 |

Suppose:

$$
x_3=\text{machine}
$$

The sequence becomes:

> I love machine

---

### Step 4

The model computes:

$$
P(x_4\mid\text{I,love,machine})
$$

Suppose it generates:

$$
x_4=\text{learning}
$$

Now:

> I love machine learning

---

### Step 5

The model may assign a high probability to:

$$
\texttt{<END>}
$$

If:

$$
x_5=\texttt{<END>}
$$

the generation stops.

Final generated sequence:

> I love machine learning

---

# 8. Why Can Different Runs Produce Different Sequences?

Suppose at the first step:

$$
P(\text{I})=0.6
$$

and:

$$
P(\text{You})=0.4
$$

One run might produce:

$$
x_1=\text{I}
$$

while another produces:

$$
x_1=\text{You}
$$

Now the contexts are different:

$$
P(x_2\mid\text{I})
\neq
P(x_2\mid\text{You})
$$

Therefore the subsequent distributions can also become different.

This gives:

$$
\boxed{
\text{Different Samples}
\rightarrow
\text{Different Contexts}
\rightarrow
\text{Different Future Tokens}
}
$$

Hence one trained model can generate many different sequences.

---

# 9. Sampling vs. Greedy Selection

Sampling should be distinguished from always selecting the most probable token.

## Greedy Selection

Choose:

$$
x^{\langle t+1\rangle}
=
\arg\max_x
P(x\mid x^{\langle 1:t\rangle})
$$

For example:

| Token | Probability |
|---|---:|
| cat | 0.50 |
| dog | 0.30 |
| bird | 0.15 |
| car | 0.05 |

Greedy selection always chooses:

> cat

For a fixed model state, the result is deterministic.

## Sampling

Sampling instead uses the whole probability distribution:

$$
x^{\langle t+1\rangle}
\sim
P(x\mid x^{\langle 1:t\rangle})
$$

Therefore, `dog` or `bird` can also be selected.

The key distinction is:

$$
\boxed{
\text{Greedy}
=
\text{Always choose maximum probability}
}
$$

$$
\boxed{
\text{Sampling}
=
\text{Draw from the probability distribution}
}
$$

---

# 10. Why Sampling Produces a Novel Sequence

The model learned conditional probabilities such as:

$$
P(x_1)
$$

$$
P(x_2\mid x_1)
$$

$$
P(x_3\mid x_1,x_2)
$$

and so on.

The probability of a complete sequence is:

$$
P(x_{1:T})
=
\prod_{t=1}^{T}
P(x_t\mid x_{1:t-1})
$$

Sampling each token sequentially therefore produces a sample from the learned sequence distribution.

Conceptually:

$$
\boxed{
\text{Token-by-Token Sampling}
\rightarrow
\text{Sample from the Sequence Distribution}
}
$$

This is why the model can produce a sequence that was not explicitly present in the training data.

---

# 11. Role of `<START>` and `<END>`

Two special tokens are useful for controlling generation.

## `<START>`

Marks the beginning of generation:

$$
\texttt{<START>}
\rightarrow
x_1
$$

## `<END>`

Marks the end of generation:

$$
x_T
\rightarrow
\texttt{<END>}
$$

Therefore:

$$
\boxed{
\texttt{<START>}
=
\text{begin generation}
}
$$

$$
\boxed{
\texttt{<END>}
=
\text{stop generation}
}
$$

Without an explicit stopping condition, the generation process would have no natural endpoint.

---

# 12. Training vs. Sampling

The distinction between training and generation is important.

### Training

The ground-truth sequence is known.

For example:

> I love machine learning

The model predicts the next token and compares it with the known target:

$$
x_t
\rightarrow
\hat y_t
\rightarrow
\text{compare with target}
\rightarrow
L_t
$$

The model parameters are updated using backpropagation.

### Sampling

The correct future token is unknown.

Instead:

$$
x_t
\rightarrow
P(x_{t+1}\mid x_{1:t})
\rightarrow
\text{sample }x_{t+1}
$$

The generated token is then fed back into the model.

Therefore:

$$
\boxed{
\text{Training}
=
\text{Learn the Distribution}
}
$$

$$
\boxed{
\text{Sampling}
=
\text{Use the Learned Distribution to Generate}
}
$$

---

# 13. Complete Generation Pipeline

The complete process can be summarized as:

$$
\boxed{
\texttt{<START>}
\rightarrow
\text{RNN}
\rightarrow
\text{Softmax}
\rightarrow
P(x_1)
\rightarrow
\text{Sample }x_1
}
$$

Then:

$$
\boxed{
x_1
\rightarrow
\text{RNN}
\rightarrow
P(x_2\mid x_1)
\rightarrow
\text{Sample }x_2
}
$$

and:

$$
\boxed{
x_{1:2}
\rightarrow
\text{RNN}
\rightarrow
P(x_3\mid x_{1:2})
\rightarrow
\text{Sample }x_3
}
$$

Continue until:

$$
x_t=\texttt{<END>}
$$

---

# 14. Final Mental Model

The core mechanism is:

$$
\boxed{
\text{Context}
\rightarrow
\text{RNN Hidden State}
\rightarrow
\text{Softmax Distribution}
\rightarrow
\text{Sample Token}
\rightarrow
\text{Feed Back}
\rightarrow
\text{Repeat}
}
$$

Mathematically:

$$
\boxed{
x^{\langle t+1\rangle}
\sim
P
\left(
x^{\langle t+1\rangle}
\mid
x^{\langle 1:t\rangle}
\right)
}
$$

and the resulting sequence follows:

$$
\boxed{
P(x_{1:T})
=
\prod_{t=1}^{T}
P(x_t\mid x_{1:t-1})
}
$$

The essential idea is:

> **A trained RNN language model generates a novel sequence by repeatedly sampling the next token from its learned conditional probability distribution and feeding that sampled token back as the context for the next timestep.**

# Vanishing Gradients with RNNs

Vanilla RNNs can struggle to learn **long-term dependencies** because gradients are repeatedly propagated through the recurrent connections during Backpropagation Through Time (BPTT).

The core chain is:

$$
\boxed{
\text{RNN Recurrence}
\rightarrow
\text{BPTT}
\rightarrow
\text{Repeated Jacobian Multiplication}
\rightarrow
\text{Vanishing / Exploding Gradients}
}
$$

For vanishing gradients, the result is:

$$
\boxed{
\text{Weak Gradient}
\rightarrow
\text{Weak Learning Signal}
\rightarrow
\text{Difficulty Learning Long-Term Dependencies}
}
$$

---

# 1. Why Does the Problem Occur?

A vanilla RNN computes:

$$
a_t
=
\tanh
\left(
W_{aa}a_{t-1}
+
W_{ax}x_t
+
b_a
\right)
$$

The hidden states are recurrently connected:

$$
a_1
\rightarrow
a_2
\rightarrow
a_3
\rightarrow
\cdots
\rightarrow
a_T
$$

During the forward pass, information flows from earlier timesteps to later timesteps.

During BPTT, the gradient flows in the opposite direction:

$$
L_T
\rightarrow
a_T
\rightarrow
a_{T-1}
\rightarrow
\cdots
\rightarrow
a_1
$$

Therefore, a gradient that needs to travel a long temporal distance must pass through many recurrent transformations.

---

# 2. Gradient Through Multiple Timesteps

Suppose we want:

$$
\frac{\partial a_T}{\partial a_t}
$$

Using the chain rule:

$$
\frac{\partial a_T}{\partial a_t}
=
\frac{\partial a_T}{\partial a_{T-1}}
\frac{\partial a_{T-1}}{\partial a_{T-2}}
\cdots
\frac{\partial a_{t+1}}{\partial a_t}
$$

Define the Jacobian at timestep $k$ as:

$$
J_k
=
\frac{\partial a_k}{\partial a_{k-1}}
$$

Then:

$$
\boxed{
\frac{\partial a_T}{\partial a_t}
=
J_TJ_{T-1}\cdots J_{t+1}
}
$$

The critical point is:

> The gradient is a **product of many Jacobians**.

This repeated multiplication is the mathematical source of the problem.

---

# 3. Jacobian of a Vanilla RNN

Consider:

$$
a_t=\tanh(z_t)
$$

where:

$$
z_t
=
W_{aa}a_{t-1}
+
W_{ax}x_t
+
b_a
$$

Using the chain rule:

$$
\frac{\partial a_t}{\partial a_{t-1}}
=
\frac{\partial a_t}{\partial z_t}
\frac{\partial z_t}{\partial a_{t-1}}
$$

For the $\tanh$ activation:

$$
\frac{d}{dz}\tanh(z)
=
1-\tanh^2(z)
$$

Therefore:

$$
\frac{\partial a_t}{\partial a_{t-1}}
=
D_tW_{aa}
$$

where:

$$
D_t
=
\operatorname{diag}
\left(
1-a_t^2
\right)
$$

Thus:

$$
\boxed{
J_t=D_tW_{aa}
}
$$

and:

$$
\boxed{
\frac{\partial a_T}{\partial a_t}
=
(D_TW_{aa})
(D_{T-1}W_{aa})
\cdots
(D_{t+1}W_{aa})
}
$$

So the gradient must repeatedly pass through both:

$$
W_{aa}
$$

and the derivative of the activation function.

---

# 4. Simple Scalar Example

To see the phenomenon clearly, consider a scalar RNN:

$$
a_t=\tanh(wa_{t-1}+\cdots)
$$

Then:

$$
\frac{\partial a_t}{\partial a_{t-1}}
=
w(1-a_t^2)
$$

Define:

$$
J_t=w(1-a_t^2)
$$

The gradient through several timesteps becomes:

$$
\frac{\partial a_T}{\partial a_t}
=
\prod_{k=t+1}^{T}J_k
$$

Suppose approximately:

$$
|J_k|\approx0.5
$$

for every timestep.

After 10 recurrent transitions:

$$
\left|
\frac{\partial a_T}{\partial a_t}
\right|
\approx
0.5^{10}
=
9.77\times10^{-4}
$$

After 50 transitions:

$$
0.5^{50}
\approx
8.88\times10^{-16}
$$

The gradient is now essentially zero.

Therefore:

$$
\boxed{
|J_k|<1
\text{ repeatedly}
\Rightarrow
\text{exponential gradient decay}
}
$$

This is the **vanishing gradient problem**.

---

# 5. Why Can $\tanh$ Make the Problem Worse?

The derivative of $\tanh$ is:

$$
1-\tanh^2(z)
$$

Its range is:

$$
0<1-\tanh^2(z)\le1
$$

When $|z|$ becomes large, $\tanh(z)$ saturates:

$$
\tanh(z)\approx\pm1
$$

Therefore:

$$
1-\tanh^2(z)\approx0
$$

For example:

$$
\tanh(3)\approx0.995
$$

so:

$$
1-\tanh^2(3)\approx0.0099
$$

Thus the activation derivative itself can strongly shrink the gradient.

The recurrent gradient contains both effects:

$$
\boxed{
J_t=D_tW_{aa}
}
$$

so repeated multiplication can become strongly contractive.

---

# 6. Role of $W_{aa}$

To simplify the intuition, temporarily ignore the activation derivative.

Then:

$$
\frac{\partial a_T}{\partial a_t}
\approx
W_{aa}^{T-t}
$$

If the magnitude of the recurrent transformation is less than $1$, repeated multiplication causes the gradient to shrink.

For a scalar example:

$$
w=0.8
$$

then:

$$
0.8^{10}\approx0.107
$$

but:

$$
0.8^{50}\approx1.43\times10^{-5}
$$

Therefore:

$$
\boxed{
\|W_{aa}\|<1
\text{ across many timesteps}
\Rightarrow
\text{gradient tends to vanish}
}
$$

---

# 7. Eigenvalue Interpretation

For the simplified repeated multiplication:

$$
W_{aa}^k
$$

consider an eigenvalue $\lambda$ of $W_{aa}$.

Along the corresponding eigen-direction, repeated multiplication behaves roughly like:

$$
\lambda^k
$$

Therefore:

### If

$$
|\lambda|<1
$$

then:

$$
\lambda^k\rightarrow0
$$

and the gradient vanishes.

### If

$$
|\lambda|>1
$$

then:

$$
|\lambda|^k\rightarrow\infty
$$

and the gradient can explode.

### If

$$
|\lambda|\approx1
$$

the corresponding signal can be preserved for much longer.

Thus:

$$
\boxed{
\text{Magnitude of eigenvalues}
\rightarrow
\text{Behavior of repeated recurrent transformations}
}
$$

---

# 8. Singular Value Interpretation

Singular values provide another useful way to understand how a transformation changes the magnitude of vectors.

For a Jacobian with singular values:

$$
\sigma_1,\sigma_2,\dots,\sigma_n
$$

a singular value indicates how a particular direction is scaled.

If:

$$
\sigma<1
$$

that direction is contracted.

If:

$$
\sigma>1
$$

that direction is expanded.

If:

$$
\sigma\approx1
$$

that direction is approximately preserved.

After repeated multiplication:

$$
J_TJ_{T-1}\cdots J_{t+1}
$$

small contraction factors can make the gradient extremely small.

Therefore:

$$
\boxed{
\text{Repeated contraction}
\rightarrow
\text{Vanishing Gradient}
}
$$

---

# 9. Why Does This Cause a Long-Term Dependency Problem?

Consider a sequence where an early token is important for a much later prediction.

Conceptually:

$$
x_1
\rightarrow
a_1
\rightarrow
a_2
\rightarrow
\cdots
\rightarrow
a_T
\rightarrow
L_T
$$

To learn the relationship between $x_1$ and the later loss, the gradient must travel through many timesteps:

$$
\frac{\partial L_T}{\partial a_1}
=
\frac{\partial L_T}{\partial a_T}
\frac{\partial a_T}{\partial a_{T-1}}
\cdots
\frac{\partial a_2}{\partial a_1}
$$

If this product becomes tiny:

$$
\frac{\partial L_T}{\partial a_1}
\approx0
$$

then the early timestep receives almost no learning signal from the later loss.

Therefore:

$$
\boxed{
\text{Long Temporal Distance}
\rightarrow
\text{Many Jacobian Products}
\rightarrow
\text{Tiny Gradient}
}
$$

This makes it difficult for vanilla RNNs to learn long-term dependencies.

---

# 10. Short-Term vs. Long-Term Dependencies

Suppose:

$$
a_1\rightarrow a_2\rightarrow\cdots\rightarrow a_{100}
$$

A loss at timestep 100 influencing $a_{90}$ only needs to pass through approximately 10 recurrent transitions:

$$
\frac{\partial a_{100}}{\partial a_{90}}
=
J_{100}\cdots J_{91}
$$

But influencing $a_1$ requires:

$$
\frac{\partial a_{100}}{\partial a_1}
=
J_{100}\cdots J_2
$$

The second product is much longer.

Therefore:

$$
\boxed{
\text{Short-term dependency}
\rightarrow
\text{easier to learn}
}
$$

while:

$$
\boxed{
\text{Long-term dependency}
\rightarrow
\text{harder to learn}
}
$$

This is the practical consequence of vanishing gradients.

---

# 11. Forward Information vs. Backward Gradient

These two concepts should not be confused.

### Forward

Information is propagated through the hidden states:

$$
a_1
\rightarrow
a_2
\rightarrow
\cdots
\rightarrow
a_T
$$

### Backward

The learning signal is propagated through the same recurrent connections in reverse:

$$
\frac{\partial L}{\partial a_T}
\rightarrow
\frac{\partial L}{\partial a_{T-1}}
\rightarrow
\cdots
\rightarrow
\frac{\partial L}{\partial a_1}
$$

Vanishing gradients specifically mean:

$$
\boxed{
\text{The backward learning signal becomes extremely small}
}
$$

It does **not** simply mean:

> "The hidden state contains no information."

This distinction is important when understanding why LSTM changes the recurrent structure.

---

# 12. Effect on Parameter Learning

RNN parameters are shared across timesteps.

For example:

$$
W_{aa}
$$

is reused at every timestep.

Therefore:

$$
\frac{\partial L}{\partial W_{aa}}
=
\sum_t
\frac{\partial L}{\partial W_{aa}^{(t)}}
$$

However, the contribution of a distant loss to an earlier timestep may become extremely small because of vanishing gradients.

Therefore:

$$
\boxed{
\text{Vanishing Gradient}
\rightarrow
\text{Weak contribution from distant dependencies}
}
$$

The model tends to receive stronger learning signals from nearby timesteps than from very distant ones.

---

# 13. Vanishing vs. Exploding Gradients

Both phenomena come from the same fundamental mechanism:

$$
\boxed{
\text{Repeated Multiplication Through Time}
}
$$

The difference is the magnitude of the repeated factors.

### Vanishing Gradient

$$
\|J_t\|<1
$$

repeated many times:

$$
\prod_t\|J_t\|
\rightarrow0
$$

### Exploding Gradient

$$
\|J_t\|>1
$$

repeated many times:

$$
\prod_t\|J_t\|
\rightarrow\infty
$$

Therefore:

$$
\boxed{
\begin{aligned}
\text{Repeated factors}<1
&\rightarrow
\text{Vanishing}
\\
\text{Repeated factors}>1
&\rightarrow
\text{Exploding}
\end{aligned}
}
$$

---

# 14. Why Gradient Clipping Helps Exploding Gradients

Suppose the gradient becomes extremely large:

$$
\|g\|\gg1
$$

Gradient clipping limits its magnitude.

A common form is:

$$
\tilde g
=
g
\min
\left(
1,
\frac{\tau}{\|g\|}
\right)
$$

where $\tau$ is the maximum allowed norm.

This prevents a huge gradient from causing an excessively large parameter update.

Thus:

$$
\boxed{
\text{Gradient Clipping}
\rightarrow
\text{Controls Exploding Gradients}
}
$$

But suppose instead:

$$
\|g\|=10^{-12}
$$

Clipping does not make it larger.

Therefore:

$$
\boxed{
\text{Gradient Clipping does not solve Vanishing Gradients}
}
$$

It primarily addresses the exploding-gradient side of the problem.

---

# 15. Complete Causal Chain

Starting from the vanilla RNN:

$$
a_t
=
\tanh
\left(
W_{aa}a_{t-1}
+
W_{ax}x_t
+
b_a
\right)
$$

BPTT gives:

$$
\frac{\partial a_T}{\partial a_t}
=
J_TJ_{T-1}\cdots J_{t+1}
$$

where:

$$
J_t=D_tW_{aa}
$$

Therefore, gradient propagation involves repeated multiplication:

$$
\boxed{
\text{Many Recurrent Jacobians}
}
$$

If the transformations repeatedly contract the gradient:

$$
\boxed{
\text{Repeated Contraction}
\rightarrow
\text{Gradient $\rightarrow0$}
}
$$

Then:

$$
\boxed{
\text{Vanishing Gradient}
}
$$

which leads to:

$$
\boxed{
\text{Weak Learning Signal for Distant Timesteps}
}
$$

and consequently:

$$
\boxed{
\text{Difficulty Learning Long-Term Dependencies}
}
$$

---

# 16. Final Mental Model

The most important chain to remember is:

$$
\boxed{
\text{RNN Recurrence}
}
$$

$$
\downarrow
$$

$$
\boxed{
\text{BPTT}
}
$$

$$
\downarrow
$$

$$
\boxed{

}

\downarrow
$$

$$
\boxed{
\text{Factors Smaller Than 1}
}
$$

$$

\boxed{

}
$$

The core mathematical idea is:

$$
\boxed{
\frac{\partial a_T}{\partial a_t}
=

The core practical idea is:

$$
\boxed{
\text{The farther a dependency is across time, the more recurrent transformations its gradient must pass through.}
}
$$

And this leads directly to the motivation for gated recurrent architectures:

$$
\boxed{

\rightarrow

# Gated Recurrent Unit (GRU)

The **Gated Recurrent Unit (GRU)** is a gated recurrent architecture designed to improve the ability of recurrent neural networks to preserve useful information over long sequences.

A vanilla RNN updates its hidden state using a nonlinear transformation:

$$
a_t
=
\tanh
\left(
W_{aa}a_{t-1}
+
W_{ax}x_t
+
b_a
\right)
$$

Because the same recurrent transformation is applied repeatedly, gradients can become very small during BPTT.

GRU introduces **gates** that allow the network to learn:

- when to keep previous information;
- when to update the hidden state;
- how much previous information should be used when constructing new information.

The core idea is:

$$
\boxed{
\text{GRU}
=
\text{Recurrent State}
+
\text{Learned Gating}
}
$$

---

# 1. Motivation: Why Do We Need GRU?

For a vanilla RNN:

$$
a_t
=
\tanh
\left(
W_{aa}a_{t-1}
+
W_{ax}x_t
+
b_a
\right)
$$

The hidden state is repeatedly transformed:

$$
a_1
\rightarrow
a_2
\rightarrow
\cdots
\rightarrow
a_T
$$

During BPTT, the gradient becomes a product of many Jacobians:

$$
\frac{\partial a_T}{\partial a_t}
=
J_TJ_{T-1}\cdots J_{t+1}
$$

If these transformations repeatedly shrink the gradient, then:

$$
\left\|
\frac{\partial a_T}{\partial a_t}
\right\|
\rightarrow0
$$

This makes long-term dependencies difficult to learn.

GRU addresses this by introducing a mechanism that allows the hidden state to be **copied forward more directly** when the model decides that old information is still useful.

The key equation is:

$$
\boxed{
a_t
=
\Gamma_u^{\langle t\rangle}
\odot
\tilde a_t
+
\left(
1-\Gamma_u^{\langle t\rangle}
\right)
\odot
a_{t-1}
}
$$

This equation is the central idea of GRU.

---

# 2. Components of a GRU

At timestep $t$, a GRU uses:

- input $x_t$;
- previous hidden state $a_{t-1}$;
- update gate $\Gamma_u^t$;
- reset gate $\Gamma_r^t$;
- candidate hidden state $\tilde a_t$;
- final hidden state $a_t$.

![GRU Architecture](https://media.springernature.com/lw685/springer-static/image/art%3A10.1007%2Fs11042-023-15571-y/MediaObjects/11042_2023_15571_Fig2_HTML.png)

The architecture can be conceptually summarized as:

$$
\boxed{
a_{t-1},x_t
\rightarrow
\begin{cases}
\Gamma_u^t\\
\Gamma_r^t
\end{cases}
\rightarrow
\tilde a_t
\rightarrow
a_t
}
$$

There are two learned gates:

$$
\boxed{
\Gamma_u^t
=
\text{Update Gate}
}
$$

$$
\boxed{
\Gamma_r^t
=
\text{Reset Gate}
}
$$

---

# 3. Update Gate

The update gate is:

$$
\boxed{
\Gamma_u^{\langle t\rangle}
=
\sigma
\left(
W_u
\left[
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right]
+
b_u
\right)
}
$$

Because sigmoid is used:

$$
0<\Gamma_u^{\langle t\rangle}<1
$$

The gate is a vector, so each hidden-state dimension can make its own update decision.

The update gate determines how much the new candidate should replace the old state.

The final state is:

$$
a_t
=
\Gamma_u^t\odot\tilde a_t
+
(1-\Gamma_u^t)\odot a_{t-1}
$$

Therefore:

### If

$$
\Gamma_u^t\approx0
$$

then:

$$
a_t\approx a_{t-1}
$$

The old state is mostly preserved.

### If

$$
\Gamma_u^t\approx1
$$

then:

$$
a_t\approx\tilde a_t
$$

The new candidate is mostly written into the state.

Thus:

$$
\boxed{
\Gamma_u^t
=
\text{How much new information should replace old information}
}
$$

---

# 4. Reset Gate

The reset gate is:

$$
\boxed{
\Gamma_r^{\langle t\rangle}
=
\sigma
\left(
W_r
\left[
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right]
+
b_r
\right)
}
$$

The reset gate controls how much of the previous hidden state is used when constructing the candidate state.

This happens through:

$$
\Gamma_r^t\odot a_{t-1}
$$

### If

$$
\Gamma_r^t\approx1
$$

then:

$$
\Gamma_r^t\odot a_{t-1}
\approx
a_{t-1}
$$

The previous state strongly influences the candidate.

### If

$$
\Gamma_r^t\approx0
$$

then:

$$
\Gamma_r^t\odot a_{t-1}
\approx0
$$

The candidate is constructed almost independently of the previous state.

Therefore:

$$
\boxed{
\Gamma_r^t
=
\text{How much previous state should influence the candidate}
}
$$

---

# 5. Candidate Hidden State

The candidate state is:

$$
\boxed{
\tilde a^{\langle t\rangle}
=
\tanh
\left(
W_a
\left[
\Gamma_r^{\langle t\rangle}
\odot
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right]
+
b_a
\right)
}
$$

The candidate represents:

> What the hidden state could look like if we decided to update it using the current input and selected information from the previous state.

It is important to distinguish:

$$
\boxed{
\tilde a_t
=
\text{Candidate new state}
}
$$

from:

$$
\boxed{
a_t
=
\text{Final hidden state}
}
$$

The update gate decides how much of $\tilde a_t$ is actually written into $a_t$.

---

# 6. Final Hidden-State Update

The final GRU equation is:

$$
\boxed{
a_t
=
\Gamma_u^t\odot\tilde a_t
+
(1-\Gamma_u^t)\odot a_{t-1}
}
$$

This can be interpreted as a learned interpolation between:

$$
\text{new candidate}
$$

and:

$$
\text{old state}
$$

More explicitly:

$$
\underbrace{
\Gamma_u^t\odot\tilde a_t
}_{\text{new information}}
+
\underbrace{
(1-\Gamma_u^t)\odot a_{t-1}
}_{\text{preserved information}}
$$

Therefore:

$$
\boxed{
\text{New State}
=
\text{Selected New Information}
+
\text{Selected Old Information}
}
$$

This is the fundamental difference from a vanilla RNN.

---

# 7. Complete Forward Pass

At timestep $t$, the GRU performs the following steps.

## Step 1: Compute Update Gate

$$
\Gamma_u^t
=
\sigma
\left(
W_u[a_{t-1},x_t]+b_u
\right)
$$

## Step 2: Compute Reset Gate

$$
\Gamma_r^t
=
\sigma
\left(
W_r[a_{t-1},x_t]+b_r
\right)
$$

## Step 3: Compute Candidate State

$$
\tilde a_t
=
\tanh
\left(
W_a
[
\Gamma_r^t\odot a_{t-1},
x_t
]
+b_a
\right)
$$

## Step 4: Compute Final Hidden State

$$
a_t
=
\Gamma_u^t\odot\tilde a_t
+
(1-\Gamma_u^t)\odot a_{t-1}
$$

Thus:

$$
\boxed{
(a_{t-1},x_t)
\rightarrow
\Gamma_u^t,\Gamma_r^t
\rightarrow
\tilde a_t
\rightarrow
a_t
}
$$

---

# 8. Tensor Shapes

Suppose:

$$
x_t\in\mathbb R^{n_x}
$$

and:

$$
a_{t-1}\in\mathbb R^{n_a}
$$

After concatenation:

$$
[a_{t-1},x_t]
\in
\mathbb R^{n_a+n_x}
$$

Therefore:

$$
W_u,W_r,W_a
\in
\mathbb R^{n_a\times(n_a+n_x)}
$$

and:

$$
b_u,b_r,b_a
\in
\mathbb R^{n_a}
$$

The outputs have dimension:

$$
\Gamma_u^t,\Gamma_r^t,\tilde a_t,a_t
\in
\mathbb R^{n_a}
$$

The three parameter sets are:

$$
(W_u,b_u)
$$

$$
(W_r,b_r)
$$

$$
(W_a,b_a)
$$

---

# 9. Small Numerical Example

Consider a scalar GRU hidden state.

Suppose:

$$
a_{t-1}=0.8
$$

and the candidate state is:

$$
\tilde a_t=-0.2
$$

Let:

$$
\Gamma_u=0.1
$$

Then:

$$
a_t
=
0.1(-0.2)
+
(1-0.1)(0.8)
$$

$$
=
-0.02+0.72
$$

$$
\boxed{
a_t=0.70
}
$$

The previous state was:

$$
0.8
$$

and the new candidate was:

$$
-0.2
$$

but the model only used $10\%$ of the candidate.

Therefore, the state remains close to the previous state.

The GRU has effectively decided:

> The old information is still important.

---

# 10. What Happens When the Update Gate Is Large?

Keep:

$$
a_{t-1}=0.8
$$

and:

$$
\tilde a_t=-0.2
$$

but now:

$$
\Gamma_u=0.9
$$

Then:

$$
a_t
=
0.9(-0.2)
+
0.1(0.8)
$$

$$
=
-0.18+0.08
$$

$$
\boxed{
a_t=-0.10
}
$$

The hidden state changes substantially.

Therefore:

$$
\boxed{
\Gamma_u\approx1
\rightarrow
\text{strong update}
}
$$

while:

$$
\boxed{
\Gamma_u\approx0
\rightarrow
\text{strong preservation}
}
$$

---

# 11. Small Numerical Example for the Reset Gate

Suppose:

$$
a_{t-1}=0.8
$$

and:

$$
\Gamma_r=0.05
$$

Then the previous-state contribution entering the candidate is:

$$
\Gamma_r a_{t-1}
=
0.05(0.8)
=
0.04
$$

The candidate therefore receives almost no information from the previous hidden state.

Conceptually:

$$
\boxed{
\Gamma_r\approx0
\rightarrow
\text{ignore previous state when constructing candidate}
}
$$

This is different from the update gate.

The reset gate affects the **candidate construction**.

The update gate affects the **final state update**.

---

# 12. Update Gate vs. Reset Gate

This distinction should be memorized conceptually, not just symbolically.

| Gate | Main Question | Operates On |
|---|---|---|
| Update gate $\Gamma_u$ | How much new information should replace the old state? | Final state update |
| Reset gate $\Gamma_r$ | How much old state should influence the candidate? | Candidate construction |

Therefore:

$$
\boxed{
\Gamma_r
\rightarrow
\text{Candidate}
}
$$

while:

$$
\boxed{
\Gamma_u
\rightarrow
\text{Final Hidden State}
}
$$

This is one of the most important distinctions in GRU.

---

# 13. Why Does GRU Help with Vanishing Gradients?

The key equation is:

$$
a_t
=
\Gamma_u^t\odot\tilde a_t
+
(1-\Gamma_u^t)\odot a_{t-1}
$$

Consider the direct state-preservation path:

$$
a_{t-1}
\rightarrow
(1-\Gamma_u^t)\odot a_{t-1}
\rightarrow
a_t
$$

The direct derivative of this branch with respect to $a_{t-1}$ is:

$$
(1-\Gamma_u^t)
$$

ignoring for the moment the dependence of the gates themselves on $a_{t-1}$.

If:

$$
\Gamma_u^t\approx0
$$

then:

$$
1-\Gamma_u^t\approx1
$$

and therefore:

$$
\boxed{
\frac{\partial a_t}{\partial a_{t-1}}
\approx1
}
$$

along this direct path.

This is fundamentally different from the vanilla RNN, where the state must pass through the recurrent nonlinear transformation:

$$
a_t=\tanh(W_{aa}a_{t-1}+\cdots)
$$

The GRU therefore provides a path through time that can behave more like an identity connection.

---

# 14. Direct Gradient Path

Suppose:

$$
\delta_t
=
\frac{\partial L}{\partial a_t}
$$

From:

$$
a_t
=
\Gamma_u^t\odot\tilde a_t
+
(1-\Gamma_u^t)\odot a_{t-1}
$$

the direct gradient to the previous state is:

$$
\boxed{
\left.
\frac{\partial L}{\partial a_{t-1}}
\right|_{\text{direct}}
=
\delta_t
\odot
(1-\Gamma_u^t)
}
$$

If:

$$
\Gamma_u^t\approx0
$$

then:

$$
\left.
\frac{\partial L}{\partial a_{t-1}}
\right|_{\text{direct}}
\approx
\delta_t
$$

So the gradient can be passed backward with little attenuation through this path.

This is the central mathematical intuition behind gated recurrence.

---

# 15. Backward Pass Through the Update Equation

Let:

$$
\delta_t
=
\frac{\partial L}{\partial a_t}
$$

and:

$$
a_t
=
\Gamma_u
\odot
\tilde a_t
+
(1-\Gamma_u)
\odot
a_{t-1}
$$

## Gradient to the Candidate

$$
\boxed{
\frac{\partial L}{\partial\tilde a_t}
=
\delta_t
\odot
\Gamma_u
}
$$

The candidate receives more gradient when the update gate is large.

## Gradient to the Update Gate

Since:

$$
\frac{\partial a_t}{\partial\Gamma_u}
=
\tilde a_t-a_{t-1}
$$

we obtain:

$$
\boxed{
\frac{\partial L}{\partial\Gamma_u}
=
\delta_t
\odot
(\tilde a_t-a_{t-1})
}
$$

This tells us how the loss changes with respect to the decision of how much to update.

## Direct Gradient to the Previous State

From the old-state branch:

$$
\boxed{
\left.
\frac{\partial L}{\partial a_{t-1}}
\right|_{\text{direct}}
=
\delta_t
\odot
(1-\Gamma_u)
}
$$

This direct path is one of the most important differences from vanilla RNNs.

---

# 16. Backward Through the Candidate

The candidate is:

$$
\tilde a_t
=
\tanh(z_t)
$$

where:

$$
z_t
=
W_a
[
\Gamma_r\odot a_{t-1},
x_t
]
+b_a
$$

We already have:

$$
\delta_{\tilde a}
=
\delta_t\odot\Gamma_u
$$

Through $\tanh$:

$$
\boxed{
\delta_z
=
\delta_{\tilde a}
\odot
(1-\tilde a_t^2)
}
$$

This is the gradient entering the candidate preactivation.

---

# 17. Backward Through the Reset Gate

Let:

$$
q_t
=
\Gamma_r\odot a_{t-1}
$$

Then:

$$
\delta_q
=
W_a^T\delta_z
$$

The gradient with respect to the reset gate is:

$$
\boxed{
\frac{\partial L}{\partial\Gamma_r}
=
\delta_q
\odot
a_{t-1}
}
$$

The candidate also contributes gradient back to the previous hidden state:

$$
\boxed{
\left.
\frac{\partial L}{\partial a_{t-1}}
\right|_{\text{candidate}}
=
\delta_q
\odot
\Gamma_r
}
$$

---

# 18. Backward Through the Update-Gate Network

The update gate is:

$$
\Gamma_u
=
\sigma(z_u)
$$

with:

$$
z_u
=
W_u[a_{t-1},x_t]+b_u
$$

We have:

$$
\delta_{\Gamma_u}
=
\delta_t
\odot
(\tilde a_t-a_{t-1})
$$

Since:

$$
\frac{d\sigma(z)}{dz}
=
\sigma(z)(1-\sigma(z))
$$

we obtain:

$$
\boxed{
\delta_{z_u}
=
\delta_{\Gamma_u}
\odot
\Gamma_u(1-\Gamma_u)
}
$$

Then:

$$
\boxed{
\frac{\partial L}{\partial W_u}
=
\delta_{z_u}
[a_{t-1},x_t]^T
}
$$

and:

$$
\boxed{
\frac{\partial L}{\partial b_u}
=
\delta_{z_u}
}
$$

---

# 19. Backward Through the Reset-Gate Network

Similarly:

$$
\Gamma_r
=
\sigma(z_r)
$$

with:

$$
z_r
=
W_r[a_{t-1},x_t]+b_r
$$

We obtain:

$$
\boxed{
\delta_{z_r}
=
\delta_{\Gamma_r}
\odot
\Gamma_r(1-\Gamma_r)
}
$$

Then:

$$
\boxed{
\frac{\partial L}{\partial W_r}
=
\delta_{z_r}
[a_{t-1},x_t]^T
}
$$

and:

$$
\boxed{
\frac{\partial L}{\partial b_r}
=
\delta_{z_r}
}
$$

---

# 20. Gradient of Candidate Parameters

Recall:

$$
z_t
=
W_a
[
\Gamma_r\odot a_{t-1},
x_t
]
+
b_a
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial W_a}
=
\delta_z
[
\Gamma_r\odot a_{t-1},
x_t
]^T
}
$$

and:

$$
\boxed{
\frac{\partial L}{\partial b_a}
=
\delta_z
}
$$

---

# 21. Total Gradient to $a_{t-1}$

This is important because $a_{t-1}$ receives gradient from multiple paths.

At minimum:

1. directly through the final state equation;
2. through the candidate;
3. through the update gate;
4. through the reset gate.

Therefore the total gradient is a sum of multiple contributions:

$$
\boxed{
\frac{\partial L}{\partial a_{t-1}}
=
\text{Direct Path}
+
\text{Candidate Path}
+
\text{Gate Paths}
}
$$

The direct path is particularly important:

$$
\boxed{
\delta_t(1-\Gamma_u)
}
$$

because it can remain strong when:

$$
\Gamma_u\approx0
$$

This provides the mechanism that helps gated RNNs preserve long-term information and gradient flow.

---

# 22. Why the Gate Values Are Learned

The gates are not manually specified.

They are learned using parameters:

$$
W_u,b_u
$$

and:

$$
W_r,b_r
$$

For example:

$$
\Gamma_u
=
\sigma
\left(
W_u[a_{t-1},x_t]+b_u
\right)
$$

During training, BPTT computes:

$$
\frac{\partial L}{\partial W_u}
$$

and:

$$
\frac{\partial L}{\partial W_r}
$$

so gradient descent updates these parameters.

Therefore the model learns:

> Under which contexts should information be preserved or replaced?

This makes the gating mechanism **data-dependent and task-dependent**.

---

# 23. One-Step Numerical Backward Example

Suppose:

$$
a_{t-1}=0.8
$$

$$
\tilde a_t=-0.2
$$

$$
\Gamma_u=0.1
$$

and:

$$
\frac{\partial L}{\partial a_t}=1
$$

From:

$$
\frac{\partial L}{\partial\tilde a_t}
=
\delta_t\Gamma_u
$$

we get:

$$
\frac{\partial L}{\partial\tilde a_t}
=
1(0.1)
=
0.1
$$

For the update gate:

$$
\frac{\partial L}{\partial\Gamma_u}
=
1(-0.2-0.8)
$$

so:

$$
\boxed{
\frac{\partial L}{\partial\Gamma_u}
=
-1
}
$$

The direct path to the previous state is:

$$
\left.
\frac{\partial L}{\partial a_{t-1}}
\right|_{\text{direct}}
=
1(1-0.1)
$$

thus:

$$
\boxed{
\left.
\frac{\partial L}{\partial a_{t-1}}
\right|_{\text{direct}}
=
0.9
}
$$

The old state has therefore received a strong direct gradient signal.

# 24. GRU vs. Vanilla RNN

Vanilla RNN:

$$
a_t
=
\tanh
\left(
W_{aa}a_{t-1}
+
W_{ax}x_t
+
b_a
\right)
$$

GRU:

$$
a_t
=
\Gamma_u\odot\tilde a_t
+
(1-\Gamma_u)\odot a_{t-1}
$$

The essential conceptual difference is:

$$
\boxed{
\text{Vanilla RNN}
\rightarrow
\text{Always transform the previous state}
}
$$

while:

$$
\boxed{
\text{GRU}
\rightarrow
\text{Learn how much to preserve and how much to update}
}
$$

# 25. GRU and Long-Term Memory

Suppose an important piece of information appears at timestep $t$.

The model may decide that it should remain relevant for many future timesteps.

It can learn:

$$
\Gamma_u^{t+1}\approx0
$$

then:

$$
a_{t+1}\approx a_t
$$

and again:

$$
\Gamma_u^{t+2}\approx0
$$

so:

$$
a_{t+2}\approx a_{t+1}
$$

and therefore approximately:

$$
a_{t+2}\approx a_t
$$

This can continue across multiple timesteps.

Conceptually:

$$
\boxed{
a_t
\rightarrow
a_{t+1}
\rightarrow
a_{t+2}
\rightarrow
\cdots
}
$$

with the state being preserved whenever the model decides that the old information remains useful.

---

# 26. Important Precision: GRU Does Not Eliminate Vanishing Gradients Completely

It is important not to say:

> "GRU solves vanishing gradients."

A more precise statement is:

$$
\boxed{
\text{GRU reduces the difficulty of learning long-term dependencies}
}
$$

because it introduces gated paths that can preserve state and gradient more effectively.

The gradient can still vanish because:

- the gates themselves are nonlinear;
- the candidate path still contains $\tanh$;
- the gate values may not remain near the ideal regime.

Therefore:

$$
\boxed{
\text{GRU improves gradient flow}
\neq
\text{GRU guarantees no vanishing gradient}
}
$$

\boxed{
\Gamma_r^t
=

\right)
}
$$

### Candidate State

$$
\boxed{
\tilde a_t
=

\left(

=

+

# Long Short-Term Memory (LSTM)

The **Long Short-Term Memory (LSTM)** is a gated recurrent neural network architecture designed to better preserve information across long sequences.

A vanilla RNN repeatedly applies a nonlinear transformation:

$$
a_t
=
\tanh
\left(
W_{aa}a_{t-1}
+
W_{ax}x_t
+
b_a
\right)
$$

During BPTT, the gradient must pass through many recurrent transformations:

$$
\frac{\partial a_T}{\partial a_t}
=
J_TJ_{T-1}\cdots J_{t+1}
$$

This can lead to vanishing gradients and make long-term dependencies difficult to learn.

LSTM introduces a separate **cell state** and several **gates** that control how information is:

- forgotten;
- written into memory;
- exposed to the hidden state.

The central idea is:

$$
\boxed{
\text{LSTM}
=
\text{Cell State}
+
\text{Hidden State}
+
\text{Gated Information Flow}
}
$$

---

# 1. Why Do We Need LSTM?

Consider a sequence where an early piece of information becomes important much later.

For example:

> The keys on the table in the old room are broken.

To predict `are`, the model needs to preserve the fact that:

> `keys` is plural.

The words:

> on the table in the old room

appear between the subject and the verb.

A vanilla RNN must continuously transform its hidden state while carrying this information:

$$
a_1
\rightarrow
a_2
\rightarrow
\cdots
\rightarrow
a_T
$$

During BPTT, the corresponding learning signal must travel backward through the same sequence.

LSTM introduces a dedicated memory pathway:

$$
c_1
\rightarrow
c_2
\rightarrow
\cdots
\rightarrow
c_T
$$

The crucial difference is that the cell state can preserve information through a controlled additive update:

$$
\boxed{
c_t
=
\Gamma_f^t\odot c_{t-1}
+
\Gamma_u^t\odot\tilde c_t
}
$$

This allows the model to learn when to preserve old information instead of completely replacing it at every timestep.

---

# 2. Two States in LSTM

LSTM maintains two recurrent states.

## Cell State

$$
c_t
$$

The cell state is the main memory pathway.

Conceptually:

$$
\boxed{
c_t
=
\text{information that can be preserved over time}
}
$$

## Hidden State

$$
a_t
$$

The hidden state is the current output representation of the LSTM.

It is used to:

- produce predictions;
- pass information to the next timestep;
- expose selected information from the cell state.

Therefore:

$$
\boxed{
c_t
=
\text{memory pathway}
}
$$

while:

$$
\boxed{
a_t
=
\text{current hidden representation}
}
$$

The distinction between these two states is fundamental to understanding LSTM.

---

# 3. LSTM Architecture

![LSTM Architecture](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRB96rWEuYeSU_fleBiaMmK7M7QBHev3vM4VjumFrLQpPILMuNYz8mnMCew&s=10)

At timestep $t$, the LSTM receives:

$$
x_t
$$

and the previous states:

$$
a_{t-1},\qquad c_{t-1}
$$

It produces:

$$
a_t,\qquad c_t
$$

The information flow can be summarized as:

$$
\boxed{
(a_{t-1},c_{t-1},x_t)
\rightarrow
\text{Gates + Candidate}
\rightarrow
c_t
\rightarrow
a_t
}
$$

The main components are:

$$
\Gamma_f
\quad
\Gamma_u
\quad
\tilde c_t
\quad
\Gamma_o
$$

corresponding to:

- forget gate;
- update/input gate;
- candidate cell state;
- output gate.

---

# 4. Forget Gate

The forget gate determines how much of the previous cell state should be preserved.

$$
\boxed{
\Gamma_f^t
=
\sigma
\left(
W_f
[a_{t-1},x_t]
+
b_f
\right)
}
$$

Because the sigmoid function is used:

$$
0<\Gamma_f^t<1
$$

The gate acts element-wise on:

$$
c_{t-1}
$$

through:

$$
\Gamma_f^t\odot c_{t-1}
$$

### If

$$
\Gamma_f^t\approx1
$$

then:

$$
\Gamma_f^t\odot c_{t-1}
\approx
c_{t-1}
$$

Most old information is preserved.

### If

$$
\Gamma_f^t\approx0
$$

then:

$$
\Gamma_f^t\odot c_{t-1}
\approx0
$$

Most old information is forgotten.

Therefore:

$$
\boxed{
\Gamma_f
=
\text{How much old memory should survive?}
}
$$

---

# 5. Update Gate

The update gate determines how much new information should be written into the cell state.

$$
\boxed{
\Gamma_u^t
=
\sigma
\left(
W_u
[a_{t-1},x_t]
+
b_u
\right)
}
$$

Again:

$$
0<\Gamma_u^t<1
$$

The gate controls the amount of the candidate state that will enter the cell state:

$$
\Gamma_u^t\odot\tilde c_t
$$

### If

$$
\Gamma_u^t\approx1
$$

the candidate is strongly written.

### If

$$
\Gamma_u^t\approx0
$$

the candidate is mostly ignored.

Therefore:

$$
\boxed{
\Gamma_u
=
\text{How much new information should be written?}
}
$$

---

# 6. Candidate Cell State

The candidate cell state represents the new information that could be added to memory.

$$
\boxed{
\tilde c_t
=
\tanh
\left(
W_c
[a_{t-1},x_t]
+
b_c
\right)
}
$$

The candidate is not yet the final memory.

It is better understood as:

> "What new information could the memory contain at this timestep?"

The update gate decides how much of this candidate is actually written.

Thus:

$$
\boxed{
\tilde c_t
=
\text{candidate new memory}
}
$$

while:

$$
\boxed{
c_t
=
\text{final memory after gating}
}
$$

---

# 7. Cell State Update

The core LSTM memory equation is:

$$
\boxed{
c_t
=
\Gamma_f^t\odot c_{t-1}
+
\Gamma_u^t\odot\tilde c_t
}
$$

This equation combines two sources:

$$
\underbrace{
\Gamma_f^t\odot c_{t-1}
}_{\text{preserved old memory}}
+
\underbrace{
\Gamma_u^t\odot\tilde c_t
}_{\text{selected new memory}}
$$

Therefore:

$$
\boxed{
\text{New Memory}
=
\text{Kept Old Memory}
+
\text{Written New Information}
}
$$

This is the central mechanism of LSTM.

---

# 8. Output Gate

After updating the cell state, the LSTM determines which part of the memory should be exposed as the hidden state.

The output gate is:

$$
\boxed{
\Gamma_o^t
=
\sigma
\left(
W_o
[a_{t-1},x_t]
+
b_o
\right)
}
$$

The hidden state is then:

$$
\boxed{
a_t
=
\Gamma_o^t
\odot
\tanh(c_t)
}
$$

### If

$$
\Gamma_o^t\approx1
$$

more of the current memory is exposed.

### If

$$
\Gamma_o^t\approx0
$$

less of the memory is exposed.

Therefore:

$$
\boxed{
\Gamma_o
=
\text{How much memory should be exposed as the hidden state?}
}
$$

---

# 9. Complete Forward Pass

The complete LSTM forward computation at timestep $t$ is:

## Step 1 — Forget Gate

$$
\Gamma_f^t
=
\sigma
\left(
W_f[a_{t-1},x_t]+b_f
\right)
$$

## Step 2 — Update Gate

$$
\Gamma_u^t
=
\sigma
\left(
W_u[a_{t-1},x_t]+b_u
\right)
$$

## Step 3 — Candidate Cell State

$$
\tilde c_t
=
\tanh
\left(
W_c[a_{t-1},x_t]+b_c
\right)
$$

## Step 4 — Cell State Update

$$
\boxed{
c_t
=
\Gamma_f^t\odot c_{t-1}
+
\Gamma_u^t\odot\tilde c_t
}
$$

## Step 5 — Output Gate

$$
\Gamma_o^t
=
\sigma
\left(
W_o[a_{t-1},x_t]+b_o
\right)
$$

## Step 6 — Hidden State

$$
\boxed{
a_t
=
\Gamma_o^t\odot\tanh(c_t)
}
$$

The complete flow is:

$$
\boxed{
(a_{t-1},c_{t-1},x_t)
\rightarrow
\Gamma_f,\Gamma_u,\Gamma_o,\tilde c_t
\rightarrow
c_t
\rightarrow
a_t
}
$$

---

# 10. Tensor Shapes

Let:

$$
x_t\in\mathbb R^{n_x}
$$

and:

$$
a_{t-1},c_{t-1}\in\mathbb R^{n_a}
$$

The concatenated input is:

$$
[a_{t-1},x_t]
\in
\mathbb R^{n_a+n_x}
$$

Therefore:

$$
W_f,W_u,W_o,W_c
\in
\mathbb R^{n_a\times(n_a+n_x)}
$$

and:

$$
b_f,b_u,b_o,b_c
\in
\mathbb R^{n_a}
$$

The states and gates all have hidden dimension $n_a$:

$$
a_t,c_t,\tilde c_t
\in
\mathbb R^{n_a}
$$

$$
\Gamma_f,\Gamma_u,\Gamma_o
\in
\mathbb R^{n_a}
$$

---

# 11. Small Numerical Example

Consider a scalar LSTM.

Suppose:

$$
c_{t-1}=0.8
$$

The model produces:

$$
\Gamma_f=0.9
$$

$$
\Gamma_u=0.2
$$

and:

$$
\tilde c_t=0.5
$$

Then:

$$
c_t
=
0.9(0.8)
+
0.2(0.5)
$$

$$
=
0.72+0.10
$$

Therefore:

$$
\boxed{
c_t=0.82
}
$$

The previous memory contributed:

$$
0.72
$$

while the new candidate contributed:

$$
0.10
$$

So the LSTM mostly preserved the old memory while adding a small amount of new information.

---

# 12. When the LSTM Wants to Forget

Suppose:

$$
\Gamma_f=0
$$

Then:

$$
c_t
=
0\odot c_{t-1}
+
\Gamma_u\odot\tilde c_t
$$

Therefore:

$$
\boxed{
c_t
=
\Gamma_u\odot\tilde c_t
}
$$

The old memory is removed.

Thus:

$$
\boxed{
\Gamma_f\approx0
\rightarrow
\text{Forget old memory}
}
$$

---

# 13. When the LSTM Wants to Write New Information

Suppose:

$$
\Gamma_u=1
$$

Then:

$$
c_t
=
\Gamma_f\odot c_{t-1}
+
\tilde c_t
$$

The candidate is fully written into the memory.

Thus:

$$
\boxed{
\Gamma_u\approx1
\rightarrow
\text{Strongly write new information}
}
$$

---

# 14. When the LSTM Wants to Preserve Memory

Consider:

$$
\Gamma_f\approx1
$$

and:

$$
\Gamma_u\approx0
$$

Then:

$$
c_t
\approx
c_{t-1}
$$

The cell state is almost copied forward:

$$
c_{t-1}
\rightarrow
c_t
\rightarrow
c_{t+1}
\rightarrow
\cdots
$$

This is the key mechanism that allows information to remain relevant over many timesteps.

---

# 15. Subject–Verb Agreement Example

Consider:

> The keys on the table are broken.

The important information is:

> `keys` is plural.

Conceptually, the LSTM can behave like:

```text
"The keys"
     ↓
important grammatical information
     ↓
stored in cell state
     ↓
"on the table"
     ↓
forget gate preserves useful memory
     ↓
"are"
     ↓
hidden state exposes relevant information
     ↓
prediction
```

The important point is that LSTM does not receive a predefined rule:

> plural subject → plural verb

Instead, training gradually changes the parameters so that useful information for future prediction becomes represented in the hidden/cell states.

---

# 16. How Does LSTM Help the Gradient Flow?

This is the central mathematical motivation.

Recall:

$$
c_t
=
\Gamma_f^t\odot c_{t-1}
+
\Gamma_u^t\odot\tilde c_t
$$

Ignoring the dependence of the gates themselves on $c_{t-1}$ for the moment, the direct derivative through the memory path is:

$$
\boxed{
\frac{\partial c_t}{\partial c_{t-1}}
\approx
\Gamma_f^t
}
$$

Suppose:

$$
\Gamma_f^t\approx1
$$

Then:

$$
\frac{\partial c_t}{\partial c_{t-1}}
\approx1
$$

Across many timesteps:

$$
\frac{\partial c_T}{\partial c_t}
\approx
\prod_{k=t+1}^{T}
\Gamma_f^k
$$

If the forget gates remain close to $1$:

$$
\Gamma_f^k\approx1
$$

then:

$$
\prod_{k=t+1}^{T}
\Gamma_f^k
\approx1
$$

Therefore the gradient can travel through the cell-state pathway with much less attenuation than in a vanilla RNN.

The key idea is:

$$
\boxed{
\text{Forget Gate near 1}
\rightarrow
\text{Memory Path near Identity}
\rightarrow
\text{Better Gradient Flow}
}
$$

---

# 17. Why Is This Different from Vanilla RNN?

Vanilla RNN:

$$
a_t
=
\tanh(W_{aa}a_{t-1}+W_{ax}x_t+b_a)
$$

The state is repeatedly passed through:

$$
W_{aa}
$$

and:

$$
\tanh
$$

The corresponding gradient therefore involves repeated nonlinear transformations.

LSTM instead has:

$$
c_t
=
\Gamma_f c_{t-1}
+
\Gamma_u\tilde c_t
$$

If the model wants to preserve memory:

$$
\Gamma_f\approx1
$$

and:

$$
\Gamma_u\approx0
$$

giving:

$$
c_t\approx c_{t-1}
$$

Thus the cell state has an approximately identity-like path through time.

---

# 18. Backward Through the Cell-State Update

Let:

$$
\delta_{c_t}
=
\frac{\partial L}{\partial c_t}
$$

From:

$$
c_t
=
\Gamma_f c_{t-1}
+
\Gamma_u\tilde c_t
$$

we obtain several gradient branches.

## Gradient to Previous Cell State

The direct contribution is:

$$
\boxed{
\left.
\frac{\partial L}{\partial c_{t-1}}
\right|_{\text{direct}}
=
\delta_{c_t}
\odot
\Gamma_f
}
$$

This is the important memory-path gradient.

## Gradient to Forget Gate

$$
\boxed{
\frac{\partial L}{\partial\Gamma_f}
=
\delta_{c_t}
\odot
c_{t-1}
}
$$

## Gradient to Update Gate

$$
\boxed{
\frac{\partial L}{\partial\Gamma_u}
=
\delta_{c_t}
\odot
\tilde c_t
}
$$

## Gradient to Candidate

$$
\boxed{
\frac{\partial L}{\partial\tilde c_t}
=
\delta_{c_t}
\odot
\Gamma_u
}
$$

---

# 19. Backward Through the Candidate

The candidate is:

$$
\tilde c_t
=
\tanh(z_c)
$$

where:

$$
z_c
=
W_c[a_{t-1},x_t]+b_c
$$

Since:

$$
\delta_{\tilde c_t}
=
\delta_{c_t}\odot\Gamma_u
$$

the gradient through $\tanh$ is:

$$
\boxed{
\delta_{z_c}
=
\delta_{\tilde c_t}
\odot
(1-\tilde c_t^2)
}
$$

Then:

$$
\boxed{
\frac{\partial L}{\partial W_c}
=
\delta_{z_c}
[a_{t-1},x_t]^T
}
$$

and:

$$
\boxed{
\frac{\partial L}{\partial b_c}
=
\delta_{z_c}
}
$$

---

# 20. Backward Through the Forget Gate

The forget gate is:

$$
\Gamma_f
=
\sigma(z_f)
$$

where:

$$
z_f
=
W_f[a_{t-1},x_t]+b_f
$$

We have:

$$
\delta_{\Gamma_f}
=
\delta_{c_t}\odot c_{t-1}
$$

Through the sigmoid:

$$
\boxed{
\delta_{z_f}
=
\delta_{\Gamma_f}
\odot
\Gamma_f(1-\Gamma_f)
}
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial W_f}
=
\delta_{z_f}
[a_{t-1},x_t]^T
}
$$

and:

$$
\boxed{
\frac{\partial L}{\partial b_f}
=
\delta_{z_f}
}
$$

---

# 21. Backward Through the Update Gate

Similarly:

$$
\Gamma_u
=
\sigma(z_u)
$$

where:

$$
z_u
=
W_u[a_{t-1},x_t]+b_u
$$

We have:

$$
\delta_{\Gamma_u}
=
\delta_{c_t}
\odot
\tilde c_t
$$

Then:

$$
\boxed{
\delta_{z_u}
=
\delta_{\Gamma_u}
\odot
\Gamma_u(1-\Gamma_u)
}
$$

and:

$$
\boxed{
\frac{\partial L}{\partial W_u}
=
\delta_{z_u}
[a_{t-1},x_t]^T
}
$$

$$
\boxed{
\frac{\partial L}{\partial b_u}
=
\delta_{z_u}
}
$$

---

# 22. Backward Through the Output Gate

The hidden state is:

$$
a_t
=
\Gamma_o
\odot
\tanh(c_t)
$$

Let:

$$
\delta_{a_t}
=
\frac{\partial L}{\partial a_t}
$$

The gradient to the output gate is:

$$
\boxed{
\frac{\partial L}{\partial\Gamma_o}
=
\delta_{a_t}
\odot
\tanh(c_t)
}
$$

The gradient flowing from the output branch into the cell state is:

$$
\boxed{
\left.
\frac{\partial L}{\partial c_t}
\right|_{\text{output path}}
=
\delta_{a_t}
\odot
\Gamma_o
\odot
\left(
1-\tanh^2(c_t)
\right)
}
$$

This contribution is combined with any other gradient arriving at $c_t$.

---

# 23. Total Gradient Flow

The cell state receives gradient from the hidden-state output path.

Then that gradient can propagate backward through:

$$
c_t
\rightarrow
c_{t-1}
\rightarrow
c_{t-2}
\rightarrow
\cdots
$$

At each timestep, the direct cell-state pathway contains:

$$
\Gamma_f
$$

Therefore:

$$
\boxed{
\frac{\partial c_T}{\partial c_t}
\approx
\prod_{k=t+1}^{T}
\Gamma_f^k
}
$$

when focusing on the direct memory pathway.

This is the fundamental mathematical reason LSTM can preserve long-term information and provide a more stable gradient pathway.

---

# 24. Important Precision

LSTM does **not** guarantee that gradients never vanish.

The correct interpretation is:

$$
\boxed{
\text{LSTM provides a mechanism that can preserve information and gradient much better than vanilla RNN}
}
$$

The gradient can still be affected by:

- gate values;
- nonlinearities;
- other gradient paths.

Therefore:

$$
\boxed{
\text{Better Gradient Flow}
\neq
\text{No Vanishing Gradient at All}
}
$$

---

# 25. GRU vs. LSTM

The main distinction is structural.

### GRU

One main recurrent state:

$$
a_t
$$

with:

$$
\Gamma_u,\Gamma_r
$$

and:

$$
a_t
=
\Gamma_u\tilde a_t
+
(1-\Gamma_u)a_{t-1}
$$

### LSTM

Two states:

$$
a_t,\qquad c_t
$$

with:

$$
\Gamma_f,\Gamma_u,\Gamma_o
$$

and:

$$
c_t
=
\Gamma_f c_{t-1}
+
\Gamma_u\tilde c_t
$$

Thus:

$$
\boxed{
\text{GRU}
=
\text{Gated Hidden-State Update}
}
$$

while:

$$
\boxed{
\text{LSTM}
=
\text{Separate Memory Cell + Hidden State + Gates}
}
$$

---

# 26. The Three Questions of an LSTM

A useful way to interpret the whole cell is to ask three questions.

### 1. What should I forget?

$$
\Gamma_f
$$

Controls old memory.

### 2. What should I write?

$$
\Gamma_u
$$

Controls new information.

### 3. What should I expose?

$$
\Gamma_o
$$

Controls how much memory becomes the hidden state.

Therefore:

$$
\boxed{
\text{Forget}
\rightarrow
\text{Write}
\rightarrow
\text{Expose}
}
$$

---

# 27. The Central Memory Equation

The most important equation is:

$$
\boxed{
c_t
=
\underbrace{
\Gamma_f^t\odot c_{t-1}
}_{\text{keep old memory}}
+
\underbrace{
\Gamma_u^t\odot\tilde c_t
}_{\text{add new information}}
}
$$

If the model wants to preserve memory:

$$
\Gamma_f^t\approx1,
\qquad
\Gamma_u^t\approx0
$$

then:

$$
c_t\approx c_{t-1}
$$

This is the central mechanism behind LSTM's ability to maintain information over long periods.

---

# 28. Final Mental Model

The complete LSTM mechanism can be summarized as:

$$
\boxed{
(a_{t-1},x_t)
\rightarrow
\text{Forget Gate}
}
$$

$$
\boxed{
(a_{t-1},x_t)
\rightarrow
\text{Update Gate + Candidate}
}
$$

Then:

$$
\boxed{
c_{t-1}
\overset{\Gamma_f}{\longrightarrow}
\text{preserved memory}
}
$$

and:

$$
\boxed{
\tilde c_t
\overset{\Gamma_u}{\longrightarrow}
\text{new memory}
}
$$

which combine into:

$$
\boxed{
c_t
=
\Gamma_f c_{t-1}
+
\Gamma_u\tilde c_t
}
$$

Then:

$$
\boxed{
c_t
\overset{\Gamma_o}{\longrightarrow}
a_t
}
$$

The overall causal chain is:

$$
\boxed{
\text{Vanilla RNN}
\rightarrow
\text{Vanishing Gradient}
\rightarrow
\text{Need for Better Memory}
}
$$

$$
\downarrow
$$

$$
\boxed{
\text{Cell State}
+
\text{Forget Gate}
+
\text{Update Gate}
+
\text{Output Gate}
}
$$

$$
\downarrow
$$

$$
\boxed{
\text{Controlled Memory Flow}
}
$$

$$
\downarrow
$$

$$
\boxed{
\text{Better Long-Term Information and Gradient Flow}
}
$$

The essential idea is:

> **LSTM does not simply "remember more." It learns when to forget old memory, when to write new information, and what part of its memory should be exposed as the current hidden state.**

The mathematical core is:

$$
\boxed{
c_t
=
\Gamma_f^t\odot c_{t-1}
+
\Gamma_u^t\odot\tilde c_t
}
$$

and the key gradient insight is:

$$
\boxed{
\frac{\partial c_t}{\partial c_{t-1}}
\approx
\Gamma_f^t
}
$$

so when the forget gate stays near $1$, the cell-state pathway can behave approximately like an identity path through time.

# Bidirectional Recurrent Neural Network (Bidirectional RNN)

## 1. Motivation

A standard Recurrent Neural Network processes a sequence in one direction:

$$
x^{\langle1\rangle}
\rightarrow
x^{\langle2\rangle}
\rightarrow
\cdots
\rightarrow
x^{\langle T\rangle}
$$

At timestep \(t\), its hidden state is computed as:

$$
a^{\langle t\rangle}
=
f
\left(
a^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

Therefore, the representation at timestep \(t\) can use information from the **past**, but not from future timesteps.

However, many sequence tasks need context from both sides of the current position.

For example:

> The man went to the bank to deposit money.

To understand the meaning of **bank**, the words

> to deposit money

provide important information. This information occurs **after** the word `bank`.

A standard left-to-right RNN cannot use this future context when computing the representation of `bank`.

This motivates the Bidirectional RNN:

$$
\boxed{
\text{Past Context}
+
\text{Future Context}
}
$$

---

## 2. Core Idea

A Bidirectional RNN uses **two RNNs**:

1. A forward RNN that processes the sequence from left to right.
2. A backward RNN that processes the sequence from right to left.

The forward hidden state is:

$$
\overrightarrow{a}^{\langle t\rangle}
$$

and the backward hidden state is:

$$
\overleftarrow{a}^{\langle t\rangle}
$$

They are then combined:

$$
\boxed{
a^{\langle t\rangle}
=
\left[
\overrightarrow{a}^{\langle t\rangle};
\overleftarrow{a}^{\langle t\rangle}
\right]
}
$$

where \(;\) denotes concatenation.

Therefore, the representation at timestep \(t\) contains information from both directions.

The basic idea is:

$$
\boxed{
\text{Forward RNN}
+
\text{Backward RNN}
\rightarrow
\text{Bidirectional Representation}
}
$$

![Bidirectional RNN Architecture](https://media.geeksforgeeks.org/wp-content/uploads/20230512185249/Bidirectional-Recurrent-Neural-Network.webp)
---

## 3. Forward RNN

The forward RNN processes the sequence normally:

$$
x^{\langle1\rangle}
\rightarrow
x^{\langle2\rangle}
\rightarrow
\cdots
\rightarrow
x^{\langle T\rangle}
$$

Its hidden state is computed as:

$$
z_{\rightarrow}^{\langle t\rangle}
=
W_{a\rightarrow a}
\overrightarrow{a}^{\langle t-1\rangle}
+
W_{a\rightarrow x}
x^{\langle t\rangle}
+
b_{\rightarrow}
$$

$$
\overrightarrow{a}^{\langle t\rangle}
=
\tanh
\left(
z_{\rightarrow}^{\langle t\rangle}
\right)
$$

Therefore:

$$
\overrightarrow{a}^{\langle t\rangle}
=
f_{\rightarrow}
\left(
\overrightarrow{a}^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

The forward hidden state can use:

$$
x^{\langle1\rangle},
x^{\langle2\rangle},
\ldots,
x^{\langle t\rangle}
$$

So it represents the **left context** of the current timestep.

---

## 4. Backward RNN

The backward RNN processes the same sequence in the opposite direction:

$$
x^{\langle T\rangle}
\rightarrow
x^{\langle T-1\rangle}
\rightarrow
\cdots
\rightarrow
x^{\langle1\rangle}
$$

Its recurrence can be written as:

$$
z_{\leftarrow}^{\langle t\rangle}
=
W_{a\leftarrow a}
\overleftarrow{a}^{\langle t+1\rangle}
+
W_{a\leftarrow x}
x^{\langle t\rangle}
+
b_{\leftarrow}
$$

$$
\overleftarrow{a}^{\langle t\rangle}
=
\tanh
\left(
z_{\leftarrow}^{\langle t\rangle}
\right)
$$

Therefore:

$$
\overleftarrow{a}^{\langle t\rangle}
=
f_{\leftarrow}
\left(
\overleftarrow{a}^{\langle t+1\rangle},
x^{\langle t\rangle}
\right)
$$

The backward hidden state can use:

$$
x^{\langle t\rangle},
x^{\langle t+1\rangle},
\ldots,
x^{\langle T\rangle}
$$

So it represents the **right context** of the current timestep.

---

## 5. Combining the Two Directions

At timestep \(t\), we have:

$$
\overrightarrow{a}^{\langle t\rangle}
$$

and:

$$
\overleftarrow{a}^{\langle t\rangle}
$$

We concatenate them:

$$
\boxed{
a^{\langle t\rangle}
=
\left[
\overrightarrow{a}^{\langle t\rangle};
\overleftarrow{a}^{\langle t\rangle}
\right]
}
$$

Suppose each direction has hidden dimension \(n_a\):

$$
\overrightarrow{a}^{\langle t\rangle}
\in
\mathbb{R}^{n_a}
$$

$$
\overleftarrow{a}^{\langle t\rangle}
\in
\mathbb{R}^{n_a}
$$

Then the combined representation has dimension:

$$
a^{\langle t\rangle}
\in
\mathbb{R}^{2n_a}
$$

Thus:

$$
\boxed{
n_a^{BRNN}=2n_a
}
$$

when the two hidden states have the same dimension and are concatenated.

---

## 6. Output Layer

After combining the two directions, the output at timestep \(t\) can be computed as:

$$
z_y^{\langle t\rangle}
=
W_y a^{\langle t\rangle}
+
b_y
$$

and:

$$
\hat y^{\langle t\rangle}
=
g_y
\left(
z_y^{\langle t\rangle}
\right)
$$

Since:

$$
a^{\langle t\rangle}
\in
\mathbb{R}^{2n_a}
$$

the output weight matrix has shape:

$$
W_y
\in
\mathbb{R}^{n_y\times 2n_a}
$$

where \(n_y\) is the output dimension.

---

## 7. Information Flow

The essential information flow is:

$$
x^{\langle t\rangle}
\rightarrow
\begin{cases}
\overrightarrow{a}^{\langle t\rangle}
\\
\overleftarrow{a}^{\langle t\rangle}
\end{cases}
$$

then:

$$
\left[
\overrightarrow{a}^{\langle t\rangle};
\overleftarrow{a}^{\langle t\rangle}
\right]
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
$$

More conceptually:

$$
\boxed{
\text{Left Context}
+
\text{Current Input}
+
\text{Right Context}
\rightarrow
\text{Prediction}
}
$$

This is the fundamental advantage of a Bidirectional RNN.

---

## 8. What Information Does Each Hidden State Contain?

The forward hidden state satisfies:

$$
\overrightarrow{a}^{\langle t\rangle}
=
f
\left(
x^{\langle1\rangle},
\ldots,
x^{\langle t\rangle}
\right)
$$

conceptually.

The backward hidden state satisfies:

$$
\overleftarrow{a}^{\langle t\rangle}
=
f
\left(
x^{\langle t\rangle},
\ldots,
x^{\langle T\rangle}
\right)
$$

conceptually.

Therefore:

$$
a^{\langle t\rangle}
=
\left[
\overrightarrow{a}^{\langle t\rangle};
\overleftarrow{a}^{\langle t\rangle}
\right]
$$

can depend on the entire sequence.

This gives:

$$
\boxed{
a^{\langle t\rangle}
=
f
\left(
x^{\langle1\rangle},
\ldots,
x^{\langle T\rangle}
\right)
}
$$

in the sense that the representation at \(t\) can incorporate both left and right context.

---

## 9. Example: Context From Both Sides

Consider:

> The man went to the bank to deposit money.

Suppose we want to understand the word:

> bank

The forward RNN has processed:

$$
\text{The man went to the bank}
$$

so it knows the **left context**.

The backward RNN can process:

$$
\text{to deposit money}
$$

so it knows the **right context**.

Thus the representation for `bank` becomes:

$$
\boxed{
\text{bank representation}
=
\text{Left Context}
+
\text{Right Context}
}
$$

The right context strongly suggests the financial meaning of `bank`.

The important point is not the specific word `bank`, but the general mechanism:

> A token can be interpreted using information that occurs both before and after it.

---

## 10. Why Bidirectionality Is Useful

Bidirectional RNNs are especially useful when the meaning of the current element depends on its surrounding context.

Typical applications include:

- Named Entity Recognition (NER)
- Part-of-Speech tagging
- Sequence labeling
- Text classification at the token level
- Other offline sequence understanding tasks

For example:

$$
[\text{The},\text{students},\text{are},\text{studying}]
$$

A sequence-labeling model may produce:

$$
[\text{DET},\text{NOUN},\text{VERB},\text{VERB}]
$$

The prediction for each token can benefit from information on both sides.

---

## 11. Why Not Simply Use a Larger RNN?

Increasing the hidden size of a standard RNN changes its capacity, but it does not change the direction of information flow.

A standard RNN still has:

$$
x^{\langle1\rangle}
\rightarrow
x^{\langle2\rangle}
\rightarrow
x^{\langle3\rangle}
\rightarrow
\cdots
$$

Therefore, at timestep \(t\), it cannot use future inputs that have not yet been processed.

A Bidirectional RNN explicitly introduces the reverse information flow:

$$
x^{\langle T\rangle}
\rightarrow
x^{\langle T-1\rangle}
\rightarrow
\cdots
\rightarrow
x^{\langle1\rangle}
$$

Thus the key change is not simply model size.

It is:

$$
\boxed{
\text{Change in information flow}
}
$$

rather than:

$$
\boxed{
\text{Only increase model capacity}
}
$$

---

## 12. Parameter Sharing

A standard RNN uses the same parameters at every timestep.

The same principle applies within each direction of a BRNN.

For the forward RNN:

$$
W_{\rightarrow},b_{\rightarrow}
$$

are shared across all timesteps.

For the backward RNN:

$$
W_{\leftarrow},b_{\leftarrow}
$$

are shared across all timesteps.

Therefore:

$$
\boxed{
\text{Forward parameters are shared across time}
}
$$

and:

$$
\boxed{
\text{Backward parameters are shared across time}
}
$$

However, the forward and backward directions generally have **different parameter sets**.

So:

$$
W_{\rightarrow}
\neq
W_{\leftarrow}
$$

in general.

---

## 13. Tensor Shapes

Assume:

- Batch size: \(B\)
- Sequence length: \(T\)
- Input dimension: \(n_x\)
- Hidden size per direction: \(n_a\)
- Output dimension: \(n_y\)

Using the batch-first convention:

$$
X
\in
\mathbb{R}^{B\times T\times n_x}
$$

### Forward hidden states

$$
\overrightarrow A
\in
\mathbb{R}^{B\times T\times n_a}
$$

### Backward hidden states

$$
\overleftarrow A
\in
\mathbb{R}^{B\times T\times n_a}
$$

### Concatenated representation

$$
A
\in
\mathbb{R}^{B\times T\times 2n_a}
$$

### Output

$$
\hat Y
\in
\mathbb{R}^{B\times T\times n_y}
$$

Therefore:

$$
\boxed{
(B,T,n_x)
\rightarrow
(B,T,n_a)
}
$$

for the forward direction,

and:

$$
\boxed{
(B,T,n_x)
\rightarrow
(B,T,n_a)
}
$$

for the backward direction.

Then:

$$
\boxed{
(B,T,n_a)
\oplus
(B,T,n_a)
\rightarrow
(B,T,2n_a)
\rightarrow
(B,T,n_y)
}
$$

where \(\oplus\) denotes concatenation along the hidden dimension.

---

## 14. Sequence Length Does Not Double

A common misunderstanding is that because the model processes the sequence twice, the sequence length becomes \(2T\).

That is incorrect.

The sequence still contains:

$$
T
$$

timesteps.

What doubles is the representation size when the two hidden states are concatenated:

$$
n_a
\rightarrow
2n_a
$$

Therefore:

$$
\boxed{
T\text{ stays the same}
}
$$

while:

$$
\boxed{
n_a\text{ becomes }2n_a
}
$$

for equal-sized forward and backward hidden states.

---

## 15. Many-to-Many Architecture

Bidirectional RNNs are especially natural for many-to-many tasks:

$$
x^{\langle1\rangle},
\ldots,
x^{\langle T\rangle}
\rightarrow
\hat y^{\langle1\rangle},
\ldots,
\hat y^{\langle T\rangle}
$$

At every timestep:

$$
\hat y^{\langle t\rangle}
$$

is predicted using:

$$
\left[
\overrightarrow a^{\langle t\rangle};
\overleftarrow a^{\langle t\rangle}
\right]
$$

This makes BRNN well suited to tasks where each position must be interpreted using surrounding context.

---

## 16. Loss

The loss formulation itself does not fundamentally change.

For a many-to-many task:

$$
L
=
\sum_{t=1}^{T}
L^{\langle t\rangle}
$$

For classification at each timestep, for example:

$$
L^{\langle t\rangle}
=
-\sum_k
y_k^{\langle t\rangle}
\log
\hat y_k^{\langle t\rangle}
$$

Therefore:

$$
\boxed{
\text{BRNN changes the representation mechanism, not the basic loss principle}
}
$$

---

## 17. Backpropagation Through a Bidirectional RNN

Because the model contains two recurrent computation paths, gradients propagate through both directions.

At timestep \(t\):

$$
a^{\langle t\rangle}
=
\left[
\overrightarrow a^{\langle t\rangle};
\overleftarrow a^{\langle t\rangle}
\right]
$$

Therefore, the gradient with respect to the combined representation is split into two parts:

$$
\frac{\partial L}
{\partial a^{\langle t\rangle}}
\rightarrow
\begin{cases}
\frac{\partial L}
{\partial\overrightarrow a^{\langle t\rangle}}
\\[4pt]
\frac{\partial L}
{\partial\overleftarrow a^{\langle t\rangle}}
\end{cases}
$$

The first part propagates backward through the forward RNN.

The second part propagates backward through the backward RNN.

Thus:

$$
\boxed{
\text{BPTT through forward RNN}
+
\text{BPTT through backward RNN}
}
$$

The underlying principle remains the same as BPTT for the vanilla RNN: the unrolled recurrent graph is differentiated backward through time.

---

## 18. Does Bidirectional RNN Solve Vanishing Gradients?

No.

This distinction is important.

The primary purpose of a Bidirectional RNN is:

$$
\boxed{
\text{Access to future context}
}
$$

It is not:

$$
\boxed{
\text{Elimination of vanishing gradients}
}
$$

Each direction is still an RNN.

Therefore, each direction can still suffer from:

$$
\text{Vanishing Gradients}
$$

or:

$$
\text{Exploding Gradients}
$$

LSTM and GRU specifically introduce gated mechanisms to improve the propagation of information and gradients over long time intervals.

So:

$$
\boxed{
\text{BRNN} \neq \text{solution to vanishing gradients}
}
$$

---

## 19. The Streaming / Online Limitation

The biggest limitation of Bidirectional RNNs is their dependence on future information.

Suppose data arrives sequentially:

$$
x^{\langle1\rangle}
\rightarrow
x^{\langle2\rangle}
\rightarrow
x^{\langle3\rangle}
\rightarrow
\cdots
$$

At time \(t=3\), suppose we only know:

$$
x^{\langle1\rangle},
x^{\langle2\rangle},
x^{\langle3\rangle}
$$

The backward hidden state at \(t=3\) would require information such as:

$$
x^{\langle4\rangle},
x^{\langle5\rangle},
\ldots
$$

But those values do not exist yet.

Therefore, a full Bidirectional RNN cannot naturally compute the complete representation at \(t=3\) in a strict real-time setting.

This leads to:

$$
\boxed{
\text{BRNN is naturally an offline sequence model}
}
$$

rather than a strictly causal online model.

---

## 20. Offline vs. Online Processing

### Offline

The complete sequence is available:

$$
x^{\langle1\rangle},
x^{\langle2\rangle},
\ldots,
x^{\langle T\rangle}
$$

Therefore both directions can operate:

$$
\text{Past}
+
\text{Future}
\rightarrow
\text{Representation}
$$

This is ideal for Bidirectional RNNs.

### Online / Streaming

Only the prefix is available:

$$
x^{\langle1\rangle},
\ldots,
x^{\langle t\rangle}
$$

Future values are unknown.

Therefore:

$$
\text{Future Context}
=
\text{Unavailable}
$$

This makes a strict Bidirectional RNN unsuitable for fully causal streaming inference.

---

## 21. Example of the Streaming Problem

Suppose a speech signal arrives continuously.

At time \(t\), the system wants to produce an output immediately.

A forward RNN can process:

$$
x^{\langle t\rangle}
\rightarrow
\overrightarrow a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
$$

without waiting for future input.

A Bidirectional RNN would ideally also need:

$$
\overleftarrow a^{\langle t\rangle}
$$

which depends on future observations.

Therefore the system must either:

- wait for future input,
- use a limited look-ahead window,
- or use a causal architecture instead.

The core limitation remains:

$$
\boxed{
\text{Bidirectional processing requires future context}
}
$$

---

## 22. Bidirectional RNN vs. Vanilla RNN

| Aspect | Vanilla RNN | Bidirectional RNN |
|---|---|---|
| Number of directions | 1 | 2 |
| Information used | Past + current | Past + current + future |
| Hidden state | \(a_t\) | \([\overrightarrow a_t;\overleftarrow a_t]\) |
| Hidden size after concatenation | \(n_a\) | \(2n_a\) |
| Future context | No | Yes |
| Suitable for offline sequence labeling | Yes | Yes, often better |
| Strict streaming inference | Suitable | Not naturally suitable |
| Vanishing gradient | Possible | Still possible |
| Main purpose | Temporal recurrence | Bidirectional context |

---

## 23. Bidirectional RNN vs. Larger Hidden State

Increasing the hidden size:

$$
n_a\rightarrow2n_a
$$

does not create future information.

The model would still operate:

$$
x_1\rightarrow x_2\rightarrow\cdots\rightarrow x_T
$$

A Bidirectional RNN instead creates an additional information path:

$$
x_T\rightarrow x_{T-1}\rightarrow\cdots\rightarrow x_1
$$

Therefore:

$$
\boxed{
\text{Larger RNN}
\neq
\text{Bidirectional RNN}
}
$$

The difference is fundamentally about **direction of information flow**.

---

## 24. Complete Mathematical Picture

For the forward direction:

$$
\overrightarrow{a}^{\langle t\rangle}
=
f_{\rightarrow}
\left(
\overrightarrow{a}^{\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

For the backward direction:

$$
\overleftarrow{a}^{\langle t\rangle}
=
f_{\leftarrow}
\left(
\overleftarrow{a}^{\langle t+1\rangle},
x^{\langle t\rangle}
\right)
$$

Combine:

$$
a^{\langle t\rangle}
=
\left[
\overrightarrow{a}^{\langle t\rangle};
\overleftarrow{a}^{\langle t\rangle}
\right]
$$

Output:

$$
z_y^{\langle t\rangle}
=
W_y a^{\langle t\rangle}+b_y
$$

$$
\hat y^{\langle t\rangle}
=
g_y
\left(
z_y^{\langle t\rangle}
\right)
$$

Sequence loss:

$$
L
=
\sum_{t=1}^{T}
L^{\langle t\rangle}
$$

Therefore the entire architecture can be summarized as:

$$
\boxed{
\text{Input Sequence}
\rightarrow
\begin{cases}
\text{Forward RNN}\\
\text{Backward RNN}
\end{cases}
\rightarrow
\text{Concatenation}
\rightarrow
\text{Prediction}
}
$$

---

## 25. Comparison with LSTM and GRU

Bidirectionality and gated recurrence solve **different problems**.

A Bidirectional RNN changes:

$$
\boxed{
\text{Direction of information flow}
}
$$

LSTM and GRU change:

$$
\boxed{
\text{How information is stored and propagated through time}
}
$$

Therefore they can be combined.

For example, a Bidirectional LSTM contains:

$$
\boxed{
\text{Forward LSTM}
+
\text{Backward LSTM}
}
$$

Similarly, a Bidirectional GRU can contain:

$$
\boxed{
\text{Forward GRU}
+
\text{Backward GRU}
}
$$

So:

$$
\text{Bidirectional}
$$

is a property of the **architecture direction**, while:

$$
\text{LSTM / GRU}
$$

describes the **type of recurrent cell**.

---

## 26. Important Conceptual Distinction

There are two independent questions:

### Question 1: Which direction should information flow?

- One direction → Vanilla RNN
- Two directions → Bidirectional RNN

### Question 2: How should memory be maintained?

- Simple recurrence → Vanilla RNN
- Gated memory → LSTM / GRU

Therefore we can have:

$$
\boxed{
\text{Bidirectional RNN}
}
$$

$$
\boxed{
\text{Bidirectional LSTM}
}
$$

$$
\boxed{
\text{Bidirectional GRU}
}
$$

These are not contradictory concepts.

---

The complete causal chain is:

$$

}
$$

↓

$$

}
$$

↓

}
$$

}

\boxed{
\text{Future context requires access to the full sequence}
}
$$

Therefore:

$$
\boxed{
\text{Bidirectional RNN}
=
\text{Past Context}

\text{Future Context}
}
$$

but:

\boxed{
\text{Bidirectional RNN}
\text{ is not naturally causal for strict streaming inference}
}
$$

# Deep Recurrent Neural Networks (Deep RNNs)

## 1. Motivation

A standard Recurrent Neural Network processes a sequence using a single recurrent layer:

$$
x^{\langle t\rangle}
\rightarrow
a^{\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
$$

A single recurrent layer must simultaneously:

- process the input;
- maintain temporal information;
- build a representation;
- support the final prediction.

A natural way to increase the representational capacity is to **stack multiple recurrent layers**:

$$
\boxed{
\text{Input}
\rightarrow
\text{RNN Layer 1}
\rightarrow
\text{RNN Layer 2}
\rightarrow
\text{RNN Layer 3}
\rightarrow
\text{Output}
}
$$

The main idea is therefore:

$$
\boxed{
\text{Deep RNN}
=
\text{Multiple Recurrent Layers}
}
$$

Each layer can learn a different level of representation while preserving temporal recurrence.

---

## 2. Deep RNN vs. Vanilla RNN

A Vanilla RNN can be represented as:

$$
x^{\langle t\rangle}
\rightarrow
a^{[1]\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
$$

A Deep RNN can be represented as:

$$
x^{\langle t\rangle}
\rightarrow
a^{[1]\langle t\rangle}
\rightarrow
a^{[2]\langle t\rangle}
\rightarrow
a^{[3]\langle t\rangle}
\rightarrow
\hat y^{\langle t\rangle}
$$

Here:

$$
[1],[2],[3]
$$

denote **network layers**, while:

$$
\langle t\rangle
$$

denotes the **timestep**.

Therefore:

$$
\boxed{
[\,l\,]=\text{layer index}
}
$$

and:

$$
\boxed{
\langle t\rangle=\text{time index}
}
$$

This distinction is fundamental.

---

## 3. Deep RNN Architecture

Consider a Deep RNN with three recurrent layers.

At timestep \(t\):

### Layer 1

$$
a^{[1]\langle t\rangle}
=
f^{[1]}
\left(
a^{[1]\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

### Layer 2

$$
a^{[2]\langle t\rangle}
=
f^{[2]}
\left(
a^{[2]\langle t-1\rangle},
a^{[1]\langle t\rangle}
\right)
$$

### Layer 3

$$
a^{[3]\langle t\rangle}
=
f^{[3]}
\left(
a^{[3]\langle t-1\rangle},
a^{[2]\langle t\rangle}
\right)
$$

Finally:

$$
\hat y^{\langle t\rangle}
=
g_y
\left(
a^{[3]\langle t\rangle}
\right)
$$

Thus, at a single timestep:

$$
\boxed{
x_t
\rightarrow
a_t^{[1]}
\rightarrow
a_t^{[2]}
\rightarrow
a_t^{[3]}
\rightarrow
\hat y_t
}
$$

while every recurrent layer also carries information through time.

---

## 4. Deep RNN Architecture Diagram

![Deep RNN Architecture](https://tiefenauer.github.io/assets/img/articles/ml/dl_5/deep-rnn.png)

The important structure to observe is that the network has two dimensions:

$$
\boxed{
\text{Time}
\times
\text{Depth}
}
$$

Horizontal connections represent temporal recurrence, while vertical connections represent stacked recurrent layers.

---

## 5. Depth vs. Sequence Length

This is one of the most important concepts in Deep RNNs.

Suppose:

$$
T=10
$$

and the network has:

$$
L=3
$$

recurrent layers.

Then:

- sequence length = \(10\);
- network depth = \(3\).

The model does **not** have 30 timesteps.

Instead, each of the 3 layers processes the same 10 timesteps.

Therefore:

$$
\boxed{
T=\text{Sequence Length}
}
$$

and:

$$
\boxed{
L=\text{Network Depth}
}
$$

These are independent quantities.

---

## 6. Two Types of Connections

Deep RNNs contain two fundamental types of dependencies.

### Temporal dependency

Within the same layer:

$$
a^{[l]\langle t-1\rangle}
\rightarrow
a^{[l]\langle t\rangle}
$$

This is the standard recurrent connection.

It allows the network to maintain memory across time.

### Depth dependency

Between layers at the same timestep:

$$
a^{[l-1]\langle t\rangle}
\rightarrow
a^{[l]\langle t\rangle}
$$

This allows higher layers to process representations produced by lower layers.

Therefore:

$$
\boxed{
\text{Deep RNN}
=
\text{Temporal Recurrence}
+
\text{Depth}
}
$$

---

## 7. General Deep RNN Equation

For a generic recurrent layer \(l\):

$$
\boxed{
a^{[l]\langle t\rangle}
=
f^{[l]}
\left(
a^{[l]\langle t-1\rangle},
a^{[l-1]\langle t\rangle}
\right)
}
$$

with the input layer defined as:

$$
a^{[0]\langle t\rangle}
=
x^{\langle t\rangle}
$$

Therefore:

$$
a^{[1]\langle t\rangle}
=
f^{[1]}
\left(
a^{[1]\langle t-1\rangle},
x^{\langle t\rangle}
\right)
$$

and for \(l>1\):

$$
a^{[l]\langle t\rangle}
=
f^{[l]}
\left(
a^{[l]\langle t-1\rangle},
a^{[l-1]\langle t\rangle}
\right)
$$

This is the central mathematical structure of a Deep RNN.

---

## 8. Forward Propagation

Consider a vanilla tanh RNN.

### Layer 1

$$
z^{[1]\langle t\rangle}
=
W_{aa}^{[1]}
a^{[1]\langle t-1\rangle}
+
W_{ax}^{[1]}
x^{\langle t\rangle}
+
b_a^{[1]}
$$

$$
a^{[1]\langle t\rangle}
=
\tanh
\left(
z^{[1]\langle t\rangle}
\right)
$$

### Layer 2

$$
z^{[2]\langle t\rangle}
=
W_{aa}^{[2]}
a^{[2]\langle t-1\rangle}
+
W_{a1}^{[2]}
a^{[1]\langle t\rangle}
+
b_a^{[2]}
$$

$$
a^{[2]\langle t\rangle}
=
\tanh
\left(
z^{[2]\langle t\rangle}
\right)
$$

### Layer 3

$$
z^{[3]\langle t\rangle}
=
W_{aa}^{[3]}
a^{[3]\langle t-1\rangle}
+
W_{a2}^{[3]}
a^{[2]\langle t\rangle}
+
b_a^{[3]}
$$

$$
a^{[3]\langle t\rangle}
=
\tanh
\left(
z^{[3]\langle t\rangle}
\right)
$$

Then:

$$
\hat y^{\langle t\rangle}
=
g_y
\left(
W_y a^{[3]\langle t\rangle}
+
b_y
\right)
$$

---

## 9. Understanding One Timestep

At timestep \(t\), the information flows vertically:

$$
x_t
\rightarrow
a_t^{[1]}
\rightarrow
a_t^{[2]}
\rightarrow
a_t^{[3]}
\rightarrow
\hat y_t
$$

However, each layer also receives its own previous hidden state:

$$
a_{t-1}^{[1]}
\rightarrow
a_t^{[1]}
$$

$$
a_{t-1}^{[2]}
\rightarrow
a_t^{[2]}
$$

$$
a_{t-1}^{[3]}
\rightarrow
a_t^{[3]}
$$

Therefore:

$$
\boxed{
a_t^{[l]}
\text{ depends on both time and depth}
}
$$

More explicitly:

$$
a_t^{[l]}
=
f
\left(
a_{t-1}^{[l]},
a_t^{[l-1]}
\right)
$$

---

## 10. Hierarchical Representation

The motivation for stacking recurrent layers is similar to deep networks in general.

Lower layers can learn relatively basic patterns, while higher layers can combine those patterns into more abstract representations.

Conceptually:

$$
\boxed{
\text{Low-level temporal patterns}
}
$$

↓

$$
\boxed{
\text{Higher-level temporal patterns}
}
$$

↓

$$
\boxed{
\text{More abstract sequence representation}
}
$$

For example, in a speech-related task, one might intuitively think of the layers as learning progressively richer temporal representations.

This is an intuition rather than a strict rule: the model is not explicitly told what each layer must represent.

The important principle is:

$$
\boxed{
\text{Depth allows hierarchical representation learning}
}
$$

---

## 11. Tensor Shapes

Assume:

- batch size: \(B\)
- sequence length: \(T\)
- input dimension: \(n_x\)
- hidden size of layer 1: \(n_1\)
- hidden size of layer 2: \(n_2\)
- hidden size of layer 3: \(n_3\)
- output dimension: \(n_y\)

Using the batch-first convention:

$$
X
\in
\mathbb R^{B\times T\times n_x}
$$

Layer 1 output:

$$
A^{[1]}
\in
\mathbb R^{B\times T\times n_1}
$$

Layer 2 output:

$$
A^{[2]}
\in
\mathbb R^{B\times T\times n_2}
$$

Layer 3 output:

$$
A^{[3]}
\in
\mathbb R^{B\times T\times n_3}
$$

Final output:

$$
\hat Y
\in
\mathbb R^{B\times T\times n_y}
$$

Therefore:

$$
\boxed{
(B,T,n_x)
\rightarrow
(B,T,n_1)
\rightarrow
(B,T,n_2)
\rightarrow
(B,T,n_3)
\rightarrow
(B,T,n_y)
}
$$

The sequence length \(T\) remains unchanged across the stacked layers.

## 12. Weight Shapes

For layer 1:

$$
W_{ax}^{[1]}
\in
\mathbb R^{n_1\times n_x}
$$

$$
W_{aa}^{[1]}
\in
\mathbb R^{n_1\times n_1}
$$

For layer 2:

$$
W_{a1}^{[2]}
\in
\mathbb R^{n_2\times n_1}
$$

$$
W_{aa}^{[2]}
\in
\mathbb R^{n_2\times n_2}
$$

For layer 3:

$$
W_{a2}^{[3]}
\in
\mathbb R^{n_3\times n_2}
$$

$$
W_{aa}^{[3]}
\in
\mathbb R^{n_3\times n_3}
$$

Biases:

$$
b_a^{[l]}
\in
\mathbb R^{n_l}
$$

The important idea is:

$$
\boxed{
\text{Each layer has its own recurrent parameters}
}
$$

and:

$$
\boxed{
\text{Parameters are shared across timesteps within that layer}
}
$$

---

## 13. Parameter Sharing

Suppose layer \(2\) has parameters:

$$
W_{aa}^{[2]},
W_{a1}^{[2]},
b_a^{[2]}
$$

These parameters are reused at every timestep:

$$
t=1,2,\ldots,T
$$

Thus:

$$
W_{aa}^{[2]}
\text{ at }t=1
=
W_{aa}^{[2]}
\text{ at }t=2
=
\cdots
$$

This is temporal parameter sharing.

However, different layers have different parameter sets:

$$
W^{[1]}
\neq
W^{[2]}
\neq
W^{[3]}
$$

in general.

Therefore:

$$
\boxed{
\text{Share parameters across time}
}
$$

but:

$$
\boxed{
\text{Do not share parameters across depth}
}
$$

unless explicitly designed that way.

## 14. Loss

For a many-to-many task:

$$
\hat y^{\langle1\rangle},
\hat y^{\langle2\rangle},
\ldots,
\hat y^{\langle T\rangle}
$$

may each have a target:

$$
y^{\langle1\rangle},
y^{\langle2\rangle},
\ldots,
y^{\langle T\rangle}
$$

The total loss can be:

$$
\boxed{
L
=
\sum_{t=1}^{T}
L^{\langle t\rangle}
}
$$

For example, with cross-entropy:

$$
L^{\langle t\rangle}
=
-\sum_k
y_k^{\langle t\rangle}
\log
\hat y_k^{\langle t\rangle}
$$

The basic loss principle remains unchanged.

The important difference is that the prediction depends on the representation produced by multiple recurrent layers.

---

## 15. Backpropagation in a Deep RNN

A Vanilla RNN has a recurrent dependency through time:

$$
a_{t-1}
\rightarrow
a_t
\rightarrow
a_{t+1}
$$

A Deep RNN has both:

$$
\boxed{
\text{Temporal dependency}
}
$$

and:

$$
\boxed{
\text{Depth dependency}
}
$$

Therefore, backpropagation must propagate through both dimensions of the computation graph.

Conceptually:

$$
\boxed{
\text{Deep RNN Backpropagation}
=
\text{Backpropagation Through Time}
+
\text{Backpropagation Through Layers}
}
$$

---

## 16. Gradient Flow Through Depth

Suppose the network has three layers and we have a loss at timestep \(t\):

$$
L_t
$$

The gradient first reaches the top layer:

$$
L_t
\rightarrow
a_t^{[3]}
$$

Then it propagates downward:

$$
a_t^{[3]}
\rightarrow
a_t^{[2]}
\rightarrow
a_t^{[1]}
$$

This is the ordinary deep-network backpropagation dimension.

---

## 17. Gradient Flow Through Time

At the same time, every recurrent layer has temporal dependencies.

For example:

$$
a_t^{[2]}
\rightarrow
a_{t-1}^{[2]}
\rightarrow
a_{t-2}^{[2]}
\rightarrow
\cdots
$$

Thus gradient can also propagate backward through time.

Therefore the gradient has to navigate a graph with both:

$$
\boxed{
\text{vertical paths}
}
$$

and:

$$
\boxed{
\text{horizontal recurrent paths}
}
$$

---

## 18. Multiple Gradient Contributions

Consider:

$$
a_t^{[2]}
$$

This hidden state can influence:

1. the next timestep of layer 2;
2. the upper layer 3 at the same timestep.

Therefore:

$$
\frac{\partial L}{\partial a_t^{[2]}}
$$

can receive contributions from both paths.

Conceptually:

$$
\boxed{
\frac{\partial L}{\partial a_t^{[l]}}
=
\text{gradient from upper layer}
+
\text{gradient from future timestep}
}
$$

plus any local contribution when applicable.

This is the same computational-graph principle used throughout backpropagation: gradients from multiple paths are summed.

---

## 19. Backpropagation Example

Consider a 3-layer RNN at timestep \(t\):

$$
x_t
\rightarrow
a_t^{[1]}
\rightarrow
a_t^{[2]}
\rightarrow
a_t^{[3]}
\rightarrow
L_t
$$

The gradient from the loss first goes:

$$
L_t
\rightarrow
a_t^{[3]}
\rightarrow
a_t^{[2]}
\rightarrow
a_t^{[1]}
$$

But because each layer is recurrent, there are also temporal paths:

$$
a_t^{[3]}
\rightarrow
a_{t-1}^{[3]}
$$

$$
a_t^{[2]}
\rightarrow
a_{t-1}^{[2]}
$$

$$
a_t^{[1]}
\rightarrow
a_{t-1}^{[1]}
$$

Hence:

$$
\boxed{
\text{Backpropagation}
\text{ propagates both downward and backward through time}
}
$$

---

## 20. Vanishing Gradients in Deep RNNs

Vanishing gradients are already a problem for ordinary RNNs because gradients can shrink through many timesteps.

For example:

$$
\frac{\partial a_T}{\partial a_t}
=
\prod_{k=t+1}^{T}
\frac{\partial a_k}{\partial a_{k-1}}
$$

Deep RNNs add another source of repeated transformations.

The gradient may also have to pass through:

$$
a^{[L]}
\rightarrow
a^{[L-1]}
\rightarrow
\cdots
\rightarrow
a^{[1]}
$$

Thus a deep and long sequence can create a particularly long gradient path.

Conceptually:

$$
\boxed{
\text{Long sequence}
+
\text{Large depth}
\rightarrow
\text{Longer gradient paths}
}
$$

This can make optimization more difficult.

---

## 21. Deep RNN Does Not Solve Vanishing Gradients

The purpose of adding depth is mainly:

$$
\boxed{
\text{Richer hierarchical representations}
}
$$

not:

$$
\boxed{
\text{Eliminate vanishing gradients}
}
$$

A Deep Vanilla RNN can still experience:

$$
\text{Vanishing Gradients}
$$

and:

$$
\text{Exploding Gradients}
$$

This is one reason gated recurrent architectures such as LSTM and GRU are important.

---

## 22. Deep RNN vs. Long Sequence

These two concepts must not be confused.

### Case A

$$
L=1,\qquad T=100
$$

This is a shallow RNN with a long sequence.

### Case B

$$
L=5,\qquad T=20
$$

This is a deep RNN with a shorter sequence.

The two quantities describe different structural properties:

$$
\boxed{
L=\text{depth}
}
$$

$$
\boxed{
T=\text{sequence length}
}
$$

Neither one determines the other.

---

## 23. Deep RNN vs. Bidirectional RNN

These are two different architectural ideas.

### Deep RNN

Changes:

$$
\boxed{
\text{Depth}
}
$$

by stacking:

$$
\text{RNN}_1
\rightarrow
\text{RNN}_2
\rightarrow
\text{RNN}_3
$$

### Bidirectional RNN

Changes:

$$
\boxed{
\text{Direction of information flow}
}
$$

by using:

$$
\text{Forward RNN}
+
\text{Backward RNN}
$$

Therefore:

$$
\boxed{
\text{Deep}
\neq
\text{Bidirectional}
}
$$

They are independent architectural properties.

---

## 24. Deep + Bidirectional

The two ideas can be combined.

For example:

$$
\boxed{
\text{Bidirectional Deep RNN}
}
$$

can contain multiple layers, where each layer has forward and backward recurrent paths.

Conceptually:

$$
\text{Layer 1}
=
\text{Forward RNN}
+
\text{Backward RNN}
$$

↓

$$
\text{Layer 2}
=
\text{Forward RNN}
+
\text{Backward RNN}
$$

and so on.

Therefore:

$$
\boxed{
\text{Bidirectionality}
\text{ controls direction}
}
$$

while:

$$
\boxed{
\text{Depth}
\text{ controls number of stacked layers}
}
$$

---

## 25. Deep RNN + LSTM / GRU

Depth is also independent of the recurrent cell type.

We can stack:

$$
\text{LSTM}_1
\rightarrow
\text{LSTM}_2
\rightarrow
\text{LSTM}_3
$$

or:

$$
\text{GRU}_1
\rightarrow
\text{GRU}_2
\rightarrow
\text{GRU}_3
$$

Thus:

$$
\boxed{
\text{Deep}
=
\text{architecture depth}
}
$$

while:

$$
\boxed{
\text{RNN / LSTM / GRU}
=
\text{recurrent cell type}
}
$$

---

## 26. Connection to CNNs

In CNNs, we often stack layers:

$$
X
\rightarrow
\text{Conv}_1
\rightarrow
\text{Conv}_2
\rightarrow
\text{Conv}_3
$$

Higher layers learn richer representations from lower-layer representations.

Deep RNNs follow a similar hierarchical principle:

$$
X
\rightarrow
\text{RNN}_1
\rightarrow
\text{RNN}_2
\rightarrow
\text{RNN}_3
$$

The difference is that each RNN layer also contains temporal recurrence.

Therefore:

$$
\boxed{
\text{Deep CNN}
=
\text{Depth}
+
\text{Spatial Structure}
}
$$

while:

$$
\boxed{
\text{Deep RNN}
=
\text{Depth}
+
\text{Temporal Recurrence}
}
$$

Here:

```python
num_layers=3
```

\boxed{
\text{3 recurrent layers}
}
$$

It does **not** mean:

$$
\boxed{
\text{sequence length}=3
}
$$

For example:

```python
x.shape = (32, 100, 128)
```

means:

```python
num_layers=3
```

defines the network depth independently.

## 29. Complete Training Process

The training pipeline of a Deep RNN follows the same fundamental learning process as other neural networks.

### Forward pass

For every timestep:

$$
x_t
\rightarrow
a_t^{[1]}
\rightarrow
a_t^{[2]}
\rightarrow
\cdots
\rightarrow
a_t^{[L]}
\rightarrow
\hat y_t
$$

### Loss

$$
L
=
\sum_t L_t
$$

### Backward pass

Gradient propagates:

1. through the stacked layers;
2. through the recurrent timesteps.

Conceptually:

$$
\boxed{
\text{Backpropagation Through Depth}
+
\text{Backpropagation Through Time}
}
$$

### Parameter update

Finally:

$$
\theta
\leftarrow
\theta
-
\alpha
\nabla_\theta L
$$

The optimization rule itself remains the same basic gradient-descent principle.
