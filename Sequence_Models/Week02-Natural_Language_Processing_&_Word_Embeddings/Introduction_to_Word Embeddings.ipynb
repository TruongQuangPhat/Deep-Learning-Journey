# Word Representation

## 1. Motivation: Why Do We Need Word Representation?

Neural networks cannot directly operate on discrete words such as `"king"`, `"queen"`, or `"coffee"`. We first need to map each word into a numerical representation.

Let the vocabulary be:

$$
\mathcal{V} = \{w_1, w_2, \ldots, w_V\}
$$

where $V = |\mathcal{V}|$ is the vocabulary size.

The fundamental problem is therefore:

$$
\boxed{
\text{Word}
\rightarrow
\text{Numerical Representation}
}
$$

A simple solution is **one-hot representation**, but one-hot mainly represents the identity of a word. It does not naturally encode relationships between words. This motivates **word embeddings**, where each word is represented by a dense vector in a lower-dimensional continuous space.

---

## 2. One-Hot Representation

For a vocabulary of size $V$, each word can be represented by a one-hot vector:

$$
o_i \in \mathbb{R}^{V}
$$

where exactly one element is $1$, corresponding to the position of word $w_i$, and all other elements are $0$.

For example, suppose:

$$
\mathcal{V}
=
\{
\text{I},
\text{love},
\text{deep},
\text{learning},
\text{NLP}
\}
$$

Then:

$$
o_{\text{I}}
=
\begin{bmatrix}
1\\
0\\
0\\
0\\
0
\end{bmatrix}
$$

while:

$$
o_{\text{deep}}
=
\begin{bmatrix}
0\\
0\\
1\\
0\\
0
\end{bmatrix}
$$

Thus one-hot answers:

> **Which word is this?**

However, it does not naturally express:

> **How is this word related to other words?**

For example, `king` and `queen` are semantically related, but their one-hot vectors are simply two different basis vectors:

$$
o_{\text{king}} \neq o_{\text{queen}}
$$

The representation itself gives no useful notion of semantic similarity.

Therefore:

$$
\boxed{
\text{One-hot}
=
\text{Identity Representation}
}
$$

rather than:

$$
\boxed{
\text{One-hot}
=
\text{Semantic Representation}
}
$$

This limitation motivates the use of word embeddings.

---

## 3. Word Embedding

Instead of representing a word in the $V$-dimensional vocabulary space, we map it to a much lower-dimensional dense vector:

$$
e_i \in \mathbb{R}^{d}
$$

where typically:

$$
d \ll V
$$

For example:

$$
V = 50{,}000,
\qquad
d = 300
$$

Then instead of representing a word using a 50,000-dimensional one-hot vector, we use a 300-dimensional dense vector.

Conceptually:

$$
\boxed{
\text{Word}
\rightarrow
\text{Dense Vector}
}
$$

The important point is that the embedding is **learned**.

The coordinates of an embedding vector do not necessarily correspond to explicit human-interpretable concepts such as:

* dimension 1 = human;
* dimension 2 = animal;
* dimension 3 = positive.

Instead, the entire vector represents a word as a point in a learned $d$-dimensional space.

For example:

$$
e_{\text{king}},
e_{\text{queen}},
e_{\text{man}},
e_{\text{woman}}
\in
\mathbb{R}^{d}
$$

Their locations can capture useful relationships.

Therefore:

$$
\boxed{
\text{Embedding}
=
\text{Learned Dense Representation}
}
$$

---

## 4. Embedding Matrix and Embedding Lookup

All word embeddings can be stored in a single embedding matrix:

$$
\boxed{
E \in \mathbb{R}^{d \times V}
}
$$

where each column corresponds to one vocabulary item:

$$
E
=
\begin{bmatrix}
| & | & & |\\
e_1 & e_2 & \cdots & e_V\\
| & | & & |
\end{bmatrix}
$$

Thus:

$$
e_i \in \mathbb{R}^{d}
$$

is the embedding vector of word $w_i$.

Now consider the one-hot vector:

$$
o_i \in \mathbb{R}^{V}
$$

The corresponding embedding can be obtained by:

$$
\boxed{
e_i = Eo_i
}
$$

The dimensions are:

$$
(d \times V)(V \times 1)
=
(d \times 1)
$$

so:

$$
e_i \in \mathbb{R}^{d}
$$

### Why does this multiplication perform a lookup?

Suppose:

$$
E
=
\begin{bmatrix}
| & | & |\\
e_1 & e_2 & e_3\\
| & | & |
\end{bmatrix}
$$

and word $w_2$ has one-hot representation:

$$
o_2
=
\begin{bmatrix}
0\\
1\\
0
\end{bmatrix}
$$

Then:

$$
Eo_2
=
0e_1 + 1e_2 + 0e_3
=
e_2
$$

Therefore, multiplying by a one-hot vector simply selects one column of $E$.

This operation is called an **embedding lookup**.

In practice, deep learning frameworks normally implement this lookup directly instead of explicitly constructing a one-hot vector and multiplying by $E$. The mathematical formulation

$$
e_i = Eo_i
$$

is nevertheless useful for understanding the mechanism.

---

## 5. Why Can Embeddings Capture Semantic Relationships?

The key question is not only:

> What is an embedding?

but also:

> **Why can the embedding space contain semantic information?**

The embedding matrix is not manually filled with semantic knowledge. It is a set of trainable parameters learned through the task objective.

The overall mechanism can be viewed as:

$$
\boxed{
\text{Context}
\rightarrow
\text{Prediction Objective}
\rightarrow
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\text{Embedding Update}
}
$$

Suppose two words frequently appear in similar contexts:

> I drink **coffee**.

> I drink **tea**.

During training, the model repeatedly encounters contexts in which these words play similar roles.

Because the training objective generates related learning signals, the embeddings of these words can receive related updates.

Over many training examples:

$$
\boxed{
\text{Similar Contexts}
\rightarrow
\text{Related Updates}
\rightarrow
\text{Similar Embeddings}
}
$$

Therefore, semantic structure can **emerge from the training objective** rather than being explicitly programmed into the embedding matrix.

A useful causal chain is:

$$
\boxed{
\text{Language Patterns}
\rightarrow
\text{Training Signal}
\rightarrow
\text{Parameter Updates}
\rightarrow
\text{Embedding Geometry}
}
$$

The important point is that semantic structure is learned from data.

---

## 6. Similarity in the Embedding Space

Once words are represented as vectors, we can compare them geometrically.

A common similarity measure is **cosine similarity**:

$$
\boxed{
\operatorname{sim}(u,v)
=
\frac{u^\top v}
{\|u\|\,\|v\|}
}
$$

where:

* $u^\top v$ is the dot product;
* $|u|$ and $|v|$ are the vector norms.

Cosine similarity measures how similar the directions of two vectors are.

If:

$$
\operatorname{sim}(u,v) \approx 1
$$

then the vectors point in very similar directions.

In a useful embedding space, words with related meanings or usage patterns may therefore have similar vectors.

A classic conceptual example is:

$$
\boxed{
e_{\text{king}}
-
e_{\text{man}}
+
e_{\text{woman}}
\approx
e_{\text{queen}}
}
$$

This should be understood as an illustration of how relationships can be represented geometrically, rather than as an exact universal law for every embedding model.

The deeper idea is:

$$
\boxed{
\text{Embedding Space Can Encode Relationships}
}
$$

rather than simply encoding word identities.

---

## 7. Embeddings Are Trainable Parameters

The embedding matrix:

$$
E \in \mathbb{R}^{d \times V}
$$

is part of the model's trainable parameters.

During training, it can be updated using gradient descent:

$$
\boxed{
E
\leftarrow
E
-
\alpha
\frac{\partial L}{\partial E}
}
$$

where:

* $L$ is the loss;
* $\alpha$ is the learning rate;
* $\frac{\partial L}{\partial E}$ is the gradient with respect to the embedding matrix.

Therefore, an embedding is not merely a fixed preprocessing step.

A typical training pipeline is:

$$
\text{Input Tokens}
\rightarrow
\text{Embedding}
\rightarrow
\text{Sequence Model}
\rightarrow
\hat{y}
\rightarrow
L
$$

During backpropagation:

$$
L
\rightarrow
\text{Sequence Model}
\rightarrow
\text{Embedding}
$$

so the embedding parameters can be updated jointly with the sequence model.

This connects directly to the earlier principle of backpropagation:

$$
\boxed{
\text{Training Objective}
\rightarrow
\text{Gradient}
\rightarrow
\text{Parameter Update}
}
$$

The embedding layer is simply another trainable component in this computational graph.

---

## 8. Tensor Representation of a Sequence

Suppose we have:

* batch size: $B$;
* sequence length: $T$;
* vocabulary size: $V$;
* embedding dimension: $d$.

After tokenization, a batch of sequences can be represented by token IDs:

$$
\boxed{
X \in \mathbb{R}^{B \times T}
}
$$

Each element $X_{b,t}$ is the token ID at timestep $t$ of sample $b$.

The embedding layer maps every token ID to a $d$-dimensional vector:

$$
\boxed{
(B,T)
\rightarrow
(B,T,d)
}
$$

For example, if:

$$
d = 300
$$

then:

$$
X \in \mathbb{R}^{B \times T}
$$

becomes:

$$
E(X) \in \mathbb{R}^{B \times T \times 300}
$$

Each timestep now contains a continuous vector:

$$
x_t \in \mathbb{R}^{d}
$$

instead of a discrete token ID.

This gives the direct bridge:

$$
\boxed{
\text{Discrete Token IDs}
\rightarrow
\text{Continuous Vectors}
\rightarrow
\text{Sequence Model}
}
$$

---

## 9. From Word Representation to Sequence Modeling

Once embeddings are understood, the overall NLP pipeline becomes:

$$
\boxed{
\text{Text}
\rightarrow
\text{Tokenization}
\rightarrow
\text{Token IDs}
\rightarrow
\text{Embedding}
\rightarrow
\text{RNN/LSTM/GRU}
\rightarrow
\text{Prediction}
}
$$

The responsibilities of these components are different.

The **embedding layer** answers:

> How should each individual token be represented numerically?

The **RNN/LSTM/GRU** answers:

> How should information from these token representations be combined and propagated across the sequence?

Thus:

$$
\boxed{
\text{Token Representation}
\rightarrow
\text{Sequence Representation}
\rightarrow
\text{Prediction}
}
$$

For sentiment classification:

$$
\text{"This movie is amazing"}
$$

is first converted into token IDs, then into:

$$
[e_1,e_2,e_3,e_4]
$$

The sequence model processes:

$$
e_1
\rightarrow
e_2
\rightarrow
e_3
\rightarrow
e_4
$$

and finally produces a prediction:

$$
\hat{y} = \text{Positive}
$$

For sequence labeling such as Named Entity Recognition:

$$
x_1,x_2,\ldots,x_T
\rightarrow
\hat{y}_1,\hat{y}_2,\ldots,\hat{y}_T
$$

so the model produces an output at each timestep.

The important distinction is:

$$
\boxed{
\text{Embedding}
=
\text{Representation of Individual Tokens}
}
$$

while:

$$
\boxed{
\text{RNN/LSTM/GRU}
=
\text{Modeling Relationships Across Time}
}
$$

---

# Final Mental Model

The complete process is:

$$
\boxed{
\text{Token}
\rightarrow
\text{Token ID}
\rightarrow
\text{Embedding Vector}
\rightarrow
\text{Sequence Model}
\rightarrow
\text{Prediction}
}
$$

The most important distinction is:

$$
\boxed{
\text{One-hot}
=
\text{Identity Representation}
}
$$

while:

$$
\boxed{
\text{Embedding}
=
\text{Learned Dense Representation}
}
$$

One-hot tells the model:

> **Which token is this?**

Embedding instead provides a learned continuous representation whose geometry can capture useful statistical and semantic relationships.

The learning process can be summarized as:

$$
\boxed{
\text{Training Objective}
\rightarrow
\text{Gradient}
\rightarrow
\text{Embedding Update}
\rightarrow
\text{Useful Embedding Geometry}
}
$$

Therefore, the core idea of Word Representation is:

$$
\boxed{
\text{Convert Discrete Linguistic Symbols into a Learned Continuous Space}
}
$$

This gives sequence models such as RNNs, LSTMs, and GRUs numerical representations of words that they can process and combine over time.


# Using Word Embeddings

## 1. Motivation: Why Use Pre-trained Word Embeddings?

After learning how to represent words as embedding vectors, the next question is:

> **How can we reuse a word representation learned from a large corpus for a new NLP task?**

This is the core idea of **using word embeddings for transfer learning**.

A new NLP task may have only a relatively small labeled dataset. For example, a Named Entity Recognition (NER) system may require labeled sentences where each token has an entity tag:

$$
[
\text{Sally},
\text{Johnson},
\text{works},
\text{at},
\text{Google}
]
$$

with targets such as:

$$
[
\text{B-PER},
\text{I-PER},
\text{O},
\text{O},
\text{B-ORG}
]
$$

The problem is that a small labeled dataset does not provide enough information to learn good representations of every word from scratch.

Meanwhile, a huge amount of unlabeled text is available:

$$
\text{Books}
+
\text{Wikipedia}
+
\text{Web Pages}
+
\text{Articles}
+\cdots
$$

We can therefore separate the problem into two stages:

$$
\boxed{
\text{Learn General Word Representation}
}
$$

followed by:

$$
\boxed{
\text{Transfer It to a Specific NLP Task}
}
$$

The overall idea is:

$$
\boxed{
\text{Large Unlabeled Corpus}
\rightarrow
\text{Learn Word Embeddings}
\rightarrow
\text{Transfer to Downstream Task}
}
$$

This is the main role of **Using Word Embeddings**: not learning another representation from scratch, but reusing a representation that has already learned useful information from a much larger dataset.

---

## 2. The Basic Transfer Learning Process

Suppose we have a pretrained embedding matrix:

$$
E_{\text{pretrained}}
\in
\mathbb{R}^{d \times V}
$$

where:

* $V$ is the vocabulary size;
* $d$ is the embedding dimension;
* each column contains the embedding of one word.

There are three main stages.

### Step 1 — Learn or Obtain Pre-trained Embeddings

The embedding matrix can be learned from a very large corpus.

Conceptually:

$$
\text{Large Corpus}
\rightarrow
\text{Training Objective}
\rightarrow
E_{\text{pretrained}}
$$

Instead of training embeddings ourselves every time, we can also use embeddings that have already been trained on a large corpus.

Therefore:

$$
\boxed{
\text{Pre-trained Embeddings}
=
\text{Knowledge learned from a large corpus}
}
$$

The important point is that the downstream task does not have to start with completely random word representations.

---

### Step 2 — Transfer the Embeddings to the New Task

Suppose the new task is NER.

Instead of representing each token only using a one-hot vector:

$$
o_t \in \mathbb{R}^{V}
$$

we use its embedding:

$$
e_t
=
E_{\text{pretrained}}o_t
$$

where:

$$
e_t \in \mathbb{R}^{d}
$$

The sequence then becomes:

$$
e_1,e_2,\ldots,e_T
$$

and is passed to a sequence model:

$$
\boxed{
e_1,e_2,\ldots,e_T
\rightarrow
\text{RNN/LSTM/GRU}
\rightarrow
\hat{y}_1,\hat{y}_2,\ldots,\hat{y}_T
}
$$

The downstream model therefore starts with a representation that has already been learned from a much larger source of text.

---

### Step 3 — Optional Fine-tuning

After transferring the embeddings, we have two possibilities.

We can keep the embeddings fixed:

$$
E = E_{\text{pretrained}}
$$

or continue training them on the downstream task:

$$
\boxed{
E_{\text{pretrained}}
\rightarrow
E_{\text{updated}}
}
$$

If we fine-tune, the embedding matrix receives gradients from the downstream loss:

$$
E
\leftarrow
E
-
\alpha
\frac{\partial L}{\partial E}
$$

Whether fine-tuning is useful depends on the amount of downstream data.

For a very small labeled dataset, keeping the pretrained embedding relatively fixed can be preferable because aggressively updating a large number of embedding parameters may cause the representation to overfit the small task-specific dataset.

With more downstream data, fine-tuning can allow the embedding space to adapt to the new task.

The conceptual workflow is therefore:

$$
\boxed{
\text{Pre-trained Embedding}
\rightarrow
\text{Transfer}
\rightarrow
\text{Optional Fine-tuning}
}
$$

---

## 3. Why Does This Improve Generalization?

The key benefit of pretrained embeddings is that they allow the downstream model to exploit relationships learned from a much larger corpus.

Consider the words:

$$
\text{apple},
\quad
\text{orange}
$$

With one-hot representation:

$$
o_{\text{apple}}
=
[1,0,\ldots]
$$

and:

$$
o_{\text{orange}}
=
[0,1,\ldots]
$$

The two vectors are simply different basis vectors.

There is no natural notion that these words are related.

In an embedding space, however:

$$
e_{\text{apple}},
e_{\text{orange}}
\in
\mathbb{R}^{d}
$$

can have similar geometric properties because the two words occur in related contexts.

For example:

$$
\operatorname{sim}
\left(
e_{\text{apple}},
e_{\text{orange}}
\right)
>
\operatorname{sim}
\left(
e_{\text{apple}},
e_{\text{king}}
\right)
$$

in a suitable embedding space.

This gives the downstream model a better basis for generalization.

The causal chain is:

$$
\boxed{
\text{Large Unlabeled Corpus}
\rightarrow
\text{Rich Word Representation}
\rightarrow
\text{Transfer}
\rightarrow
\text{Better Generalization}
}
$$

The important advantage is therefore not simply dimensionality reduction.

The deeper advantage is:

$$
\boxed{
\text{Related Words Can Share Useful Representational Structure}
}
$$

This is difficult with one-hot vectors because one-hot treats words as essentially independent identities.

---

## 4. Why Are Word Embeddings Especially Useful for Small NLP Datasets?

Suppose a downstream task contains relatively little labeled data.

If we train everything from scratch:

$$
\boxed{
\text{Small Dataset}
\rightarrow
\text{Learn Word Representation}
+
\text{Learn Task}
}
$$

the model has to solve two problems simultaneously:

1. learn useful representations of words;
2. learn how those representations should be used for the task.

That is inefficient when labeled data is scarce.

With pretrained embeddings:

$$
\boxed{
\text{Large Corpus}
\rightarrow
\text{Learn General Representation}
}
$$

and then:

$$
\boxed{
\text{Small Labeled Dataset}
\rightarrow
\text{Learn Task-Specific Behavior}
}
$$

So the downstream task can focus more on learning the task itself rather than rediscovering general linguistic structure.

This is the central motivation for transfer learning with word embeddings.

---

## 5. Using Embeddings with an RNN/LSTM/GRU

This connects directly to the sequence models from Week 1.

Suppose the input sequence is:

$$
x_1,x_2,\ldots,x_T
$$

After embedding:

$$
x_t
\rightarrow
e_t
$$

where:

$$
e_t \in \mathbb{R}^{d}
$$

The sequence model then computes:

$$
a_t = f(a_{t-1},e_t)
$$

For an RNN, this might be:

$$
a_t
=
g
\left(
W_{aa}a_{t-1}
+
W_{ax}e_t
+
b_a
\right)
$$

and then:

$$
\hat{y}_t
=
g_y
\left(
W_{ya}a_t
+
b_y
\right)
$$

So the complete computation becomes:

$$
\boxed{
\text{Token}
\rightarrow
\text{Embedding}
\rightarrow
\text{Recurrent State}
\rightarrow
\text{Prediction}
}
$$

For a many-to-many task such as NER:

$$
[e_1,e_2,\ldots,e_T]
\rightarrow
\text{LSTM/GRU}
\rightarrow
[\hat{y}_1,\hat{y}_2,\ldots,\hat{y}_T]
$$

For a many-to-one task such as sentiment classification:

$$
[e_1,e_2,\ldots,e_T]
\rightarrow
\text{LSTM/GRU}
\rightarrow
\hat{y}
$$

Therefore the roles remain clearly separated:

$$
\boxed{
\text{Embedding}
=
\text{Represent Each Token}
}
$$

$$
\boxed{
\text{RNN/LSTM/GRU}
=
\text{Model the Sequence}
}
$$

$$
\boxed{
\text{Task Head}
=
\text{Produce the Prediction}
}
$$

---

## 6. Embedding Transfer vs. Training the Entire NLP Model

It is important not to confuse **transfer of embeddings** with transferring an entire neural network.

In the topic being studied, the central transferable component is:

$$
\boxed{
E_{\text{pretrained}}
}
$$

The downstream model can still be newly initialized:

$$
E_{\text{pretrained}}
\rightarrow
\text{New LSTM}
\rightarrow
\text{New Output Layer}
$$

So we are reusing a useful representation while allowing the downstream model to learn task-specific behavior.

This gives a useful mental distinction:

$$
\boxed{
\text{Transfer Representation}
\neq
\text{Transfer Entire Model}
}
$$

The same general transfer-learning principle appears throughout deep learning: a representation learned from a large dataset can become useful for another task with less data.

---

# Final Mental Model

The complete workflow is:

$$
\boxed{
\text{Large Unlabeled Corpus}
\rightarrow
\text{Pre-trained Word Embeddings}
\rightarrow
\text{Transfer}
\rightarrow
\text{Downstream NLP Task}
}
$$

For a sequence model:

$$
\boxed{
\text{Text}
\rightarrow
\text{Token IDs}
\rightarrow
\text{Pre-trained Embeddings}
\rightarrow
\text{RNN/LSTM/GRU}
\rightarrow
\text{Prediction}
}
$$

And when the downstream dataset is small:

$$
\boxed{
\text{Pre-trained Embedding}
\rightarrow
\text{Less Need to Learn Word Representations from Scratch}
}
$$

The deepest idea is:

$$
\boxed{
\text{Learn General Linguistic Representations Once}
\rightarrow
\text{Reuse Them Across NLP Tasks}
}
$$

while fine-tuning provides an optional adaptation step:

$$
\boxed{
\text{Pre-trained Representation}
\rightarrow
\text{Task-Specific Adaptation}
}
$$

Thus, **Word Representation** asks:

> *How can we represent a word as a useful vector?*

while **Using Word Embeddings** asks:

> *How can we reuse that learned representation to improve a new NLP task, especially when the new task has limited labeled data?*


# Properties of Word Embeddings

## 1. From Words to Points in a Continuous Space

A word embedding represents each word as a dense vector:

$$
e_w\in\mathbb R^d
$$

where \(d\) is the embedding dimension.

Instead of treating every word as an independent identity, embeddings place words in a shared continuous space:

$$
\boxed{
\text{Word}
\rightarrow
\text{Point in Embedding Space}
}
$$

For example:

$$
e_{\text{king}},
e_{\text{queen}},
e_{\text{man}},
e_{\text{woman}}
\in\mathbb R^d
$$

The important information is not necessarily contained in any individual dimension. Rather, the **entire vector and its position relative to other vectors** can encode useful linguistic information.

This is fundamentally different from one-hot representation:

$$
\boxed{
\text{One-hot}
=
\text{Identity of a word}
}
$$

while:

$$
\boxed{
\text{Embedding}
=
\text{Learned representation of a word}
}
$$

The resulting embedding space can therefore contain meaningful geometric structure.

---

## 2. Similar Words Tend to Have Similar Embeddings

One important property is:

$$
\boxed{
\text{Similar Words}
\rightarrow
\text{Similar Embeddings}
}
$$

However, "similar" should not be interpreted only as dictionary-level synonymy.

Word embeddings are learned from language usage and context. If two words frequently occur in similar contexts, their representations can become similar.

For example:

> I drink coffee.

> I drink tea.

The words `coffee` and `tea` occur in related contexts, so a learned embedding space may contain:

$$
e_{\text{coffee}}
\approx
e_{\text{tea}}
$$

according to an appropriate similarity measure.

A common measure is cosine similarity:

$$
\boxed{
\operatorname{sim}(u,v)
=
\frac{u^\top v}
{\|u\|\|v\|}
}
$$

If:

$$
\operatorname{sim}(u,v)\approx1
$$

then the two vectors point in similar directions.

Therefore, embeddings can form semantic neighborhoods in which words with related usage patterns tend to be represented near one another.

The key idea is:

$$
\boxed{
\text{Embedding Similarity}
\approx
\text{Similarity in Learned Usage Patterns}
}
$$

rather than an exact dictionary definition of similarity.

---

## 3. Embeddings Can Encode Relationships Between Words

A deeper property is that embeddings can represent not only similarity but also **relationships**.

Consider the classic example:

$$
\boxed{
e_{\text{king}}
-
e_{\text{man}}
+
e_{\text{woman}}
\approx
e_{\text{queen}}
}
$$

Equivalently:

$$
e_{\text{king}}
-
e_{\text{man}}
\approx
e_{\text{queen}}
-
e_{\text{woman}}
$$

The important idea is not that this equation must be exactly true for every embedding model.

The important idea is that the **difference between two vectors can represent a relationship**.

For example:

$$
e_{\text{king}}-e_{\text{man}}
$$

can be viewed as a direction associated with a particular semantic transformation.

If a similar transformation exists elsewhere:

$$
e_{\text{queen}}-e_{\text{woman}}
$$

the two directions may be approximately aligned.

Thus:

$$
\boxed{
\text{Relationship Between Words}
\rightarrow
\text{Geometric Relationship Between Vectors}
}
$$

This is more powerful than simply saying that two words are close to each other.

A useful embedding space can therefore encode:

$$
\boxed{
\text{Similarity}
+
\text{Relationships}
}
$$

---

## 4. Why Does This Structure Appear?

The geometric structure is not manually programmed into the embedding matrix.

Instead, it emerges through training.

Suppose the embedding matrix is:

$$
E\in\mathbb R^{d\times V}
$$

and its parameters are learned using some prediction objective.

The training process can be summarized as:

$$
\boxed{
\text{Context}
\rightarrow
\text{Prediction Objective}
\rightarrow
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\text{Embedding Update}
}
$$

Now consider two words that repeatedly appear in similar contexts.

Because they participate in related prediction problems, the gradients used to update their embeddings can also contain related patterns.

After many updates:

$$
\boxed{
\text{Similar Contexts}
\rightarrow
\text{Related Updates}
\rightarrow
\text{Similar Embeddings}
}
$$

This gives the key causal chain:

$$
\boxed{
\text{Language Patterns}
\rightarrow
\text{Training Signal}
\rightarrow
\text{Parameter Updates}
\rightarrow
\text{Embedding Geometry}
}
$$

Therefore, semantic structure is not a mysterious property that embeddings automatically possess.

It is a consequence of learning from the statistical structure of language.

---

## 5. Dense and Low-Dimensional Representation

Another important property is that embeddings are typically **dense** and **lower-dimensional** than the vocabulary space.

For vocabulary size \(V\):

$$
o_w\in\mathbb R^V
$$

is a one-hot representation.

An embedding instead uses:

$$
e_w\in\mathbb R^d
$$

where typically:

$$
d\ll V
$$

For example:

$$
V=50,000,\qquad d=300
$$

Therefore:

$$
\text{One-hot}
\rightarrow
\text{50,000 dimensions}
$$

while:

$$
\text{Embedding}
\rightarrow
\text{300 dimensions}
$$

The embedding is also dense, meaning that many or most coordinates can contain non-zero values.

Hence:

$$
\boxed{
\text{One-hot}
=
\text{Sparse + High-dimensional}
}
$$

while:

$$
\boxed{
\text{Embedding}
=
\text{Dense + Lower-dimensional}
}
$$

However, dimensionality reduction alone is not what makes an embedding useful.

The important combination is:

$$
\boxed{
\text{Compact Representation}
+
\text{Learned Structure}
}
$$

A random 300-dimensional vector would not automatically provide useful semantic information.

---

## 6. Geometry Becomes Meaningful

Once words are represented in a shared vector space, geometric operations become useful.

For example:

$$
\operatorname{sim}(e_i,e_j)
$$

can measure how similar two word representations are.

This gives a useful relationship:

$$
\boxed{
\text{Geometry}
\rightarrow
\text{Useful Linguistic Information}
}
$$

This is something one-hot representations do not naturally provide.

For two distinct one-hot vectors:

$$
o_i^\top o_j=0,
\qquad i\neq j
$$

so every pair of different words is orthogonal under the standard dot product.

In contrast, embeddings can produce different degrees of similarity:

$$
\operatorname{sim}(e_i,e_j)
\in[-1,1]
$$

for cosine similarity.

Therefore the embedding space contains a much richer notion of "how words relate."

---

## 7. An Important Limitation: Embeddings Reflect Their Training Data

The properties of an embedding are learned from data.

Therefore:

$$
\boxed{
\text{Training Data}
\rightarrow
\text{Learned Associations}
\rightarrow
\text{Embedding Geometry}
}
$$

This is a strength because the model can learn useful linguistic patterns.

But it is also a limitation because biases or undesirable associations present in the training corpus can become encoded in the embedding space.

Thus an embedding should not automatically be considered a perfectly objective representation of language.

A useful mental model is:

$$
\boxed{
\text{Embedding}
=
\text{Learned Statistical Representation of Language Data}
}
$$

not:

$$
\boxed{
\text{Embedding}
=
\text{Perfect Representation of Human Meaning}
}
$$

---

# Final Mental Model

The most important properties can be summarized as:

$$
\boxed{
\text{Words}
\rightarrow
\text{Vectors}
\rightarrow
\text{Geometric Structure}
}
$$

From this structure, we can obtain:

$$
\boxed{
\text{Similarity}
+
\text{Semantic Relationships}
+
\text{Generalization}
}
$$

The core causal chain is:

$$
\boxed{
\text{Language Statistics}
\rightarrow
\text{Training Signal}
\rightarrow
\text{Embedding Updates}
\rightarrow
\text{Embedding Geometry}
}
$$

Therefore, the central idea behind the properties of word embeddings is:

$$
\boxed{
\text{A Good Embedding Space Preserves Useful Structure of Language}
}
$$

The two most important properties to remember are:

$$
\boxed{
\text{Similar Contexts}
\rightarrow
\text{Similar Embeddings}
}
$$

and:

$$
\boxed{
\text{Relationships Between Words}
\rightarrow
\text{Relationships Between Vectors}
}
$$

These properties are what make embeddings much more useful than one-hot representations for downstream NLP tasks.

# Embedding Matrix

## 1. What Is the Embedding Matrix?

After representing each word as an embedding vector:

$$
e_i \in \mathbb{R}^d
$$

we need a compact way to store the embeddings of all words in the vocabulary.

Let the vocabulary be:

$$
\mathcal{V}
=
\{w_1,w_2,\ldots,w_V\}
$$

where $V$ is the vocabulary size.

We collect all word embeddings into a single matrix:

$$
\boxed{
E \in \mathbb{R}^{d \times V}
}
$$

The matrix can be written as:

$$
E
=
\begin{bmatrix}
| & | & & |\\
e_1 & e_2 & \cdots & e_V\\
| & | & & |
\end{bmatrix}
$$

Therefore:

$$
\boxed{
E_{:,i}=e_i
}
$$

That is, the **$i$-th column of $E$** is the embedding vector of the $i$-th word in the vocabulary.

The two dimensions have clear meanings:

$$
\boxed{
\underbrace{d}_{\text{embedding dimension}}
\times
\underbrace{V}_{\text{vocabulary size}}
}
$$

So:

- $d$ determines the size of each word vector;
- $V$ determines how many word embeddings are stored.

---

## 2. Why Is the Matrix $d \times V$?

Each individual embedding has:

$$
e_i \in \mathbb{R}^d
$$

so it contains $d$ values.

There are $V$ words, and we place each embedding as one **column**.

Therefore:

$$
\boxed{
E \in \mathbb{R}^{d \times V}
}
$$

For example, if:

$$
V=5,
\qquad
d=3
$$

then:

$$
E \in \mathbb{R}^{3 \times 5}
$$

and:

$$
E
=
\begin{bmatrix}
e_{1,1} & e_{1,2} & e_{1,3} & e_{1,4} & e_{1,5}\\
e_{2,1} & e_{2,2} & e_{2,3} & e_{2,4} & e_{2,5}\\
e_{3,1} & e_{3,2} & e_{3,3} & e_{3,4} & e_{3,5}
\end{bmatrix}
$$

The first column is $e_1$, the second is $e_2$, and so on.

Thus:

$$
\boxed{
\text{One column}
=
\text{One word embedding}
}
$$

This is the most important structural fact to remember about the embedding matrix.

---

## 3. Embedding Lookup

Suppose word $w_i$ is represented by a one-hot vector:

$$
o_i \in \mathbb{R}^{V}
$$

The embedding vector can be obtained by:

$$
\boxed{
e_i=Eo_i
}
$$

The dimensions are:

$$
E \in \mathbb{R}^{d \times V}
$$

and:

$$
o_i \in \mathbb{R}^{V \times 1}
$$

so:

$$
(d \times V)(V \times 1)
=
(d \times 1)
$$

which gives:

$$
e_i \in \mathbb{R}^d
$$

### Why does this work?

Suppose:

$$
E
=
\begin{bmatrix}
| & | & |\\
e_1 & e_2 & e_3\\
| & | & |
\end{bmatrix}
$$

and:

$$
o_2
=
\begin{bmatrix}
0\\
1\\
0
\end{bmatrix}
$$

Then:

$$
Eo_2
=
0e_1+1e_2+0e_3
=
e_2
$$

So the one-hot vector simply selects one column of $E$.

Therefore:

$$
\boxed{
\text{One-hot Vector}
\rightarrow
\text{Select a Column of }E
\rightarrow
\text{Embedding Vector}
}
$$

This operation is called an **embedding lookup**.

In practical implementations, we normally do not explicitly construct a large one-hot vector. Given a token ID $i$, the framework can directly retrieve the corresponding embedding:

$$
e_i=E_{:,i}
$$

The mathematical formulation $e_i=Eo_i$ is mainly useful for understanding the mechanism.

---

## 4. The Embedding Matrix Is Trainable

A critical point is that:

$$
\boxed{
E \text{ is a trainable parameter matrix}
}
$$

It is not simply a fixed lookup table.

During training, the loss produces gradients with respect to $E$:

$$
\frac{\partial L}{\partial E}
$$

and gradient descent updates the matrix:

$$
\boxed{
E
\leftarrow
E
-
\alpha
\frac{\partial L}{\partial E}
}
$$

Therefore, the values inside the embedding matrix are learned from data.

For example:

$$
E^{(0)}
\rightarrow
E^{(1)}
\rightarrow
E^{(2)}
\rightarrow
\cdots
\rightarrow
E^{(*)}
$$

where $E^{(*)}$ is the learned embedding matrix after training.

Consequently, every word embedding:

$$
e_i
$$

can also change during training.

This leads to the important chain:

$$
\boxed{
\text{Training Signal}
\rightarrow
\text{Gradient}
\rightarrow
\text{Embedding Update}
}
$$

and eventually:

$$
\boxed{
\text{Embedding Updates}
\rightarrow
\text{Meaningful Embedding Geometry}
}
$$

---

## 5. How Does a Specific Word Receive a Gradient?

Suppose the current input token is $w_i$.

The forward computation is:

$$
o_i
\rightarrow
e_i=Eo_i
\rightarrow
\text{Sequence Model}
\rightarrow
\hat{y}
\rightarrow
L
$$

During backpropagation:

$$
L
\rightarrow
\frac{\partial L}{\partial e_i}
\rightarrow
\frac{\partial L}{\partial E}
$$

Since $e_i$ corresponds to one specific column of $E$, the learning signal updates the embedding associated with that word.

Conceptually:

$$
\boxed{
\text{Word Appears}
\rightarrow
\text{Its Embedding Is Used}
\rightarrow
\text{It Receives Gradient}
\rightarrow
\text{Its Embedding Is Updated}
}
$$

If the same word appears many times in the training set, its embedding participates in many computations and can receive learning signals from many occurrences.

This is how the embedding matrix gradually adapts to the training objective.

---

## 6. Parameter Count

Because:

$$
E \in \mathbb{R}^{d \times V}
$$

the number of trainable parameters in the embedding matrix is:

$$
\boxed{
|E|=dV
}
$$

For example:

$$
V=50{,}000
$$

and:

$$
d=300
$$

then:

$$
|E|
=
50{,}000 \times 300
=
15{,}000{,}000
$$

So the embedding matrix contains:

$$
\boxed{
15\text{ million parameters}
}
$$

This explains why vocabulary size $V$ and embedding dimension $d$ both matter for memory consumption and model size.

---

## 7. Embedding Matrix for a Whole Sequence

At one timestep:

$$
o_t \in \mathbb{R}^{V}
$$

is mapped to:

$$
e_t=Eo_t \in \mathbb{R}^d
$$

For an entire sequence:

$$
o_1,o_2,\ldots,o_T
$$

we obtain:

$$
e_1,e_2,\ldots,e_T
$$

Therefore:

$$
\boxed{
\text{Sequence of Tokens}
\rightarrow
\text{Sequence of Embedding Vectors}
}
$$

At batch level, suppose token IDs are:

$$
X \in \mathbb{R}^{B \times T}
$$

where:

- $B$ = batch size;
- $T$ = sequence length.

After the embedding lookup:

$$
\boxed{
(B,T)
\rightarrow
(B,T,d)
}
$$

For example:

$$
B=32,
\qquad
T=50,
\qquad
d=300
$$

gives:

$$
X:(32,50)
$$

and:

$$
E(X):(32,50,300)
$$

Thus every token in every sequence is replaced by a 300-dimensional vector.

This tensor can then be passed into a recurrent model:

$$
(B,T,d)
\rightarrow
\text{RNN/LSTM/GRU}
$$

---

# Final Mental Model

The embedding matrix is best understood as a **trainable table of word representations**:

$$
\boxed{
E \in \mathbb{R}^{d \times V}
}
$$

with:

$$
\boxed{
E_{:,i}=e_i
}
$$

and the mathematical lookup:

$$
\boxed{
e_i=Eo_i
}
$$

The complete learning process is:

$$
\boxed{
\text{Token ID}
\rightarrow
\text{Embedding Lookup}
\rightarrow
e_i
\rightarrow
\text{Sequence Model}
\rightarrow
L
\rightarrow
\frac{\partial L}{\partial E}
\rightarrow
\text{Update }E
}
$$

The most important facts to remember are:

$$
\boxed{
E \in \mathbb{R}^{d \times V}
}
$$

$$
\boxed{
\text{Each column of }E\text{ is one word embedding}
}
$$

$$
\boxed{
|E|=dV
}
$$

and:

$$
\boxed{
\text{Embedding Matrix}
=
\text{Trainable Word Representation Table}
}
$$

This matrix is the bridge between **discrete token IDs** and the **continuous vectors** consumed by RNNs, LSTMs, and GRUs.
