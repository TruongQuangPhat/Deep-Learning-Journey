# What is Neural Style Transfer?

**Neural Style Transfer (NST)** is a technique that generates a new image by combining the:

$$
\boxed{
\text{Content of One Image}
+
\text{Style of Another Image}
\rightarrow
\text{Generated Image}
}
$$

For example:

- **Content image**: a photo of a building.
- **Style image**: a painting.
- **Generated image**: preserves the structure and content of the building while adopting the visual style of the painting.

The key idea is to use a pretrained CNN to obtain:

$$
\boxed{
\text{Content Representation}
}
$$

and:

$$
\boxed{
\text{Style Representation}
}
$$

Then optimize the generated image $G$ so that it simultaneously matches the content of the content image and the style of the style image.

Thus:

$$
\boxed{
\text{Content Preservation}
+
\text{Style Matching}
\rightarrow
\text{Generated Image}
}
$$

# What are deep ConvNets learning?

**Deep Convolutional Networks (ConvNets)** learn a **hierarchy of feature representations**. Rather than directly learning high-level concepts from raw pixels, the network gradually builds increasingly complex features across its layers.

The overall progression can be viewed as:

$$
\boxed{
\text{Pixels}
\rightarrow
\text{Edges}
\rightarrow
\text{Textures / Simple Patterns}
\rightarrow
\text{Parts}
\rightarrow
\text{Objects}
}
$$

---

## 1. Shallow Layers Learn Low-Level Features

The early layers have relatively small receptive fields and tend to learn simple visual patterns such as:

- edges;
- orientations;
- colors;
- simple local patterns.

Conceptually:

$$
\text{Image}
\rightarrow
\text{Low-Level Features}
$$

A convolutional filter may respond strongly to a particular type of edge at different spatial locations.

This is possible because convolution uses **shared parameters**.

---

## 2. Deeper Layers Compose Simpler Features

Later layers receive the feature maps produced by earlier layers and combine simpler features into increasingly complex patterns:

$$
\text{Edges}
\rightarrow
\text{Textures}
\rightarrow
\text{Parts}
$$

For example:

$$
\text{Edge Features}
\rightarrow
\text{Eye-like Structure}
$$

or:

$$
\text{Simple Shapes}
\rightarrow
\text{Face Structure}
$$

As the network becomes deeper, the receptive field also becomes larger, allowing a neuron to use information from a larger region of the original image.

---

## 3. Deep Layers Learn High-Level Semantic Features

At deeper layers, the representation becomes increasingly abstract and semantic.

The network can combine lower-level structures into meaningful object-level patterns:

$$
\text{Parts}
\rightarrow
\text{Object-Level Features}
$$

For example:

$$
\text{Eyes}
+
\text{Nose}
+
\text{Mouth}
\rightarrow
\text{Face}
$$

or:

$$
\text{Wheels}
+
\text{Body Shape}
\rightarrow
\text{Car}
$$

Therefore:

$$
\boxed{
\text{Shallow Layers}
\rightarrow
\text{Low-Level Features}
}
$$

$$
\boxed{
\text{Deep Layers}
\rightarrow
\text{High-Level Semantic Features}
}
$$

---

## 4. Feature Representations Become More Abstract

The network does not simply preserve a copy of the original image at every layer.

Instead, each layer produces a new representation:

$$
a^{[1]}
\rightarrow
a^{[2]}
\rightarrow
a^{[3]}
\rightarrow
\cdots
\rightarrow
a^{[L]}
$$

Each representation is constructed from the previous one.

Thus:

$$
\boxed{
\text{Low-Level Features}
\rightarrow
\text{Mid-Level Features}
\rightarrow
\text{High-Level Features}
}
$$

The deeper the representation, the more abstract and task-relevant it can become.

---

## 5. What Is the Network Actually Learning?

The network learns its filters and parameters so that the resulting representations become useful for the final task.

For example, in classification:

$$
X
\rightarrow
a^{[1]}
\rightarrow
a^{[2]}
\rightarrow
\cdots
\rightarrow
a^{[L]}
\rightarrow
\hat y
$$

Backpropagation updates the parameters so that the learned representations ultimately help the network make accurate predictions.

Importantly, the network is **not explicitly programmed** with rules such as:

> "This filter must detect an edge."

or:

> "This filter must detect an eye."

Those filters are learned from the data.

The concepts such as edges, textures, parts, and objects are useful interpretations of the hierarchy of representations that a deep ConvNet often develops.

---

## 6. Core Mental Model

The key idea is:

$$
\boxed{
\text{Pixels}
\rightarrow
\text{Edges}
\rightarrow
\text{Textures}
\rightarrow
\text{Parts}
\rightarrow
\text{Objects}
}
$$

More generally:

$$
\boxed{
\text{Simple Features}
\rightarrow
\text{Composed Features}
\rightarrow
\text{Semantic Features}
}
$$

Therefore:

> **Deep ConvNets learn a hierarchy of representations, where shallow layers tend to capture simple visual patterns and deeper layers progressively combine them into more complex and semantically meaningful features.**

# Cost Function

In **Neural Style Transfer**, the goal is to generate an image $G$ that simultaneously preserves the content of a **content image** $C$ and matches the style of a **style image** $S$.

The total cost is:

$$
\boxed{
J(G)
=
\alpha J_{\text{content}}(G,C)
+
\beta J_{\text{style}}(S,G)
}
$$

where:

- $G$: generated image;
- $C$: content image;
- $S$: style image;
- $\alpha$: weight of the content term;
- $\beta$: weight of the style term.

The key difference from normal neural network training is that the pretrained CNN is kept **fixed**, while the generated image $G$ itself is optimized.

---

## 1. Content Cost

The content cost measures how well the generated image preserves the content of the content image.

Let:

$$
a^{[l]}(C)
$$

be the activation of the content image at layer $l$, and:

$$
a^{[l]}(G)
$$

be the activation of the generated image at the same layer.

A common content cost is:

$$
\boxed{
J_{\text{content}}(G,C)
=
\frac{1}{2}
\left\|
a^{[l]}(G)-a^{[l]}(C)
\right\|_F^2
}
$$

The intuition is that if the activations are similar at a sufficiently deep layer, the two images tend to contain similar high-level content.

Therefore:

$$
J_{\text{content}}\downarrow
\Rightarrow
\text{Generated Image preserves the content}
$$

---

## 2. Style Cost

Content cost alone cannot ensure that the generated image has the desired visual style.

Style is represented through relationships between feature channels.

Suppose the activation at layer $l$ has shape:

$$
a^{[l]}
\in
\mathbb R^{n_H\times n_W\times n_C}
$$

Reshape it into:

$$
F\in
\mathbb R^{(n_Hn_W)\times n_C}
$$

Then compute the **Gram Matrix**:

$$
\boxed{
G=F^TF
}
$$

so that:

$$
G\in
\mathbb R^{n_C\times n_C}
$$

Each element is:

$$
G_{ij}
=
\sum_kF_{ki}F_{kj}
$$

which measures the correlation between feature channels $i$ and $j$.

The Gram Matrix therefore captures statistical relationships between visual features, which are useful for representing style.

---

## 3. Style Cost at One Layer

Let:

$$
G^{[l](S)}
$$

be the Gram Matrix of the style image and:

$$
G^{[l](G)}
$$

be the Gram Matrix of the generated image.

A style cost at layer $l$ can be defined as:

$$
\boxed{
J_{\text{style}}^{[l]}(S,G)
=
\frac{1}{(2n_Hn_Wn_C)^2}
\left\|
G^{[l](G)}
-
G^{[l](S)}
\right\|_F^2
}
$$

The objective is:

$$
J_{\text{style}}^{[l]}\downarrow
$$

so that the feature correlations of the generated image become similar to those of the style image.

---

## 4. Style Cost Across Multiple Layers

Different CNN layers capture different levels of visual structure.

Therefore style can be measured across multiple layers:

$$
\boxed{
J_{\text{style}}(S,G)
=
\sum_l
\lambda^{[l]}
J_{\text{style}}^{[l]}(S,G)
}
$$

where:

$$
\lambda^{[l]}
$$

controls the contribution of each layer.

This allows the style representation to capture information at multiple levels, from lower-level textures and patterns to more complex visual structures.

---

## 5. Total Cost Function

The complete objective is:

$$
\boxed{
J(G)
=
\alpha J_{\text{content}}(G,C)
+
\beta J_{\text{style}}(S,G)
}
$$

We want to find:

$$
\boxed{
G^*
=
\arg\min_GJ(G)
}
$$

Thus the generated image is optimized to satisfy two competing objectives:

$$
\boxed{
\text{Content Preservation}
}
$$

and:

$$
\boxed{
\text{Style Matching}
}
$$

---

## 6. Optimization

The pretrained CNN parameters are fixed:

$$
\theta=\text{constant}
$$

Instead, the generated image is optimized.

Starting from an initial image:

$$
G_0
$$

we repeatedly perform:

$$
G
\rightarrow
\text{Fixed CNN}
\rightarrow
J(G)
$$

then compute:

$$
\frac{\partial J}{\partial G}
$$

and update:

$$
\boxed{
G
\leftarrow
G-\eta
\frac{\partial J}{\partial G}
}
$$

where $\eta$ is the learning rate.

Therefore, Neural Style Transfer optimizes the **pixels of the generated image**, not the parameters of the pretrained CNN.

---

## 7. Comparison with Normal Neural Network Training

### Normal Neural Network Training

$$
X
\rightarrow
f(X;\theta)
\rightarrow
L
\rightarrow
\frac{\partial L}{\partial\theta}
\rightarrow
\theta
$$

The parameters are updated:

$$
\theta
\leftarrow
\theta-\eta
\frac{\partial L}{\partial\theta}
$$

### Neural Style Transfer

$$
G
\rightarrow
f(G;\theta_{\text{fixed}})
\rightarrow
J(G)
\rightarrow
\frac{\partial J}{\partial G}
\rightarrow
G
$$

The CNN parameters remain fixed:

$$
\boxed{
\theta_{\text{fixed}}
}
$$

while the generated image is updated:

$$
\boxed{
G
\leftarrow
G-\eta\frac{\partial J}{\partial G}
}
$$

This is one of the most important conceptual differences in Neural Style Transfer.

---

## 8. Role of $\alpha$ and $\beta$

The total cost is:

$$
J(G)
=
\alpha J_{\text{content}}
+
\beta J_{\text{style}}
$$

### Large $\alpha$

The optimization places more emphasis on preserving content:

$$
\text{Content Preservation}\uparrow
$$

### Large $\beta$

The optimization places more emphasis on matching style:

$$
\text{Style Matching}\uparrow
$$

Therefore:

$$
\boxed{
\frac{\alpha}{\beta}
\rightarrow
\text{Content–Style Trade-off}
}
$$

---

## 9. Core Mental Model

The Cost Function of Neural Style Transfer can be summarized as:

$$
\boxed{
\text{Total Cost}
=
\text{Content Cost}
+
\text{Style Cost}
}
$$

or:

$$
\boxed{
J(G)
=
\alpha J_{\text{content}}
+
\beta J_{\text{style}}
}
$$

where:

$$
J_{\text{content}}
\rightarrow
\text{Preserve Content}
$$

and:

$$
J_{\text{style}}
\rightarrow
\text{Match Style}
$$

Style is represented using **Gram Matrices**, which capture correlations between feature channels.

Finally, instead of updating the pretrained CNN:

$$
\theta
$$

we directly optimize:

$$
\boxed{
G
\leftarrow
G-\eta
\frac{\partial J}{\partial G}
}
$$

Thus the essential idea is:

> **A fixed pretrained ConvNet provides the feature representations, while the generated image is optimized so that its content features match the content image and its style statistics match the style image.**

# Content Cost Function

In **Neural Style Transfer**, the **Content Cost Function** measures how different the content of the generated image is from the content image.

The goal is:

$$
\boxed{
\text{Generated Image}
\approx
\text{Content Image}
}
$$

in terms of their **content representation**.

Let:

- $C$: content image;
- $G$: generated image;
- $a^{[l](C)}$: activation of the content image at layer $l$;
- $a^{[l](G)}$: activation of the generated image at layer $l$.

---

## 1. Using a Pretrained ConvNet

Both images are passed through the **same pretrained ConvNet**:

```mermaid
flowchart LR
    A["Content Image C"] --> B["Pretrained ConvNet"]
    C["Generated Image G"] --> D["Pretrained ConvNet"]

    B --> E["Activation at Layer l"]
    D --> F["Activation at Layer l"]

    E --> G["Content Cost"]
    F --> G
```

The pretrained ConvNet is kept **fixed**.

We use its activation at a selected layer to represent the content of each image.

---

## 2. Why Compare Activations Instead of Pixels?

A direct pixel-wise comparison would require:

$$
G\approx C
$$

pixel by pixel.

But this is not what we want in Neural Style Transfer because the generated image should be allowed to have a completely different visual appearance while preserving the same content.

Instead, we compare feature representations:

$$
a^{[l](G)}
\quad\text{and}\quad
a^{[l](C)}
$$

At a relatively deep layer, the activations capture more high-level information such as:

$$
\text{Shapes}
\rightarrow
\text{Parts}
\rightarrow
\text{Objects}
$$

rather than only low-level pixels.

Therefore, similar activations indicate that the generated image preserves important aspects of the content.

---

## 3. Content Cost Function

The content cost can be defined as:

$$
\boxed{
J_{\text{content}}(G,C)
=
\frac{1}{2}
\left\|
a^{[l](G)}
-
a^{[l](C)}
\right\|_F^2
}
$$

where:

$$
\|\cdot\|_F
$$

is the **Frobenius norm**.

If the activation has shape:

$$
a^{[l]}
\in
\mathbb R^{n_H\times n_W\times n_C}
$$

then the loss is the sum of squared differences across all activation values:

$$
J_{\text{content}}
=
\frac12
\sum_{i,j,k}
\left(
a_{ijk}^{[l](G)}
-
a_{ijk}^{[l](C)}
\right)^2
$$

---

## 4. Interpretation

If:

$$
J_{\text{content}}\approx0
$$

then:

$$
a^{[l](G)}
\approx
a^{[l](C)}
$$

meaning the generated image has a similar content representation to the content image at layer $l$.

If:

$$
J_{\text{content}}\uparrow
$$

the generated image becomes more different from the content image in that feature space.

Therefore the optimization tries to minimize:

$$
\boxed{
J_{\text{content}}\downarrow
}
$$

so that the generated image preserves the content.

---

## 5. Simple Numerical Example

Suppose:

$$
a^{[l](C)}
=
\begin{bmatrix}
1 & 2\\
3 & 4
\end{bmatrix}
$$

and:

$$
a^{[l](G)}
=
\begin{bmatrix}
1.2 & 1.8\\
3.1 & 3.9
\end{bmatrix}
$$

Then:

$$
a^{[l](G)}-a^{[l](C)}
=
\begin{bmatrix}
0.2 & -0.2\\
0.1 & -0.1
\end{bmatrix}
$$

Therefore:

$$
J_{\text{content}}
=
\frac12
\left(
0.2^2+(-0.2)^2+0.1^2+(-0.1)^2
\right)
$$

$$
=
\frac12(0.04+0.04+0.01+0.01)
$$

$$
=
0.05
$$

A smaller value indicates that the two feature representations are more similar.

---

## 6. Optimization of the Generated Image

In Neural Style Transfer, the pretrained CNN parameters are fixed:

$$
\boxed{
\theta_{\text{CNN}}=\text{constant}
}
$$

The variable being optimized is the generated image:

$$
G
$$

The computational graph is:

$$
G
\rightarrow
\text{Fixed ConvNet}
\rightarrow
a^{[l](G)}
\rightarrow
J_{\text{content}}
$$

Backpropagation computes:

$$
\frac{\partial J_{\text{content}}}{\partial G}
$$

and updates the generated image:

$$
\boxed{
G
\leftarrow
G-\eta
\frac{\partial J_{\text{content}}}{\partial G}
}
$$

Thus the gradient changes the pixels of $G$ so that its feature representation becomes closer to that of the content image.

---

## 7. Core Mental Model

The complete idea is:

$$
\boxed{
C
\rightarrow
a^{[l](C)}
}
$$

and:

$$
\boxed{
G
\rightarrow
a^{[l](G)}
}
$$

then compare them:

$$
\boxed{
J_{\text{content}}
=
\frac12
\left\|
a^{[l](G)}
-
a^{[l](C)}
\right\|_F^2
}
$$

The objective is:

$$
\boxed{
J_{\text{content}}\downarrow
\Rightarrow
\text{Generated Image preserves Content}
}
$$

The essential point is:

> **Content Cost does not compare the two images pixel by pixel. It compares their feature representations at a selected layer of a fixed pretrained ConvNet, encouraging the generated image to preserve the content represented by that layer.**

# Style Cost Function

In **Neural Style Transfer**, the **Style Cost Function** measures how different the visual style of the generated image is from the style image.

The objective is:

$$
\boxed{
\text{Generated Image}
\rightarrow
\text{Match the Style of the Style Image}
}
$$

Unlike the **Content Cost**, which compares feature activations directly, the Style Cost compares the **correlations between feature channels**.

---

## 1. Style Representation

Suppose a pretrained ConvNet produces an activation at layer $l$:

$$
a^{[l]}
\in
\mathbb R^{n_H\times n_W\times n_C}
$$

where:

- $n_H$: activation height;
- $n_W$: activation width;
- $n_C$: number of feature channels.

Each channel can respond to different visual patterns such as edges, textures, or other learned features.

To represent style, we are interested in:

> **How frequently different feature channels are activated together across the image.**

Therefore, style is represented through **feature correlations** rather than exact spatial positions.

---

## 2. Gram Matrix

First reshape the activation:

$$
a^{[l]}
\rightarrow
F
$$

where:

$$
F\in
\mathbb R^{(n_Hn_W)\times n_C}
$$

Each row corresponds to a spatial location, while each column corresponds to a feature channel.

The **Gram Matrix** is:

$$
\boxed{
G=F^TF
}
$$

and therefore:

$$
G\in
\mathbb R^{n_C\times n_C}
$$

Each element is:

$$
G_{ij}
=
\sum_k F_{ki}F_{kj}
$$

which measures the correlation between feature channels $i$ and $j$.

Thus:

$$
\boxed{
\text{Gram Matrix}
\rightarrow
\text{Feature Correlations}
}
$$

---

## 3. Why Does the Gram Matrix Represent Style?

Suppose a style image contains a certain texture.

The exact location of that texture is not necessarily important for style.

Instead, we care about the statistical relationship between visual patterns:

$$
\text{Feature}_1
\leftrightarrow
\text{Feature}_2
$$

$$
\text{Feature}_1
\leftrightarrow
\text{Feature}_3
$$

and so on.

The Gram Matrix summarizes these relationships.

Therefore two images can have different spatial arrangements but still have similar style representations if their feature correlations are similar.

This gives the key distinction:

$$
\boxed{
\text{Content}
\rightarrow
\text{Feature Activations}
}
$$

$$
\boxed{
\text{Style}
\rightarrow
\text{Feature Correlations}
}
$$

---

## 4. Style Representation of the Two Images

Let the style image be $S$ and the generated image be $G$.

At layer $l$:

$$
F^{[l](S)}
$$

is the feature matrix of the style image.

Its Gram Matrix is:

$$
G^{[l](S)}
=
(F^{[l](S)})^T
F^{[l](S)}
$$

Similarly, for the generated image:

$$
G^{[l](G)}
=
(F^{[l](G)})^T
F^{[l](G)}
$$

The goal is:

$$
\boxed{
G^{[l](G)}
\approx
G^{[l](S)}
}
$$

so that the generated image has similar feature correlations to the style image.

---

## 5. Style Cost at One Layer

A common formulation is:

$$
\boxed{
J_{\text{style}}^{[l]}(S,G)
=
\frac{1}{(2n_Hn_Wn_C)^2}
\left\|
G^{[l](G)}
-
G^{[l](S)}
\right\|_F^2
}
$$

The loss becomes smaller when the two Gram Matrices become more similar:

$$
J_{\text{style}}^{[l]}\downarrow
\Rightarrow
G^{[l](G)}
\approx
G^{[l](S)}
$$

---

## 6. Why Not Compare Activations Directly?

Suppose a texture appears at position:

$$
(x_1,y_1)
$$

in the style image but at:

$$
(x_2,y_2)
$$

in the generated image.

A direct comparison:

$$
a^{[l](G)}
\approx
a^{[l](S)}
$$

would strongly depend on the exact spatial positions.

But for style, we generally care more about:

> **Which visual features appear together and how strongly they correlate.**

The Gram Matrix removes the emphasis on exact spatial arrangement and focuses on feature statistics.

Thus:

$$
\boxed{
\text{Style}
\rightarrow
\text{Statistical Feature Relationships}
}
$$

rather than:

$$
\boxed{
\text{Style}
\rightarrow
\text{Exact Spatial Arrangement}
}
$$

---

## 7. Style Cost Across Multiple Layers

Different layers capture different levels of visual information.

For example:

$$
\text{Early Layers}
\rightarrow
\text{Low-Level Textures}
$$

$$
\text{Middle Layers}
\rightarrow
\text{Patterns and Structures}
$$

$$
\text{Deeper Layers}
\rightarrow
\text{Higher-Level Visual Structure}
$$

Therefore style cost is often computed across multiple layers:

$$
\boxed{
J_{\text{style}}(S,G)
=
\sum_l
\lambda^{[l]}
J_{\text{style}}^{[l]}(S,G)
}
$$

where:

$$
\lambda^{[l]}
$$

controls the contribution of layer $l$.

This allows the generated image to match style at multiple levels.

---

## 8. Forward Propagation

For the style image:

$$
S
\rightarrow
\text{Fixed ConvNet}
\rightarrow
a^{[l](S)}
\rightarrow
F^{[l](S)}
\rightarrow
G^{[l](S)}
$$

For the generated image:

$$
G
\rightarrow
\text{Fixed ConvNet}
\rightarrow
a^{[l](G)}
\rightarrow
F^{[l](G)}
\rightarrow
G^{[l](G)}
$$

Then:

$$
G^{[l](G)}
\leftrightarrow
G^{[l](S)}
\rightarrow
J_{\text{style}}^{[l]}
$$

The important point is:

$$
\boxed{
\text{Style Cost compares Gram Matrices, not raw pixels}
}
$$

---

## 9. Optimization of the Generated Image

In Neural Style Transfer, the pretrained CNN parameters remain fixed:

$$
\boxed{
\theta_{\text{CNN}}=\text{fixed}
}
$$

The generated image $G$ is the variable being optimized.

We compute:

$$
\frac{\partial J_{\text{style}}}{\partial G}
$$

and update:

$$
\boxed{
G
\leftarrow
G-\eta
\frac{\partial J_{\text{style}}}{\partial G}
}
$$

Thus the style cost produces gradients that modify the generated image so that its feature correlations become closer to those of the style image.

---

## 10. Style Cost vs. Content Cost

The key difference is:

### Content Cost

$$
\boxed{
J_{\text{content}}
=
\frac12
\left\|
a^{[l](G)}
-
a^{[l](C)}
\right\|_F^2
}
$$

It compares feature activations directly.

### Style Cost

$$
\boxed{
J_{\text{style}}^{[l]}
\propto
\left\|
G^{[l](G)}
-
G^{[l](S)}
\right\|_F^2
}
$$

It compares correlations between feature channels.

Therefore:

$$
\boxed{
\text{Content}
=
\text{What features are present}
}
$$

$$
\boxed{
\text{Style}
=
\text{How features correlate}
}
$$

---

## 11. Core Mental Model

The entire Style Cost computation can be summarized as:

$$
\boxed{
\text{Style Image}
\rightarrow
\text{CNN Features}
\rightarrow
\text{Gram Matrix}
}
$$

and:

$$
\boxed{
\text{Generated Image}
\rightarrow
\text{CNN Features}
\rightarrow
\text{Gram Matrix}
}
$$

then:

$$
\boxed{
\text{Compare Gram Matrices}
\rightarrow
\text{Style Cost}
}
$$

The objective is:

$$
\boxed{
J_{\text{style}}\downarrow
\Rightarrow
\text{Generated Image matches the Style Image}
}
$$

The key insight is:

> **Style Cost does not try to reproduce the style image pixel by pixel or preserve its exact spatial arrangement. It encourages the generated image to have similar correlations among learned visual features, which allows it to reproduce the texture and visual characteristics of the style image.**
