# Why Look at Case Studies?

## 1. Core Idea

We study successful CNN architectures not to memorize every layer, but to understand **reusable architecture design principles**.

Each architecture was created to solve a particular limitation of earlier CNNs.

The progression in this part of the course can be viewed as:

```text
LeNet-5
-> establish the basic CNN pattern

AlexNet
-> scale CNNs to large real-world image datasets

VGG
-> build deeper networks using simple repeated 3x3 convolutions

ResNet
-> make very deep networks easier to optimize

Inception
-> process multiple spatial scales efficiently

MobileNet
-> reduce computation for resource-constrained devices

EfficientNet
-> scale depth, width, and image resolution systematically
```

The goal is therefore:

```text
Do not memorize architectures.

Understand:
why each architecture was needed
and
what reusable idea it introduced.
```

## 2. What to Learn from Each Architecture

For every architecture, focus on:

```text
1. What problem existed before it?

2. What bottleneck was the architecture trying to solve?

3. What is the central design idea?

4. How does information flow through the network?

5. How do spatial resolution and channel depth change?

6. Where are parameters and computation concentrated?

7. Why does the new design improve optimization or efficiency?

8. Which ideas are still useful in modern architectures?
```

This makes architecture study much more useful than simply memorizing layer configurations.

## 3. Evolution of CNN Architecture Design

### 3.1. LeNet-5 — Basic CNN Design

Core pattern:

```mermaid
flowchart LR
    N0["Conv"]
    N1["Pool"]
    N2["Conv"]
    N3["Pool"]
    N4["Fully Connected"]
    N5["Output"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5
```

LeNet-5 demonstrated the fundamental CNN principles:

```text
local connectivity
+
parameter sharing
+
hierarchical feature extraction
+
spatial downsampling
```

It established the basic structure that later CNN architectures would scale and improve.

### 3.2. AlexNet — Scaling CNNs

Core idea:

```text
large CNN
+
ReLU
+
GPU training
+
data augmentation
+
dropout
```

AlexNet showed that CNNs could scale successfully to large natural-image datasets such as ImageNet.

Important lesson:

```text
CNN architecture
+
large datasets
+
sufficient computation
+
good optimization techniques
=
powerful visual recognition
```

AlexNet was an important transition from relatively small CNNs to large-scale deep learning.

### 3.3. VGG — Simple and Deep

Core idea:

```text
Use small 3x3 convolutions
and stack them repeatedly.
```

Instead of relying on large heterogeneous kernels, VGG uses a systematic pattern:

```mermaid
flowchart LR
    N0["3x3 Conv"]
    N1["ReLU"]
    N2["3x3 Conv"]
    N3["ReLU"]
    N4["Pool"]
    N0 --> N1 --> N2 --> N3 --> N4
```

Stacked $3\times3$ convolutions provide:

```text
larger effective receptive fields
+
more nonlinear transformations
+
greater depth
+
a simple repeated architecture
```

VGG demonstrated that very simple building blocks can create powerful deep feature hierarchies.

### 3.4. ResNet — Making Very Deep Networks Trainable

Core idea:

$$
\boxed{
H(x)=x+F(x)
}
$$

Instead of forcing a block to learn the complete mapping $H(x)$ directly, the residual branch learns:

$$
F(x)=H(x)-x
$$

while the shortcut provides $x$ directly.

Architecture:

```mermaid
flowchart LR
    I["x"]
    A(("+"))
    M0["Residual Branch<br/>F(x)"]
    I --> M0
    M0 --> A
    I --> A
    O["H(x)"]
    A --> O
```

This gives:

```text
easy identity mapping
+
feature reuse
+
short information paths
+
better gradient propagation
```

If:

$$
F(x)\approx0
$$

then:

$$
H(x)\approx x
$$

so an unnecessary residual block can behave approximately as identity.

This makes very deep networks much easier to optimize.

### 3.5. Network in Network and $1\times1$ Convolution — Channel Transformation

A $1\times1$ convolution performs a learned transformation across channels at every spatial location:

$$
H\times W\times C_{in}
\rightarrow
H\times W\times C_{out}
$$

Its main uses include:

```text
channel mixing
+
dimensionality reduction
+
channel expansion
+
additional nonlinear transformations
+
computational bottlenecks
```

This becomes especially important in Inception, ResNet bottlenecks, and many later architectures.

### 3.6. Inception — Multi-Scale Feature Extraction

Different visual patterns can occur at different spatial scales.

Instead of choosing only one convolution size, Inception processes several operations in parallel.

Conceptually:

```mermaid
flowchart LR
    I["Input"]
    C["Concatenate"]
    B0_0["1x1 Conv"]
    I --> B0_0
    B0_0 --> C
    B1_0["1x1 Conv"]
    I --> B1_0
    B1_1["3x3 Conv"]
    B1_0 --> B1_1
    B1_1 --> C
    B2_0["1x1 Conv"]
    I --> B2_0
    B2_1["5x5 Conv"]
    B2_0 --> B2_1
    B2_1 --> C
    B3_0["Pool"]
    I --> B3_0
    B3_1["1x1 Conv"]
    B3_0 --> B3_1
    B3_1 --> C
```

This allows the network to learn:

```text
small-scale features
+
medium-scale features
+
large-scale features
```

at the same stage.

However, large convolutions can be computationally expensive.

Therefore, $1\times1$ convolutions are used as bottlenecks before expensive convolutions.

Main lesson:

```text
multi-scale processing
+
parallel branches
+
1x1 bottlenecks
=
strong features with controlled computation
```

### 3.7. MobileNet — Efficient Convolution

As CNNs became more powerful, another problem became important:

```text
How can we reduce computation
for mobile and resource-constrained devices?
```

MobileNet addresses this using **depthwise separable convolution**.

Instead of a standard convolution performing spatial and channel mixing simultaneously:

```text
Standard Conv:

spatial processing
+
channel mixing
```

MobileNet separates the operation into:

```text
Depthwise Conv
-> spatial processing independently per channel

Pointwise 1x1 Conv
-> mix information across channels
```

Conceptually:

```mermaid
flowchart LR
    N0["Input"]
    N1["Depthwise Conv"]
    N2["Pointwise 1x1 Conv"]
    N3["Output"]
    N0 --> N1 --> N2 --> N3
```

This significantly reduces:

```text
parameters
+
FLOPs
+
computational cost
```

while maintaining useful representational capacity.

Main lesson:

> **Expensive convolution can sometimes be factorized into cheaper specialized operations.**

### 3.8. EfficientNet — Systematic Model Scaling

Once efficient CNN blocks exist, another question appears:

```text
If we have more computational resources,
how should we make the model larger?
```

There are three main ways to scale a CNN:

```text
Depth:
add more layers

Width:
add more channels

Resolution:
use larger input images
```

A naive approach might increase only one dimension.

EfficientNet instead proposes **compound scaling**:

```text
scale depth
+
scale width
+
scale input resolution
in a coordinated way
```

The central idea is:

> Do not scale CNN dimensions independently and arbitrarily.

Instead, balance:

$$
\text{depth}
,\quad
\text{width}
,\quad
\text{resolution}
$$

according to available computational resources.

Main lesson:

```text
efficient architecture
+
balanced scaling
=
better accuracy-computation trade-off
```

## 4. Design Principles Learned from the Case Studies

These architectures introduce several reusable ideas.

### Depth

```text
VGG:
greater depth can create stronger hierarchical representations
```

### Skip Connections

```text
ResNet:
very deep networks need easier information and gradient paths
```

### Channel Transformation

```text
1x1 Conv:
mix and reshape channel representations efficiently
```

### Multi-Scale Processing

```text
Inception:
different receptive-field sizes can be processed in parallel
```

### Efficient Factorization

```text
MobileNet:
separate spatial processing from channel mixing
```

### Systematic Scaling

```text
EfficientNet:
scale depth, width, and resolution together
```

These ideas are more important than memorizing the exact number of layers in each architecture.

# Classic Networks

## 1. Core Idea

In Course 4 Week 2, Classic Networks usually refer to three important CNN architectures:

```text
LeNet-5 -> AlexNet -> VGG
```

They represent three important stages in the evolution of CNN design:

```text
LeNet-5  -> basic CNN for small grayscale images and document recognition
AlexNet  -> large CNN for ImageNet-scale natural image classification
VGG      -> deeper CNN built from repeated 3x3 convolution blocks
```

The goal is not to memorize every layer, but to understand the architecture design principles:

```text
1. What problem was the model designed for?
2. What bottleneck did it address?
3. Why was the architecture designed that way?
4. How do spatial size and channel depth change?
5. Where are most parameters located?
6. Which ideas are still useful today?
```

General parameter formulas:

For a standard convolution:

$$
\text{Params}
=
f_h \times f_w \times C_{\text{in}} \times C_{\text{out}}
+
C_{\text{out}}
$$

For a grouped convolution:

$$
\text{Params}
=
f_h \times f_w
\times
\frac{C_{\text{in}}}{g}
\times
C_{\text{out}}
+
C_{\text{out}}
$$

where:

$$
g = \text{number of groups}
$$

For a fully connected layer:

$$
\text{Params}
=
n_{\text{in}} \times n_{\text{out}}
+
n_{\text{out}}
$$

For modern MaxPool or AvgPool:

$$
\text{Params} = 0
$$

However, original LeNet-5 uses trainable subsampling layers, so its S2 and S4 layers have trainable parameters.


## 2. LeNet-5

## 2.1. Design Intuition

LeNet-5 was designed for handwritten digit and character recognition, especially in document recognition.

The key intuition is:

```text
Images have local structure.

Small visual patterns such as edges, curves, and strokes
can appear at different positions.

Therefore, use small filters to scan the image instead of
using fully connected layers from the start.
```

Design pipeline:

```mermaid
flowchart LR
    N0["Local pattern detection"]
    N1["Local downsampling"]
    N2["Higher-level feature extraction"]
    N3["Classification"]
    N0 --> N1 --> N2 --> N3
```

LeNet-5 should be understood as:

```mermaid
flowchart LR
    N0["Image"]
    N1["Detect small strokes, edges, and curves"]
    N2["Reduce spatial size"]
    N3["Combine local features into higher-level patterns"]
    N4["Classify the digit"]
    N0 --> N1 --> N2 --> N3 --> N4
```

Important differences from modern simplified LeNet:

```text
1. S2 and S4 are trainable subsampling layers,
   not standard parameter-free MaxPool or AvgPool.

2. C3 is partially connected,
   not a full convolution.

3. The output layer uses fixed RBF prototype vectors,
   not a modern Dense + Softmax classifier.
```


## 2.2. Original Architecture

```text
Input: 32 x 32 x 1

C1: Convolution
-> 6 feature maps
-> 28 x 28
-> kernel 5 x 5

S2: Trainable subsampling
-> 6 feature maps
-> 14 x 14
-> each map has one trainable coefficient and one bias

C3: Convolution with partial connections
-> 16 feature maps
-> 10 x 10
-> kernel 5 x 5
-> not every output map connects to all 6 input maps

S4: Trainable subsampling
-> 16 feature maps
-> 5 x 5
-> each map has one trainable coefficient and one bias

C5: Convolution
-> 120 feature maps
-> 1 x 1
-> kernel 5 x 5 over all 16 S4 maps

F6: Fully connected
-> 84 units

Output:
-> 10 Euclidean RBF units
-> each unit compares F6 with a fixed 84-dimensional prototype
```

![Lenet-5](https://www.philschmid.de/static/blog/getting-started-with-cnn-by-calculating-lenet-layer-manually/lenet-5.svg)


## 2.3. Shape Flow

```mermaid
flowchart LR
    N0["32 x 32 x 1"]
    N1["C1: 28 x 28 x 6"]
    N2["S2: 14 x 14 x 6"]
    N3["C3: 10 x 10 x 16"]
    N4["S4: 5 x 5 x 16"]
    N5["C5: 1 x 1 x 120"]
    N6["F6: 84"]
    N7["Output: 10"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6 --> N7
```

Convolution output size:

$$
H_{\text{out}}
=
\left\lfloor
\frac{H + 2p - f}{s}
\right\rfloor
+1
$$

For C1:

$$
\frac{32-5}{1}+1=28
$$

For C3:

$$
\frac{14-5}{1}+1=10
$$

For C5:

```text
Input spatial size = 5 x 5
Kernel size        = 5 x 5
Stride             = 1

Output spatial size = 1 x 1
```

Therefore:

$$
5\times5\times16
\rightarrow
1\times1\times120
$$

Although C5 is formally a convolutional layer, its kernel covers the entire spatial extent of S4.

For this fixed input size, it therefore behaves similarly to a fully connected transformation:

$$
400 \rightarrow 120
$$

because:

$$
5\times5\times16=400
$$


## 2.4. Parameter Count

| Layer | Output Shape | Parameter Reasoning | Trainable Parameters |
|---|---:|---:|---:|
| C1 | $28 \times 28 \times 6$ | $6(5\times5\times1+1)$ | 156 |
| S2 | $14 \times 14 \times 6$ | $6\times2$ | 12 |
| C3 | $10 \times 10 \times 16$ | $60\times5\times5+16$ | 1,516 |
| S4 | $5 \times 5 \times 16$ | $16\times2$ | 32 |
| C5 | $1 \times 1 \times 120$ | $120(5\times5\times16+1)$ | 48,120 |
| F6 | 84 | $120\times84+84$ | 10,164 |
| Output RBF | 10 | 10 fixed prototypes of dimension 84 | 0 trainable |

Total trainable parameters:

$$
156
+
12
+
1516
+
32
+
48120
+
10164
=
60000
$$

Therefore:

```text
Original LeNet-5 = 60,000 trainable parameters
```

### Important RBF Detail

The output layer contains:

$$
10\times84=840
$$

prototype values.

However, these prototype vectors are fixed rather than learned by backpropagation.

Therefore:

```text
840 RBF prototype values
≠
840 trainable parameters
```

The original trainable parameter count remains:

$$
\boxed{60{,}000}
$$

If the original RBF output is replaced by a modern Dense classifier with bias:

$$
84\times10+10
=
850
$$

then, while keeping the rest of the original architecture unchanged:

$$
60000+850
=
60850
$$

So:

```text
LeNet-5 with modern Dense output
but original C3/S2/S4 structure
= 60,850 trainable parameters
```

Important C3 correction:

```text
Original C3 is partially connected.

C3 original:
1,516 parameters

C3 simplified full convolution:
5 x 5 x 6 x 16 + 16
= 2,416 parameters
```


## 2.5. Why LeNet-5 Matters

LeNet-5 established the basic CNN design pattern:

```text
local receptive fields
weight sharing
subsampling
hierarchical feature extraction
classifier
```

It shows that images should not be treated as flat vectors from the beginning.

Instead:

```text
preserve spatial structure
-> extract local features
-> progressively combine features
-> reduce spatial resolution
-> classify using higher-level representations
```

This is the fundamental design philosophy behind convolutional neural networks.


## 2.6. Common Mistakes

| Mistake | Correct Understanding |
|---|---|
| Counting C3 as $5\times5\times6\times16+16$ | That is the simplified full-convolution version; original C3 has 1,516 parameters |
| Treating S2/S4 as parameter-free MaxPool | Original S2/S4 are trainable subsampling layers |
| Counting the 840 RBF prototype values as trainable parameters | The original RBF prototypes are fixed; original LeNet-5 has 60,000 trainable parameters |
| Treating the output as modern Softmax | Original LeNet-5 uses Euclidean RBF output units |
| Confusing C5 with a normal FC layer | C5 is formally a convolution over a $5\times5\times16$ input |
| Memorizing only the layer flow | The main idea is local feature extraction, parameter sharing, hierarchical composition, and downsampling |


## 3. AlexNet

## 3.1. Design Intuition

AlexNet was designed for ImageNet-scale natural image classification.

Compared with MNIST-like digit recognition, ImageNet is much harder:

```text
high-resolution RGB images
1,000 object classes
millions of labeled examples
complex backgrounds
diverse objects
large variation in viewpoint, lighting, and scale
```

The design intuition is:

```text
LeNet showed that CNNs work on small visual problems.

Now scale CNNs to large natural images using:

large-scale data
+ GPUs
+ deeper networks
+ ReLU
+ dropout
+ augmentation
```

AlexNet is not only an architecture.

It is also a large-scale CNN training recipe:

```text
large-scale dataset
+ deep CNN
+ ReLU
+ GPU implementation
+ local response normalization
+ overlapping max pooling
+ data augmentation
+ dropout
```


## 3.2. Original Architecture

AlexNet has 8 layers with learnable weights:

```text
5 convolutional layers
3 fully connected layers
```

Overall flow:

```text
Input
-> Conv1 -> ReLU -> LRN -> MaxPool
-> Conv2 -> ReLU -> LRN -> MaxPool
-> Conv3 -> ReLU
-> Conv4 -> ReLU
-> Conv5 -> ReLU -> MaxPool
-> FC6 -> ReLU -> Dropout
-> FC7 -> ReLU -> Dropout
-> FC8 -> Softmax
```

Layer specifications:

| Layer | Specification |
|---|---|
| Conv1 | 96 kernels, $11\times11\times3$, stride 4 |
| Conv2 | 256 kernels, $5\times5\times48$, groups = 2 |
| Conv3 | 384 kernels, $3\times3\times256$, groups = 1 |
| Conv4 | 384 kernels, $3\times3\times192$, groups = 2 |
| Conv5 | 256 kernels, $3\times3\times192$, groups = 2 |
| FC6 | 4096 units |
| FC7 | 4096 units |
| FC8 | 1000 units |

Important grouped-convolution detail:

```text
Original AlexNet was distributed across two GPUs.

Conv2, Conv4, and Conv5 use grouped connections.

Conv3 is a full convolution and connects across all
feature maps from Conv2.
```

In modern grouped-convolution notation:

```text
Conv1: groups = 1
Conv2: groups = 2
Conv3: groups = 1
Conv4: groups = 2
Conv5: groups = 2
```

A crucial distinction is:

```text
total input channels
!=
channels seen by one kernel
```

For Conv2:

```text
Input tensor:
96 channels

groups = 2

Channels seen by each kernel:
96 / 2 = 48
```

For Conv4 and Conv5:

```text
Input tensor:
384 channels

groups = 2

Channels seen by each kernel:
384 / 2 = 192
```

Therefore:

```text
48 and 192 are kernel depths / channels per group.

They are not the total channel counts of the input tensors.
```

## 3.3. Shape Flow

A common canonical shape flow uses:

```text
Input: 224 x 224 x 3
Conv1 padding: 2
```

Then:

```mermaid
flowchart TD
    N0["Input<br/>224 x 224 x 3"]
    N1["Conv1<br/>11x11, stride 4, padding 2, 96 filters<br/>55 x 55 x 96"]
    N2["ReLU + LRN"]
    N3["MaxPool<br/>3x3, stride 2<br/>27 x 27 x 96"]
    N4["Conv2<br/>5x5, padding 2, 256 filters, groups = 2<br/>27 x 27 x 256"]
    N5["ReLU + LRN"]
    N6["MaxPool<br/>3x3, stride 2<br/>13 x 13 x 256"]
    N7["Conv3<br/>3x3, padding 1, 384 filters, groups = 1<br/>13 x 13 x 384"]
    N8["ReLU"]
    N9["Conv4<br/>3x3, padding 1, 384 filters, groups = 2<br/>13 x 13 x 384"]
    N10["ReLU"]
    N11["Conv5<br/>3x3, padding 1, 256 filters, groups = 2<br/>13 x 13 x 256"]
    N12["ReLU"]
    N13["MaxPool<br/>3x3, stride 2<br/>6 x 6 x 256"]
    N14["Flatten<br/>9216"]
    N15["FC6<br/>4096"]
    N16["FC7<br/>4096"]
    N17["FC8<br/>1000"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6 --> N7 --> N8 --> N9 --> N10 --> N11 --> N12 --> N13 --> N14 --> N15 --> N16 --> N17
```

### Input-Size Convention

There is a well-known historical inconsistency around AlexNet's first-layer input size.

A common modern canonical convention is:

```text
Input = 224 x 224
kernel = 11 x 11
stride = 4
padding = 2
```

which produces:

$$
\left\lfloor
\frac{224+2(2)-11}{4}
\right\rfloor+1
=
55
$$

Another common historical implementation convention is:

```text
Input = 227 x 227
kernel = 11 x 11
stride = 4
padding = 0
```

which also gives:

$$
\frac{227-11}{4}+1
=
55
$$

For these notes, we use:

```text
224 x 224 input
+
padding = 2
```

because it preserves the canonical:

```text
55 -> 27 -> 13 -> 6
```

shape flow.

### Spatial Size

A compressed view is:

```text
224 -> 55 -> 27 -> 13 -> 6
```

More precisely:

```text
Input    : 224
Conv1    : 55
Pool1    : 27
Conv2    : 27
Pool2    : 13
Conv3    : 13
Conv4    : 13
Conv5    : 13
Pool3    : 6
```

Therefore:

```text
Spatial resolution progressively decreases,
but it does not decrease after every convolution.
```

### Channel Depth

```text
3
-> 96
-> 256
-> 384
-> 384
-> 256
```

Therefore:

```text
Channel depth generally increases in the early and middle layers,
but Conv5 reduces channels from 384 to 256.
```

Main pattern:

```text
Spatial resolution is progressively reduced.

Channel depth becomes much larger than the RGB input depth.

High-level convolutional features are finally flattened.

The fully connected classifier head is extremely large.
```

### Important Grouped-Convolution Corrections

Conv2 receives:

$$
96
$$

total channels.

However:

$$
\frac{96}{2}=48
$$

$$

$$

$$

\frac{58631144}{60965224}
\approx
0.962
$$

Therefore:

```text
Approximately 96.2% of AlexNet's parameters
are in FC6, FC7, and FC8.
```

This reveals one major weakness of classic CNN architectures:

```text
Flatten a relatively large feature map
-> enormous fully connected classifier
-> huge parameter count
```


## 3.6. Important Training Components

### ReLU

AlexNet uses:

$$
\operatorname{ReLU}(z)
=
\max(0,z)
$$

ReLU helped train the network much faster than saturating nonlinearities such as sigmoid or tanh.

Core intuition:

```text
simple computation
+
non-saturating positive region
+
better optimization for deep networks
```

### Local Response Normalization

LRN is applied after Conv1 and Conv2:

```mermaid
flowchart LR
    A1["Conv1"] --> B1["ReLU"] --> C1["LRN"] --> D1["MaxPool"]
    A2["Conv2"] --> B2["ReLU"] --> C2["LRN"] --> D2["MaxPool"]
```

LRN:

```text
operates across nearby feature maps
does not change tensor shape
has no trainable parameters
```

Its purpose was to create competition among strong activations at the same spatial location.

LRN is rarely used in modern CNNs.

### Overlapping Max Pooling

AlexNet uses:

```text
MaxPool:
kernel = 3 x 3
stride = 2
```

Since:

$$
3>2
$$

neighboring pooling windows overlap.

Therefore, it is called:

```text
overlapping max pooling
```

### Dropout

Dropout is applied after FC6 and FC7:

```mermaid
flowchart LR
    A1["FC6"] --> B1["ReLU"] --> C1["Dropout"]
    A2["FC7"] --> B2["ReLU"] --> C2["Dropout"]
```

The original dropout probability is:

$$
p=0.5
$$

It helps reduce overfitting in the extremely large fully connected classifier.

### Data Augmentation

Important augmentation strategies included:

```text
random crops
horizontal reflections
RGB intensity perturbations
```

AlexNet therefore succeeded not only because of its architecture, but because of the complete training recipe.


## 3.7. Why AlexNet Matters

AlexNet showed that CNNs could scale to large natural-image classification when combined with:

```text
large-scale data
GPU training
ReLU
data augmentation
dropout
large model capacity
```

Its historical importance is not simply:

```text
AlexNet is deeper than LeNet.
```

The deeper idea is:

```text
CNNs can scale from small handwritten-digit problems
to large real-world natural-image recognition
when enough data and computation are available.
```

AlexNet demonstrated the effectiveness of end-to-end feature learning:

```mermaid
flowchart LR
    N0["Raw pixels"]
    N1["Learned low-level features"]
    N2["Learned mid-level features"]
    N3["Learned high-level features"]
    N4["Classification"]
    N0 --> N1 --> N2 --> N3 --> N4
```

instead of relying primarily on hand-crafted visual features.

| Mistake | Correct Understanding |
|---|---|
| Saying AlexNet has only 5 layers | It has 5 convolutional + 3 fully connected = 8 learned layers |
| Counting Conv2 as $5\times5\times96\times256+256$ | Conv2 receives 96 channels, but each kernel sees only 48 because `groups = 2` |
| Saying Conv2 has only 48 input channels | The full tensor has 96 channels; 48 is the kernel depth / channels per group |
| Counting Conv4/Conv5 as full convolutions | They use `groups = 2` |
| Saying Conv4/Conv5 receive only 192 channels | They receive 384 total channels; each kernel sees 192 |
| Treating Conv3 as grouped | Conv3 is a full convolution with `groups = 1` |
| Adding pooling after Conv3 or Conv4 | Conv3, Conv4, and Conv5 are consecutive convolutional layers |
| Ignoring LRN | Original AlexNet includes LRN after Conv1 and Conv2 |
| Treating LRN as BatchNorm | Their normalization mechanisms are fundamentally different |
| Ignoring overlapping pooling | AlexNet uses $3\times3$ MaxPool with stride 2 |
| Thinking pooling is trainable | MaxPool has 0 trainable parameters |
| Thinking AlexNet succeeded only because it was deeper | Data, GPUs, ReLU, augmentation, and dropout were also crucial |
| Ignoring FC layers | Approximately 96% of the parameters are in FC6-FC8 |
| Saying spatial size decreases after every convolution | Conv2-Conv5 mostly preserve spatial resolution |
| Saying channel depth always increases | Conv5 reduces channels from 384 to 256 |
| Assuming $224\to55$ with stride 4 requires no explanation | The canonical 224 convention uses padding 2; 227 with no padding is another common historical convention |
| Always putting Softmax inside a modern PyTorch model | With `CrossEntropyLoss`, the model should normally return raw logits |


## 4. VGG-16

## 4.1. Design Intuition

After AlexNet, the natural question was:

```text
If large CNNs work,
can deeper CNNs work even better?
```

VGG answered with a clean design principle:

```text
Use very small 3x3 filters.
Stack them repeatedly.
Increase depth systematically.
```

The main intuition is:

```text
A simple block, repeated many times,
can create a powerful deep representation.
```

Compared with AlexNet, VGG makes the architecture much more systematic.

Instead of mixing large convolution sizes such as:

```text
11x11
5x5
3x3
```

VGG mainly uses:

```text
3x3 convolution
stride = 1
padding = 1
```

throughout the convolutional backbone.

The design philosophy can be summarized as:

```text
small convolution kernels
+
deep stacking
+
repeated simple blocks
+
gradual downsampling
=
strong hierarchical representations
```

## 4.2. Why Use Many 3x3 Convolutions?

One of the most important ideas in VGG is that several small $3\times3$ convolutions can provide the same **theoretical receptive-field size** as a single larger convolution.

Assume:

```text
kernel size = 3x3
stride = 1
dilation = 1
```

Then:

```text
1 Conv 3x3
-> receptive field 3x3

2 Conv 3x3
-> receptive field 5x5

3 Conv 3x3
-> receptive field 7x7
```

However, this does **not** mean that:

```text
two 3x3 convolutions
=
one 5x5 convolution
```

as mathematical functions.

It only means that a neuron after two stacked $3\times3$ convolutions can depend on a $5\times5$ region of the original input.

Similarly, a neuron after three stacked $3\times3$ convolutions can depend on a $7\times7$ region.

### 4.2.1. Receptive Field Intuition

Consider one spatial dimension first.

Suppose the original input is:

```text
x1 x2 x3 x4 x5 x6 x7 ...
```

A convolution with kernel size 3 computes one output from three neighboring input positions.

For example:

```mermaid
flowchart TD
    N0["x3, x4, x5"]
    N1["y4"]
    N0 --> N1
```

Therefore, $y_4$ has a receptive field of size:

$$
3
$$

Now apply another convolution with kernel size 3.

Suppose:

```mermaid
flowchart TD
    N0["y3, y4, y5"]
    N1["z4"]
    N0 --> N1
```

Each $y$ already depends on three positions in the original input:

```text
y3 depends on:
x2 x3 x4

y4 depends on:
x3 x4 x5

y5 depends on:
x4 x5 x6
```

Therefore, $z_4$ indirectly depends on the union:

```text
x2 x3 x4 x5 x6
```

which contains:

$$
5
$$

input positions.

Therefore:

```text
3 -> 3
```

produces an effective receptive field of:

$$
5
$$

in one dimension.

In two dimensions:

$$
\boxed{
3\times3
\rightarrow
3\times3
\Rightarrow
5\times5
\text{ receptive field}
}
$$

### 4.2.2. Why Does the Receptive Field Grow by 2?

A $3\times3$ kernel extends one pixel away from its center in every spatial direction.

Conceptually:

```text
First 3x3 convolution:

radius = 1

receptive-field width:
1 + 1 + 1 = 3
```

After another $3\times3$ convolution, the network can look one additional input position farther in each direction:

```text
radius = 2

receptive-field width:
2 + 1 + 2 = 5
```

After a third convolution:

```text
radius = 3

receptive-field width:
3 + 1 + 3 = 7
```

Therefore:

```text
1 Conv 3x3 -> 3x3
2 Conv 3x3 -> 5x5
3 Conv 3x3 -> 7x7
4 Conv 3x3 -> 9x9
```

Each additional $3\times3$ convolution increases the receptive-field width by:

$$
3-1=2
$$

The reason it is not:

$$
3+3=6
$$

is that neighboring receptive fields overlap.

The second convolution does not introduce three completely new pixels in each direction. It extends the previous receptive field by only one position on the left and one position on the right:

$$
3+1+1=5
$$

### 4.2.3. General Receptive Field Formula

For consecutive convolutional layers with:

```text
stride = 1
dilation = 1
```

the receptive field follows:

$$
R_l
=
R_{l-1}
+
(k_l-1)
$$

where:

- $R_l$ is the receptive-field width after layer $l$
- $k_l$ is the kernel size of layer $l$

Initially:

$$
R_0=1
$$

For a $3\times3$ convolution:

$$
k=3
$$

so:

$$
R_l
=
R_{l-1}+2
$$

After the first convolution:

$$
R_1
=
1+2
=
3
$$

After the second convolution:

$$
R_2
=
3+2
=
5
$$

After the third convolution:

$$
R_3
=
5+2
=
7
$$

Therefore, for $n$ consecutive $3\times3$ convolutions:

$$
\boxed{
R_n=1+2n
}
$$

So:

$$
n=1
\Rightarrow
R=3
$$

$$
n=2
\Rightarrow
R=5
$$

$$
n=3
\Rightarrow
R=7
$$

In two-dimensional images:

$$
\boxed{
R_n\times R_n
=
(2n+1)\times(2n+1)
}
$$

Thus:

```text
2 stacked 3x3 convolutions
-> 5x5 receptive field

3 stacked 3x3 convolutions
-> 7x7 receptive field
```

### 4.2.4. Same Receptive Field Does Not Mean Same Operation

It is important to distinguish:

```text
same receptive-field size
```

from:

```text
same mathematical transformation
```

Consider:

```text
5x5 Conv
-> ReLU
```

versus:

```mermaid
flowchart LR
    N0["3x3 Conv"]
    N1["ReLU"]
    N2["3x3 Conv"]
    N3["ReLU"]
    N0 --> N1 --> N2 --> N3
```

Both can produce an output whose theoretical receptive field is:

$$
5\times5
$$

But they are not equivalent networks.

The stacked version contains:

```text
two convolutional transformations
+
two nonlinear activation stages
```

while the single $5\times5$ convolution contains only:

```text
one convolutional transformation
+
one nonlinear activation stage
```

The same applies to:

```text
7x7 Conv
-> ReLU
```

versus:

```mermaid
flowchart LR
    N0["3x3 Conv"]
    N1["ReLU"]
    N2["3x3 Conv"]
    N3["ReLU"]
    N4["3x3 Conv"]
    N5["ReLU"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5
```

Both have a theoretical receptive-field size of:

$$
7\times7
$$

but the deeper version performs more intermediate nonlinear transformations.

This is one reason why stacking small convolutions can produce more expressive representations.

### 4.2.5. Parameter Advantage

Stacking small $3\times3$ convolutions can also reduce the number of parameters.

Assume every convolution has the same input and output channel width:

$$
C
$$

and ignore biases.

#### One 5x5 Convolution

A single $5\times5$ convolution requires:

$$
5\times5\times C\times C
=
25C^2
$$

parameters.

#### Two 3x3 Convolutions

Each $3\times3$ convolution requires:

$$
3\times3\times C\times C
=
9C^2
$$

Two layers require:

$$
2\times9C^2
=
18C^2
$$

Therefore:

$$
18C^2<25C^2
$$

So:

```text
2 stacked 3x3 convolutions

receptive field:
5x5

parameters:
18C²
```

compared with:

```text
1 convolution 5x5

receptive field:
5x5

parameters:
25C²
```

For a $7\times7$ receptive field:

#### One 7x7 Convolution

$$
7\times7\times C\times C
=
49C^2
$$

#### Three 3x3 Convolutions

$$
3
\times
(3\times3\times C\times C)
=
27C^2
$$

Therefore:

$$
27C^2<49C^2
$$

So:

```text
3 stacked 3x3 convolutions

receptive field:
7x7

parameters:
27C²
```

compared with:

```text
1 convolution 7x7

receptive field:
7x7

parameters:
49C²
```

### 4.2.6. Why This Matters for VGG

The important comparison is:

```text
Large-kernel approach:

7x7 Conv
-> ReLU
```

versus:

```mermaid
flowchart LR
    N0["3x3 Conv"]
    N1["ReLU"]
    N2["3x3 Conv"]
    N3["ReLU"]
    N4["3x3 Conv"]
    N5["ReLU"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5
```

The VGG-style version gives:

```text
same theoretical receptive-field size
+
fewer parameters under equal channel width
+
greater depth
+
more nonlinear transformations
```

Therefore:

```text
small filters
+
deep stacking
+
large effective receptive field
+
more nonlinear transformations
+
fewer parameters than equivalent large kernels
```

is one of the central VGG design principles.

A useful distinction to remember is:

```text
Feature-map spatial size
and
receptive-field size
are different concepts.
```

For example:

```mermaid
flowchart LR
    N0["Input<br/>224 x 224"]
    N1["Conv 3x3, stride 1, padding 1<br/>feature map still 224 x 224"]
    N2["Another Conv 3x3<br/>feature map still 224 x 224"]
    N0 --> N1 --> N2
```

but the receptive field of one activation changes:

```text
after first Conv:
3x3

after second Conv:
5x5
```

Therefore:

```text
spatial tensor size can remain unchanged
while receptive field grows with depth
```

## 4.3. VGG-16 Configuration D Architecture

VGG-16 Configuration D uses:

```text
13 convolutional layers
3 fully connected layers
```

Therefore:

$$
13+3=16
$$

layers with learnable weights.

Architecture:

```text
Input: 224 x 224 x 3

Block 1:
conv3-64
ReLU
conv3-64
ReLU
MaxPool

Block 2:
conv3-128
ReLU
conv3-128
ReLU
MaxPool

Block 3:
conv3-256
ReLU
conv3-256
ReLU
conv3-256
ReLU
MaxPool

Block 4:
conv3-512
ReLU
conv3-512
ReLU
conv3-512
ReLU
MaxPool

Block 5:
conv3-512
ReLU
conv3-512
ReLU
conv3-512
ReLU
MaxPool

FC1:
4096
ReLU
Dropout

FC2:
4096
ReLU
Dropout

FC3:
1000

Softmax
```

Convolution details:

```text
kernel size = 3 x 3
stride = 1
padding = 1
```

Therefore, convolution preserves spatial size inside each block.

Using:

$$
H_{\text{out}}
=
\left\lfloor
\frac{H+2p-f}{s}
\right\rfloor+1
$$

with:

$$
f=3,\qquad p=1,\qquad s=1
$$

gives:

$$
H_{\text{out}}
=
H
$$

MaxPool details:

```text
kernel size = 2 x 2
stride = 2
```

Therefore, every MaxPool operation halves height and width for the standard VGG feature-map sizes.

The original training setup applies dropout to the first two fully connected layers with:

$$
p=0.5
$$

## 4.4. Shape Flow

Because VGG uses padding 1 for $3\times3$ convolutions, spatial size is preserved inside each block.

Spatial size decreases after MaxPool.

```mermaid
flowchart TD
    N0["Input<br/>224 x 224 x 3"]
    N1["Block 1<br/>224 x 224 x 64"]
    N2["MaxPool<br/>112 x 112 x 64"]
    N3["Block 2<br/>112 x 112 x 128"]
    N4["MaxPool<br/>56 x 56 x 128"]
    N5["Block 3<br/>56 x 56 x 256"]
    N6["MaxPool<br/>28 x 28 x 256"]
    N7["Block 4<br/>28 x 28 x 512"]
    N8["MaxPool<br/>14 x 14 x 512"]
    N9["Block 5<br/>14 x 14 x 512"]
    N10["MaxPool<br/>7 x 7 x 512"]
    N11["Flatten<br/>7 x 7 x 512 = 25088"]
    N12["FC1<br/>4096"]
    N13["FC2<br/>4096"]
    N14["FC3<br/>1000"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6 --> N7 --> N8 --> N9 --> N10 --> N11 --> N12 --> N13 --> N14
```

Spatial size:

```text
224
-> 112
-> 56
-> 28
-> 14
-> 7
```

Channel depth:

```text
3
-> 64
-> 128
-> 256
-> 512
-> 512
```

Main pattern:

```text
Preserve H and W inside a convolution block.

Apply several feature transformations
before reducing spatial resolution.

Downsample using MaxPool.

Increase channel depth as spatial resolution decreases.

Use the same simple 3x3 convolution design repeatedly.
```

This produces the general pattern:

```text
large spatial resolution
+
fewer feature channels

        ↓

smaller spatial resolution
+
more feature channels
```

Early layers focus more strongly on local spatial information.

Deeper layers represent increasingly complex and abstract feature types.

## 4.6. Why VGG Matters

VGG is important because it introduced a clean and reusable CNN design philosophy:

```text
small filters
repeated blocks
greater depth
gradual downsampling
gradual channel increase
```

It demonstrated that a powerful architecture does not necessarily require complicated heterogeneous layer designs.

Instead:

```text
simple operation
+
systematic repetition
+
greater depth
=
strong hierarchical representation
```

The repeated pattern:

```mermaid
flowchart LR
    N0["Conv 3x3"]
    N1["ReLU"]
    N2["Conv 3x3"]
    N3["ReLU"]
    N4["MaxPool"]
    N0 --> N1 --> N2 --> N3 --> N4
```

and:

```mermaid
flowchart LR
    N0["Conv 3x3"]
    N1["ReLU"]
    N2["Conv 3x3"]
    N3["ReLU"]
    N4["Conv 3x3"]
    N5["ReLU"]
    N6["MaxPool"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6
```

became an important architectural idea.

The progression can be understood as:

```text
early layers
-> edges and simple local patterns

middle layers
-> textures and larger structures

deeper layers
-> object parts and semantic representations
```

Therefore:

```mermaid
flowchart LR
    N0["Pixels"]
    N1["Edges"]
    N2["Textures"]
    N3["Shapes"]
    N4["Object parts"]
    N5["Semantic features"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5
```

VGG also demonstrated an important principle:

```text
Increasing depth can increase representation power,
even when each individual layer is very simple.
```

This idea influenced many later CNN architectures.

## 4.8. Essence

The core VGG idea is:

```text
Do not use complicated large convolution kernels.

Instead:

use small 3x3 convolutions
+
stack them deeply
+
preserve spatial resolution inside each block
+
downsample gradually
+
increase channel depth progressively
```

The most important receptive-field relationship is:

```text
1 Conv 3x3
-> receptive field 3x3

2 Conv 3x3
-> receptive field 5x5

3 Conv 3x3
-> receptive field 7x7
```

because, with stride 1 and dilation 1:

$$
R_l
=
R_{l-1}
+
(k-1)
$$

and for:

$$
k=3
$$

each additional layer increases receptive-field width by:

$$
2
$$

Therefore:

$$
\boxed{
R_n=2n+1
}
$$

for $n$ consecutive $3\times3$ convolutions.

The important distinction is:

```text
same receptive-field size
does not mean
same mathematical operation
```

Stacked small convolutions provide:

```text
large receptive field
+
greater depth
+
more nonlinear transformations
+
fewer parameters under equal channel width
```

The VGG-16 shape flow is:

```mermaid
flowchart TD
    N0["Input<br/>224 x 224 x 3"]
    N1["Block 1<br/>224 x 224 x 64"]
    N2["Pool<br/>112 x 112 x 64"]
    N3["Block 2<br/>112 x 112 x 128"]
    N4["Pool<br/>56 x 56 x 128"]
    N5["Block 3<br/>56 x 56 x 256"]
    N6["Pool<br/>28 x 28 x 256"]
    N7["Block 4<br/>28 x 28 x 512"]
    N8["Pool<br/>14 x 14 x 512"]
    N9["Block 5<br/>14 x 14 x 512"]
    N10["Pool<br/>7 x 7 x 512"]
    N11["Flatten<br/>25088"]
    N12["FC1<br/>4096"]
    N13["FC2<br/>4096"]
    N14["FC3<br/>1000"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6 --> N7 --> N8 --> N9 --> N10 --> N11 --> N12 --> N13 --> N14
```

The architectural progression from AlexNet to VGG can be summarized as:

```text
AlexNet:
CNNs can scale to ImageNet.

        ↓

VGG:
Once CNNs can scale,
build deeper representations using
simple repeated 3x3 convolution blocks.

        ↓

ResNet:
Once depth becomes very large,
solve the optimization problems
of deep plain networks.
```

Final takeaway:

> **VGG showed that a deep visual hierarchy can be built from repeated small $3\times3$ convolutions. Stacking these layers gradually enlarges the receptive field while adding depth and nonlinear transformations, making the network both systematic and expressive.**

## 6. Main Design Principles

## 6.1. LeNet-5: Use Local Image Structure

```text
Filter
=
local feature detector

Feature map
=
detector response over spatial positions
```

Core idea:

```text
Do not flatten the image immediately.

First:
learn local visual features

Then:
combine them into increasingly complex representations

Finally:
classify using the high-level representation
```

Important principles:

```text
local connectivity
parameter sharing
hierarchical feature learning
downsampling
```


## 6.2. AlexNet: Scale CNNs to Large Data

AlexNet shows that CNNs can scale with:

```text
large-scale data
GPU training
ReLU
regularization
augmentation
large model capacity
```

It also demonstrates the importance of optimization and training strategy.

Its success should be understood as:

```text
architecture
+
data
+
compute
+
optimization
+
regularization
```

However, AlexNet also reveals a weakness of classic CNNs:

```text
Flatten a large feature map
-> huge fully connected layers
-> most parameters live in the classifier
```

This weakness becomes even more obvious in VGG.


## 6.3. VGG: Build Depth with Repeated Simple Blocks

VGG shows that strong architectures can be built from a simple repeated block:

```text
Conv 3x3 -> ReLU
Conv 3x3 -> ReLU
MaxPool
```

For deeper stages:

```text
Conv 3x3 -> ReLU
Conv 3x3 -> ReLU
Conv 3x3 -> ReLU
MaxPool
```

Core idea:

```text
Depth increases representation capacity.

Small filters allow depth to increase
without requiring very large spatial kernels.

Repeated blocks make the architecture
simple and systematic.
```


# ResNet

## 1. Core Idea

ResNet stands for **Residual Network**.

It was designed to solve a key problem in very deep CNNs:

```text
Deeper networks can have stronger representation power,
but they can become harder to optimize.
```

This is called the **degradation problem**:

```text
A deeper plain network may have higher training error
than a shallower network.
```

This is not simply overfitting. In overfitting:

```text
training error is low
validation/test error is high
```

In degradation:

```text
training error itself becomes worse
```

ResNet solves this by using:

```text
residual learning
shortcut connections
better information flow
better gradient flow
```

## 2. The Formula $H(x) = F(x) + x$

Suppose a block receives input:

$$
x
$$

The desired output mapping is:

$$
H(x)
$$

In a plain network, stacked layers must learn the whole mapping directly:

$$
x \rightarrow H(x)
$$

ResNet rewrites this mapping as:

$$
H(x) = F(x) + x
$$

So:

$$
F(x) = H(x) - x
$$

Meaning:

```text
x    = input representation
H(x) = desired output mapping
F(x) = residual correction
```

The word **residual** means:

```text
the difference between the desired output and the current input
```

So ResNet does not force the block to learn the full transformation from scratch. It asks the block to learn:

```text
What should be changed or added to the input?
```

## 3. Intuition Behind Residual Learning

The core intuition is:

```text
It is easier to learn a correction than to relearn the whole representation.
```

If the input representation is already good, the desired mapping may be close to identity:

$$
H(x) \approx x
$$

In a plain network, the block must learn:

$$
H(x) = x
$$

This means a stack of layers must learn identity mapping from scratch.

In ResNet:

$$
H(x) = F(x) + x
$$

If the desired mapping is identity, then the block only needs:

$$
F(x) \approx 0
$$

So:

$$
H(x) = F(x) + x \approx x
$$

Key idea:

```text
Learning zero residual is easier than learning identity mapping from scratch.
```

If the block is useful:

```text
F(x) learns a meaningful transformation.
```

If the block is not useful:

```text
F(x) can become close to zero,
so the input passes through almost unchanged.
```

## 4. Plain Block vs Residual Block

A plain block learns:

$$
y = H(x)
$$

Flow:

```text
x
-> Conv
-> BN
-> ReLU
-> Conv
-> BN
-> ReLU
-> y
```

A residual block learns:

$$
y = F(x) + x
$$

Flow:

```text
x ----------------------+
                        |
Conv -> BN -> ReLU -> Conv -> BN
                        |
                        + -> Add -> ReLU
```

Main difference:

```text
Plain block:
learn the full output directly.

Residual block:
learn the residual correction and add the input back.
```

![ResNet Architecture](https://api.wandb.ai/files/mostafaibrahim17/images/projects/37042936/3a93c9fd.png)

## 5. Why Shortcut Connections Help

For a residual block:

$$
y = F(x) + x
$$

During backpropagation:

$$
\frac{\partial y}{\partial x}
=
\frac{\partial F(x)}{\partial x}
+
I
$$

The term $I$ comes from the identity shortcut.

Intuition:

```text
Gradient can flow through the main path F(x),
but it also has a direct shortcut path through x.
```

So shortcut connections help:

```text
1. Forward information flow
2. Backward gradient flow
3. Optimization of very deep networks
```

ResNet does not magically remove all optimization problems, but it makes very deep networks much easier to train.

## 6. Residual Block in CNNs

In CNNs, $x$ is not a scalar. It is a tensor.

Example:

$$
x \in \mathbb{R}^{56 \times 56 \times 64}
$$

The main path learns:

$$
F(x) \in \mathbb{R}^{56 \times 56 \times 64}
$$

The shortcut keeps:

$$
x \in \mathbb{R}^{56 \times 56 \times 64}
$$

Then ResNet adds them element-wise:

$$
y = F(x) + x
$$

Important rule:

```text
F(x) and x must have the same shape before addition.
```

Valid addition:

```text
F(x): 56 x 56 x 64
x:    56 x 56 x 64
y:    56 x 56 x 64
```

Invalid addition:

```text
F(x): 28 x 28 x 128
x:    56 x 56 x 64
```

When shapes differ, use a projection shortcut:

$$
y = F(x) + W_sx
$$

where $W_s$ is usually a $1 \times 1$ convolution.

Example:

```text
x:      56 x 56 x 64
W_s x:  28 x 28 x 128
F(x):   28 x 28 x 128
y:      28 x 28 x 128
```

## 7. Identity Block and Convolutional Block

### 7.1. Identity Block

An identity block is used when input and output have the same shape.

Example:

```text
Input:  56 x 56 x 64
Output: 56 x 56 x 64
```

Formula:

$$
y = F(x) + x
$$

Shortcut:

```text
identity shortcut
no extra trainable parameters
```

Use case:

```text
Keep spatial size and channel depth unchanged.
```

### 7.2. Convolutional Block

A convolutional block is used when spatial size or channel depth changes.

Example:

```text
Input:  56 x 56 x 64
Output: 28 x 28 x 128
```

Formula:

$$
y = F(x) + W_sx
$$

Shortcut:

```text
1x1 convolution
often with stride 2 for downsampling
```

Use case:

```text
Change spatial size or channel depth while keeping addition valid.
```

## 8. Basic Block

The **Basic Block** is used in:

```text
ResNet-18
ResNet-34
```

Architecture:

```mermaid
flowchart LR
    I["x"]
    A(("Add"))
    M0["Conv 3x3"]
    I --> M0
    M1["BatchNorm"]
    M0 --> M1
    M2["ReLU"]
    M1 --> M2
    M3["Conv 3x3"]
    M2 --> M3
    M4["BatchNorm"]
    M3 --> M4
    M4 --> A
    I --> A
    O["ReLU"]
    A --> O
```

Formula:

$$
y = \text{ReLU}(F(x) + x)
$$

The residual function is:

$$
F(x)
=
\text{BN}_2(
\text{Conv}_{3 \times 3}(
\text{ReLU}(
\text{BN}_1(
\text{Conv}_{3 \times 3}(x)
)
)
)
)
$$

A Basic Block has:

```mermaid
flowchart LR
    N0["2 convolution layers"]
    N1["Conv 3x3"]
    N2["Conv 3x3"]
    N0 --> N1 --> N2
```

If shape is unchanged:

```text
shortcut = identity
```

If shape changes:

```text
shortcut = 1x1 convolution
```

## 9. Basic Block Parameter Reasoning

Assume input and output both have $C$ channels.

Ignoring BatchNorm and bias:

One $3 \times 3$ convolution has:

$$
3 \times 3 \times C \times C = 9C^2
$$

Two $3 \times 3$ convolutions have:

$$
9C^2 + 9C^2 = 18C^2
$$

So a Basic Block has approximately:

$$
18C^2
$$

convolution weights if the shortcut is identity.

If projection shortcut is needed:

$$
1 \times 1 \times C_{\text{in}} \times C_{\text{out}}
$$

## 10. Bottleneck Block

The **Bottleneck Block** is used in:

```text
ResNet-50
ResNet-101
ResNet-152
```

Architecture:

```mermaid
flowchart LR
    I["x"]
    A(("Add"))
    M0["Conv 1x1"]
    I --> M0
    M1["BatchNorm"]
    M0 --> M1
    M2["ReLU"]
    M1 --> M2
    M3["Conv 3x3"]
    M2 --> M3
    M4["BatchNorm"]
    M3 --> M4
    M5["ReLU"]
    M4 --> M5
    M6["Conv 1x1"]
    M5 --> M6
    M7["BatchNorm"]
    M6 --> M7
    M7 --> A
    I --> A
    O["ReLU"]
    A --> O
```

Main idea:

```mermaid
flowchart LR
    N0["1x1 Conv<br/>reduce channels"]
    N1["3x3 Conv<br/>process spatial features"]
    N2["1x1 Conv<br/>expand channels"]
    N0 --> N1 --> N2
```

Example:

```mermaid
flowchart LR
    N0["Input<br/>256 channels"]
    N1["1x1 Conv<br/>reduce to 64 channels"]
    N2["3x3 Conv<br/>process 64 channels"]
    N3["1x1 Conv<br/>expand back to 256 channels"]
    N0 --> N1 --> N2 --> N3
```

It is called a bottleneck because:

```text
The middle of the block has fewer channels.
```

Purpose:

```text
Reduce computation in the expensive 3x3 convolution
while still allowing very deep networks.
```

## 11. Bottleneck Block Parameter Reasoning

Assume:

```text
input channels = C_in
bottleneck width = C_b
output channels = 4C_b
```

Main path:

```text
1x1 conv: C_in -> C_b
3x3 conv: C_b -> C_b
1x1 conv: C_b -> 4C_b
```

Ignoring BatchNorm and bias:

First $1 \times 1$ convolution:

$$
C_{\text{in}}C_b
$$

Middle $3 \times 3$ convolution:

$$
9C_b^2
$$

Final $1 \times 1$ convolution:

$$
C_b \times 4C_b = 4C_b^2
$$

Total main path:

$$
C_{\text{in}}C_b + 9C_b^2 + 4C_b^2
$$

So:

$$
\text{Params}
=
C_{\text{in}}C_b + 13C_b^2
$$

If projection shortcut is needed:

$$
C_{\text{in}} \times 4C_b
$$

Example: first bottleneck block in ResNet-50 conv2_x.

```text
input channels = 64
bottleneck width = 64
output channels = 256
```

Main path:

$$
1 \times 1: 64 \times 64 = 4096
$$

$$
3 \times 3: 9 \times 64 \times 64 = 36864
$$

$$
1 \times 1: 64 \times 256 = 16384
$$

Projection shortcut:

$$
1 \times 1: 64 \times 256 = 16384
$$

Total convolution weights:

$$
4096 + 36864 + 16384 + 16384 = 73728
$$

Key idea:

```text
The output can have many channels,
but the expensive 3x3 convolution runs on fewer channels.
```

## 12. General ResNet Architecture

For ImageNet-style input:

```text
Input: 224 x 224 x 3
```

ResNet starts with a stem:

```mermaid
flowchart LR
    N0["Conv 7x7<br/>64 filters, stride 2, padding 3"]
    N1["BatchNorm"]
    N2["ReLU"]
    N3["MaxPool 3x3<br/>stride 2"]
    N0 --> N1 --> N2 --> N3
```

Shape:

```mermaid
flowchart LR
    N0["224 x 224 x 3"]
    N1["Conv 7x7<br/>stride 2"]
    N2["112 x 112 x 64"]
    N3["MaxPool 3x3<br/>stride 2"]
    N4["56 x 56 x 64"]
    N0 --> N1 --> N2 --> N3 --> N4
```

Then there are four main stages:

```text
conv2_x
conv3_x
conv4_x
conv5_x
```

General shape flow:

```text
Input:
224 x 224 x 3

Stem:
112 x 112 x 64
56 x 56 x 64

conv2_x:
56 x 56

conv3_x:
28 x 28

conv4_x:
14 x 14

conv5_x:
7 x 7

Global Average Pooling:
1 x 1

FC:
1000 classes
```

Main pattern:

```text
Spatial size decreases.
Channel depth increases.
Each stage is built from residual blocks.
```

Unlike VGG, ResNet usually avoids a huge Flatten + FC classifier and uses Global Average Pooling before the final FC layer.

## 13. ResNet-34

ResNet-34 uses **Basic Blocks**.

Block distribution:

```text
[3, 4, 6, 3]
```

Meaning:

```text
conv2_x: 3 Basic Blocks
conv3_x: 4 Basic Blocks
conv4_x: 6 Basic Blocks
conv5_x: 3 Basic Blocks
```

Each Basic Block has 2 convolution layers.

Layer count:

```text
conv1:   1 layer
conv2_x: 3 blocks x 2 conv = 6 layers
conv3_x: 4 blocks x 2 conv = 8 layers
conv4_x: 6 blocks x 2 conv = 12 layers
conv5_x: 3 blocks x 2 conv = 6 layers
FC:      1 layer

Total = 1 + 6 + 8 + 12 + 6 + 1 = 34 layers
```

Shape flow:

```mermaid
flowchart TD
    N0["Input<br/>224 x 224 x 3"]
    N1["Conv7x7, 64, stride 2<br/>112 x 112 x 64"]
    N2["MaxPool<br/>56 x 56 x 64"]
    N3["conv2_x<br/>3 Basic Blocks, 64 channels<br/>56 x 56 x 64"]
    N4["conv3_x<br/>4 Basic Blocks, 128 channels<br/>28 x 28 x 128"]
    N5["conv4_x<br/>6 Basic Blocks, 256 channels<br/>14 x 14 x 256"]
    N6["conv5_x<br/>3 Basic Blocks, 512 channels<br/>7 x 7 x 512"]
    N7["Global Average Pooling<br/>512"]
    N8["FC<br/>1000"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6 --> N7 --> N8
```

## 14. ResNet-50

ResNet-50 uses **Bottleneck Blocks**.

Block distribution:

```text
[3, 4, 6, 3]
```

Each Bottleneck Block has 3 convolution layers.

Layer count:

```text
conv1:   1 layer
conv2_x: 3 blocks x 3 conv = 9 layers
conv3_x: 4 blocks x 3 conv = 12 layers
conv4_x: 6 blocks x 3 conv = 18 layers
conv5_x: 3 blocks x 3 conv = 9 layers
FC:      1 layer

Total = 1 + 9 + 12 + 18 + 9 + 1 = 50 layers
```

Shape flow:

```mermaid
flowchart TD
    N0["Input<br/>224 x 224 x 3"]
    N1["Conv7x7, 64, stride 2<br/>112 x 112 x 64"]
    N2["MaxPool<br/>56 x 56 x 64"]
    N3["conv2_x<br/>3 Bottleneck Blocks<br/>56 x 56 x 256"]
    N4["conv3_x<br/>4 Bottleneck Blocks<br/>28 x 28 x 512"]
    N5["conv4_x<br/>6 Bottleneck Blocks<br/>14 x 14 x 1024"]
    N6["conv5_x<br/>3 Bottleneck Blocks<br/>7 x 7 x 2048"]
    N7["Global Average Pooling<br/>2048"]
    N8["FC<br/>1000"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6 --> N7 --> N8
```

Because bottleneck expansion factor is 4:

```text
64  -> 256
128 -> 512
256 -> 1024
512 -> 2048
```

Typical parameter count:

| Part | Parameters |
|---|---:|
| Stem | 9,536 |
| conv2_x | 215,808 |
| conv3_x | 1,219,584 |
| conv4_x | 7,098,368 |
| conv5_x | 14,964,736 |
| FC | 2,049,000 |
| Total | 25,557,032 |

Therefore:

```text
ResNet-50 ≈ 25.6M parameters
```

Important observation:

```text
ResNet-50 is deeper than ResNet-34,
but its parameter count does not explode
because it uses bottleneck blocks.
```

## 16. Why ResNet Uses Global Average Pooling

Older CNNs such as VGG often use:

```text
Flatten -> large fully connected layers
```

This creates many parameters.

ResNet usually uses:

```mermaid
flowchart LR
    N0["7 x 7 feature map"]
    N1["Global Average Pooling"]
    N2["Feature vector"]
    N3["FC classifier"]
    N0 --> N1 --> N2 --> N3
```

$$
2048 \times 1000 + 1000 = 2049000
$$

This avoids the extremely large classifier head used in older CNNs.

## 17. Why ResNet Matters

ResNet changed the design philosophy of deep networks.

Before ResNet:

```text
Make networks stronger by stacking more layers.
```

After ResNet:

```text
Make networks deeper by improving information flow and optimization.
```

Core contribution:

```text
Residual learning + shortcut connections make very deep CNNs trainable.
```

ResNet shows that depth is useful only when optimization remains stable.

## 18. Common Mistakes

| Mistake | Correct Understanding |
|---|---|
| Thinking ResNet is just a deeper CNN | ResNet is built from residual blocks with shortcut connections |
| Thinking $F(x)$ is the final output | The output is $F(x) + x$ |
| Not understanding $F(x)=H(x)-x$ | $F(x)$ is the residual correction |
| Thinking shortcut always has no parameters | Identity shortcut has no parameters, projection shortcut has parameters |
| Adding tensors with different shapes | Main path and shortcut must have the same shape before addition |
| Forgetting projection shortcut | Use $1 \times 1$ conv when spatial size or channels change |
| Confusing BasicBlock and Bottleneck | ResNet-18/34 use BasicBlock; ResNet-50+ use Bottleneck |
| Counting layers incorrectly | BasicBlock has 2 conv layers; Bottleneck has 3 conv layers; also count conv1 and FC |
| Thinking ResNet uses a huge Flatten like VGG | ResNet usually uses Global Average Pooling |
| Saying skip connection only solves vanishing gradient | More accurately, it improves optimization and information flow |

## 19. Essence

The central formula is:

$$
H(x) = F(x) + x
$$

Equivalent form:

$$
F(x) = H(x) - x
$$

Meaning:

```text
H(x) = desired mapping
x    = input representation
F(x) = residual correction
```

Core intuition:

```text
Instead of learning the whole output from scratch,
ResNet learns the correction that should be added to the input.
```

If the block is useful:

```text
F(x) learns a meaningful transformation.
```

If the block does not need to change the input:

```text
F(x) ≈ 0
```

Then:

$$
H(x) = F(x) + x \approx x
$$

So the block behaves like identity mapping.

Final takeaway:

```text
ResNet makes very deep CNNs trainable by learning residual corrections and using shortcut connections for better information and gradient flow.
```

# Why ResNets Work?

## 1. The Problem with Very Deep Plain Networks

A natural assumption is:

```text
more layers
-> greater representation capacity
-> better performance
```

However, very deep **plain networks** can suffer from the **degradation problem**:

```text
deeper network
-> harder optimization
-> higher training error
```

This is different from overfitting.

```text
Overfitting:
training error decreases
validation error increases

Degradation:
even training error may increase
```

The important question is:

> If a shallower network already works well, why can a deeper network not simply reproduce the same function?

In principle, the newly added layers could learn:

$$
H(x)=x
$$

the identity mapping.

The problem is that a stack of nonlinear layers may be **capable** of representing identity, but gradient-based optimization may have difficulty finding it.

This motivates residual learning.

## 2. Residual Learning

A plain block tries to learn:

$$
H(x)
$$

directly.

ResNet rewrites the desired mapping as:

$$
\boxed{
H(x)=x+F(x)
}
$$

where:

$$
\boxed{
F(x)=H(x)-x
}
$$

The architecture becomes:

```mermaid
flowchart LR
    I["x"]
    A(("+"))
    M0["Residual branch<br/>F(x)"]
    I --> M0
    M0 --> A
    I --> A
    O["H(x) = x + F(x)"]
    A --> O
```

The shortcut provides $x$ directly.

Therefore, the learned branch focuses on $F(x)$.

## 3. What Does "Learning the Residual" Really Mean?

A common example is:

$$
x=10
$$

and:

$$
H(x)=12
$$

so:

$$
F(x)=12-10=2
$$

It is misleading to interpret this as:

```text
Learning 2 is easier than learning 12.
```

That is not the main idea.

A neural network is not learning a single numeric transition:

$$
10\rightarrow12
$$

It is learning a function over high-dimensional representations.

A better interpretation is:

```mermaid
flowchart TD
    subgraph P["Plain block"]
        P0["x"] --> P1["Learned layers preserve useful information in x"]
        P1 --> P2["Learned layers also modify x"]
        P2 --> P3["H(x)"]
    end

    subgraph R["Residual block"]
        R0["x"] --> R1["Learn modification<br/>F(x)"] --> R2(("+"))
        R0 --> R2
        R2 --> R3["H(x)"]
    end
```

Suppose:

$$
H(x)=x+\Delta(x)
$$

where the current representation $x$ is already useful and only requires refinement.

A plain network must learn the whole function:

$$
x+\Delta(x)
$$

whereas ResNet already provides $x$ through the shortcut.

Therefore:

$$
F(x)=\Delta(x)
$$

The residual branch can focus on the required correction.

The key idea is therefore:

> **ResNet does not work because a numerically smaller value is automatically easier to learn. It works because the identity component $x$ is explicitly provided by the architecture.**

## 4. Why Identity Mapping Becomes Easy

Suppose the desired transformation is approximately identity:

$$
H(x)\approx x
$$

Then:

$$
F(x)=H(x)-x\approx0
$$

For a residual block:

$$
H(x)=x+F(x)
$$

if:

$$
F(x)\approx0
$$

then:

$$
H(x)\approx x
$$

Therefore:

```text
If the residual block learns something useful:
F(x) contributes a useful correction.

If the residual block is unnecessary:
F(x) ≈ 0
and the block becomes approximately identity.
```

This makes increasing network depth much safer.

A newly added residual block does not necessarily need to transform the representation significantly.

It can simply behave approximately as:

$$
x\rightarrow x
$$

## 5. Feature Refinement Interpretation

Residual learning can be understood as:

```text
current representation
+
learned correction
=
improved representation
```

For example:

```text
Current representation:
"This probably contains a face."

Residual branch:
"Strengthen relationships between
eyes, nose, and mouth."

Updated representation:
"This is very likely a face."
```

A residual block therefore behaves naturally as a **feature refinement unit** rather than forcing every block to reconstruct an entirely new representation.

However, $F(x)$ is **not required to be small**.

If a large transformation is necessary, the residual branch can still learn a large $F(x)$.

## 6. Backpropagation Through a Residual Block

Consider:

$$
y=x+F(x)
$$

Differentiating with respect to $x$:

$$
\boxed{
\frac{\partial y}{\partial x}
=
I+
\frac{\partial F(x)}{\partial x}
}
$$

For a scalar example:

$$
F(x)=ax
$$

Then:

$$
y=x+ax
$$

and:

$$
\frac{dy}{dx}
=
1+a
$$

Compare this with a plain block:

$$
y=ax
$$

whose derivative is:

$$
\frac{dy}{dx}=a
$$

Therefore:

```text
Plain block:

gradient multiplier = a


Residual block:

gradient multiplier = 1 + a
```

If:

$$
a=0.01
$$

then:

```text
Plain:
gradient multiplier = 0.01

Residual:
gradient multiplier = 1.01
```

The identity shortcut contributes the term:

$$
1
$$

or, in the vector case:

$$
I
$$

This provides a direct gradient path.

## 7. Concrete Backpropagation Example

Consider three residual blocks:

$$
x_{l+1}=x_l+F_l(x_l)
$$

with:

$$
F_1(x)=0.1x
$$

$$
F_2(x)=-0.2x
$$

$$
F_3(x)=0.05x
$$

Forward:

```text
x0 = 2

Block 1:
x1 = x0 + 0.1x0
   = 2.2

Block 2:
x2 = x1 - 0.2x1
   = 1.76

Block 3:
x3 = x2 + 0.05x2
   = 1.848
```

Suppose:

$$
t=2
$$

and:

$$
L=\frac12(x_3-t)^2
$$

Then:

$$
\frac{\partial L}{\partial x_3}
=
1.848-2
=
-0.152
$$

Backward through Block 3:

$$
\frac{\partial x_3}{\partial x_2}
=
1+0.05
=
1.05
$$

so:

$$
\frac{\partial L}{\partial x_2}
=
-0.152(1.05)
=
-0.1596
$$

Backward through Block 2:

$$
\frac{\partial x_2}{\partial x_1}
=
1-0.2
=
0.8
$$

so:

$$
\frac{\partial L}{\partial x_1}
=
-0.1596(0.8)
=
-0.12768
$$

Backward through Block 1:

$$
\frac{\partial x_1}{\partial x_0}
=
1+0.1
=
1.1
$$

so:

$$
\frac{\partial L}{\partial x_0}
=
-0.12768(1.1)
=
-0.140448
$$

Therefore:

```text
Gradient at the output:
0.152

Gradient after 3 residual blocks:
0.140448
```

The gradient remains in approximately the same magnitude range.

## 8. Plain Network vs ResNet Gradient

For a plain deep network:

$$
x_{l+1}=F_l(x_l)
$$

the gradient contains products of Jacobians:

$$
\boxed{
\frac{\partial L}{\partial x_0}
=
\frac{\partial L}{\partial x_L}
\prod_l J_{F_l}
}
$$

If many Jacobians have magnitude below 1, repeated multiplication can make gradients extremely small.

For example:

$$
0.5^{100}
\approx0
$$

For a residual network:

$$
x_{l+1}=x_l+F_l(x_l)
$$

each block has Jacobian:

$$
\boxed{
I+J_{F_l}
}
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial x_0}
=
\frac{\partial L}{\partial x_L}
\prod_l
(I+J_{F_l})
}
$$

If:

$$
J_{F_l}\approx0
$$

then:

$$
I+J_{F_l}\approx I
$$

so gradient propagation can remain much more stable.

This is why residual connections provide a **short gradient path** through very deep networks.

However, this does not mathematically guarantee that every arbitrarily deep ResNet will always train successfully.

Optimization still depends on factors such as:

```text
initialization
BatchNorm
activation functions
optimizer
learning rate
architecture
data
```

The correct statement is:

> **Residual connections make optimization of very deep networks substantially easier; they do not eliminate every possible gradient problem.**

## 9. Information Flow Across Many Residual Blocks

From:

$$
x_{l+1}=x_l+F_l(x_l)
$$

we obtain:

$$
x_1=x_0+F_0(x_0)
$$

$$
x_2
=
x_0
+
F_0(x_0)
+
F_1(x_1)
$$

and after many blocks:

$$
\boxed{
x_L
=
x_0
+
\sum_{l=0}^{L-1}F_l(x_l)
}
$$

Conceptually:

```text
deep representation
=
earlier representation
+
correction
+
correction
+
correction
+
...
```

The original representation has a direct path through the network rather than being completely replaced at every layer.

## 10. Identity Shortcut

When:

$$
\operatorname{shape}(F(x))
=
\operatorname{shape}(x)
$$

the block can use:

$$
\boxed{
y=F(x)+x
}
$$

Example:

```text
x:
(N, 64, 56, 56)

F(x):
(N, 64, 56, 56)

Output:
(N, 64, 56, 56)
```

The shortcut contains no trainable parameters:

```mermaid
flowchart LR
    I["x"]
    A(("+"))
    M0["Conv → ... → F(x)"]
    I --> M0
    M0 --> A
    I --> A
    O["Output"]
    A --> O
```

Therefore:

$$
\text{shortcut parameters}=0
$$

## 11. Projection Shortcut When Shapes Differ

At a stage transition, the shapes may differ.

For example:

$$
x:
(N,64,56,56)
$$

while:

$$
F(x):
(N,128,28,28)
$$

We cannot compute:

$$
F(x)+x
$$

because their shapes are different.

Instead, transform the shortcut:

$$
\boxed{
y=F(x)+W_sx
}
$$

Usually:

```text
1x1 Conv
stride = 2
```

is used.

Example:

```mermaid
flowchart LR
    I["Input<br/>(N, 64, 56, 56)"]

    I --> M1["Main branch<br/>3x3 Conv, stride 2"]
    M1 --> M2["..."]
    M2 --> M3["(N, 128, 28, 28)"]

    I --> S1["Shortcut branch<br/>1x1 Conv, stride 2"]
    S1 --> S2["(N, 128, 28, 28)"]

    M3 --> A(("Add"))
    S2 --> A
```

Now:

$$
\operatorname{shape}(F(x))
=
\operatorname{shape}(W_sx)
$$

so addition becomes possible:

$$
y=F(x)+W_sx
$$

The $1\times1$ convolution performs two jobs:

```text
stride 2:
56x56 -> 28x28

out_channels = 128:
64 -> 128 channels
```

Unlike an identity shortcut, a projection shortcut has trainable parameters.

## 12. Why Does the Shortcut Span Two Convolutions?

There is no mathematical rule saying that a shortcut must span exactly two convolutional layers.

The correct principle is:

> **A shortcut skips over one residual function $F(x)$.**

The number of layers inside $F(x)$ is an architectural choice.

In ResNet-18 and ResNet-34, the **BasicBlock** uses:

```mermaid
flowchart LR
    I["x"] --> C1["3x3 Conv"] --> B1["BN"] --> R1["ReLU"]
    R1 --> C2["3x3 Conv"] --> B2["BN"] --> A(("Add"))
    I --> A
    A --> R2["ReLU"]
```

Therefore:

$$
F(x)
=
\text{Conv}_2(
\operatorname{ReLU}(
\text{Conv}_1(x)
))
$$

The shortcut therefore appears to "skip two convolutions."

The important unit is the **residual block**, not the number two.

## 13. Why Not One Convolution per Residual Block?

It is possible to define:

$$
F(x)=\text{Conv}(x)
$$

and use:

$$
y=x+F(x)
$$

This is still residual learning.

However, the residual branch would be relatively shallow.

With two convolutions and a nonlinear activation:

$$
F(x)
=
W_2\sigma(W_1x)
$$

the residual branch can learn a richer nonlinear correction.

Two convolutions provide a useful balance:

```text
enough transformation capacity
+
short residual path
+
frequent shortcut connections
```

## 14. Why Not Skip Ten Convolutions at Once?

It is also mathematically possible:

```mermaid
flowchart LR
    I["x"]
    A(("+"))
    M0["Conv"]
    I --> M0
    M1["Conv"]
    M0 --> M1
    M2["..."]
    M1 --> M2
    M3["Conv<br/>(10 layers)"]
    M2 --> M3
    M3 --> A
    I --> A
    O["Output"]
    A --> O
```

But the residual branch itself becomes very deep.

Its Jacobian contains:

$$
J_{F}
=
J_{10}J_9\cdots J_1
$$

so optimization inside the residual branch can again become difficult.

Using smaller residual blocks gives frequent identity paths:

```mermaid
flowchart TD
    A["Conv"] --> B["Conv"] --> C(("+"))
    C --> D["Conv"] --> E["Conv"] --> F(("+"))
    F --> G["Conv"] --> H["Conv"] --> I(("+"))
    I --> J["..."]
```

This makes both feature propagation and gradient propagation easier.

## 15. Bottleneck Blocks Use Three Convolutions

The fact that ResNet-50/101/152 use three convolutions per residual block confirms that "two convolutions" is not a requirement.

A Bottleneck block uses:

```mermaid
flowchart LR
    I["x"]
    A(("+"))
    M0["1x1 Conv"]
    I --> M0
    M1["3x3 Conv"]
    M0 --> M1
    M2["1x1 Conv"]
    M1 --> M2
    M2 --> A
    I --> A
    O["Output"]
    A --> O
```

The roles are:

```text
1x1 Conv:
reduce channels

3x3 Conv:
process spatial information

1x1 Conv:
expand channels
```

Therefore:

```text
ResNet-18 / 34:
2 Conv layers per residual block

ResNet-50 / 101 / 152:
3 Conv layers per bottleneck block
```

The real principle remains:

$$
\boxed{
H(x)=x+F(x)
}
$$

## 16. Common Mistakes

| Mistake | Correct Understanding |
|---|---|
| ResNet was created mainly to solve overfitting | Its central motivation was the optimization/degradation problem of very deep networks |
| Degradation means overfitting | Degradation can increase training error itself |
| Residual learning means "learning 2 is easier than learning 12" | The shortcut provides the identity component $x$, allowing the branch to focus on $H(x)-x$ |
| ResNet learns $H(x)$ directly | The residual branch learns $F(x)=H(x)-x$ |
| The block output is $F(x)$ | The output is $x+F(x)$ |
| $F(x)$ must always be small | It may be small when the desired mapping is near identity, but this is not a constraint |
| Skip connections completely solve vanishing gradients | They significantly improve gradient paths but do not mathematically eliminate all gradient problems |
| Shortcut always contains zero parameters | Identity shortcuts do; projection shortcuts can contain a $1\times1$ convolution |
| $F(x)$ and $x$ can always be added | Their shapes must match first |
| Skip connections must span exactly two Conv layers | Two Conv layers are the BasicBlock design of ResNet-18/34, not a mathematical requirement |
| More Conv layers inside one residual block are always better | Very deep residual branches reduce the advantage of frequent short identity paths |
| Skip connection removes the residual layers | Both the residual branch and shortcut participate in the computation |

## 17. Essence

The central ResNet equation is:

$$
\boxed{
H(x)=x+F(x)
}
$$

A plain network asks:

```text
What complete new representation H(x)
should I produce from x?
```

A residual network asks:

```text
The current representation x is already available.

What correction F(x)
should I add to it?
```

If the desired mapping is near identity:

$$
H(x)\approx x
$$

then:

$$
F(x)\approx0
$$

so the block can easily behave approximately as identity.

For backpropagation:

$$
y=x+F(x)
$$

gives:

$$
\boxed{
\frac{\partial y}{\partial x}
=
I+J_F
}
$$

instead of only:

$$
J_F
$$

as in a plain transformation.

Therefore, residual connections provide:

```text
easy identity mapping
+
feature reuse
+
learned refinement
+
direct information paths
+
shorter gradient paths
```

The shortcut does **not** need to span exactly two convolutions.

Instead:

```text
shortcut
=
skip over one residual function F(x)
```

BasicBlock:

```text
2 Conv layers
-> ResNet-18 / 34
```

Bottleneck block:

```text
3 Conv layers
-> ResNet-50 / 101 / 152
```

Final takeaway:

> **ResNets work because they explicitly preserve the current representation through an identity path and let the learned branch focus on residual corrections. The same shortcut also introduces an identity term into backpropagation, making information and gradients much easier to propagate through very deep networks.**


# Networks in Networks and $1\times1$ Convolutions

## 1. Core Idea

A $1\times1$ convolution may look trivial spatially, but it performs an important operation across the **channel dimension**.

Suppose the input is:

$$
X\in\mathbb{R}^{H\times W\times C_{in}}
$$

At each spatial position $(i,j)$, we have a channel vector:

$$
X_{i,j,:}\in\mathbb{R}^{C_{in}}
$$

A $1\times1$ convolution applies a learned transformation to this vector:

$$
\boxed{
z_{i,j}=WX_{i,j,:}+b
}
$$

where:

$$
W\in\mathbb{R}^{C_{out}\times C_{in}}
$$

Therefore:

$$
\boxed{
H\times W\times C_{in}
\rightarrow
H\times W\times C_{out}
}
$$

with stride $1$.

The spatial dimensions remain unchanged, while the channel dimension can change.

## 2. What Does a $1\times1$ Filter Actually See?

A single $1\times1$ filter has shape:

$$
1\times1\times C_{in}
$$

It does **not** see only one number.

It sees:

```text
one spatial location
+
all input channels
```

For example:

$$
28\times28\times192
$$

At one spatial position:

$$
X_{i,j,:}
=
[x_1,x_2,\ldots,x_{192}]
$$

A single $1\times1$ filter computes:

$$
z_{i,j}
=
\sum_{c=1}^{192}
w_cx_c+b
$$

If we use 32 filters:

$$
28\times28\times192
\rightarrow
28\times28\times32
$$

Therefore:

> **The number of $1\times1$ filters determines the number of output channels.**

## 3. $1\times1$ Convolution as a Fully Connected Layer per Spatial Position

A useful interpretation is:

```mermaid
flowchart LR
    N0["Channel vector"]
    N1["Linear transformation"]
    N2["New channel vector"]
    N0 --> N1 --> N2
```

For example:

$$
\mathbb{R}^{192}
\rightarrow
\mathbb{R}^{32}
$$

is performed independently at every spatial location.

The same weights are shared across all locations.

Therefore, a $1\times1$ convolution behaves similarly to a small fully connected layer applied to every pixel location.

Unlike flattening followed by a fully connected layer, it:

```text
preserves spatial structure
+
shares parameters across locations
```

## 4. Channel Mixing

A $1\times1$ convolution is mainly used for **channel mixing**.

Suppose different channels represent:

```text
channel 1 -> vertical edge
channel 2 -> horizontal edge
channel 3 -> texture
channel 4 -> color pattern
channel 5 -> object-part feature
```

A new output channel can learn:

$$
z
=
w_1f_1+w_2f_2+\cdots+w_Cf_C
$$

Therefore, the network can learn useful combinations of existing features.

The important distinction is:

```text
3x3 Conv:
spatial mixing + channel mixing

1x1 Conv:
channel mixing only
```

With stride $1$, a $1\times1$ convolution does not increase the spatial receptive field.

## 5. Adding Nonlinearity

A $1\times1$ convolution alone performs a linear channel transformation:

$$
Z=WX+b
$$

But architectures commonly use:

```mermaid
flowchart LR
    N0["1x1 Conv"]
    N1["BatchNorm"]
    N2["ReLU"]
    N0 --> N1 --> N2
```

giving:

$$
A=\operatorname{ReLU}(WX+b)
$$

This creates a nonlinear transformation at every spatial location.

Stacking these operations increases the representational power of the network without changing spatial resolution.

## 6. Network in Network

The **Network in Network** idea replaces a simple local linear transformation with a richer local nonlinear transformation.

Traditional convolution:

```text
local receptive field
-> linear filter
-> activation
```

Network in Network:

```mermaid
flowchart LR
    N0["Local receptive field"]
    N1["Transformation"]
    N2["Nonlinear transformation"]
    N3["Transformation"]
    N0 --> N1 --> N2 --> N3
```

This idea is commonly associated with **MLPconv**, where $1\times1$ convolutions are used to build small neural transformations inside a larger CNN.

For example:

```mermaid
flowchart LR
    N0["3x3 Conv"]
    N1["ReLU"]
    N2["1x1 Conv"]
    N3["ReLU"]
    N4["1x1 Conv"]
    N5["ReLU"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5
```

At each spatial location, the $1\times1$ layers behave like an MLP operating across channels.

Thus:

```text
Network in Network
=
a richer local neural network
inside the larger CNN
```

## 7. Dimensionality Reduction

One of the most important uses of $1\times1$ convolution is reducing channel dimension before an expensive convolution.

Suppose:

$$
X\in\mathbb{R}^{28\times28\times192}
$$

and we want to apply a $5\times5$ convolution with 256 filters.

Without dimensionality reduction:

$$
5\times5\times192\times256
=
1{,}228{,}800
$$

weights.

Now insert a $1\times1$ convolution:

```mermaid
flowchart TD
    N0["28x28x192"]
    N1["1x1 Conv<br/>32 filters"]
    N2["28x28x32"]
    N3["5x5 Conv<br/>256 filters"]
    N4["28x28x256"]
    N0 --> N1 --> N2 --> N3 --> N4
```

The $1\times1$ layer requires:

$$
1\times1\times192\times32
=
6{,}144
$$

weights.

The $5\times5$ convolution now requires:

$$
5\times5\times32\times256
=
204{,}800
$$

Total:

$$
6{,}144+204{,}800
=
210{,}944
$$

Compared with:

$$
1{,}228{,}800
$$

without the bottleneck.

Therefore, $1\times1$ convolution can drastically reduce computation and parameter count.

## 8. Learned Projection, Not Channel Deletion

When reducing:

$$
192\rightarrow32
$$

the model does not simply keep 32 channels and delete the others.

Each new output channel is:

$$
z_k
=
\sum_{c=1}^{192}
W_{k,c}x_c
$$

Therefore:

```mermaid
flowchart TD
    N0["192 original feature dimensions"]
    N1["Learned projection"]
    N2["32 new feature dimensions"]
    N0 --> N1 --> N2
```

Each output channel can combine information from **all 192 input channels**.

This can be interpreted as learned dimensionality reduction.

However, overly aggressive compression can still remove useful information.

## 9. $1\times1$ Convolution Can Reduce, Preserve, or Expand Channels

The output dimension depends entirely on the number of filters.

### Reduce

$$
256\rightarrow64
$$

Used for:

```text
bottlenecks
computation reduction
```

### Preserve

$$
256\rightarrow256
$$

Used for:

```text
channel transformation
additional nonlinear capacity
```

### Expand

$$
64\rightarrow256
$$

Used for:

```text
increasing feature dimension
creating richer representations
```

Therefore, a $1\times1$ convolution is not only a channel-reduction operation.

## 10. Parameter Count

For:

$$
C_{in}
$$

input channels and:

$$
C_{out}
$$

filters, each kernel has:

$$
1\times1\times C_{in}
$$

parameters.

Therefore:

$$
\boxed{
\text{Parameters}
=
C_{in}C_{out}+C_{out}
}
$$

including bias.

Ignoring bias:

$$
\boxed{
C_{in}C_{out}
}
$$

For example:

$$
192\rightarrow32
$$

requires:

$$
192\times32+32
=
6{,}176
$$

parameters.

## 11. Connection to Inception

Inception uses $1\times1$ convolutions extensively as bottlenecks.

Instead of:

```text
high-dimensional input
-> expensive 3x3 or 5x5 Conv
```

it uses:

```mermaid
flowchart LR
    N0["High-dimensional input"]
    N1["1x1 Conv"]
    N2["Lower-dimensional representation"]
    N3["Expensive spatial Conv"]
    N0 --> N1 --> N2 --> N3
```

This reduces:

```text
parameters
FLOPs
computation
```

while still allowing learned combinations of all input channels.

## 12. Connection to ResNet

### Projection Shortcut

When residual and shortcut shapes differ:

$$
(N,64,56,56)
\rightarrow
(N,128,28,28)
$$

ResNet can use:

```text
1x1 Conv
stride = 2
```

to transform the shortcut:

$$
(N,64,56,56)
\rightarrow
(N,128,28,28)
$$

This allows:

$$
F(x)+W_sx
$$

to be computed.

The $1\times1$ convolution can simultaneously:

```text
change channel dimension
+
change spatial resolution through stride
```

### Bottleneck Block

ResNet-50/101/152 use:

```mermaid
flowchart LR
    N0["1x1 Conv"]
    N1["3x3 Conv"]
    N2["1x1 Conv"]
    N0 --> N1 --> N2
```

Conceptually:

```text
1x1:
reduce channels

3x3:
perform spatial processing cheaply

1x1:
expand channels again
```

For example:

```mermaid
flowchart LR
    N0["256"]
    N1["64"]
    N2["64"]
    N3["256"]
    N0 --> N1 --> N2 --> N3
```

This is the **bottleneck architecture**.

## 13. $1\times1$ vs $3\times3$ Convolution

| Property | $1\times1$ Conv | $3\times3$ Conv |
|---|---|---|
| Spatial neighborhood | One location | $3\times3$ neighborhood |
| Mix channels | Yes | Yes |
| Increase receptive field with stride 1 | No | Yes |
| Change channel dimension | Yes | Yes |
| Computational cost | Low | Higher |
| Useful for bottlenecks | Excellent | Less suitable |
| Spatial feature extraction | Limited | Strong |

They are complementary operations rather than replacements for one another.

## 14. Common Mistakes

| Mistake | Correct Understanding |
|---|---|
| A $1\times1$ Conv processes only one number | It processes one spatial position across all $C_{in}$ channels |
| A $1\times1$ filter has only one parameter | Its shape is $1\times1\times C_{in}$ |
| $1\times1$ Conv mixes neighboring pixels | With stride 1, it only mixes channels at the same spatial position |
| $1\times1$ Conv increases receptive field | It does not increase spatial receptive field with stride 1 |
| It is only used to reduce channels | It can reduce, preserve, or increase channels |
| Channel reduction simply deletes features | It learns combinations of all input channels |
| It is identical to a normal FC layer | It behaves similarly per location but shares weights spatially and preserves $H,W$ |
| Network in Network means placing a complete CNN inside another CNN | It refers to richer local nonlinear transformations, especially MLPconv-style operations |

## 15. Essence

A $1\times1$ convolution performs:

$$
\boxed{
X_{i,j,:}
\in
\mathbb{R}^{C_{in}}
\rightarrow
WX_{i,j,:}+b
\in
\mathbb{R}^{C_{out}}
}
$$

independently at every spatial location using shared weights.

Therefore:

$$
\boxed{
H\times W\times C_{in}
\rightarrow
H\times W\times C_{out}
}
$$

Its main purposes are:

```text
channel mixing
+
nonlinear feature transformation
+
channel reduction
+
channel expansion
+
computational bottlenecks
+
dimension matching
```

The key sentence to remember is:

> **A $1\times1$ convolution does not primarily learn spatial relationships; it performs a learned transformation across channels independently at every spatial location.**

And the Network in Network idea is:

> **Instead of using only a simple linear filter at each local region, use a richer local nonlinear network to learn more expressive features.**

# Inception Network Motivation

## 1. Core Problem

A key CNN design question is:

```text
Which operation should we use?

1x1 Conv?
3x3 Conv?
5x5 Conv?
Pooling?
```

Different filter sizes capture different spatial patterns.

```text
1x1 Conv
-> channel transformation

3x3 Conv
-> medium local patterns

5x5 Conv
-> larger spatial patterns

Pooling
-> local aggregation and robustness
```

There is no single filter size that is always optimal.

Inception therefore proposes:

> **Instead of choosing only one operation, apply several useful operations in parallel and let the network learn how to combine their features.**

## 2. Multi-Scale Processing

A basic Inception module applies several branches to the same input:

```mermaid
flowchart LR
    I["Input"]
    C["Concatenate"]
    B0_0["1x1 Conv"]
    I --> B0_0
    B0_0 --> C
    B1_0["3x3 Conv"]
    I --> B1_0
    B1_0 --> C
    B2_0["5x5 Conv"]
    I --> B2_0
    B2_0 --> C
    B3_0["Pool"]
    I --> B3_0
    B3_0 --> C
```

Each branch captures information at a different spatial scale.

Conceptually:

```text
small receptive field
+
medium receptive field
+
large receptive field
+
pooled representation
```

Afterward, the branch outputs are concatenated along the **channel dimension**.

For example:

```text
1x1 branch -> 64 channels
3x3 branch -> 128 channels
5x5 branch -> 32 channels
Pool branch -> 32 channels
```

The output depth becomes:

$$
64+128+32+32=256
$$

If all branches preserve spatial size:

$$
28\times28\times256
$$

is the final output.

Therefore, Inception performs:

> **multi-scale feature extraction at the same network stage.**

## 3. The Computational Problem

The naive Inception design is powerful but computationally expensive.

Suppose the input is:

$$
28\times28\times192
$$

and we apply a $5\times5$ convolution with 32 filters.

The approximate number of multiplications is:

$$
28\times28\times32\times5\times5\times192
$$

$$
=
120{,}422{,}400
$$

or roughly:

$$
120M
$$

operations.

Running several expensive convolution branches in parallel can therefore cause computation to grow rapidly.

So Inception needs to solve two problems:

```text
1. Capture features at multiple scales.

2. Prevent computation from becoming too expensive.
```

## 4. $1\times1$ Convolution as a Bottleneck

The solution is to use $1\times1$ convolutions before expensive spatial convolutions.

Instead of:

```mermaid
flowchart TD
    N0["28x28x192"]
    N1["5x5 Conv<br/>32 filters"]
    N2["28x28x32"]
    N0 --> N1 --> N2
```

use:

```mermaid
flowchart TD
    N0["28x28x192"]
    N1["1x1 Conv<br/>16 filters"]
    N2["28x28x16"]
    N3["5x5 Conv<br/>32 filters"]
    N4["28x28x32"]
    N0 --> N1 --> N2 --> N3 --> N4
```

The $1\times1$ convolution reduces:

$$
192\rightarrow16
$$

channels before the expensive $5\times5$ convolution.

This creates a **bottleneck representation**.

## 5. Computational Saving

The $1\times1$ convolution requires approximately:

$$
28\times28\times192\times16
$$

$$
=
2{,}408{,}448
$$

operations.

The following $5\times5$ convolution requires:

$$
28\times28\times16\times32\times5\times5
$$

$$
=
10{,}035{,}200
$$

operations.

Total:

$$
2.4M+10.0M
\approx
12.4M
$$

Compare:

```text
Direct 5x5 Conv:

≈ 120.4M operations


1x1 bottleneck
+
5x5 Conv:

≈ 12.4M operations
```

The computational cost is reduced dramatically.

## 6. Why Does the Bottleneck Still Work?

Reducing:

$$
192\rightarrow16
$$

does not mean simply deleting 176 channels.

A $1\times1$ convolution computes learned combinations of all input channels:

$$
z_k
=
\sum_{c=1}^{192}
W_{k,c}x_c
$$

Therefore:

```mermaid
flowchart TD
    N0["192-dimensional representation"]
    N1["Learned projection"]
    N2["16-dimensional representation"]
    N0 --> N1 --> N2
```

The network learns which combinations of channels are useful before performing the expensive spatial convolution.

However, excessive compression can still remove useful information, so bottleneck width is an architectural design choice.

## 7. Practical Inception Module

The naive module:

```mermaid
flowchart LR
    I["Input"]
    C["Concatenate"]
    B0_0["1x1 Conv"]
    I --> B0_0
    B0_0 --> C
    B1_0["3x3 Conv"]
    I --> B1_0
    B1_0 --> C
    B2_0["5x5 Conv"]
    I --> B2_0
    B2_0 --> C
    B3_0["Pool"]
    I --> B3_0
    B3_0 --> C
```

becomes the computationally practical version:

```mermaid
flowchart LR
    I["Input"]
    C["Concatenate"]
    B0_0["1x1 Conv"]
    I --> B0_0
    B0_0 --> C
    B1_0["1x1 Conv"]
    I --> B1_0
    B1_1["3x3 Conv"]
    B1_0 --> B1_1
    B1_1 --> C
    B2_0["1x1 Conv"]
    I --> B2_0
    B2_1["5x5 Conv"]
    B2_0 --> B2_1
    B2_1 --> C
    B3_0["Pool"]
    I --> B3_0
    B3_1["1x1 Conv"]
    B3_0 --> B3_1
    B3_1 --> C
```

The $1\times1$ convolutions control channel depth and reduce the cost of the expensive branches.

## 8. Why Use $1\times1$ After Pooling?

Pooling can preserve the number of channels.

For example:

$$
28\times28\times192
\rightarrow
28\times28\times192
$$

with same-padding pooling.

If these 192 channels are concatenated directly with all other branches, output depth may grow too quickly.

Therefore:

```mermaid
flowchart LR
    N0["Pooling"]
    N1["1x1 Conv"]
    N0 --> N1
```

can reduce:

$$
192\rightarrow32
$$

before concatenation.

Thus, $1\times1$ convolutions also help control the final output depth of the Inception module.

## 9. What Inception Solves

The first problem is architectural choice:

```text
Should we use:

1x1?
3x3?
5x5?
Pooling?
```

Inception answers:

```text
Use several operations in parallel.
```

The second problem is computational cost:

```text
Parallel spatial convolutions
can become very expensive.
```

Inception answers:

```text
Use 1x1 bottleneck convolutions
before expensive operations.
```

Therefore:

$$
\boxed{
\text{Inception}
=
\text{multi-scale processing}
+
\text{computational bottlenecks}
}
$$

## 10. Connection to Previous Concepts

The progression is:

```text
VGG
-> repeated small convolutions

ResNet
-> easier optimization of very deep networks

1x1 Convolution
-> efficient channel transformation

Inception
-> parallel multi-scale processing
   with 1x1 bottlenecks
```

Inception makes especially strong use of the $1\times1$ convolution because it allows the architecture to remain computationally practical.

## 11. Common Mistakes

| Mistake | Correct Understanding |
|---|---|
| Inception manually chooses the best kernel size | Several operations are applied in parallel |
| Inception branches run sequentially | They operate in parallel on the same input |
| $1\times1$ Conv mainly extracts spatial patterns | Its main role here is channel transformation and dimensionality reduction |
| $1\times1$ Conv simply deletes channels | It learns combinations of all input channels |
| Concatenation happens along spatial dimensions | Branches are concatenated along the channel dimension |
| Branch outputs can have arbitrary $H,W$ | Their spatial dimensions must match before concatenation |
| Inception only focuses on multi-scale features | Computational efficiency is also a central motivation |
| Bottlenecks preserve all information perfectly | They are learned compressions and can lose information if overly aggressive |

## 12. Essence

The original question is:

```text
Which operation should a CNN layer use?

1x1 Conv?
3x3 Conv?
5x5 Conv?
Pooling?
```

Inception answers:

```text
Do not choose only one.

Use several operations in parallel.
```

This provides:

```text
multi-scale feature extraction
```

but creates:

```text
high computational cost
```

so Inception introduces:

```mermaid
flowchart LR
    N0["1x1 Conv"]
    N1["Reduce channels"]
    N2["Expensive spatial Conv"]
    N0 --> N1 --> N2
```

The practical architecture is:

```mermaid
flowchart LR
    I["Input"]
    C["Concatenate"]
    B0_0["1x1"]
    I --> B0_0
    B0_0 --> C
    B1_0["1x1"]
    I --> B1_0
    B1_1["3x3"]
    B1_0 --> B1_1
    B1_1 --> C
    B2_0["1x1"]
    I --> B2_0
    B2_1["5x5"]
    B2_0 --> B2_1
    B2_1 --> C
    B3_0["Pool"]
    I --> B3_0
    B3_1["1x1"]
    B3_0 --> B3_1
    B3_1 --> C
```

The key takeaway is:

> **Inception extracts features at multiple spatial scales in parallel, while using $1\times1$ bottleneck convolutions to control channel dimensions and prevent computational cost from exploding.**


# Inception Network

## 1. Core Idea

The central idea of Inception is:

> **Instead of forcing a CNN layer to choose only one operation, process the same input through several operations in parallel and concatenate their outputs.**

Different operations capture different types of information:

```text
1x1 Conv
-> channel transformation

3x3 Conv
-> medium-scale spatial features

5x5 Conv
-> larger-scale spatial features

Pooling
-> local aggregation and robust features
```

Therefore:

$$
\boxed{
\text{Inception}
=
\text{multi-scale processing}
+
\text{computational efficiency}
}
$$

The architecture tries to answer two questions:

```text
1. Which receptive-field size should we use?

2. How can we use multiple receptive fields
   without making computation explode?
```

## 2. Inception v1 Module

The original Inception architecture was introduced in **GoogLeNet / Inception v1**.

Suppose the input is:

$$
X\in\mathbb{R}^{28\times28\times192}
$$

A practical Inception module contains four parallel branches:

```mermaid
flowchart LR
    A[Input]

    A --> B1["1×1 Conv"]
    A --> B2["1×1 Conv"]
    A --> B3["1×1 Conv"]
    A --> B4["MaxPool"]

    B2 --> C2["3×3 Conv"]
    B3 --> C3["5×5 Conv"]
    B4 --> C4["1×1 Conv"]

    B1 --> D["Concatenate"]
    C2 --> D
    C3 --> D
    C4 --> D
```

All branches receive the **same input**.

They operate in parallel and their outputs are concatenated along the channel dimension.

![Inception Module](https://miro.medium.com/v2/resize:fit:1400/1*uW81y16b-ptBDV8SIT1beQ.png)

### Branch 1 — $1\times1$ Convolution

Example:

$$
28\times28\times192
\rightarrow
28\times28\times64
$$

This branch performs learned channel transformation:

$$
\mathbb{R}^{192}
\rightarrow
\mathbb{R}^{64}
$$

at each spatial location.

It does not expand the spatial receptive field.

Its main purpose is:

```text
channel mixing
+
cheap nonlinear feature transformation
```

### Branch 2 — $1\times1\rightarrow3\times3$

Example:

$$
28\times28\times192
\rightarrow
28\times28\times96
\rightarrow
28\times28\times128
$$

Flow:

```mermaid
flowchart TD
    N0["192 channels"]
    N1["1x1 Conv"]
    N2["96 channels"]
    N3["3x3 Conv"]
    N4["128 channels"]
    N0 --> N1 --> N2 --> N3 --> N4
```

The first $1\times1$ convolution reduces the number of channels before the expensive $3\times3$ convolution.

### Branch 3 — $1\times1\rightarrow5\times5$

Example:

$$
28\times28\times192
\rightarrow
28\times28\times16
\rightarrow
28\times28\times32
$$

Flow:

```mermaid
flowchart TD
    N0["192 channels"]
    N1["1x1 Conv"]
    N2["16 channels"]
    N3["5x5 Conv"]
    N4["32 channels"]
    N0 --> N1 --> N2 --> N3 --> N4
```

The $1\times1$ convolution acts as a **bottleneck** before the computationally expensive $5\times5$ convolution.

### Branch 4 — Pooling $\rightarrow1\times1$

Example:

$$
28\times28\times192
\rightarrow
28\times28\times192
\rightarrow
28\times28\times32
$$

Flow:

```mermaid
flowchart TD
    N0["Input"]
    N1["MaxPool"]
    N2["192 channels"]
    N3["1x1 Conv"]
    N4["32 channels"]
    N0 --> N1 --> N2 --> N3 --> N4
```

Pooling itself normally preserves channel depth.

The following $1\times1$ convolution prevents the pooling branch from contributing too many channels to the final concatenation.

## 3. Concatenation

Suppose the four branches produce:

```text
1x1 branch        -> 64 channels
3x3 branch        -> 128 channels
5x5 branch        -> 32 channels
Pooling branch    -> 32 channels
```

All outputs preserve:

$$
H=W=28
$$

Therefore their channels can be concatenated:

$$
64+128+32+32
=
256
$$

Final output:

$$
\boxed{
28\times28\times256
}
$$

Conceptually:

```mermaid
flowchart LR
    B1["Branch 1<br/>64 channels"] --> C["Concatenate"]
    B2["Branch 2<br/>128 channels"] --> C
    B3["Branch 3<br/>32 channels"] --> C
    B4["Branch 4<br/>32 channels"] --> C
    C --> O["256 channels"]
```

Important distinction:

```text
Inception:
combine branches by concatenation

ResNet:
combine paths by element-wise addition
```

For Inception concatenation:

```text
H and W must match.

Channel counts may differ.
```

## 4. The Computational Problem

The naive Inception design would apply:

```text
1x1
3x3
5x5
pooling
```

directly to the full input tensor.

Suppose:

$$
X\in\mathbb{R}^{28\times28\times192}
$$

and the $5\times5$ branch uses 32 filters.

One filter has:

$$
5\times5\times192
$$

weights.

Each output activation therefore requires:

$$
5\times5\times192
=
4800
$$

multiplications.

The output contains:

$$
28\times28\times32
$$

activations.

Total:

$$
28\times28\times32\times5\times5\times192
$$

$$
=
120{,}422{,}400
$$

or approximately:

$$
\boxed{120.4M}
$$

multiplications.

The general convolution cost is approximately:

$$
\boxed{
H_{out}W_{out}C_{out}
k_hk_wC_{in}
}
$$

Therefore:

```text
large kernel
×
many input channels
×
many output channels
×
many spatial positions

=
huge computation
```

The problem becomes worse because Inception executes several branches in parallel.

## 5. $1\times1$ Convolution as a Bottleneck

Instead of:

```mermaid
flowchart TD
    N0["28x28x192"]
    N1["5x5 Conv"]
    N2["28x28x32"]
    N0 --> N1 --> N2
```

Inception uses:

```mermaid
flowchart TD
    N0["28x28x192"]
    N1["1x1 Conv<br/>192 → 16"]
    N2["28x28x16"]
    N3["5x5 Conv<br/>16 → 32"]
    N4["28x28x32"]
    N0 --> N1 --> N2 --> N3 --> N4
```

The expensive spatial convolution now processes only:

$$
16
$$

input channels rather than:

$$
192
$$

channels.

### Computation of the $1\times1$ Layer

$$
28\times28\times192\times16
$$

$$
\approx2.4M
$$

### Computation of the $5\times5$ Layer

$$
28\times28\times16\times32\times25
$$

$$
\approx10M
$$

Total:

$$
2.4M+10M
=
12.4M
$$

Compare:

```text
Direct 5x5:

≈ 120.4M


1x1 bottleneck + 5x5:

≈ 12.4M
```

So adding an extra cheap layer can make the entire operation much cheaper.

The reusable design principle is:

$$
\boxed{
\text{cheap channel projection}
\rightarrow
\text{expensive spatial operation}
}
$$

## 6. Why Does the Bottleneck Still Work?

Reducing:

$$
192\rightarrow16
$$

does not mean:

```text
keep 16 original channels
delete the other 176
```

Instead, each new channel is a learned combination:

$$
z_k
=
\sum_{c=1}^{192}
W_{k,c}x_c
$$

Therefore:

```mermaid
flowchart TD
    N0["192-dimensional representation"]
    N1["Learned projection"]
    N2["16-dimensional representation"]
    N0 --> N1 --> N2
```

Every output channel can contain information from **all 192 input channels**.

However:

> A bottleneck is still a compression operation.

If the bottleneck is too narrow, useful information can be lost.

Therefore, bottleneck width is an architectural design choice.

## 7. Why Keep a Standalone $1\times1$ Branch?

The $1\times1$ convolution is not only a dimensionality-reduction tool.

The standalone branch:

```mermaid
flowchart TD
    N0["Input"]
    N1["1x1 Conv"]
    N0 --> N1
```

learns useful transformations directly across channels.

Therefore:

```text
1x1 before 3x3 / 5x5:
mainly bottleneck + channel transformation

standalone 1x1:
direct channel-wise feature learning
```

## 8. Why Include Pooling as a Parallel Branch?

Convolution and pooling have different inductive behaviors.

Convolution:

```text
learned spatial transformation
```

Pooling:

```text
fixed local aggregation
```

Using both allows the module to preserve:

```text
learned spatial features
+
pooled robust features
```

The network can later decide which representations are useful.

## 9. GoogLeNet — Full Inception v1 Network

The complete Inception v1 architecture is called **GoogLeNet**.

High-level flow:

```mermaid
flowchart TD
    N0["Input<br/>224 x 224 x 3"]
    N1["Initial Conv / Pool"]
    N2["Inception 3a"]
    N3["Inception 3b"]
    N4["MaxPool"]
    N5["Inception 4a"]
    N6["Inception 4b"]
    N7["Inception 4c"]
    N8["Inception 4d"]
    N9["Inception 4e"]
    N10["MaxPool"]
    N11["Inception 5a"]
    N12["Inception 5b"]
    N13["Global Average Pooling"]
    N14["Classifier"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6 --> N7 --> N8 --> N9 --> N10 --> N11 --> N12 --> N13 --> N14
```

The network contains nine Inception modules:

```text
3a
3b

4a
4b
4c
4d
4e

5a
5b
```

A major architectural characteristic is that GoogLeNet avoids the huge fully connected classifier used in networks such as VGG.

Instead, it uses Global Average Pooling near the end.

## 10. Global Average Pooling

Suppose the final representation is:

$$
H\times W\times C
$$

Global Average Pooling computes:

$$
z_c
=
\frac{1}{HW}
\sum_{i=1}^{H}
\sum_{j=1}^{W}
A_{i,j,c}
$$

for every channel.

Therefore:

$$
H\times W\times C
\rightarrow
C
$$

Example:

```mermaid
flowchart TD
    N0["7 x 7 x 1024"]
    N1["Global Average Pooling"]
    N2["1024-dimensional vector"]
    N0 --> N1 --> N2
```

This greatly reduces the number of classifier parameters compared with:

```text
Flatten
-> huge FC
-> huge FC
-> classifier
```

## 11. Auxiliary Classifiers

GoogLeNet also introduced **Auxiliary Classifiers**.

At the time, deep networks were difficult to optimize and residual connections had not yet been introduced as the standard solution.

Without auxiliary supervision:

```text
early layers
---------------------------------> Main Loss
```

The gradient from the main classifier must pass through many transformations.

GoogLeNet therefore attaches small classification heads to intermediate layers.

Conceptually:

```mermaid
flowchart TD
    I["Input"] --> A["Inception"]
    A --> M1["More layers"] --> B["Inception"]
    B --> M2["More layers"] --> MC["Main Classifier"] --> ML["Main Loss"]

    A --> AC1["Auxiliary Classifier 1"] --> AL1["Aux Loss 1"]
    B --> AC2["Auxiliary Classifier 2"] --> AL2["Aux Loss 2"]
```

In original GoogLeNet, the two auxiliary heads are attached around intermediate Inception modules 4a and 4d.

## 12. How Auxiliary Classifiers Work

An auxiliary classifier is a small classification network attached to an intermediate feature map.

Conceptually:

```mermaid
flowchart TD
    N0["Intermediate Feature Map"]
    N1["Average Pooling"]
    N2["1x1 Conv"]
    N3["Fully Connected"]
    N4["ReLU"]
    N5["Dropout"]
    N6["Classifier"]
    N7["Auxiliary Loss"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6 --> N7
```

It predicts the same target classes as the final classifier.

Instead of optimizing only:

$$
L=L_{\text{main}}
$$

training uses approximately:

$$
\boxed{
L
=
L_{\text{main}}
+
0.3L_{\text{aux1}}
+
0.3L_{\text{aux2}}
}
$$

Therefore, for an early parameter $W$:

$$
\frac{\partial L}{\partial W}
=
\frac{\partial L_{\text{main}}}{\partial W}
+
0.3
\frac{\partial L_{\text{aux1}}}{\partial W}
+
0.3
\frac{\partial L_{\text{aux2}}}{\partial W}
$$

The early layers can receive:

```text
long gradient path from main classifier
+
shorter gradient paths from auxiliary classifiers
```

During inference:

```text
Auxiliary classifiers are removed.

Only the main classifier is used.
```

## 13. What Auxiliary Classifiers Actually Help With

The original motivation was:

```text
improve gradient signals
+
encourage useful intermediate features
+
help optimization
```

However, later experiments by the Inception authors suggested that auxiliary classifiers did not mainly improve the early convergence phase.

Their benefit appeared more strongly later in training.

Therefore, an important modern interpretation is:

```text
Auxiliary classifiers
=
additional supervision
+
regularization
```

rather than simply:

```text
a solution for vanishing gradients
```

## 14. Auxiliary Classifier vs Residual Connection

These mechanisms should not be confused.

### Auxiliary Classifier

```mermaid
flowchart LR
    A["Intermediate Features"] --> B["Auxiliary Head"] --> C["Aux Loss"]
```

It adds:

$$
\boxed{\text{additional supervision}}
$$

### Residual Connection

```mermaid
flowchart LR
    I["x"]
    A(("+"))
    M0["F(x)"]
    I --> M0
    M0 --> A
    I --> A
    O["F(x) + x"]
    A --> O
```

It adds:

$$
\boxed{\text{direct information and gradient path}}
$$

Comparison:

| Auxiliary Classifier | Residual Connection |
|---|---|
| Adds another loss | Adds another feature path |
| Provides intermediate supervision | Provides an identity shortcut |
| Mostly a training mechanism | Part of forward computation |
| Removed during inference | Remains during inference |
| Gradient comes from extra classifier | Gradient flows directly through shortcut |
| Does not define the block output | Changes output to $F(x)+x$ |

Therefore:

$$
\boxed{
\text{Auxiliary Classifier}
\neq
\text{Residual Connection}
}
$$

## 15. Evolution of the Inception Family

The general evolution is:

```mermaid
flowchart TD
    A["Inception v1<br/>Parallel multi-scale processing"]
    --> B["Inception v2<br/>Convolution factorization"]
    --> C["Inception v3<br/>Stronger factorization<br/>+ improved training techniques"]
    --> D["Inception v4<br/>Deeper and more systematic architecture"]
    --> E["Inception-ResNet<br/>Inception transformations<br/>+ residual connections"]
```

There is an important naming nuance:

> **The terms Inception v2 and v3 are not used perfectly consistently across all historical papers and implementations.**

A useful study convention is:

```text
Inception v1
= original GoogLeNet

Inception v2
= redesigned / factorized Inception generation

Inception v3
= cumulative improved version with
  further factorization and training refinements

Inception v4
= deeper standardized architecture
```

The design evolution is more important than memorizing the labels.

## 16. Inception v2 — Replace $5\times5$ with Two $3\times3$

One major improvement is:

$$
5\times5
\rightarrow
3\times3
\rightarrow
3\times3
$$

Two stacked $3\times3$ convolutions have a theoretical receptive field of approximately:

$$
5\times5
$$

for stride 1 and dilation 1.

Assume equal channel width $C$.

### One $5\times5$

$$
25C^2
$$

weights.

### Two $3\times3$

$$
9C^2+9C^2
=
18C^2
$$

Ratio:

$$
\frac{18}{25}
=
0.72
$$

So the simplified saving is approximately:

$$
\boxed{28\%}
$$

In addition, two convolutions introduce two nonlinear transformations:

```mermaid
flowchart LR
    subgraph A["5x5"]
        A1["Conv"] --> A2["ReLU"]
    end

    subgraph B["3x3 + 3x3"]
        B1["Conv"] --> B2["ReLU"] --> B3["Conv"] --> B4["ReLU"]
    end
```

Therefore the factorized version can be both:

```text
cheaper
+
more expressive
```

## 17. Is This Idea Taken from VGG?

There is a strong conceptual connection.

VGG demonstrated that multiple small $3\times3$ convolutions can replace larger receptive-field operations while increasing network depth.

For example:

$$
3\times3
\rightarrow
3\times3
$$

has approximately a:

$$
5\times5
$$

theoretical receptive field.

And:

$$
3\times3
\rightarrow
3\times3
\rightarrow
3\times3
$$

has approximately a:

$$
7\times7
$$

theoretical receptive field.

Therefore:

```text
VGG:
small kernels
+
greater depth

Inception:
factorize expensive large kernels
into smaller operations
```

However, it is not accurate to simply say:

```text
Inception copied VGG.
```

A better statement is:

> **VGG popularized the effectiveness of stacking small convolutions, while later Inception architectures developed the same broad principle further as explicit convolution factorization for computational efficiency.**

## 18. Inception v3 — Asymmetric Factorization

The next step asks:

> Can $3\times3$ itself be factorized further?

Instead of:

$$
3\times3
$$

use:

$$
1\times3
\rightarrow
3\times1
$$

Conceptually:

```mermaid
flowchart TD
    N0["3x3 Conv"]
    N1["Factorize"]
    N2["1x3 Conv"]
    N3["3x1 Conv"]
    N0 --> N1 --> N2 --> N3
```

Assuming equal channel widths:

### Original $3\times3$

$$
9C^2
$$

### Factorized Version

$$
3C^2+3C^2
=
6C^2
$$

Ratio:

$$
\frac{6}{9}
=
\frac23
$$

giving approximately:

$$
\boxed{33\%}
$$

saving under the simplified assumptions.

The same principle can be used for larger kernels:

$$
7\times7
\rightarrow
1\times7
\rightarrow
7\times1
$$

This is called **asymmetric convolution factorization**.

The general principle becomes:

$$
\boxed{
n\times n
\rightarrow
1\times n
\rightarrow
n\times1
}
$$

## 19. Better Grid-Size Reduction

CNNs eventually need to reduce:

$$
H,W
$$

while usually increasing:

$$
C
$$

A naive downsampling pipeline might be:

```text
Conv
-> narrow intermediate representation
-> downsample
```

which can create a representational bottleneck.

Later Inception architectures instead use parallel reduction:

```mermaid
flowchart LR
    I["Input"]
    C["Concatenate"]
    B0_0["Conv<br/>stride 2"]
    I --> B0_0
    B0_0 --> C
    B1_0["Pool<br/>stride 2"]
    I --> B1_0
    B1_0 --> C
```

Both branches reduce spatial resolution:

$$
H,W\downarrow
$$

but their outputs are concatenated:

$$
C\uparrow
$$

This preserves more representational capacity during downsampling.

## 20. Additional Inception v3 Improvements

The architecture commonly called **Inception v3** combines several ideas:

```text
5x5 factorization
+
asymmetric convolution factorization
+
efficient grid-size reduction
+
Batch Normalization
+
RMSProp
+
label smoothing
+
improved auxiliary classifier
```

So Inception v3 should not be understood as just:

```text
replace one kernel
```

but as a more comprehensive redesign of both:

```text
architecture
+
training procedure
```

## 21. Inception v4

Inception v4 makes the architecture deeper and more systematic.

Instead of one generic repeated Inception block, it organizes the network into standardized modules:

```mermaid
flowchart TD
    N0["Stem"]
    N1["Inception-A"]
    N2["Inception-A"]
    N3["..."]
    N4["Reduction-A"]
    N5["Inception-B"]
    N6["Inception-B"]
    N7["..."]
    N8["Reduction-B"]
    N9["Inception-C"]
    N10["Inception-C"]
    N11["..."]
    N12["Global Average Pooling"]
    N13["Classifier"]
    N0 --> N1 --> N2 --> N3 --> N4 --> N5 --> N6 --> N7 --> N8 --> N9 --> N10 --> N11 --> N12 --> N13
```

Main module families:

```text
Inception-A
Inception-B
Inception-C

Reduction-A
Reduction-B
```

The A/B/C modules operate at different spatial resolutions.

Reduction modules perform controlled downsampling between stages.

The main improvement is:

```text
cleaner architecture
+
deeper architecture
+
more systematic block design
```

Important:

> **Pure Inception v4 does not use ResNet residual connections.**

Residual connections belong to the separate **Inception-ResNet** architectures introduced alongside it.

## 22. Inception-ResNet

After ResNet demonstrated the effectiveness of residual learning, the Inception design was combined with residual connections.

This produces:

$$
\boxed{
\text{Inception-ResNet}
}
$$

![Inception-ResNet Architecture](https://miro.medium.com/v2/resize:fit:1200/1*CYRgf1i1q_4hx5AcdcaSEg.jpeg)

A normal Inception transformation computes features through several branches:

```text
             ┌-> Branch A ──┐
             │              │
x -----------├-> Branch B ──┤
             │              ├-> Concatenate
             └-> Branch C ──┘
```

Pure Inception would use the concatenated representation as the next representation.

Inception-ResNet instead transforms the concatenated output into a residual:

```text
x ------------------------------------------┐
                                           │
        ┌-> Branch A ──┐                    │
        │              │                    │
x ------├-> Branch B ──┼-> Concatenate      │
        │              │        │           │
        └-> Branch C ──┘        ▼           │
                            1x1 Conv         │
                                │            │
                                ▼            │
                               F(x)          │
                                │            │
                                └----------->+
                                             │
                                             ▼
                                           y
```

The output becomes:

$$
\boxed{
y=x+F(x)
}
$$

Therefore:

```text
Inception contributes:
multi-scale feature extraction

ResNet contributes:
residual learning
+
direct information path
+
direct gradient path
```

## 23. Why Is a $1\times1$ Conv Used Before Residual Addition?

Suppose the Inception branches produce:

```text
Branch A -> 32 channels
Branch B -> 32 channels
Branch C -> 64 channels
```

After concatenation:

$$
32+32+64
=
128
$$

channels.

But suppose the shortcut input has:

$$
256
$$

channels.

Then:

$$
128\neq256
$$

so element-wise addition is impossible.

A $1\times1$ convolution performs:

$$
128\rightarrow256
$$

giving:

$$
\operatorname{shape}(F(x))
=
\operatorname{shape}(x)
$$

Then:

$$
F(x)+x
$$

is valid.

Conceptually:

```mermaid
flowchart LR
    X["x"] --> A["Inception branches"]
    A --> B["Concatenate"]
    B --> C["1x1 Conv"]
    C --> D["Match channel depth"]
    D --> E["F(x)"]
    E --> G(("+"))
    X --> G
    G --> O["Output"]
```

## 24. Residual Scaling in Inception-ResNet

Inception branches can be very wide.

Therefore the magnitude of:

$$
F(x)
$$

may become large.

Directly computing:

$$
x+F(x)
$$

can make optimization unstable in very wide residual architectures.

A useful solution is:

$$
\boxed{
y=x+\alpha F(x)
}
$$

where:

$$
0<\alpha<1
$$

Conceptually:

```mermaid
flowchart LR
    X["x"]
    X --> A["Inception branches"] --> F["F(x)"] --> S["Multiply by α"] --> SF["αF(x)"] --> G(("+"))
    X --> G
    G --> O["x + αF(x)"]
```

The residual branch becomes a controlled correction to the existing representation.

## 25. Inception vs ResNet vs Inception-ResNet

### Inception

```mermaid
flowchart LR
    I["Input"]
    C["Concatenate"]
    B0_0["Branch A"]
    I --> B0_0
    B0_0 --> C
    B1_0["Branch B"]
    I --> B1_0
    B1_0 --> C
    B2_0["Branch C"]
    I --> B2_0
    B2_0 --> C
    B3_0["Branch D"]
    I --> B3_0
    B3_0 --> C
```

Core principle:

```text
multi-scale parallel feature extraction
```

### ResNet

```mermaid
flowchart LR
    I["x"]
    A(("+"))
    M0["F(x)"]
    I --> M0
    M0 --> A
    I --> A
    O["F(x) + x"]
    A --> O
```

Core principle:

```text
residual learning
+
identity shortcut
```

### Inception-ResNet

```mermaid
flowchart LR
    X["x"] --> B["Inception branches"]
    B --> C["Concatenate"]
    C --> P["Projection"]
    P --> F["F(x)"]
    F --> A(("+"))
    X --> A
    A --> O["Output"]
```

Core principle:

```text
multi-scale processing
+
residual learning
```

## 26. VGG vs Inception vs ResNet

| Architecture | Main Design Principle |
|---|---|
| VGG | Stack simple small $3\times3$ convolutions |
| Inception | Process multiple spatial scales in parallel |
| ResNet | Learn residual corrections through shortcut connections |
| Inception-ResNet | Combine multi-scale Inception transformations with residual learning |

Conceptually:

```text
VGG
-> depth through repeated simple blocks

Inception
-> width through parallel multi-scale branches

ResNet
-> depth through easier residual optimization

Inception-ResNet
-> multi-scale transformations
   +
   residual shortcuts
```

## 27. Evolution of the Inception Family

The entire progression can be summarized as:

```mermaid
flowchart TD
    S0["Inception v1"]
    P0["Problem:<br/>Which kernel size should we use?"]
    Q0["Solution:<br/>Parallel 1x1 / 3x3 / 5x5 / pooling"]
    S0 --> P0 --> Q0
    S1["Inception v2"]
    P1["Problem:<br/>Large convolutions are expensive."]
    Q1["Solution:<br/>5x5 → 3x3 → 3x3"]
    S1 --> P1 --> Q1
    Q0 --> S1
    S2["Inception v3"]
    P2["Problem:<br/>Can convolution be factorized further?"]
    Q2["Solution:<br/>nxn → 1xn → nx1<br/>+ better grid reduction<br/>+ improved regularization/training"]
    S2 --> P2 --> Q2
    Q1 --> S2
    S3["Inception v4"]
    P3["Problem:<br/>Inception architecture became complex."]
    Q3["Solution:<br/>Standardized Inception-A / B / C<br/>+ Reduction-A / B<br/>+ deeper systematic architecture"]
    S3 --> P3 --> Q3
    Q2 --> S3
    S4["Inception-ResNet"]
    P4["Problem:<br/>Can Inception gain the optimization advantages of residual learning?"]
    Q4["Solution:<br/>Inception branches → concatenate → 1x1 projection → residual addition"]
    S4 --> P4 --> Q4
    Q3 --> S4
```

There is also a parallel historical evolution in deep-network optimization:

```text
GoogLeNet:

deep network
-> auxiliary classifiers
-> intermediate supervision / regularization


ResNet:

deep network
-> identity shortcut
-> direct information and gradient paths


Inception-ResNet:

Inception feature extraction
+
ResNet residual optimization
```

## 28. Common Mistakes

| Mistake | Correct Understanding |
|---|---|
| Inception selects one best kernel size | It uses several operations in parallel |
| Inception branches operate sequentially | They operate in parallel |
| Branch outputs are added | Pure Inception concatenates them |
| All branch tensor shapes must be identical | Only $H,W$ must match for channel concatenation |
| $1\times1$ Conv is only a bottleneck | It also directly learns channel transformations |
| $1\times1$ Conv simply deletes channels | It learns projections across all input channels |
| $5\times5$ directly processes the full input depth in practical Inception | A $1\times1$ bottleneck normally reduces channels first |
| Pooling creates no channel-growth problem | Pooling preserves channels, so a $1\times1$ projection is useful afterward |
| Two $3\times3$ layers are mathematically identical to one $5\times5$ | They only have a similar theoretical receptive field; their parameterization and nonlinearities differ |
| Inception v2 simply copied VGG | They share the small-kernel principle, but Inception develops it as explicit computational factorization |
| Auxiliary Classifier is a skip connection | It is an additional supervised loss branch |
| Auxiliary Classifiers remain during inference | They are removed during inference |
| Auxiliary Classifiers only solve vanishing gradients | They also provide intermediate supervision and have a regularization effect |
| Inception v4 uses residual connections | Pure Inception v4 does not |
| Inception and ResNet merge branches in the same way | Inception concatenates; ResNet adds |
| Inception-ResNet directly adds concatenated branches to $x$ | A projection is normally used to match the residual depth first |
| Increasing channels restores spatial information lost by downsampling | More channels increase feature capacity but do not recreate lost spatial detail |

## 29. Essence

The fundamental Inception block is:

```mermaid
flowchart LR
    I["Input"]
    C["Concatenate"]
    B0_0["1x1"]
    I --> B0_0
    B0_0 --> C
    B1_0["1x1"]
    I --> B1_0
    B1_1["3x3"]
    B1_0 --> B1_1
    B1_1 --> C
    B2_0["1x1"]
    I --> B2_0
    B2_1["5x5"]
    B2_0 --> B2_1
    B2_1 --> C
    B3_0["Pool"]
    I --> B3_0
    B3_1["1x1"]
    B3_0 --> B3_1
    B3_1 --> C
```

Its two original design principles are:

$$
\boxed{
\text{multi-scale parallel feature extraction}
}
$$

and:

$$
\boxed{
\text{$1\times1$ bottlenecks for computational efficiency}
}
$$

The architecture later evolves through convolution factorization:

$$
5\times5
\rightarrow
3\times3
\rightarrow
3\times3
$$

then:

$$
3\times3
\rightarrow
1\times3
\rightarrow
3\times1
$$

and more generally:

$$
n\times n
\rightarrow
1\times n
\rightarrow
n\times1
$$

Auxiliary Classifiers provide intermediate supervision during GoogLeNet training:

$$
L
=
L_{\text{main}}
+
0.3L_{\text{aux1}}
+
0.3L_{\text{aux2}}
$$

but are removed during inference.

Finally, Inception is combined with residual learning:

$$
\boxed{
y=x+\alpha F_{\text{Inception}}(x)
}
$$

giving:

```text
Inception:
rich multi-scale feature extraction

+

ResNet:
direct information and gradient paths
```

The central takeaway is:

> **Inception builds rich multi-scale representations by processing the same input through parallel branches, controls computation through $1\times1$ bottlenecks and convolution factorization, and later evolves into Inception-ResNet by combining these transformations with residual shortcuts.**


# MobileNet

## 1. Core Idea

MobileNet was designed for a very practical goal:

> **Make CNNs much lighter and cheaper so they can run efficiently on mobile and edge devices.**

The key idea of MobileNet is not mainly a complicated architecture.

Instead, it comes from replacing one expensive operation:

$$
\text{Standard Convolution}
\quad \longrightarrow \quad
\text{Depthwise Separable Convolution}
$$

So the most important part of MobileNet is understanding:

```text
What does standard convolution do?
What is depthwise convolution?
What is pointwise 1x1 convolution?
Why is the new operation much cheaper?
```

## 2. The Problem with Standard Convolution

Suppose the input has shape:

$$
H \times W \times C_{in}
$$

and we want an output of shape:

$$
H' \times W' \times C_{out}
$$

using a kernel of size:

$$
K \times K
$$

In a **standard convolution**, each output channel uses one filter of size:

$$
K \times K \times C_{in}
$$

Since there are:

$$
C_{out}
$$

such filters, the total number of parameters is:

$$
K^2 C_{in} C_{out}
$$

ignoring bias.

The computation cost is also large, approximately:

$$
H'W'K^2C_{in}C_{out}
$$

The key point is:

> **A standard convolution performs spatial filtering and channel mixing at the same time.**

## 3. Standard Convolution Performs Two Jobs at Once

A standard convolution actually does two things simultaneously.

### Spatial filtering

It scans a local neighborhood with a $K \times K$ kernel and learns patterns such as:

```text
edges
textures
corners
local visual structures
```

### Channel mixing

It also combines information from multiple input channels to produce new output channels.

So we can summarize standard convolution as:

$$
\boxed{
\text{Standard Conv}
=
\text{spatial filtering}
+
\text{channel mixing}
}
$$

MobileNet asks:

> **What if we separate these two jobs?**

## 4. Depthwise Separable Convolution

This is the core idea of MobileNet.

Instead of one standard convolution, MobileNet uses two steps:

```text
Depthwise Convolution
        ↓
Pointwise 1×1 Convolution
```

So:

$$
\boxed{
\text{Depthwise Separable Conv}
=
\text{Depthwise Conv}
+
\text{Pointwise Conv}
}
$$

The following image gives a helpful visual intuition for these two operations:

![Depthwise and pointwise convolution illustration](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRiauSOxyOyo3cJRHjx0o6yZIi5wqM9kwh7L8vJ0vIW1zt_opRGJwA52qJw&s=10)

## 5. What is Depthwise Convolution?

Suppose the input is:

$$
H \times W \times C_{in}
$$

In **depthwise convolution**, we do **not** use filters of size:

$$
K \times K \times C_{in}
$$

as in standard convolution.

Instead, we use:

- one separate filter for each input channel
- each filter has size:

$$
K \times K \times 1
$$

If the input has:

$$
C_{in}
$$

channels, then we use:

$$
C_{in}
$$

depthwise filters.

For example, if the input is:

```text
H × W × 32
```

then depthwise convolution uses:

```text
32 filters, each of size 3 × 3 × 1
```

Each filter processes only one input channel.

So conceptually:

```text
Channel 1 -> filtered independently
Channel 2 -> filtered independently
...
Channel 32 -> filtered independently
```

Therefore:

> **Depthwise convolution performs spatial filtering only.**

## 6. Important Point: Depthwise Conv Does Not Mix Channels

This is the most important idea to remember.

After depthwise convolution:

- each channel has been spatially processed
- but channels are still treated independently

There is no mixing like:

```text
channel 1 + channel 5 + channel 20
```

inside the depthwise operation.

So:

```text
Depthwise Conv
= spatial processing per channel
= no channel mixing
```

This means depthwise convolution alone is not enough.

We still need another step to combine channel information.

## 7. What is Pointwise Convolution?

After depthwise convolution, MobileNet applies a:

$$
1 \times 1 \text{ convolution}
$$

called:

$$
\boxed{\text{Pointwise Convolution}}
$$

Suppose the depthwise output has shape:

$$
H \times W \times C_{in}
$$

Then a pointwise convolution can transform it into:

$$
H \times W \times C_{out}
$$

Each pointwise filter has size:

$$
1 \times 1 \times C_{in}
$$

and there are:

$$
C_{out}
$$

such filters.

Its role is:

> **to mix information across channels.**

So we can summarize:

$$
\boxed{
\text{Pointwise Conv}
=
\text{channel mixing}
}
$$

## 8. Why This Separation Matters

Now we can compare the two views.

### Standard convolution

```text
spatial filtering
+
channel mixing
```

are done together in one expensive operation.

### Depthwise separable convolution

```text
Depthwise Conv
-> spatial filtering

Pointwise 1×1 Conv
-> channel mixing
```

So MobileNet separates one heavy operation into two cheaper operations.

This is the main conceptual shift.

## 9. Parameter Comparison

### Standard convolution

The number of parameters is:

$$
K^2 C_{in} C_{out}
$$

### Depthwise separable convolution

It has two parts.

#### Depthwise part

$$
K^2 C_{in}
$$

#### Pointwise part

$$
C_{in} C_{out}
$$

So the total becomes:

$$
\boxed{
K^2 C_{in}
+
C_{in} C_{out}
}
$$

Compared with standard convolution:

$$
\boxed{
K^2 C_{in} C_{out}
}
$$

this is much cheaper.

## 10. Concrete Example

Suppose we have:

```text
3 × 3 Conv
32 -> 64
```

### Standard convolution

Parameters:

$$
3 \times 3 \times 32 \times 64
=
18{,}432
$$

### Depthwise separable convolution

#### Depthwise

$$
3 \times 3 \times 32
=
288
$$

#### Pointwise

$$
1 \times 1 \times 32 \times 64
=
2{,}048
$$

Total:

$$
288 + 2{,}048 = 2{,}336
$$

So we get:

```text
Standard Conv:              18,432 parameters
Depthwise Separable Conv:    2,336 parameters
```

That is only about:

$$
\frac{2336}{18432} \approx 0.127
$$

or roughly:

```text
12.7%
```

So the cost is reduced by about:

$$
8\times
$$

## 11. Understanding `groups` in PyTorch

To understand how depthwise convolution is implemented, we need to understand `groups`.

### Standard convolution

Normally in PyTorch:

```python
groups = 1
```

This means each filter can see all input channels.

### Grouped convolution

If:

```python
groups > 1
```

then the input channels are split into groups.

For example:

```python
in_channels = 32
groups = 4
```

Then the 32 channels are divided into 4 groups, with 8 channels per group.

### Depthwise convolution

Depthwise convolution is the special case where:

```python
groups = in_channels
```

For example:

```python
in_channels = 32
groups = 32
```

Then:

- each group contains exactly 1 input channel
- each filter operates on only 1 channel

That is exactly depthwise convolution.

In PyTorch, a depthwise convolution is often written as:

```python
nn.Conv2d(
    in_channels=32,
    out_channels=32,
    kernel_size=3,
    padding=1,
    groups=32,
)
```

Then the pointwise convolution is:

```python
nn.Conv2d(
    in_channels=32,
    out_channels=64,
    kernel_size=1,
)
```

So we can conclude:

$$
\boxed{
\text{Depthwise Conv}
=
\text{Grouped Conv with } groups = in\_channels
}
$$

## 12. Essence

The key change introduced by MobileNet is:

### Before

```text
Standard Conv
= spatial filtering + channel mixing
```

### MobileNet

```text
Depthwise Conv
= spatial filtering

Pointwise 1×1 Conv
= channel mixing
```

So:

$$
\boxed{
\text{Standard Conv}
\longrightarrow
\text{Depthwise Separable Conv}
}
$$

Because of this change, MobileNet greatly reduces:

- the number of parameters
- the amount of computation
- the overall model cost

while still keeping useful representational power.

## 13. Final Takeaway

If you only want to remember the most important part of MobileNet, keep these five points:

```text
1. Standard convolution performs spatial filtering and channel mixing together.

2. MobileNet separates these two roles.

3. Depthwise convolution performs spatial filtering independently on each channel.

4. Pointwise 1×1 convolution mixes information across channels.

5. Depthwise separable convolution is much cheaper than standard convolution.
```

That is the core idea of **MobileNet**.

# MobileNet Architecture

## 1. From Depthwise Separable Convolution to MobileNet V1

The main architectural idea of **MobileNet V1** is to replace most expensive standard convolutions with **depthwise separable convolutions**.

A depthwise separable convolution contains two operations:

```mermaid
flowchart LR
    A[Input] --> B["3×3 Depthwise Conv"]
    B --> C["1×1 Pointwise Conv"]
    C --> D[Output]
```

Their responsibilities are different:

* **Depthwise convolution** performs spatial filtering independently on each channel.
* **Pointwise $1\times1$ convolution** mixes information across channels.

Therefore:

$$
\boxed{
\text{MobileNet V1}
===================

\text{Stack of Depthwise Separable Convolution Blocks}
}
$$

The first layer is still a standard convolution, while most of the network uses depthwise separable convolution.

## 2. MobileNet V1 Basic Block

A typical MobileNet V1 block is:

```mermaid
flowchart TD
    A[Input]
    --> B["3×3 Depthwise Conv"]
    --> C[BatchNorm]
    --> D[ReLU]
    --> E["1×1 Pointwise Conv"]
    --> F[BatchNorm]
    --> G[ReLU]
    --> H[Output]
```

The two convolutions have different roles:

$$
\boxed{
\text{Depthwise Conv}
\rightarrow
\text{Spatial Processing}
}
$$

$$
\boxed{
\text{Pointwise Conv}
\rightarrow
\text{Channel Mixing}
}
$$

When spatial downsampling is required, the depthwise convolution can use:

$$
stride=2
$$

while the pointwise convolution normally uses:

$$
stride=1
$$

For example:

```mermaid
flowchart LR
    A["56×56×128"]
    --> B["3×3 Depthwise<br/>stride = 2"]
    --> C["28×28×128"]
    --> D["1×1 Pointwise<br/>128 → 256"]
    --> E["28×28×256"]
```

Thus:

$$
\boxed{
\text{Depthwise Conv can change spatial resolution}
}
$$

while:

$$
\boxed{
\text{Pointwise Conv changes channel dimension}
}
$$

## 3. MobileNet V1 Full Architecture

For an input image:

$$
224\times224\times3
$$

the architecture is approximately:

| Stage       | Operation                           | Output                 |
| ----------- | ----------------------------------- | ---------------------- |
| Input       | —                                   | $224\times224\times3$  |
| Stem        | $3\times3$ Standard Conv, $s=2$     | $112\times112\times32$ |
| Block 1     | DW + PW, $32\rightarrow64$          | $112\times112\times64$ |
| Block 2     | DW $s=2$ + PW, $64\rightarrow128$   | $56\times56\times128$  |
| Block 3     | DW + PW, $128\rightarrow128$        | $56\times56\times128$  |
| Block 4     | DW $s=2$ + PW, $128\rightarrow256$  | $28\times28\times256$  |
| Block 5     | DW + PW, $256\rightarrow256$        | $28\times28\times256$  |
| Block 6     | DW $s=2$ + PW, $256\rightarrow512$  | $14\times14\times512$  |
| Blocks 7–11 | DW + PW, $512\rightarrow512$ ×5     | $14\times14\times512$  |
| Block 12    | DW $s=2$ + PW, $512\rightarrow1024$ | $7\times7\times1024$   |
| Block 13    | DW + PW, $1024\rightarrow1024$      | $7\times7\times1024$   |
| Head        | Average Pooling                     | $1\times1\times1024$   |
| Classifier  | Fully Connected                     | Classes                |

The overall architecture is:

```mermaid
flowchart TD
    A["Input<br/>224×224×3"]
    --> B["Standard 3×3 Conv<br/>stride = 2<br/>112×112×32"]

    B --> C["Depthwise Separable Block<br/>112×112×64"]

    C --> D["DS Block<br/>stride = 2<br/>56×56×128"]

    D --> E["DS Block<br/>56×56×128"]

    E --> F["DS Block<br/>stride = 2<br/>28×28×256"]

    F --> G["DS Block<br/>28×28×256"]

    G --> H["DS Block<br/>stride = 2<br/>14×14×512"]

    H --> I["5 × DS Blocks<br/>14×14×512"]

    I --> J["DS Block<br/>stride = 2<br/>7×7×1024"]

    J --> K["DS Block<br/>7×7×1024"]

    K --> L["Average Pooling<br/>1×1×1024"]

    L --> M[Fully Connected]
    M --> N[Classes]
```

The main pattern is therefore:

$$
\boxed{
\text{Standard Conv}
\rightarrow
\text{Many Depthwise Separable Conv Blocks}
\rightarrow
\text{Average Pooling}
\rightarrow
\text{Classifier}
}
$$

## 4. Pointwise Convolution Still Dominates Computation

Depthwise convolution is extremely cheap.

Its approximate computational cost is:

$$
HWC_{in}K^2
$$

Pointwise convolution costs:

$$
HWC_{in}C_{out}
$$

When channel dimensions become large, for example:

$$
512\rightarrow512
$$

the $1\times1$ pointwise convolution becomes much more expensive than the depthwise convolution.

Therefore, MobileNet mainly shifts computation from expensive:

$$
3\times3
$$

dense convolutions toward efficient:

$$
1\times1
$$

dense convolutions.

The model becomes much cheaper overall, although pointwise convolution still accounts for most of the computation.

## 5. Limitations of MobileNet V1

The MobileNet V1 block is relatively simple:

```mermaid
flowchart LR
    A[Input]
    --> B[Depthwise]
    --> C[Pointwise]
    --> D[Output]
```

It does not yet include several mechanisms that later became important:

* residual connections;
* expansion layers;
* linear bottlenecks.

These limitations motivated **MobileNet V2**.

## 6. MobileNet V2: Main Architectural Changes

MobileNet V2 introduces two major architectural ideas:

1. **Expansion Layer**
2. **Residual / Skip Connection**

The V1 block:

```mermaid
flowchart LR
    A[Depthwise] --> B[Pointwise]
```

becomes:

```mermaid
flowchart LR
    A["1×1 Expansion"]
    --> B["3×3 Depthwise"]
    --> C["1×1 Projection"]
```

When the input and output shapes match, a residual connection is also added:

```mermaid
flowchart TD
    A[Input]
    --> B["1×1 Expansion"]
    --> C["3×3 Depthwise"]
    --> D["1×1 Projection"]
    --> E((Add))
    --> F[Output]

    A --> E
```

This structure is called an:

$$
\boxed{\text{Inverted Residual Bottleneck}}
$$

## 7. Expansion Layer

Suppose the input has:

$$
C_{in}
$$

channels.

MobileNet V2 first expands it using a $1\times1$ convolution:

$$
C_{in}
\rightarrow
tC_{in}
$$

where:

$$
t
$$

is the **expansion ratio**.

A common value is:

$$
t=6
$$

For example:

$$
32\rightarrow192
$$

because:

$$
192=6\times32
$$

```mermaid
flowchart LR
    A["Input<br/>32 channels"]
    --> B["1×1 Expansion Conv"]
    --> C["Expanded Representation<br/>192 channels"]
```

The expansion layer creates a richer high-dimensional representation before spatial processing.

## 8. Why Expand Before Depthwise Convolution?

Depthwise convolution processes each channel independently.

If the representation contains only:

$$
32
$$

channels, then the depthwise convolution has only 32 independent spatial feature channels to process.

After expansion:

$$
32\rightarrow192
$$

the depthwise convolution operates over:

$$
192
$$

channels.

This provides a richer feature space.

```mermaid
flowchart LR
    A["Low-Dimensional Representation<br/>32 channels"]
    --> B["Expansion"]
    --> C["High-Dimensional Workspace<br/>192 channels"]
    --> D["Spatial Processing<br/>Depthwise Conv"]
```

A useful mental model is:

$$
\boxed{
\text{Compact Representation}
\rightarrow
\text{Large Internal Workspace}
}
$$

## 9. Depthwise Convolution in the Expanded Space

After expansion:

$$
H\times W\times C_{in}
\rightarrow
H\times W\times tC_{in}
$$

a $3\times3$ depthwise convolution performs spatial processing.

For example:

```mermaid
flowchart LR
    A["56×56×192"]
    --> B["3×3 Depthwise Conv<br/>groups = 192"]
    --> C["56×56×192"]
```

if:

$$
stride=1
$$

The depthwise convolution:

* processes spatial information;
* keeps channels independent;
* keeps the number of channels unchanged.

## 10. Projection Layer

After depthwise convolution, MobileNet V2 uses another $1\times1$ convolution to reduce the expanded representation.

For example:

$$
192\rightarrow32
$$

```mermaid
flowchart LR
    A["Expanded Features<br/>192 channels"]
    --> B["1×1 Projection Conv"]
    --> C["Output<br/>32 channels"]
```

The complete channel transformation becomes:

$$
32
\rightarrow
192
\rightarrow
192
\rightarrow
32
$$

More generally:

$$
\boxed{
C_{in}
\rightarrow
tC_{in}
\rightarrow
tC_{in}
\rightarrow
C_{out}
}
$$

The last $1\times1$ convolution is called the **projection layer** because it projects the high-dimensional representation back into a lower-dimensional space.

## 11. Why Is It Called a Bottleneck?

The input and output representations are narrow, while the internal representation is wide.

For example:

$$
32
\rightarrow
192
\rightarrow
192
\rightarrow
32
$$

```mermaid
flowchart LR
    A["Narrow Input<br/>32"]
    --> B["Wide Internal Representation<br/>192"]
    --> C["Wide Internal Representation<br/>192"]
    --> D["Narrow Output<br/>32"]
```

This design allows:

* compact representations between blocks;
* expressive transformations inside each block.

## 12. Why Is It an Inverted Bottleneck?

A traditional ResNet bottleneck follows:

```mermaid
flowchart LR
    A["Wide<br/>256"]
    --> B["Narrow<br/>64"]
    --> C["Process<br/>64"]
    --> D["Wide<br/>256"]
```

Conceptually:

$$
\text{compress}
\rightarrow
\text{process}
\rightarrow
\text{expand}
$$

MobileNet V2 does the opposite:

```mermaid
flowchart LR
    A["Narrow<br/>32"]
    --> B["Wide<br/>192"]
    --> C["Process<br/>192"]
    --> D["Narrow<br/>32"]
```

Conceptually:

$$
\text{expand}
\rightarrow
\text{process}
\rightarrow
\text{compress}
$$

Therefore:

$$
\boxed{\text{Inverted Bottleneck}}
$$

## 13. Residual Connection

MobileNet V2 also introduces residual learning:

$$
y=F(x)+x
$$

The shortcut bypasses the entire transformation:

```mermaid
flowchart TD
    A[Input]
    --> B["1×1 Expansion"]
    --> C["3×3 Depthwise"]
    --> D["1×1 Projection"]
    --> E((Add))
    --> F[Output]

    A --> E
```

The skip connection therefore connects the **thin bottleneck representations**, not the expanded internal representation.

## 14. When Can the Residual Connection Be Used?

The identity shortcut requires compatible shapes.

Typically:

$$
stride=1
$$

and:

$$
C_{in}=C_{out}
$$

For example:

```mermaid
flowchart LR
    A["56×56×24"]
    --> B["Inverted Residual Block<br/>stride = 1"]
    --> C["56×56×24"]
    C --> D["Residual Connection Possible"]
```

But if the block changes resolution or channels:

```mermaid
flowchart LR
    A["56×56×24"]
    --> B["Block<br/>stride = 2"]
    --> C["28×28×32"]
    C --> D["No Identity Residual"]
```

So:

$$
\boxed{
\text{Use identity residual when }
stride=1
\text{ and }
C_{in}=C_{out}
}
$$

## 15. Why Expansion and Projection Work Together

MobileNet V2 must balance two goals.

### Rich internal computation

A larger representation improves the expressiveness of the transformation:

$$
32\rightarrow192
$$

### Compact storage between blocks

Keeping every intermediate tensor large would increase:

* memory usage;
* memory bandwidth;
* computational cost.

Therefore, MobileNet expands temporarily and then projects back down.

```mermaid
flowchart LR
    A["Compact Storage<br/>32 channels"]
    --> B["Large Workspace<br/>192 channels"]
    --> C["Compact Storage<br/>32 channels"]
```

A useful mental model is:

$$
\boxed{
\text{Small Storage}
\rightarrow
\text{Large Workspace}
\rightarrow
\text{Small Storage}
}
$$

## 16. Linear Bottleneck

The final projection introduces another important idea:

$$
\boxed{\text{Linear Bottleneck}}
$$

The MobileNet V2 transformation is:

```mermaid
flowchart TD
    A[Input]
    --> B["1×1 Expansion"]
    --> C[BatchNorm]
    --> D[ReLU6]

    D --> E["3×3 Depthwise"]
    --> F[BatchNorm]
    --> G[ReLU6]

    G --> H["1×1 Projection"]
    --> I[BatchNorm]
    --> J["Linear Output<br/>No ReLU"]
```

The final projection does **not** use ReLU.

## 17. Why No ReLU After the Projection?

Suppose a narrow representation is:

$$
[-2,3,-1,4]
$$

Applying ReLU gives:

$$
[0,3,0,4]
$$

Two components are completely removed.

In a low-dimensional bottleneck space, removing dimensions can destroy significant information.

Therefore:

```mermaid
flowchart LR
    A["High-Dimensional Space"]
    --> B["Nonlinear Processing<br/>ReLU6 is useful"]

    C["Low-Dimensional Bottleneck"]
    --> D["Linear Projection<br/>Preserve Information"]
```

The idea is:

$$
\boxed{
\text{Use nonlinearities in expanded space}
}
$$

but:

$$
\boxed{
\text{Keep the narrow projection linear}
}
$$

## 18. Complete MobileNet V2 Block

A complete inverted residual block can be visualized as:

```mermaid
flowchart TD
    A[Input]

    A --> B["1×1 Expansion Conv"]
    B --> C[BatchNorm]
    C --> D[ReLU6]

    D --> E["3×3 Depthwise Conv"]
    E --> F[BatchNorm]
    F --> G[ReLU6]

    G --> H["1×1 Linear Projection"]
    H --> I[BatchNorm]

    I --> J((Add))
    A --> J

    J --> K[Output]
```

The core transformation is:

$$
\boxed{
\text{Expand}
\rightarrow
\text{Depthwise}
\rightarrow
\text{Linear Project}
}
$$

with residual addition when the shapes match.

## 19. MobileNet V1 vs MobileNet V2

| Feature                          | MobileNet V1                    | MobileNet V2                 |
| -------------------------------- | ------------------------------- | ---------------------------- |
| Main primitive                   | Depthwise separable convolution | Inverted residual bottleneck |
| Expansion layer                  | No                              | Yes                          |
| Depthwise convolution            | Yes                             | Yes                          |
| Pointwise convolution            | Yes                             | Yes                          |
| Residual connection              | No                              | Yes                          |
| Expanded internal representation | No                              | Yes                          |
| Linear projection                | No                              | Yes                          |
| Linear bottleneck                | No                              | Yes                          |
| Core flow                        | DW → PW                         | Expand → DW → Project        |

The evolution is:

```mermaid
flowchart LR
    A["MobileNet V1<br/>Depthwise → Pointwise"]
    --> B["MobileNet V2<br/>Expand → Depthwise → Linear Project<br/>+ Residual"]
```

## 20. MobileNet V2 Full-Network Idea

A complete MobileNet V2 begins with a regular convolution, followed by a sequence of inverted residual bottleneck blocks.

```mermaid
flowchart TD
    A[Input]
    --> B["Standard Conv"]
    --> C["Inverted Residual Blocks"]
    --> D["More Inverted Residual Blocks"]
    --> E["Final 1×1 Conv"]
    --> F["Global Average Pooling"]
    --> G[Classifier]
```

Within a stage:

```mermaid
flowchart TD
    A["First Block of Stage"]
    --> B["May use stride = 2"]
    --> C["May change H, W and channels"]

    C --> D["Remaining Blocks"]
    --> E["stride = 1"]
    --> F["Preserve spatial size"]
```

This is similar to the stage design used in many modern CNNs:

$$
\boxed{
\text{First block may change shape}
}
$$

while:

$$
\boxed{
\text{Later blocks usually preserve shape}
}
$$

## 21. MobileNet V3

MobileNet V3 further improves MobileNet V2 while keeping the inverted residual structure.

Its main additions include:

* **Squeeze-and-Excitation blocks**
* **Hard-Swish activation**
* hardware-aware architecture optimization
* specialized **Large** and **Small** variants

A simplified MobileNet V3 block can be visualized as:

```mermaid
flowchart TD
    A[Input]
    --> B["1×1 Expansion"]
    --> C["Activation"]
    --> D["Depthwise Conv"]
    --> E["Squeeze-and-Excitation"]
    --> F["Activation"]
    --> G["1×1 Linear Projection"]
    --> H((Add))
    --> I[Output]

    A --> H
```

Not every block necessarily uses all components, but this represents the main evolution from V2.

## 22. Overall Evolution of MobileNet

The MobileNet family can be summarized as an evolution of increasingly efficient building blocks.

### Standard Convolution

A standard convolution performs:

$$
\text{Spatial Processing}
+
\text{Channel Mixing}
$$

in one operation.

### MobileNet V1

MobileNet V1 separates these responsibilities:

```mermaid
flowchart LR
    A["3×3 Depthwise<br/>Spatial Processing"]
    --> B["1×1 Pointwise<br/>Channel Mixing"]
```

Core idea:

$$
\boxed{
\text{Depthwise Separable Convolution}
}
$$

### MobileNet V2

MobileNet V2 adds a high-dimensional internal representation and residual learning:

```mermaid
flowchart LR
    A["1×1 Expand"]
    --> B["3×3 Depthwise"]
    --> C["1×1 Linear Project"]
    --> D["Residual Add"]
```

Core ideas:

$$
\boxed{
\text{Inverted Residual}
+
\text{Linear Bottleneck}
}
$$

### MobileNet V3

MobileNet V3 keeps the V2 foundation and adds channel attention and more hardware-aware components:

```mermaid
flowchart LR
    A["Inverted Residual"]
    --> B["Squeeze-and-Excitation"]
    --> C["Hard-Swish / Efficient Activations"]
    --> D["Hardware-Aware Optimization"]
```

The overall architectural evolution is:

```mermaid
flowchart LR
    A["Standard CNN<br/>Dense Convolution"]
    --> B["MobileNet V1<br/>Depthwise Separable Conv"]
    --> C["MobileNet V2<br/>Inverted Residual<br/>+ Linear Bottleneck"]
    --> D["MobileNet V3<br/>SE + Hard-Swish<br/>+ Hardware-Aware Design"]
```

The main progression is therefore:

$$
\boxed{
\text{Standard Conv}
\rightarrow
\text{Depthwise Separable Conv}
\rightarrow
\text{Inverted Residual}
\rightarrow
\text{Hardware-Aware Mobile Blocks}
}
$$


# EfficientNet

## 1. Core Idea

EfficientNet starts from a simple but important question:

> When increasing the capacity of a CNN, should we increase **depth**, **width**, or **input resolution**?

Before EfficientNet, many architectures mainly scaled only one dimension:

* increase **depth** → add more layers;
* increase **width** → add more channels;
* increase **resolution** → use larger input images.

EfficientNet argues that these three dimensions are related and should be scaled **together in a balanced way**.

The core idea is:

$$
\boxed{\text{Compound Scaling}}
$$

## 2. Three Ways to Scale a CNN

### Depth Scaling

Increase the number of layers or blocks:

$$
d\uparrow
$$

```mermaid
flowchart LR
    A["Shallow Network<br/>Few Blocks"]
    --> B["Deeper Network<br/>More Blocks"]
```

A deeper model can learn more complex hierarchical features.

### Width Scaling

Increase the number of channels:

$$
w\uparrow
$$

```mermaid
flowchart LR
    A["Feature Maps<br/>64 Channels"]
    --> B["Wider Feature Maps<br/>128 Channels"]
```

A wider model can learn more feature patterns at each stage.

### Resolution Scaling

Increase the input resolution:

$$
r\uparrow
$$

```mermaid
flowchart LR
    A["Input<br/>224×224"]
    --> B["Higher Resolution<br/>300×300"]
```

Higher-resolution images provide more spatial details.

## 3. Limitations of Single-Dimension Scaling

Each scaling method is useful, but scaling only one dimension eventually becomes inefficient.

### Increasing Depth

Deeper networks can learn richer hierarchical representations:

```mermaid
flowchart LR
    A["Edges"]
    --> B["Textures"]
    --> C["Object Parts"]
    --> D["Objects"]
```

However, excessive depth:

* increases computation;
* makes optimization harder;
* eventually gives diminishing accuracy improvements.

### Increasing Width

More channels increase representational capacity:

$$
64\rightarrow128\rightarrow256
$$

However, very wide networks require more computation and memory.

### Increasing Resolution

Higher input resolution provides more spatial information.

However, if the model remains too shallow or too narrow, it may not have enough capacity to effectively process the additional information.

## 4. Compound Scaling

Instead of asking:

> Depth **or** width **or** resolution?

EfficientNet asks:

> How should all three dimensions be scaled together when more computational resources are available?

```mermaid
flowchart TD
    A["More Computational Budget"]

    A --> B["Increase Depth"]
    A --> C["Increase Width"]
    A --> D["Increase Resolution"]

    B --> E["Balanced Larger Network"]
    C --> E
    D --> E
```

Therefore:

$$
\boxed{
\text{EfficientNet scales Depth + Width + Resolution together}
}
$$

## 5. Compound Scaling Formula

EfficientNet introduces a compound coefficient:

$$
\phi
$$

and scales the three dimensions as:

$$
\boxed{
d=\alpha^\phi
}
$$

$$
\boxed{
w=\beta^\phi
}
$$

$$
\boxed{
r=\gamma^\phi
}
$$

where:

* $d$ = depth multiplier;
* $w$ = width multiplier;
* $r$ = resolution multiplier;
* $\alpha$ = depth scaling coefficient;
* $\beta$ = width scaling coefficient;
* $\gamma$ = resolution scaling coefficient;
* $\phi$ = compound scaling coefficient.

## 6. Why Depth, Width, and Resolution Have Different Costs

The three dimensions do not affect computation equally.

### Depth

If depth increases by a factor of $d$:

$$
\text{Compute}\propto d
$$

### Width

Convolution cost depends approximately on:

$$
C_{in}C_{out}
$$

If both input and output channels increase by a factor of $w$:

$$
\text{Compute}\propto w^2
$$

### Resolution

If both height and width increase by a factor of $r$:

$$
H,W\rightarrow rH,rW
$$

then:

$$
HW\rightarrow r^2HW
$$

Therefore:

$$
\text{Compute}\propto r^2
$$

Combining them:

$$
\boxed{
\text{Compute}\propto dw^2r^2
}
$$

Substituting the compound scaling equations:

$$
\text{Compute}
\propto
\alpha^\phi
\beta^{2\phi}
\gamma^{2\phi}
$$

or:

$$
\boxed{
\text{Compute}
\propto
(\alpha\beta^2\gamma^2)^\phi
}
$$

## 7. Compound Scaling Constraint

EfficientNet selects coefficients satisfying approximately:

$$
\boxed{
\alpha\beta^2\gamma^2\approx2
}
$$

with:

$$
\alpha\ge1,\qquad
\beta\ge1,\qquad
\gamma\ge1
$$

This means that increasing:

$$
\phi\rightarrow\phi+1
$$

roughly doubles the computational cost.

For example:

$$
\phi=0
$$

represents the baseline model.

Then approximately:

$$
\phi=1\Rightarrow2\times\text{ compute}
$$

$$
\phi=2\Rightarrow4\times\text{ compute}
$$

## 8. Intuition Behind Balanced Scaling

Suppose input resolution increases.

```mermaid
flowchart TD
    A["Higher Input Resolution"]
    --> B["More Spatial Information"]

    B --> C["Need More Depth<br/>to process richer hierarchy"]
    B --> D["Need More Width<br/>to represent more features"]

    C --> E["Balanced Model Capacity"]
    D --> E
```

Higher resolution provides more details, but those details are useful only if the network also has enough:

* depth to process more complex feature hierarchies;
* width to represent more feature patterns.

Therefore:

$$
\boxed{
\text{Resolution, Depth, and Width should grow together}
}
$$

## 9. EfficientNet-B0

EfficientNet first defines a baseline network:

$$
\boxed{\text{EfficientNet-B0}}
$$

B0 is designed using neural architecture search to obtain a good balance between accuracy and computational efficiency.

The larger EfficientNet models are then produced mainly by scaling B0:

```mermaid
flowchart LR
    A["EfficientNet-B0<br/>Baseline"]
    --> B["Compound Scaling"]
    --> C["B1"]
    --> D["B2"]
    --> E["B3"]
    --> F["..."]
    --> G["B7"]
```

Therefore:

> EfficientNet-B1 to B7 are primarily scaled versions of the B0 baseline.

## 10. EfficientNet Family

The family grows as:

$$
B0\rightarrow B1\rightarrow B2\rightarrow\cdots\rightarrow B7
$$

As the model becomes larger:

$$
Depth\uparrow
$$

$$
Width\uparrow
$$

$$
Resolution\uparrow
$$

at the same time.

Typical input resolutions are:

| Model           | Input Resolution |
| --------------- | ---------------: |
| EfficientNet-B0 |            $224$ |
| EfficientNet-B1 |            $240$ |
| EfficientNet-B2 |            $260$ |
| EfficientNet-B3 |            $300$ |
| EfficientNet-B4 |            $380$ |
| EfficientNet-B5 |            $456$ |
| EfficientNet-B6 |            $528$ |
| EfficientNet-B7 |            $600$ |

The main idea is:

$$
\boxed{
B0\rightarrow B7
\Rightarrow
d\uparrow,\ w\uparrow,\ r\uparrow
}
$$

## 11. MBConv: The Main Building Block

EfficientNet uses a block based on MobileNet V2:

$$
\boxed{\text{MBConv}}
$$

which stands for:

$$
\boxed{\text{Mobile Inverted Bottleneck Convolution}}
$$

The basic structure is:

```mermaid
flowchart LR
    A[Input]
    --> B["1×1 Expansion"]
    --> C["3×3 or 5×5 Depthwise Conv"]
    --> D["Squeeze-and-Excitation"]
    --> E["1×1 Projection"]
    --> F[Output]
```

The MobileNet V2 foundation is still visible:

$$
\text{Expand}
\rightarrow
\text{Depthwise}
\rightarrow
\text{Project}
$$

EfficientNet adds:

$$
\boxed{\text{Squeeze-and-Excitation}}
$$

inside the block.

## 12. MBConv Block

A typical MBConv block is:

```mermaid
flowchart TD
    A[Input]

    A --> B["1×1 Expansion Conv"]
    B --> C["BatchNorm + Swish"]

    C --> D["3×3 / 5×5 Depthwise Conv"]
    D --> E["BatchNorm + Swish"]

    E --> F["Squeeze-and-Excitation"]

    F --> G["1×1 Projection Conv"]
    G --> H[BatchNorm]

    H --> I((Add))
    A --> I

    I --> J[Output]
```

A residual connection is used when the input and output shapes are compatible.

The main mental model is:

$$
\boxed{
\text{Expand}
\rightarrow
\text{Depthwise}
\rightarrow
\text{SE}
\rightarrow
\text{Project}
}
$$

## 13. Squeeze-and-Excitation

The Squeeze-and-Excitation block allows the network to learn:

> Which channels are more important?

Suppose the input feature tensor is:

$$
H\times W\times C
$$

### Squeeze

Global Average Pooling reduces each channel to one value:

$$
H\times W\times C
\rightarrow
1\times1\times C
$$

```mermaid
flowchart LR
    A["Feature Maps<br/>H×W×C"]
    --> B["Global Average Pooling"]
    --> C["Channel Descriptor<br/>1×1×C"]
```

### Excitation

A small channel network generates an importance weight for each channel:

$$
s_1,s_2,\dots,s_C
$$

### Scale

Each feature channel is reweighted:

$$
\tilde{x}_c=s_cx_c
$$

The complete flow is:

```mermaid
flowchart LR
    A["Feature Maps"]
    --> B["Global Average Pool"]
    --> C["Channel Descriptor"]
    --> D["Small Channel Network"]
    --> E["Channel Weights"]
    --> F["Reweight Feature Maps"]
```

This allows the network to:

* emphasize useful channels;
* suppress less useful channels.

## 14. EfficientNet-B0 Architecture

At a high level:

```mermaid
flowchart TD
    A["Input<br/>224×224×3"]
    --> B["Stem<br/>3×3 Conv"]
    --> C["MBConv1"]
    --> D["MBConv6"]
    --> E["MBConv6"]
    --> F["MBConv6"]
    --> G["MBConv6"]
    --> H["MBConv6"]
    --> I["MBConv6"]
    --> J["1×1 Conv Head"]
    --> K["Global Average Pooling"]
    --> L[Dropout]
    --> M[Fully Connected]
    --> N[Classes]
```

Here:

* `MBConv1` uses an expansion ratio around $1$;
* `MBConv6` uses an expansion ratio around $6$.

For example:

$$
C
\rightarrow
6C
$$

before the depthwise convolution.

## 15. General EfficientNet Architecture Pattern

EfficientNet follows the common CNN pattern:

```mermaid
flowchart TD
    A["High Resolution<br/>Few Channels"]
    --> B["Intermediate Stages"]
    --> C["Lower Resolution<br/>More Channels"]
    --> D["Global Average Pooling"]
    --> E[Classifier]
```

As the network becomes deeper:

$$
H,W\downarrow
$$

while:

$$
C\uparrow
$$

Conceptually:

$$
224^2\times32
\rightarrow
112^2\times24
\rightarrow
56^2\times40
\rightarrow
28^2\times80
\rightarrow
14^2\times192
\rightarrow
7^2\times320
$$

The network gradually exchanges spatial resolution for richer semantic channel representations.

## 16. Swish Activation

EfficientNet uses the **Swish** activation function:

$$
\boxed{
\operatorname{Swish}(x)=x\sigma(x)
}
$$

where:

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

The operation can be visualized as:

```mermaid
flowchart LR
    A["Input x"]
    --> B["Sigmoid σ(x)"]
    B --> C["Multiply x × σ(x)"]
    C --> D["Swish(x)"]
```

Unlike ReLU:

$$
ReLU(x)=\max(0,x)
$$

Swish is smooth and allows some negative information to pass through.

In modern PyTorch implementations, Swish is equivalent to:

```python
nn.SiLU()
```

## 17. Why EfficientNet Is Efficient

EfficientNet combines several efficient design ideas:

```mermaid
flowchart TD
    A["EfficientNet"]

    A --> B["Depthwise Separable Convolution"]
    A --> C["MBConv / Inverted Bottleneck"]
    A --> D["Squeeze-and-Excitation"]
    A --> E["Swish Activation"]
    A --> F["Compound Scaling"]

    B --> G["High Accuracy / Compute Efficiency"]
    C --> G
    D --> G
    E --> G
    F --> G
```

However, its most distinctive contribution is:

$$
\boxed{\text{Compound Scaling}}
$$

Depthwise convolution and inverted bottlenecks already came from earlier efficient CNN designs.

EfficientNet focuses strongly on the question:

> **How should an efficient CNN be scaled?**

## 18. Single-Dimension Scaling vs Compound Scaling

### Depth-Only Scaling

```mermaid
flowchart LR
    A["Baseline"]
    --> B["More Layers"]
    --> C["Same Width"]
    --> D["Same Resolution"]
```

### Width-Only Scaling

```mermaid
flowchart LR
    A["Baseline"]
    --> B["More Channels"]
    --> C["Same Depth"]
    --> D["Same Resolution"]
```

### Resolution-Only Scaling

```mermaid
flowchart LR
    A["Baseline"]
    --> B["Larger Images"]
    --> C["Same Depth"]
    --> D["Same Width"]
```

### EfficientNet Compound Scaling

```mermaid
flowchart TD
    A["Increase Computational Budget"]

    A --> B["Depth ↑"]
    A --> C["Width ↑"]
    A --> D["Resolution ↑"]

    B --> E["Balanced Scaling"]
    C --> E
    D --> E

    E --> F["EfficientNet B1 → B7"]
```

This is the main distinction:

$$
\boxed{
\text{EfficientNet scales all three dimensions together}
}
$$

## 19. Compound Scaling Example

Suppose:

$$
\alpha=1.2
$$

$$
\beta=1.1
$$

$$
\gamma=1.15
$$

For:

$$
\phi=1
$$

we obtain:

$$
d=1.2
$$

$$
w=1.1
$$

$$
r=1.15
$$

For:

$$
\phi=2
$$

we obtain:

$$
d=1.2^2
$$

$$
w=1.1^2
$$

$$
r=1.15^2
$$

The model grows in all three dimensions:

```mermaid
flowchart LR
    A["φ = 0<br/>Baseline"]
    --> B["φ = 1<br/>Depth ↑ Width ↑ Resolution ↑"]
    --> C["φ = 2<br/>Depth ↑↑ Width ↑↑ Resolution ↑↑"]
```

No dimension is scaled independently.

## 20. Relationship with MobileNet

The evolution can be understood clearly.

### MobileNet V1

```mermaid
flowchart LR
    A["Standard Conv"]
    --> B["Depthwise + Pointwise"]
```

Focus:

$$
\text{Efficient Convolution}
$$

## 21. EfficientNetV2

EfficientNet was later improved into **EfficientNetV2**.

One practical limitation of EfficientNet is that depthwise convolution may have low theoretical FLOPs but may not always achieve proportionally low hardware latency.

EfficientNetV2 introduces **Fused-MBConv**.

Instead of:

```mermaid
flowchart LR
    A["1×1 Expansion"]
    --> B["3×3 Depthwise"]
    --> C["1×1 Projection"]
```

a Fused-MBConv can use:

```mermaid
flowchart LR
    A[Input]
    --> B["3×3 Regular Conv<br/>Expansion + Spatial Processing"]
    --> C["1×1 Projection"]
    --> D[Output]
```

This may require more theoretical FLOPs but can run faster on some hardware.

Therefore:

$$
\boxed{
\text{Fewer FLOPs}
\neq
\text{Always Lower Latency}
}
$$

EfficientNetV2 uses a combination of Fused-MBConv and MBConv blocks to improve training and hardware efficiency.

## 22. Final Mental Model

EfficientNet can be understood at two levels.

### Block Level

It uses an efficient MBConv block:

```mermaid
flowchart LR
    A[Input]
    --> B[Expand]
    --> C[Depthwise]
    --> D["Squeeze-and-Excitation"]
    --> E[Project]
    --> F["Residual if possible"]
    --> G[Output]
```

The block-level essence is:

$$
\boxed{
\text{MBConv}
=============

\text{Expand}
\rightarrow
\text{Depthwise}
\rightarrow
\text{SE}
\rightarrow
\text{Project}
}
$$

### Network Level

The entire network is scaled using depth, width, and resolution together:

```mermaid
flowchart TD
    A["EfficientNet-B0"]

    A --> B["Depth × α^φ"]
    A --> C["Width × β^φ"]
    A --> D["Resolution × γ^φ"]

    B --> E["Scaled EfficientNet"]
    C --> E
    D --> E

    E --> F["B1 → B2 → ... → B7"]
```

Therefore, the essence of EfficientNet is:

$$
\boxed{
\text{Efficient Block}
+
\text{Balanced Compound Scaling}
}
$$

with:

$$
\boxed{
d=\alpha^\phi,\qquad
w=\beta^\phi,\qquad
r=\gamma^\phi
}
$$

and:

$$
\boxed{
\alpha\beta^2\gamma^2\approx2
}
$$

The key takeaway is:

> **EfficientNet does not only ask how to make a model larger. It asks how depth, width, and resolution should grow together in a balanced way as the computational budget increases.**
